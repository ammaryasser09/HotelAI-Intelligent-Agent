````python
import os
from docx import Document
from pypdf import PdfReader
````

````python
def read_txt(file_path):
    with open(file_path, "r", encoding="utf-8") as file:
        return file.read()


def read_docx(file_path):
    document = Document(file_path)
    return "\n".join(
        paragraph.text
        for paragraph in document.paragraphs
        if paragraph.text.strip()
    )


def read_pdf(file_path):
    reader = PdfReader(file_path)
    
    text = []
    
    for page in reader.pages:
        page_text = page.extract_text()
        
        if page_text:
            text.append(page_text)
    
    return "\n".join(text)
````

````python
knowledge_path = "02_Hotel_Knowledge"

hotel_knowledge = []

for folder in ["TXT", "DOCX", "PDF"]:
    
    folder_path = os.path.join(knowledge_path, folder)
    
    for file_name in os.listdir(folder_path):
        
        file_path = os.path.join(folder_path, file_name)
        
        if folder == "TXT":
            content = read_txt(file_path)
            
        elif folder == "DOCX":
            content = read_docx(file_path)
            
        elif folder == "PDF":
            content = read_pdf(file_path)
        
        hotel_knowledge.append({
            "source": file_name,
            "content": content
        })

print(f"Loaded {len(hotel_knowledge)} documents")
````

````python
def read_docx(file_path):
    try:
        document = Document(file_path)
        
        return "\n".join(
            paragraph.text
            for paragraph in document.paragraphs
            if paragraph.text.strip()
        )
    
    except Exception:
        with open(file_path, "r", encoding="utf-8") as file:
            return file.read()
````

````python
def read_pdf(file_path):
    try:
        reader = PdfReader(file_path)

        text = []

        for page in reader.pages:
            page_text = page.extract_text()

            if page_text:
                text.append(page_text)

        return "\n".join(text)

    except Exception:
        with open(file_path, "r", encoding="utf-8") as file:
            return file.read()
        
````

````python
import os

ROOT = os.getcwd()

print("=" * 70)
print("HOTEL AI - PROJECT STRUCTURE")
print("=" * 70)

def print_tree(path, prefix=""):
    items = sorted(os.listdir(path))

    for i, item in enumerate(items):
        full_path = os.path.join(path, item)

        is_last = i == len(items) - 1
        connector = "└── " if is_last else "├── "

        print(prefix + connector + item)

        if os.path.isdir(full_path):
            extension = "    " if is_last else "│   "
            print_tree(full_path, prefix + extension)


print_tree(ROOT)
````

````python
import os

knowledge_path = "02_Hotel_Knowledge"

print("=" * 70)
print("HOTEL KNOWLEDGE BASE - STRUCTURE CHECK")
print("=" * 70)

expected_structure = {
    "Hotel_Information": [
        "hotel_information.txt",
        "contact_information.txt",
        "hotel_facilities.txt"
    ],
    
    "Rooms": [
        "room_types.txt",
        "room_amenities.txt",
        "room_policies.txt"
    ],
    
    "Booking": [
        "booking_policy.txt",
        "booking_process.txt",
        "booking_modification.txt",
        "booking_confirmation.txt"
    ],
    
    "Payment": [
        "payment_methods.txt",
        "payment_policy.txt",
        "refund_policy.txt"
    ],
    
    "Cancellation": [
        "cancellation_policy.txt",
        "cancellation_process.txt"
    ],
    
    "Services": [],
    "Policies": []
}


all_correct = True

for folder, files in expected_structure.items():

    folder_path = os.path.join(
        knowledge_path,
        folder
    )

    print(f"\n📁 {folder}")

    if not os.path.exists(folder_path):
        print("   ❌ Folder does not exist")
        all_correct = False
        continue

    print("   ✅ Folder exists")

    for file_name in files:

        file_path = os.path.join(
            folder_path,
            file_name
        )

        if os.path.exists(file_path):
            size = os.path.getsize(file_path)

            if size > 0:
                print(f"   ✅ {file_name} ({size} bytes)")
            else:
                print(f"   ⚠️ {file_name} is EMPTY")
                all_correct = False

        else:
            print(f"   ❌ Missing: {file_name}")
            all_correct = False


print("\n" + "=" * 70)

if all_correct:
    print("✅ STRUCTURE CHECK PASSED")
else:
    print("⚠️ STRUCTURE NEEDS ATTENTION")
````

````python
import os

knowledge_path = "02_Hotel_Knowledge"

expected_structure = {
    "Hotel_Information": [
        "hotel_information.txt",
        "contact_information.txt",
        "hotel_facilities.txt"
    ],
    "Rooms": [
        "room_types.txt",
        "room_amenities.txt",
        "room_policies.txt"
    ],
    "Booking": [
        "booking_policy.txt",
        "booking_process.txt",
        "booking_modification.txt",
        "booking_confirmation.txt"
    ],
    "Payment": [
        "payment_methods.txt",
        "payment_policy.txt",
        "refund_policy.txt"
    ],
    "Cancellation": [
        "cancellation_policy.txt",
        "cancellation_process.txt"
    ],
    "Services": [],
    "Policies": []
}

missing = []
empty = []

for folder, files in expected_structure.items():

    folder_path = os.path.join(knowledge_path, folder)

    if not os.path.isdir(folder_path):
        missing.append(f"Folder: {folder}")
        continue

    for file_name in files:

        file_path = os.path.join(folder_path, file_name)

        if not os.path.isfile(file_path):
            missing.append(f"{folder}/{file_name}")

        elif os.path.getsize(file_path) == 0:
            empty.append(f"{folder}/{file_name}")


print("=" * 70)
print("FINAL KNOWLEDGE BASE CHECK")
print("=" * 70)

if not missing and not empty:
    print("✅ ALL REQUIRED FILES EXIST")
    print("✅ ALL FILES CONTAIN DATA")
    print("✅ STRUCTURE PASSED")
else:

    if missing:
        print("\n❌ MISSING:")
        for item in missing:
            print("   -", item)

    if empty:
        print("\n⚠️ EMPTY:")
        for item in empty:
            print("   -", item)

print("=" * 70)
````

````python
import os

knowledge_path = "02_Hotel_Knowledge"

print("=" * 80)
print("HOTEL AI - FULL KNOWLEDGE BASE AUDIT")
print("=" * 80)

total_files = 0
total_characters = 0
total_words = 0

empty_files = []
small_files = []

for root, dirs, files in os.walk(knowledge_path):

    for file_name in sorted(files):

        file_path = os.path.join(root, file_name)

        # Only analyze text files
        if not file_name.lower().endswith(".txt"):
            continue

        total_files += 1

        with open(file_path, "r", encoding="utf-8") as file:
            content = file.read()

        characters = len(content)
        words = len(content.split())

        total_characters += characters
        total_words += words

        relative_path = os.path.relpath(
            file_path,
            knowledge_path
        )

        print(f"\n📄 {relative_path}")
        print(f"   Characters: {characters}")
        print(f"   Words:      {words}")

        if characters == 0:
            empty_files.append(relative_path)

        elif characters < 200:
            small_files.append(relative_path)


print("\n" + "=" * 80)
print("SUMMARY")
print("=" * 80)

print(f"Total TXT files:       {total_files}")
print(f"Total characters:      {total_characters}")
print(f"Total words:           {total_words}")

print("\nEmpty files:")

if empty_files:
    for file in empty_files:
        print(f"❌ {file}")
else:
    print("✅ None")

print("\nVery small files (<200 characters):")

if small_files:
    for file in small_files:
        print(f"⚠️ {file}")
else:
    print("✅ None")

print("\n" + "=" * 80)
````

````python
import os
import pandas as pd

knowledge_path = "02_Hotel_Knowledge"

documents = []

for root, dirs, files in os.walk(knowledge_path):

    for file_name in sorted(files):

        if not file_name.lower().endswith(".txt"):
            continue

        file_path = os.path.join(root, file_name)

        with open(file_path, "r", encoding="utf-8") as file:
            content = file.read().strip()

        relative_path = os.path.relpath(
            file_path,
            knowledge_path
        )

        parts = relative_path.split(os.sep)

        domain = parts[0]
        source = parts[-1]

        documents.append({
            "source": source,
            "domain": domain,
            "content": content,
            "language": "English"
        })

knowledge_df = pd.DataFrame(documents)

print("Documents loaded:", len(knowledge_df))

knowledge_df.head()
````

````python
# Add unique document IDs

knowledge_df.insert(
    0,
    "document_id",
    [f"DOC{i:03d}" for i in range(1, len(knowledge_df) + 1)]
)

# Map each source file to a specific category

category_map = {
    # Booking
    "booking_confirmation.txt": "Booking",
    "booking_modification.txt": "Booking",
    "booking_policy.txt": "Booking",
    "booking_process.txt": "Booking",

    # Cancellation
    "cancellation_policy.txt": "Cancellation",
    "cancellation_process.txt": "Cancellation",

    # Hotel Information
    "contact_information.txt": "Hotel Information",
    "hotel_facilities.txt": "Hotel Facilities",
    "hotel_information.txt": "Hotel Information",

    # Payment
    "payment_methods.txt": "Payment",
    "payment_policy.txt": "Payment",
    "refund_policy.txt": "Refund",

    # Policies
    "check_in_check_out.txt": "Check-in / Check-out",
    "children_policy.txt": "Children",
    "emergency_policy.txt": "Emergency",
    "guest_conduct.txt": "Guest Conduct",
    "identification_policy.txt": "Identification",
    "lost_and_found.txt": "Lost and Found",
    "noise_policy.txt": "Noise",
    "pet_policy.txt": "Pets",
    "privacy_policy.txt": "Privacy",
    "smoking_policy.txt": "Smoking",

    # Rooms
    "room_amenities.txt": "Room Amenities",
    "room_policies.txt": "Room Policies",
    "room_types.txt": "Room Types",

    # Services
    "airport_transfer.txt": "Airport Transfer",
    "breakfast.txt": "Breakfast",
    "gym.txt": "Gym",
    "laundry.txt": "Laundry",
    "restaurant.txt": "Restaurant",
    "room_service.txt": "Room Service",
    "spa.txt": "Spa"
}

knowledge_df["category"] = knowledge_df["source"].map(category_map)

print("Documents:", len(knowledge_df))
print("\nMissing categories:")

print(
    knowledge_df[
        knowledge_df["category"].isna()
    ][["source", "domain"]]
)

knowledge_df.head()
````

````python
question_matrix = {
    "Direct": 3,
    "Paraphrased": 3,
    "Scenario": 2
}

languages = [
    "English",
    "Arabic",
    "French",
    "Russian"
]

total_questions_per_intent = sum(question_matrix.values())

total_english_questions = (
    len(intent_df) * total_questions_per_intent
)

total_all_languages = (
    total_english_questions * len(languages)
)

print("=" * 70)
print("QUESTION GENERATION MATRIX")
print("=" * 70)

print("Question Types:")

for question_type, count in question_matrix.items():
    print(f"- {question_type}: {count}")

print("\nLanguages:")

for language in languages:
    print(f"- {language}")

print("\nTotal intents:", len(intent_df))
print("Questions per intent:", total_questions_per_intent)
print("English questions:", total_english_questions)
print("Total FAQ rows:", total_all_languages)
````

````python
generation_columns = [
    "faq_id",
    "intent_id",
    "intent_name",
    "category",
    "language",
    "question_type",
    "difficulty",
    "question",
    "answer",
    "source"
]

generation_df = pd.DataFrame(columns=generation_columns)

print("=" * 70)
print("FAQ GENERATION DATASET")
print("=" * 70)

print("Columns:", len(generation_df.columns))
print("Target rows:", total_all_languages)

print("\nSchema:")
for column in generation_df.columns:
    print(f"- {column}")
````

````python
generation_plan = []

faq_counter = 1

for _, intent in intent_df.iterrows():

    for language in languages:

        for question_type, count in question_matrix.items():

            for _ in range(count):

                generation_plan.append({
                    "faq_id": f"FAQ{faq_counter:04d}",
                    "intent_id": intent["intent_id"],
                    "intent_name": intent["intent_name"],
                    "category": intent["category"],
                    "language": language,
                    "question_type": question_type,
                    "difficulty": "Easy",
                    "question": "",
                    "answer": "",
                    "source": intent["source"]
                })

                faq_counter += 1

generation_plan_df = pd.DataFrame(generation_plan)

print("=" * 70)
print("FAQ GENERATION PLAN")
print("=" * 70)

print("Total planned rows:", len(generation_plan_df))
print("Expected rows:", total_all_languages)

print("\nLanguages:")
print(generation_plan_df["language"].value_counts())

print("\nQuestion Types:")
print(generation_plan_df["question_type"].value_counts())

print("\nFirst 10 rows:")
generation_plan_df.head(10)
````

````python
# ============================================================
# IMPROVED QUESTION GENERATION PROMPT
# ============================================================

def build_question_prompt(row, existing_questions=None):

    if existing_questions is None:
        existing_questions = []

    previous_questions = "\n".join(
        f"- {q}" for q in existing_questions
    )

    if not previous_questions:
        previous_questions = "None"

    return f"""
You are generating high-quality training data for a multilingual hotel AI assistant.

HOTEL:
Nile Pearl Hotel

INTENT:
{row['intent_name']}

CATEGORY:
{row['category']}

LANGUAGE:
{row['language']}

QUESTION TYPE:
{row['question_type']}

GROUND TRUTH:
{row['ground_truth']}

SOURCE:
{row['source']}

PREVIOUS QUESTIONS FOR THIS INTENT:
{previous_questions}

TASK:
Generate ONE natural guest question that matches the specified intent.

STRICT RULES:

1. The question MUST be written entirely in {row['language']}.
2. The question MUST match the intent exactly.
3. The question MUST be answerable using the provided ground truth.
4. Do NOT introduce information that is not supported by the ground truth.
5. Do NOT change prices, times, policies, room capacities, or conditions.
6. Do NOT ask about multiple intents in the same question.
7. Do NOT answer the question.
8. Do NOT mention the intent name.
9. Do NOT copy any previous question.
10. The new question must be meaningfully different from previous questions.

QUESTION TYPE RULES:

Direct:
Ask the intent clearly and directly.

Paraphrased:
Ask the same intent using substantially different wording while preserving the exact meaning.

Scenario:
Create a realistic hotel guest situation that naturally leads to the same intent.

QUALITY REQUIREMENTS:

- Natural human wording.
- Concise.
- Relevant to a hotel guest.
- No unnecessary details.
- No invented facts.
- No repetition.

Return ONLY ONE question.
Do not include quotation marks.
Do not include explanations.

Generate the question now.
"""
````

````python
import os

# ============================================================
# HOTEL AI - FULL PROJECT STRUCTURE
# ============================================================

PROJECT_ROOT = "."

print("=" * 80)
print("HOTEL AI - CURRENT PROJECT STRUCTURE")
print("=" * 80)

def print_tree(path, prefix=""):
    
    try:
        items = sorted(
            os.listdir(path),
            key=lambda x: (not os.path.isdir(os.path.join(path, x)), x.lower())
        )
    except PermissionError:
        print(prefix + "⚠️ Permission denied")
        return

    for index, item in enumerate(items):

        full_path = os.path.join(path, item)
        is_last = index == len(items) - 1

        connector = "└── " if is_last else "├── "
        print(prefix + connector + item)

        if os.path.isdir(full_path):
            extension = "    " if is_last else "│   "
            print_tree(full_path, prefix + extension)


print_tree(PROJECT_ROOT)

print("\n" + "=" * 80)
print("IMPORTANT FILE CHECK")
print("=" * 80)

important_paths = [
    "02_Hotel_Knowledge",
    "03_Structured_Data",
    "data",
    "database",
    "scripts",
    "README.md",
    "generate_hotel_faq.ipynb"
]

for path in important_paths:

    if os.path.exists(path):
        if os.path.isdir(path):
            print(f"✅ {path}/")
        else:
            size = os.path.getsize(path)
            print(f"✅ {path} ({size:,} bytes)")
    else:
        print(f"❌ {path} - MISSING")
````

````python
# ============================================================
# INT005 - SIMILARITY FUNCTION
# ============================================================

from difflib import SequenceMatcher
import re

def normalize_text(text):
    """
    Normalize text before similarity comparison.
    """

    text = str(text).lower().strip()

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text)

    # Remove punctuation
    text = re.sub(r"[^\w\s\u0600-\u06FF\u0400-\u04FF]", "", text)

    return text


def similarity_score(text1, text2):
    """
    Calculate similarity between two texts.
    Returns value between 0 and 1.
    """

    text1 = normalize_text(text1)
    text2 = normalize_text(text2)

    if not text1 or not text2:
        return 0.0

    return SequenceMatcher(
        None,
        text1,
        text2
    ).ratio()


print("=" * 80)
print("INT005 SIMILARITY FUNCTION")
print("=" * 80)

# ------------------------------------------------------------
# Test
# ------------------------------------------------------------

test_1 = "What time is check-out?"
test_2 = "What is the standard check-out time?"

score = similarity_score(test_1, test_2)

print("Test similarity:", round(score, 3))

print("\n" + "=" * 80)
print("Status: READY")
print("=" * 80)
````

````python
# ============================================================
# INT005 - PROMPT BUILDER
# ============================================================

def build_int005_prompt(row):

    language = str(row["language"]).strip()
    question_type = str(row["question_type"]).strip()
    faq_id = str(row["faq_id"]).strip()

    ground_truth = str(row["ground_truth"]).strip()

    language_rules = {
        "English": "Write the question entirely in English.",
        "Arabic": "Write the question entirely in Arabic.",
        "French": "Write the question entirely in French.",
        "Russian": "Write the question entirely in Russian."
    }

    type_rules = {
        "Direct": (
            "Ask the information directly and clearly."
        ),
        "Paraphrased": (
            "Express the same information using different "
            "natural wording. Do not copy a direct question."
        ),
        "Scenario": (
            "Create a realistic hotel guest situation that "
            "naturally leads to asking about the same information."
        )
    }

    language_rule = language_rules.get(
        language,
        f"Write the question in {language}."
    )

    type_rule = type_rules.get(
        question_type,
        "Create a natural guest question."
    )

    prompt = f"""
You are generating a high-quality FAQ question for a hotel AI dataset.

Hotel name:
Nile Pearl Hotel

FAQ ID:
{faq_id}

Language:
{language}

Question type:
{question_type}

Ground truth:
{ground_truth}

Instructions:

1. {language_rule}

2. {type_rule}

3. The question must be about the STANDARD CHECK-OUT TIME.

4. The correct information is:
   Standard check-out time is 12:00 PM.

5. Keep the meaning consistent with the ground truth.

6. Mention "Nile Pearl Hotel" naturally when appropriate.

7. Do not invent any hotel policy or time.

8. Do not answer the question.

9. Return ONLY the question.

10. Do not add:
   - explanations
   - labels
   - quotation marks
   - bullet points
   - prefixes such as "Question:"
   - translations into another language

11. For Russian, use Cyrillic only for the Russian language.
12. For Arabic, use Arabic only for the Arabic language.
13. Do not mix Arabic, Russian, French, or English sentences.

Return exactly ONE natural guest question.

Generate now.
"""

    return prompt.strip()


print("=" * 80)
print("INT005 PROMPT BUILDER")
print("=" * 80)

# ------------------------------------------------------------
# Test with first INT005 row
# ------------------------------------------------------------

test_prompt = build_int005_prompt(
    int005_rows.iloc[0]
)

print("FAQ ID:", int005_rows.iloc[0]["faq_id"])
print("Language:", int005_rows.iloc[0]["language"])
print("Type:", int005_rows.iloc[0]["question_type"])

print("\n" + "=" * 80)
print("PROMPT CREATED SUCCESSFULLY")
print("=" * 80)

print(test_prompt)
````

````python
# ============================================================
# INT005 - VALIDATOR
# ============================================================

# ------------------------------------------------------------
# Language detection helpers
# ------------------------------------------------------------

def has_arabic(text):
    return any(
        '\u0600' <= char <= '\u06FF'
        for char in str(text)
    )


def has_cyrillic(text):
    return any(
        '\u0400' <= char <= '\u04FF'
        for char in str(text)
    )


# ------------------------------------------------------------
# Validator
# ------------------------------------------------------------

def validate_int005_question(
    question,
    language,
    question_type,
    ground_truth
):

    if question is None:
        return False, "Empty question"

    question = str(question).strip()

    # --------------------------------------------------------
    # Empty check
    # --------------------------------------------------------

    if not question:
        return False, "Empty question"

    # --------------------------------------------------------
    # Length check
    # --------------------------------------------------------

    if len(question) < 10:
        return False, "Question too short"

    if len(question) > 300:
        return False, "Question too long"

    # --------------------------------------------------------
    # Question mark check
    # --------------------------------------------------------

    if "?" not in question and "؟" not in question:
        return False, "Not formatted as a question"

    # --------------------------------------------------------
    # Check-out terminology
    # --------------------------------------------------------

    check_out_terms = {

        "English": [
            "check-out",
            "checkout",
            "check out",
            "departure"
        ],

        "Arabic": [
            "تسجيل المغادرة",
            "المغادرة",
            "الخروج"
        ],

        "French": [
            "départ",
            "check-out",
            "checkout",
            "heure de départ"
        ],

        "Russian": [
            "высел",
            "выезд",
            "врем",
            "checkout",
            "check-out"
        ]
    }

    terms = check_out_terms.get(
        language,
        []
    )

    question_lower = question.lower()

    if not any(
        term.lower() in question_lower
        for term in terms
    ):
        return False, "Not related to check-out"

    # --------------------------------------------------------
    # Language mixing
    # --------------------------------------------------------

    if language == "Russian":

        if has_arabic(question):
            return False, (
                "Arabic characters found in Russian question"
            )

    elif language == "Arabic":

        if has_cyrillic(question):
            return False, (
                "Cyrillic characters found in Arabic question"
            )

    elif language == "English":

        if has_arabic(question):
            return False, (
                "Arabic characters found in English question"
            )

        if has_cyrillic(question):
            return False, (
                "Cyrillic characters found in English question"
            )

    elif language == "French":

        if has_arabic(question):
            return False, (
                "Arabic characters found in French question"
            )

        if has_cyrillic(question):
            return False, (
                "Cyrillic characters found in French question"
            )

    # --------------------------------------------------------
    # Final validation
    # --------------------------------------------------------

    return True, "VALID"


# ============================================================
# TEST
# ============================================================

print("=" * 80)
print("INT005 VALIDATOR")
print("=" * 80)

test_question = (
    "What is the standard check-out time at Nile Pearl Hotel?"
)

valid, reason = validate_int005_question(
    question=test_question,
    language="English",
    question_type="Direct",
    ground_truth="Standard check-out time is 12:00 PM."
)

print("Test question:")
print(test_question)

print("\nValidation result:")
print("Valid:", valid)
print("Status:", reason)

print("\n" + "=" * 80)

if valid:
    print("Status: READY")
else:
    print("Status: CHECK VALIDATOR")

print("=" * 80)
````

````python
# ============================================================
# INT005 - REGENERATION FUNCTION
# ============================================================

def regenerate_int005_question(
    row,
    validation_error="",
    previous_question=""
):
    """
    Regenerate one INT005 question using Ollama
    after a validation failure.
    """

    language = str(row["language"]).strip()
    question_type = str(row["question_type"]).strip()
    ground_truth = str(row["ground_truth"]).strip()
    faq_id = str(row["faq_id"]).strip()

    prompt = f"""
You are fixing a hotel FAQ question generation failure.

Hotel:
Nile Pearl Hotel

FAQ ID:
{faq_id}

Language:
{language}

Question type:
{question_type}

Ground truth:
{ground_truth}

Previous generated question:
{previous_question}

Validation error:
{validation_error}

Generate ONE completely new question.

STRICT RULES:

- Write ONLY in {language}.
- The question must be about the standard check-out time.
- The correct check-out time is 12:00 PM.
- Do not invent another time.
- Do not answer the question.
- Do not include explanations.
- Do not include labels.
- Do not use quotation marks.
- Do not copy the previous question.
- Keep the question natural for a hotel guest.
- Return exactly ONE question.
"""

    try:

        response = requests.post(
            OLLAMA_URL,
            json={
                "model": OLLAMA_MODEL,
                "prompt": prompt.strip(),
                "stream": False,
                "options": {
                    "temperature": 0.7
                }
            },
            timeout=120
        )

        response.raise_for_status()

        result = response.json()

        generated = result.get(
            "response",
            ""
        ).strip()

        return generated

    except Exception as e:

        print(
            f"❌ Regeneration error for {faq_id}: {e}"
        )

        return ""


print("=" * 80)
print("INT005 REGENERATION FUNCTION")
print("=" * 80)

print("Function:", regenerate_int005_question.__name__)
print("Status: READY")

print("=" * 80)
````

````python
# ============================================================
# INT005 - FULL QUESTION GENERATION
# ============================================================

import time

int005_generated = []

print("=" * 80)
print("INT005 FULL QUESTION GENERATION")
print("=" * 80)

# ------------------------------------------------------------
# Generation loop
# ------------------------------------------------------------

for index, (_, row) in enumerate(
    int005_rows.iterrows(),
    start=1
):

    faq_id = str(row["faq_id"])
    language = str(row["language"])
    question_type = str(row["question_type"])
    ground_truth = str(row["ground_truth"])

    print("\n" + "=" * 80)
    print(
        f"{index}/32 | "
        f"{faq_id} | "
        f"{language} | "
        f"{question_type}"
    )
    print("=" * 80)

    generated_question = ""
    validation_status = "FAILED"
    validation_reason = ""
    attempts = 0

    # --------------------------------------------------------
    # Maximum attempts
    # --------------------------------------------------------

    max_attempts = 4

    for attempt in range(1, max_attempts + 1):

        attempts = attempt

        # ----------------------------------------------------
        # Build prompt
        # ----------------------------------------------------

        prompt = build_int005_prompt(row)

        # If previous attempt failed, add regeneration info
        if generated_question:

            prompt += f"""

Previous generated question:
{generated_question}

Previous validation error:
{validation_reason}

Generate a completely new question.
"""

        # ----------------------------------------------------
        # Call Ollama
        # ----------------------------------------------------

        try:

            response = requests.post(
                OLLAMA_URL,
                json={
                    "model": OLLAMA_MODEL,
                    "prompt": prompt,
                    "stream": False,
                    "options": {
                        "temperature": 0.7
                    }
                },
                timeout=120
            )

            response.raise_for_status()

            result = response.json()

            generated_question = (
                result.get("response", "")
                .strip()
            )

        except Exception as e:

            validation_reason = (
                f"Ollama error: {str(e)}"
            )

            generated_question = ""

            continue

        # ----------------------------------------------------
        # Clean response
        # ----------------------------------------------------

        generated_question = (
            generated_question
            .replace('"', "")
            .replace("“", "")
            .replace("”", "")
            .strip()
        )

        # ----------------------------------------------------
        # Validate
        # ----------------------------------------------------

        is_valid, reason = validate_int005_question(
            question=generated_question,
            language=language,
            question_type=question_type,
            ground_truth=ground_truth
        )

        if is_valid:

            validation_status = "VALID"
            validation_reason = "VALID"

            break

        else:

            validation_status = "FAILED"
            validation_reason = reason

    # --------------------------------------------------------
    # Store result
    # --------------------------------------------------------

    int005_generated.append({

        "faq_id": faq_id,

        "intent_id": "INT005",

        "language": language,

        "question_type": question_type,

        "question": generated_question,

        "ground_truth": ground_truth,

        "status": validation_status,

        "validation_reason": validation_reason,

        "attempts": attempts
    })

    # --------------------------------------------------------
    # Print result
    # --------------------------------------------------------

    print(
        "GENERATED:",
        generated_question
    )

    print(
        "Status:",
        validation_status,
        "| Attempts:",
        attempts
    )

    if validation_status != "VALID":

        print(
            "Reason:",
            validation_reason
        )

    time.sleep(0.2)


# ============================================================
# SUMMARY
# ============================================================

int005_generated_df = pd.DataFrame(
    int005_generated
)

valid_count = (
    int005_generated_df["status"]
    .eq("VALID")
    .sum()
)

failed_count = (
    int005_generated_df["status"]
    .ne("VALID")
    .sum()
)

non_empty_count = (
    int005_generated_df["question"]
    .fillna("")
    .astype(str)
    .str.strip()
    .ne("")
    .sum()
)

print("\n" + "=" * 80)
print("INT005 GENERATION SUMMARY")
print("=" * 80)

print("Total:", len(int005_generated_df))
print("Valid:", valid_count)
print("Failed:", failed_count)
print("Non-empty:", non_empty_count)

print("=" * 80)
````

````python
# ============================================================
# INT005 - FAILED QUESTION REGENERATION
# ============================================================

failed_rows = int005_generated_df[
    int005_generated_df["status"] != "VALID"
].copy()

print("=" * 80)
print("INT005 FAILED QUESTION REGENERATION")
print("=" * 80)

print("Failed questions:", len(failed_rows))
print("FAQ IDs:", failed_rows["faq_id"].tolist())

# ------------------------------------------------------------
# Regenerate failed questions
# ------------------------------------------------------------

for idx, row in failed_rows.iterrows():

    faq_id = row["faq_id"]
    language = row["language"]
    question_type = row["question_type"]

    print("\n" + "=" * 80)
    print("FAQ:", faq_id)
    print("Language:", language)
    print("Type:", question_type)
    print("Old question:", row["question"])

    regenerated_question = ""

    max_attempts = 5

    for attempt in range(1, max_attempts + 1):

        regenerated_question = regenerate_int005_question(
            row=row,
            validation_error=row["validation_reason"],
            previous_question=regenerated_question
        )

        regenerated_question = (
            regenerated_question
            .replace('"', "")
            .replace("“", "")
            .replace("”", "")
            .strip()
        )

        is_valid, reason = validate_int005_question(
            question=regenerated_question,
            language=language,
            question_type=question_type,
            ground_truth=row["ground_truth"]
        )

        if is_valid:

            print("FIXED:", regenerated_question)
            print("Status: VALID")
            print("Attempts:", attempt)

            int005_generated_df.loc[
                idx,
                "question"
            ] = regenerated_question

            int005_generated_df.loc[
                idx,
                "status"
            ] = "VALID"

            int005_generated_df.loc[
                idx,
                "validation_reason"
            ] = "VALID"

            int005_generated_df.loc[
                idx,
                "attempts"
            ] = row["attempts"] + attempt

            break

        else:

            print(
                f"Attempt {attempt} FAILED:",
                reason
            )

            if attempt == max_attempts:

                print(
                    "❌ FAILED TO REGENERATE"
                )

                int005_generated_df.loc[
                    idx,
                    "question"
                ] = regenerated_question

                int005_generated_df.loc[
                    idx,
                    "validation_reason"
                ] = reason

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

remaining_failed = int005_generated_df[
    int005_generated_df["status"] != "VALID"
]

print("\n" + "=" * 80)
print("INT005 REGENERATION SUMMARY")
print("=" * 80)

print(
    "Total:",
    len(int005_generated_df)
)

print(
    "Valid:",
    int005_generated_df["status"].eq("VALID").sum()
)

print(
    "Failed:",
    len(remaining_failed)
)

if len(remaining_failed) > 0:

    print(
        "\nRemaining failed FAQ IDs:",
        remaining_failed["faq_id"].tolist()
    )

else:

    print("\n✅ ALL INT005 QUESTIONS ARE VALID")

print("=" * 80)
````

````python
# ============================================================
# INT005 - FINAL LANGUAGE QUALITY CHECK
# ============================================================

language_issues = []

for _, row in int005_generated_df.iterrows():

    question = str(row["question"])
    language = row["language"]

    # --------------------------------------------------------
    # Arabic
    # --------------------------------------------------------

    if language == "Arabic":

        if has_cyrillic(question):

            language_issues.append({
                "faq_id": row["faq_id"],
                "language": language,
                "issue": "Cyrillic characters found in Arabic question",
                "question": question
            })

    # --------------------------------------------------------
    # Russian
    # --------------------------------------------------------

    elif language == "Russian":

        if has_arabic(question):

            language_issues.append({
                "faq_id": row["faq_id"],
                "language": language,
                "issue": "Arabic characters found in Russian question",
                "question": question
            })

    # --------------------------------------------------------
    # English
    # --------------------------------------------------------

    elif language == "English":

        if has_arabic(question) or has_cyrillic(question):

            language_issues.append({
                "faq_id": row["faq_id"],
                "language": language,
                "issue": "Non-English characters found in English question",
                "question": question
            })

    # --------------------------------------------------------
    # French
    # --------------------------------------------------------

    elif language == "French":

        if has_arabic(question) or has_cyrillic(question):

            language_issues.append({
                "faq_id": row["faq_id"],
                "language": language,
                "issue": "Non-French characters found in French question",
                "question": question
            })


# ============================================================
# RESULTS
# ============================================================

print("=" * 80)
print("INT005 FINAL LANGUAGE QUALITY CHECK")
print("=" * 80)

print(
    "Total questions:",
    len(int005_generated_df)
)

print(
    "Language issues:",
    len(language_issues)
)

if language_issues:

    print("\n⚠️ LANGUAGE ISSUES FOUND")

    display(
        pd.DataFrame(language_issues)
    )

else:

    print("✅ NO LANGUAGE MIXING FOUND")

print("=" * 80)
````

````python
# ============================================================
# INT005 - FINAL DUPLICATE CHECK
# ============================================================

from difflib import SequenceMatcher

def similarity(text1, text2):
    """
    Calculate similarity between two questions.
    Returns a value between 0 and 1.
    """
    return SequenceMatcher(
        None,
        str(text1).lower().strip(),
        str(text2).lower().strip()
    ).ratio()


duplicate_pairs = []

questions = (
    int005_generated_df["question"]
    .fillna("")
    .astype(str)
    .tolist()
)

faq_ids = (
    int005_generated_df["faq_id"]
    .tolist()
)

languages = (
    int005_generated_df["language"]
    .tolist()
)

question_types = (
    int005_generated_df["question_type"]
    .tolist()
)


# ------------------------------------------------------------
# Compare every pair
# ------------------------------------------------------------

for i in range(len(questions)):

    for j in range(i + 1, len(questions)):

        q1 = questions[i]
        q2 = questions[j]

        similarity_score = similarity(
            q1,
            q2
        )

        if similarity_score >= 0.85:

            duplicate_pairs.append({

                "faq_1": faq_ids[i],

                "faq_2": faq_ids[j],

                "language": languages[i],

                "question_type_1": question_types[i],

                "question_type_2": question_types[j],

                "similarity": round(
                    similarity_score,
                    3
                ),

                "question_1": q1,

                "question_2": q2
            })


# ============================================================
# RESULTS
# ============================================================

print("=" * 80)
print("INT005 FINAL DUPLICATE CHECK")
print("=" * 80)

print(
    "Total questions:",
    len(int005_generated_df)
)

print(
    "Potential duplicates:",
    len(duplicate_pairs)
)

if duplicate_pairs:

    print("\n⚠️ POTENTIAL DUPLICATES FOUND")

    duplicate_df = pd.DataFrame(
        duplicate_pairs
    )

    display(duplicate_df)

else:

    print(
        "✅ NO POTENTIAL DUPLICATES FOUND"
    )

print("=" * 80)
````

````python
# ============================================================
# INT006 - SIMILARITY FUNCTION
# ============================================================

from difflib import SequenceMatcher


def similarity(a, b):
    """
    Calculate similarity between two strings.
    Returns a value between 0 and 1.
    """
    a = str(a).lower().strip()
    b = str(b).lower().strip()

    return SequenceMatcher(None, a, b).ratio()


# ============================================================
# TEST
# ============================================================

test_a = "Can I check in early?"
test_b = "Is early check-in available?"

score = similarity(test_a, test_b)

print("=" * 80)
print("INT006 SIMILARITY FUNCTION")
print("=" * 80)

print("Test similarity:", round(score, 3))

print("\n================================================================================")
print("Status: READY")
print("================================================================================")
````

````python
# ============================================================
# INT006 - PROMPT BUILDER
# ============================================================

def build_int006_prompt(
    faq_id,
    language,
    question_type,
    ground_truth
):
    
    language_rules = {
        "English": """
Write the question entirely in English.
Do not use Arabic, Russian, French, or other non-English sentences.
""",

        "Arabic": """
Write the question entirely in Arabic.
Use Arabic script for the question.
Do not write the question in English, French, or Russian.
The hotel name "Nile Pearl Hotel" may remain in English if needed.
""",

        "French": """
Write the question entirely in French.
Do not use Arabic, Russian, or English sentences.
The hotel name "Nile Pearl Hotel" may remain in English if needed.
""",

        "Russian": """
Write the question entirely in Russian.
Use Cyrillic script for the Russian sentence.
Do not use Arabic, French, or English sentences.
The hotel name "Nile Pearl Hotel" may remain in English if needed.
"""
    }

    question_type_rules = {
        "Direct": """
Ask directly and clearly about EARLY CHECK-IN.
The question should be short and factual.
""",

        "Paraphrased": """
Ask about EARLY CHECK-IN using different natural wording.
Do not simply copy a standard direct question.
Keep exactly the same meaning.
""",

        "Scenario": """
Create a realistic hotel guest situation involving EARLY CHECK-IN,
then ask a question about whether early check-in is possible.
Do not introduce any new hotel policy.
"""
    }

    prompt = f"""
You are generating a high-quality FAQ question for a multilingual hotel AI dataset.

Hotel name:
Nile Pearl Hotel

FAQ ID:
{faq_id}

Language:
{language}

Question type:
{question_type}

Ground truth:
{ground_truth}

CORE INTENT:
early_check_in

The question MUST be about EARLY CHECK-IN.

The hotel policy states:
Early check-in is subject to room availability and hotel approval.

IMPORTANT:
- Ask about early check-in only.
- Do not ask about standard check-in time.
- Do not ask about late check-out.
- Do not ask about room availability in general.
- Do not ask about hotel approval in general.
- Do not invent a specific early check-in time.
- Do not invent additional fees.
- Do not invent guaranteed availability.
- Do not answer the question.
- Return exactly ONE question.
- Do not include explanations.
- Do not include labels.
- Do not include quotation marks.
- Do not include bullet points.
- Do not include "Question:".
- Do not mix languages.

LANGUAGE RULE:
{language_rules[language]}

QUESTION TYPE RULE:
{question_type_rules[question_type]}

Return ONLY the question.

Generate now.
"""

    return prompt.strip()


# ============================================================
# TEST PROMPT
# ============================================================

test_row = int006_df.iloc[0]

test_prompt = build_int006_prompt(
    faq_id=test_row["faq_id"],
    language=test_row["language"],
    question_type=test_row["question_type"],
    ground_truth=test_row["ground_truth"]
)

print("=" * 80)
print("INT006 PROMPT BUILDER")
print("=" * 80)

print("FAQ ID:", test_row["faq_id"])
print("Language:", test_row["language"])
print("Type:", test_row["question_type"])

print("\n" + "=" * 80)
print("PROMPT CREATED SUCCESSFULLY")
print("=" * 80)

print(test_prompt)
````

````python
# ============================================================
# INT006 - VALIDATOR
# ============================================================

import re


# ============================================================
# LANGUAGE DETECTION HELPERS
# ============================================================

def has_arabic(text):
    return bool(re.search(r"[\u0600-\u06FF]", str(text)))


def has_cyrillic(text):
    return bool(re.search(r"[\u0400-\u04FF]", str(text)))


def has_latin(text):
    return bool(re.search(r"[A-Za-z]", str(text)))


# ============================================================
# INTENT KEYWORDS
# ============================================================

EARLY_CHECKIN_KEYWORDS = {
    "English": [
        "early check-in",
        "early check in",
        "check in early",
        "arrive early",
        "before check-in",
        "before check in"
    ],

    "Arabic": [
        "تسجيل الوصول المبكر",
        "الوصول المبكر",
        "تسجيل الدخول المبكر",
        "الوصول قبل موعد",
        "تسجيل الوصول قبل"
    ],

    "French": [
        "enregistrement anticipé",
        "arrivée anticipée",
        "arriver tôt",
        "enregistrement plus tôt",
        "avant l'heure d'arrivée"
    ],

    "Russian": [
        "ранний заезд",
        "раннего заезда",
        "заселиться раньше",
        "раннее заселение",
        "заехать раньше"
    ]
}


# ============================================================
# BASIC QUESTION CHECK
# ============================================================

def is_question(text):
    text = str(text).strip()

    if not text:
        return False

    return "?" in text


# ============================================================
# INT006 VALIDATOR
# ============================================================

def validate_int006_question(
    question,
    language,
    question_type,
    ground_truth
):

    question = str(question).strip()

    # --------------------------------------------------------
    # Empty check
    # --------------------------------------------------------

    if not question:
        return False, "Empty question"

    # --------------------------------------------------------
    # Must be a question
    # --------------------------------------------------------

    if not is_question(question):
        return False, "Not formatted as a question"

    # --------------------------------------------------------
    # No answer / explanation
    # --------------------------------------------------------

    if len(question.split()) > 35:
        return False, "Question is too long"

    # --------------------------------------------------------
    # Language validation
    # --------------------------------------------------------

    if language == "English":

        if has_arabic(question):
            return False, "Arabic characters found in English question"

        if has_cyrillic(question):
            return False, "Cyrillic characters found in English question"

    elif language == "Arabic":

        if has_cyrillic(question):
            return False, "Cyrillic characters found in Arabic question"

    elif language == "French":

        if has_arabic(question):
            return False, "Arabic characters found in French question"

        if has_cyrillic(question):
            return False, "Cyrillic characters found in French question"

    elif language == "Russian":

        if has_arabic(question):
            return False, "Arabic characters found in Russian question"

    else:
        return False, f"Unsupported language: {language}"

    # --------------------------------------------------------
    # Intent validation
    # --------------------------------------------------------

    question_lower = question.lower()

    keywords = EARLY_CHECKIN_KEYWORDS[language]

    intent_match = any(
        keyword.lower() in question_lower
        for keyword in keywords
    )

    if not intent_match:
        return False, "Not related to early check-in"

    # --------------------------------------------------------
    # Wrong intent protection
    # --------------------------------------------------------

    wrong_intents = {
        "English": [
            "late check-out",
            "late checkout",
            "standard check-in time",
            "check-out time",
            "checkout time"
        ],

        "Arabic": [
            "الخروج المتأخر",
            "وقت الخروج",
            "تسجيل الوصول العادي"
        ],

        "French": [
            "départ tardif",
            "heure de départ",
            "enregistrement standard"
        ],

        "Russian": [
            "поздний выезд",
            "время выезда",
            "стандартное время заезда"
        ]
    }

    for wrong in wrong_intents[language]:

        if wrong.lower() in question_lower:
            return False, f"Wrong intent detected: {wrong}"

    # --------------------------------------------------------
    # Direct / paraphrased / scenario basic checks
    # --------------------------------------------------------

    if question_type in ["Direct", "Paraphrased", "Scenario"]:

        if not question.endswith("?"):
            return False, "Question must end with '?'"

    # --------------------------------------------------------
    # Valid
    # --------------------------------------------------------

    return True, "Valid"


# ============================================================
# TEST VALIDATOR
# ============================================================

test_question = (
    "Can I check in early at Nile Pearl Hotel?"
)

valid, reason = validate_int006_question(
    question=test_question,
    language="English",
    question_type="Direct",
    ground_truth="Early check-in is subject to room availability and hotel approval."
)

print("=" * 80)
print("INT006 VALIDATOR")
print("=" * 80)

print("Test question:")
print(test_question)

print("\nValidation result:")
print("Valid:", valid)
print("Reason:", reason)

print("=" * 80)

if valid:
    print("Status: VALID")
else:
    print("Status: INVALID")

print("=" * 80)
````

````python
# ============================================================
# INT006 - REGENERATION FUNCTION
# ============================================================

def regenerate_int006_question(
    row,
    max_attempts=5
):
    
    faq_id = row["faq_id"]
    language = row["language"]
    question_type = row["question_type"]
    ground_truth = row["ground_truth"]

    last_question = ""
    last_reason = ""

    for attempt in range(1, max_attempts + 1):

        prompt = build_int006_prompt(
            faq_id=faq_id,
            language=language,
            question_type=question_type,
            ground_truth=ground_truth
        )

        try:

            payload = {
                "model": OLLAMA_MODEL,
                "prompt": prompt,
                "stream": False,
                "options": {
                    "temperature": 0.7
                }
            }

            response = requests.post(
                OLLAMA_URL,
                json=payload,
                timeout=120
            )

            response.raise_for_status()

            data = response.json()

            question = data.get("response", "").strip()

            # ------------------------------------------------
            # Clean common unwanted prefixes
            # ------------------------------------------------

            question = re.sub(
                r"^(Question|Q)\s*:\s*",
                "",
                question,
                flags=re.IGNORECASE
            ).strip()

            # Remove surrounding quotes
            question = question.strip('"').strip("'").strip()

            last_question = question

            # ------------------------------------------------
            # Validate
            # ------------------------------------------------

            valid, reason = validate_int006_question(
                question=question,
                language=language,
                question_type=question_type,
                ground_truth=ground_truth
            )

            if valid:

                return {
                    "question": question,
                    "status": "VALID",
                    "attempts": attempt,
                    "reason": "Valid"
                }

            last_reason = reason

        except Exception as e:

            last_reason = f"Generation error: {str(e)}"

    return {
        "question": last_question,
        "status": "FAILED",
        "attempts": max_attempts,
        "reason": last_reason
    }


# ============================================================
# TEST REGENERATION
# ============================================================

test_row = int006_df.iloc[0]

print("=" * 80)
print("INT006 REGENERATION FUNCTION")
print("=" * 80)

print("FAQ ID:", test_row["faq_id"])
print("Language:", test_row["language"])
print("Type:", test_row["question_type"])

test_result = regenerate_int006_question(
    test_row,
    max_attempts=5
)

print("\nGenerated question:")
print(test_result["question"])

print("\nStatus:", test_result["status"])
print("Attempts:", test_result["attempts"])
print("Reason:", test_result["reason"])

print("=" * 80)

if test_result["status"] == "VALID":
    print("✅ REGENERATION FUNCTION READY")
else:
    print("⚠️ REGENERATION TEST FAILED")

print("=" * 80)
````

````python
# ============================================================
# INT006 - IMPROVED QUESTION GENERATION
# ============================================================

import pandas as pd
import requests
import time
import re

INTENT_ID = "INT006"

# ------------------------------------------------------------
# 1. Get INT006 rows
# ------------------------------------------------------------

int006_df = faq_generation_plan[
    faq_generation_plan["intent_id"] == INTENT_ID
].copy()

print("=" * 80)
print("INT006 IMPROVED QUESTION GENERATION")
print("=" * 80)

print(f"Total rows: {len(int006_df)}")
print()

# ------------------------------------------------------------
# 2. Ollama configuration
# ------------------------------------------------------------

OLLAMA_URL = "http://localhost:11434/api/generate"
OLLAMA_MODEL = "qwen2.5:7b"

# ------------------------------------------------------------
# 3. Ground truth
# ------------------------------------------------------------

GROUND_TRUTH = (
    "Early check-in is subject to room availability and hotel approval."
)

# ------------------------------------------------------------
# 4. Language instructions
# ------------------------------------------------------------

LANGUAGE_RULES = {
    "English": """
Write the question entirely in English.
Use natural hotel guest language.
""",

    "Arabic": """
Write the question entirely in Arabic.
Arabic script must be used.
The hotel name may remain exactly:
Nile Pearl Hotel
The question must end with:
؟
""",

    "French": """
Write the question entirely in French.
The hotel name may remain exactly:
Nile Pearl Hotel
The question must end with:
?
""",

    "Russian": """
Write the question entirely in Russian.
Use Cyrillic.
The hotel name may remain exactly:
Nile Pearl Hotel
The question must end with:
?
"""
}

# ------------------------------------------------------------
# 5. Question type instructions
# ------------------------------------------------------------

QUESTION_TYPE_RULES = {
    "Direct": """
Ask directly whether early check-in is possible.
Keep the question short and factual.
""",

    "Paraphrased": """
Ask about the same early check-in intent using different
natural wording.
Do not simply copy a Direct question.
""",

    "Scenario": """
Create a realistic hotel guest situation involving arriving
early, then ask whether early check-in is possible.
Do not invent a specific arrival time.
"""
}

# ------------------------------------------------------------
# 6. Prompt builder
# ------------------------------------------------------------

def build_int006_prompt(row):

    language = row["language"]
    question_type = row["question_type"]
    faq_id = row["faq_id"]

    prompt = f"""
You are generating ONE high-quality FAQ question
for a multilingual hotel AI assistant.

Hotel name:
Nile Pearl Hotel

FAQ ID:
{faq_id}

Language:
{language}

Question type:
{question_type}

CORE INTENT:
early_check_in

Ground truth:
{GROUND_TRUTH}


SEMANTIC RULE:

The question MUST be specifically about EARLY CHECK-IN.

The guest wants to know whether they can arrive,
check in, or access the room before the normal
check-in time.

The question must NOT be about:

- standard check-in time
- late check-out
- room availability in general
- hotel approval in general
- booking a room
- changing a reservation
- cancellation
- payment
- additional fees


DO NOT INVENT:

- a specific early check-in time
- an additional fee
- a guarantee
- a new hotel policy
- a new service


HOTEL NAME RULE:

If you mention the hotel name, write it EXACTLY as:

Nile Pearl Hotel

Never translate or modify the hotel name.

Never write:

Нил Пирл
Нил Перило
Нильский Бриллиант
Nile Перило
_hotel_name_
hôtel Nile Pearl


LANGUAGE RULE:

{LANGUAGE_RULES[language]}


QUESTION TYPE RULE:

{QUESTION_TYPE_RULES[question_type]}


QUALITY RULES:

- Return exactly ONE question.
- The output must be a question.
- Do not answer the question.
- Do not include explanations.
- Do not include labels.
- Do not include quotation marks.
- Do not include bullet points.
- Do not include "Question:".
- Do not mix languages.
- Keep the question natural and concise.


GOOD EXAMPLES:

English:
Can I request early check-in at Nile Pearl Hotel?

Arabic:
هل يمكنني تسجيل الوصول مبكرًا في Nile Pearl Hotel؟

French:
L'arrivée anticipée est-elle possible au Nile Pearl Hotel ?

Russian:
Можно ли воспользоваться ранним заездом в Nile Pearl Hotel?


FINAL SELF-CHECK:

Before returning the question, verify:

1. Exactly ONE question.
2. About early check-in.
3. Correct requested language.
4. Not about standard check-in.
5. Not about late check-out.
6. No invented time.
7. No invented fee.
8. No guarantee.
9. Hotel name, if used, is exactly Nile Pearl Hotel.
10. Correct question mark.

Return ONLY the question.

Generate now.
"""

    return prompt.strip()


# ------------------------------------------------------------
# 7. Ollama generation
# ------------------------------------------------------------

def generate_ollama_question(prompt):

    payload = {
        "model": OLLAMA_MODEL,
        "prompt": prompt,
        "stream": False,
        "options": {
            "temperature": 0.7,
            "top_p": 0.9
        }
    }

    try:

        response = requests.post(
            OLLAMA_URL,
            json=payload,
            timeout=120
        )

        response.raise_for_status()

        data = response.json()

        question = data.get("response", "").strip()

        # Remove accidental labels
        question = re.sub(
            r"^(Question|question)\s*:\s*",
            "",
            question
        ).strip()

        # Remove surrounding quotes
        question = question.strip('"').strip("'").strip()

        # Remove accidental multiple lines
        question = question.split("\n")[0].strip()

        return question

    except Exception as e:

        print(f"Ollama Error: {e}")
        return ""


# ------------------------------------------------------------
# 8. Generate all questions
# ------------------------------------------------------------

generated_questions = []

print("=" * 80)
print("GENERATING INT006 QUESTIONS")
print("=" * 80)

for index, row in int006_df.iterrows():

    faq_id = row["faq_id"]
    language = row["language"]
    question_type = row["question_type"]

    print()
    print("-" * 80)
    print(
        f"{index + 1}/{len(int006_df)} | "
        f"{faq_id} | {language} | {question_type}"
    )

    prompt = build_int006_prompt(row)

    question = generate_ollama_question(prompt)

    generated_questions.append(question)

    print("GENERATED:", question)

    time.sleep(0.5)


# ------------------------------------------------------------
# 9. Save generated questions
# ------------------------------------------------------------

int006_df["question"] = generated_questions

print()
print("=" * 80)
print("INT006 GENERATION COMPLETE")
print("=" * 80)

print("Total:", len(int006_df))
print("Non-empty:", int006_df["question"].ne("").sum())
print("Empty:", int006_df["question"].eq("").sum())
````

````python
# ============================================================
# INT006 - DUPLICATE DETECTION
# ============================================================

import pandas as pd
from difflib import SequenceMatcher

print("=" * 80)
print("INT006 DUPLICATE CHECK")
print("=" * 80)

# Make sure INT006 exists
int006_df = faq_generation_plan[
    faq_generation_plan["intent_id"] == "INT006"
].copy()

print(f"Total questions: {len(int006_df)}")

# ------------------------------------------------------------
# Similarity function
# ------------------------------------------------------------

def text_similarity(text1, text2):
    text1 = str(text1).strip().lower()
    text2 = str(text2).strip().lower()

    return SequenceMatcher(None, text1, text2).ratio()


# ------------------------------------------------------------
# Check duplicates
# ------------------------------------------------------------

duplicates = []

rows = int006_df.reset_index(drop=True)

for i in range(len(rows)):
    for j in range(i + 1, len(rows)):

        q1 = rows.loc[i, "question"]
        q2 = rows.loc[j, "question"]

        # Skip empty questions
        if not str(q1).strip() or not str(q2).strip():
            continue

        similarity = text_similarity(q1, q2)

        # Detect highly similar questions
        if similarity >= 0.85:

            duplicates.append({
                "faq_1": rows.loc[i, "faq_id"],
                "faq_2": rows.loc[j, "faq_id"],
                "language": rows.loc[i, "language"],
                "question_type_1": rows.loc[i, "question_type"],
                "question_type_2": rows.loc[j, "question_type"],
                "similarity": round(similarity, 3),
                "question_1": q1,
                "question_2": q2
            })


duplicates_df = pd.DataFrame(duplicates)

print("=" * 80)
print("INT006 DUPLICATE RESULTS")
print("=" * 80)

print(f"Potential duplicates: {len(duplicates_df)}")

if len(duplicates_df) > 0:

    print("\n⚠️ POTENTIAL DUPLICATES FOUND\n")

    display(
        duplicates_df.sort_values(
            "similarity",
            ascending=False
        ).reset_index(drop=True)
    )

else:

    print("✅ NO POTENTIAL DUPLICATES FOUND")
````

````python
# ============================================================
# INT006 - REAL DUPLICATE CHECK
# ============================================================

import pandas as pd
from difflib import SequenceMatcher

print("=" * 80)
print("INT006 REAL DUPLICATE CHECK")
print("=" * 80)

df = int006_df.copy()

# ------------------------------------------------------------
# Clean questions
# ------------------------------------------------------------

df["question"] = (
    df["question"]
    .fillna("")
    .astype(str)
    .str.strip()
)

# ------------------------------------------------------------
# Normalize text for exact duplicate detection
# ------------------------------------------------------------

def normalize_question(text):
    text = str(text).strip().lower()

    # Remove final question marks
    text = text.replace("؟", "")
    text = text.replace("?", "")

    # Normalize spaces
    text = " ".join(text.split())

    return text


df["normalized_question"] = df["question"].apply(
    normalize_question
)

# ------------------------------------------------------------
# Exact duplicates
# ------------------------------------------------------------

exact_duplicates = df[
    df.duplicated(
        subset=["language", "normalized_question"],
        keep=False
    )
].copy()

print("\nTotal questions:", len(df))
print("Exact duplicate rows:", len(exact_duplicates))

if len(exact_duplicates) == 0:

    print("✅ NO EXACT DUPLICATES")

else:

    print("⚠️ EXACT DUPLICATES FOUND\n")

    print(
        exact_duplicates[
            [
                "faq_id",
                "language",
                "question_type",
                "question"
            ]
        ].sort_values(
            ["language", "question"]
        ).to_string(index=False)
    )

# ------------------------------------------------------------
# Similarity check
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SIMILARITY CHECK")
print("=" * 80)

similar_pairs = []

for i in range(len(df)):

    for j in range(i + 1, len(df)):

        # Only compare questions in same language
        if df.iloc[i]["language"] != df.iloc[j]["language"]:
            continue

        q1 = df.iloc[i]["normalized_question"]
        q2 = df.iloc[j]["normalized_question"]

        if not q1 or not q2:
            continue

        similarity = SequenceMatcher(
            None,
            q1,
            q2
        ).ratio()

        if similarity >= 0.85:

            similar_pairs.append({
                "faq_1": df.iloc[i]["faq_id"],
                "faq_2": df.iloc[j]["faq_id"],
                "language": df.iloc[i]["language"],
                "question_type_1": df.iloc[i]["question_type"],
                "question_type_2": df.iloc[j]["question_type"],
                "similarity": round(similarity, 3),
                "question_1": df.iloc[i]["question"],
                "question_2": df.iloc[j]["question"]
            })

similar_df = pd.DataFrame(similar_pairs)

print("Potential similar pairs:", len(similar_df))

if len(similar_df) > 0:

    print("\nTop similar pairs:")

    print(
        similar_df
        .sort_values("similarity", ascending=False)
        .head(50)
        .to_string(index=False)
    )

else:

    print("✅ NO HIGH-SIMILARITY PAIRS")

print("\n" + "=" * 80)
print("DUPLICATE CHECK COMPLETE")
print("=" * 80)
````

````python
# ============================================================
# INT006 - DIVERSE GENERATION WITH DUPLICATE PREVENTION
# ============================================================

import pandas as pd
import re
import time


def normalize_question(text):
    """
    Normalize question for exact duplicate detection.
    """
    if text is None:
        return ""

    text = str(text).strip().lower()

    # Remove surrounding quotes
    text = text.strip('"').strip("'")

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text)

    return text


def is_duplicate_question(question, previous_questions):
    """
    Check exact duplicates.
    """
    q_norm = normalize_question(question)

    if not q_norm:
        return True

    previous_normalized = {
        normalize_question(q)
        for q in previous_questions
        if str(q).strip()
    }

    return q_norm in previous_normalized


def generate_int006_diverse_question(
    row,
    previous_questions,
    max_attempts=5
):
    """
    Generate one question while avoiding previous duplicates.
    """

    for attempt in range(1, max_attempts + 1):

        prompt = build_int006_diverse_prompt(row)

        # Add previous questions to the prompt
        if previous_questions:

            previous_text = "\n".join(
                f"- {q}"
                for q in previous_questions[-15:]
            )

            prompt += f"""

============================================================
QUESTIONS ALREADY GENERATED
============================================================

The following questions already exist in the dataset.

You MUST NOT copy or closely reproduce them:

{previous_text}

Generate a NEW question with a clearly different structure.

============================================================
"""

        try:

            question = generate_ollama_question(prompt)

            if question is None:
                continue

            question = str(question).strip()

            # Clean possible model formatting
            question = question.strip('"').strip("'")
            question = re.sub(
                r"^(Question|Answer)\s*:\s*",
                "",
                question,
                flags=re.IGNORECASE
            ).strip()

            # Validate
            validation = validate_int006_question(
                question,
                row
            )

            if not validation.get("valid", False):
                print(
                    f"Attempt {attempt}: "
                    f"validation failed -> "
                    f"{validation.get('reason', 'Unknown')}"
                )
                continue

            # Duplicate check
            if is_duplicate_question(
                question,
                previous_questions
            ):
                print(
                    f"Attempt {attempt}: "
                    f"duplicate detected -> {question}"
                )
                continue

            return question

        except Exception as e:

            print(
                f"Attempt {attempt}: ERROR -> {e}"
            )

        time.sleep(0.5)

    return ""


print("=" * 80)
print("INT006 DIVERSE GENERATION FUNCTION READY")
print("=" * 80)
````

````python
# ============================================================
# INT006 - FINAL CLEAN GENERATION + VALIDATION + STORAGE
# ============================================================

import os
import re
import time
import pandas as pd

INTENT_ID = "INT006"
OUTPUT_PATH = "03_Structured_Data/CSV/faq_INT006_final.csv"

print("=" * 80)
print("INT006 FINAL CLEAN GENERATION")
print("=" * 80)

# ------------------------------------------------------------
# 1. LOAD INT006 DATA
# ------------------------------------------------------------

int006_df = faq_generation_plan[
    faq_generation_plan["intent_id"] == INTENT_ID
].copy()

int006_df = int006_df.reset_index(drop=True)

print(f"Total rows: {len(int006_df)}")
print(f"Languages: {int006_df['language'].value_counts().to_dict()}")
print(f"Question types: {int006_df['question_type'].value_counts().to_dict()}")

# ------------------------------------------------------------
# 2. CHECK REQUIRED COLUMNS
# ------------------------------------------------------------

required_columns = [
    "faq_id",
    "intent_id",
    "intent_name",
    "category",
    "language",
    "question_type",
    "difficulty",
    "question",
    "answer",
    "ground_truth",
    "source"
]

missing = [
    col for col in required_columns
    if col not in int006_df.columns
]

if missing:
    raise ValueError(f"Missing columns: {missing}")

# ------------------------------------------------------------
# 3. SAFE QUESTION CLEANER
# ------------------------------------------------------------

def clean_generated_question(question):
    if question is None:
        return ""

    question = str(question).strip()

    # Remove common model prefixes
    question = re.sub(
        r"^(question|answer)\s*:\s*",
        "",
        question,
        flags=re.IGNORECASE
    )

    # Remove quotation marks
    question = question.strip('"').strip("'").strip()

    # Remove accidental markdown
    question = question.replace("**", "")
    question = question.replace("```", "")

    # Normalize spaces
    question = re.sub(r"\s+", " ", question).strip()

    return question


# ------------------------------------------------------------
# 4. QUESTION TYPE SPECIFIC DIVERSITY INSTRUCTIONS
# ------------------------------------------------------------

def diversity_instruction(language, question_type, faq_id):

    if language == "English":

        if question_type == "Direct":
            variants = {
                "FAQ0161": "Use 'Is early check-in available'.",
                "FAQ0162": "Use 'Can guests check in early'.",
                "FAQ0163": "Use 'Does the hotel allow early check-in'."
            }
            return variants.get(
                faq_id,
                "Ask directly using different natural wording."
            )

        elif question_type == "Paraphrased":
            variants = {
                "FAQ0164": "Use 'arrange early arrival'.",
                "FAQ0165": "Use 'request an earlier check-in'.",
                "FAQ0166": "Use 'check in before the usual time'."
            }
            return variants.get(
                faq_id,
                "Paraphrase naturally."
            )

        else:
            variants = {
                "FAQ0167": "Start with 'If I arrive before my scheduled check-in...'.",
                "FAQ0168": "Use a different realistic early-arrival situation."
            }
            return variants.get(
                faq_id,
                "Create a realistic early-arrival situation."
            )

    elif language == "Arabic":

        if question_type == "Direct":
            variants = {
                "FAQ0169": "استخدم: هل يتوفر تسجيل الوصول المبكر؟",
                "FAQ0170": "استخدم: هل يُسمح بتسجيل الوصول المبكر؟",
                "FAQ0171": "استخدم: هل يمكن للضيوف تسجيل الوصول مبكرًا؟"
            }
            return variants.get(
                faq_id,
                "استخدم صياغة عربية مختلفة للسؤال عن تسجيل الوصول المبكر."
            )

        elif question_type == "Paraphrased":
            variants = {
                "FAQ0172": "اسأل عن إمكانية تسجيل الوصول قبل الموعد المعتاد.",
                "FAQ0173": "استخدم صياغة تتحدث عن الوصول قبل موعد تسجيل الدخول.",
                "FAQ0174": "اسأل بطريقة مختلفة عن الدخول إلى الغرفة مبكرًا."
            }
            return variants.get(
                faq_id,
                "أعد صياغة السؤال بشكل طبيعي."
            )

        else:
            variants = {
                "FAQ0175": "ابدأ بموقف ضيف يصل إلى الفندق قبل موعد تسجيل الوصول.",
                "FAQ0176": "استخدم موقفًا مختلفًا لضيف يصل مبكرًا."
            }
            return variants.get(
                faq_id,
                "اكتب موقفًا واقعيًا متعلقًا بالوصول المبكر."
            )

    elif language == "French":

        if question_type == "Direct":
            variants = {
                "FAQ0177": "Use 'Le check-in anticipé est-il disponible'.",
                "FAQ0178": "Use 'Peut-on faire une arrivée anticipée'.",
                "FAQ0179": "Use 'L'hôtel autorise-t-il l'arrivée anticipée'."
            }
            return variants.get(
                faq_id,
                "Ask directly using different French wording."
            )

        elif question_type == "Paraphrased":
            variants = {
                "FAQ0180": "Use 'arriver avant l'heure prévue'.",
                "FAQ0181": "Use 'demander un enregistrement anticipé'.",
                "FAQ0182": "Use another natural French paraphrase."
            }
            return variants.get(
                faq_id,
                "Paraphrase naturally in French."
            )

        else:
            variants = {
                "FAQ0183": "Create a situation where the guest arrives earlier than expected.",
                "FAQ0184": "Create a different early-arrival situation."
            }
            return variants.get(
                faq_id,
                "Create a realistic French scenario."
            )

    elif language == "Russian":

        if question_type == "Direct":
            variants = {
                "FAQ0185": "Спросите: доступна ли ранняя регистрация заезда.",
                "FAQ0186": "Спросите: разрешает ли отель ранний заезд.",
                "FAQ0187": "Спросите другим естественным способом о раннем заезде."
            }
            return variants.get(
                faq_id,
                "Сформулируйте прямой вопрос о раннем заезде."
            )

        elif question_type == "Paraphrased":
            variants = {
                "FAQ0188": "Используйте формулировку о возможности заселиться раньше.",
                "FAQ0189": "Спросите о возможности раннего заезда другими словами.",
                "FAQ0190": "Используйте формулировку о прибытии до обычного времени заезда."
            }
            return variants.get(
                faq_id,
                "Перефразируйте вопрос естественно."
            )

        else:
            variants = {
                "FAQ0191": "Создайте ситуацию, когда гость приезжает раньше времени заезда.",
                "FAQ0192": "Создайте другую реалистичную ситуацию раннего прибытия."
            }
            return variants.get(
                faq_id,
                "Создайте реалистичный сценарий."
            )

    return ""


# ------------------------------------------------------------
# 5. IMPROVED PROMPT BUILDER
# ------------------------------------------------------------

def build_int006_diverse_prompt(row, previous_questions):

    faq_id = str(row["faq_id"])
    language = str(row["language"])
    question_type = str(row["question_type"])
    ground_truth = str(row["ground_truth"])

    diversity = diversity_instruction(
        language,
        question_type,
        faq_id
    )

    previous_text = "\n".join(
        f"- {q}"
        for q in previous_questions[-15:]
    )

    if language == "Arabic":
        question_mark = "؟"
        language_rule = """
اكتب السؤال باللغة العربية فقط.
يسمح بكتابة اسم الفندق Nile Pearl Hotel كما هو.
يجب أن ينتهي السؤال بعلامة الاستفهام العربية ؟
"""

    elif language == "French":
        question_mark = "?"
        language_rule = """
Write entirely in French.
The hotel name must remain exactly Nile Pearl Hotel.
End with ?.
"""

    elif language == "Russian":
        question_mark = "?"
        language_rule = """
Write entirely in Russian.
The hotel name must remain exactly Nile Pearl Hotel.
End with ?.
"""

    else:
        question_mark = "?"
        language_rule = """
Write entirely in English.
The hotel name must remain exactly Nile Pearl Hotel.
End with ?.
"""

    prompt = f"""
You are generating ONE high-quality FAQ question
for the Nile Pearl Hotel multilingual hotel AI dataset.

FAQ ID:
{faq_id}

Language:
{language}

Question type:
{question_type}

Core intent:
early_check_in

Ground truth:
{ground_truth}

IMPORTANT:
The question MUST be specifically about EARLY CHECK-IN.

It means the guest wants to know whether they can
arrive, check in, or access the room before the normal
check-in time.

DO NOT ask about:
- standard check-in time
- late check-out
- general room availability
- hotel approval in general
- booking a room
- cancellation
- payment
- fees

DO NOT invent:
- a specific time
- a fee
- a guarantee
- a new policy

HOTEL NAME:
If mentioned, write EXACTLY:
Nile Pearl Hotel

Never translate or modify the hotel name.

QUESTION TYPE:
{question_type}

DIVERSITY REQUIREMENT:
{diversity}

VERY IMPORTANT:
The question MUST be meaningfully different from previous questions.

Previous generated questions:
{previous_text if previous_text else "None"}

{language_rule}

Return ONLY ONE question.
No explanation.
No label.
No quotation marks.
No bullet points.
No "Question:".

Generate now.
"""

    return prompt


# ------------------------------------------------------------
# 6. LOCAL VALIDATOR
# ------------------------------------------------------------

def validate_int006_final_question(
    question,
    language,
    question_type,
    ground_truth
):

    if not question:
        return False, "Empty question"

    q = question.strip()

    # Must be one line
    if "\n" in q:
        return False, "Multiple lines"

    # No labels
    if re.match(r"^(question|answer)\s*:", q, re.I):
        return False, "Contains label"

    # Hotel name consistency
    bad_hotel_names = [
        "Нил Пирл",
        "Нил Перило",
        "Нильский Бриллиант",
        "Nile Перило",
        "_hotel_name_",
        "hôtel Nile Pearl"
    ]

    for bad_name in bad_hotel_names:
        if bad_name.lower() in q.lower():
            return False, "Invalid hotel name"

    # Language-specific question mark
    if language == "Arabic":
        if not q.endswith("؟"):
            return False, "Arabic question must end with ؟"
    else:
        if not q.endswith("?"):
            return False, "Question must end with ?"

    # Basic early check-in semantic validation
    lower_q = q.lower()

    if language == "English":
        keywords = [
            "early check-in",
            "early check in",
            "check in early",
            "arrive early",
            "earlier check-in",
            "earlier check in"
        ]

    elif language == "Arabic":
        keywords = [
            "تسجيل الوصول المبكر",
            "التسجيل المبكر",
            "تسجيل الدخول مبكر",
            "الوصول مبكر",
            "الدخول مبكر"
        ]

    elif language == "French":
        keywords = [
            "arrivée anticipée",
            "arrivée précoce",
            "check-in anticipé",
            "arriver plus tôt",
            "enregistrement anticipé"
        ]

    else:
        keywords = [
            "ранн",
            "заезд",
            "засел",
            "ранний заезд"
        ]

    if not any(k.lower() in lower_q for k in keywords):
        return False, "Not clearly about early check-in"

    # Prevent unrelated topics
    forbidden_patterns = [
        "late check-out",
        "late checkout",
        "standard check-in time",
        "check-out time",
        "cancellation",
        "payment",
        "fee",
        "price"
    ]

    if language == "English":
        for word in forbidden_patterns:
            if word in lower_q:
                return False, f"Contains unrelated topic: {word}"

    # Must not be absurdly long
    if len(q) > 220:
        return False, "Question too long"

    return True, "Valid"


# ------------------------------------------------------------
# 7. GENERATION FUNCTION
# ------------------------------------------------------------

def generate_unique_int006_question(
    row,
    previous_questions,
    max_attempts=7
):

    faq_id = str(row["faq_id"])
    language = str(row["language"])
    question_type = str(row["question_type"])
    ground_truth = str(row["ground_truth"])

    for attempt in range(1, max_attempts + 1):

        try:

            prompt = build_int006_diverse_prompt(
                row,
                previous_questions
            )

            raw_question = generate_ollama_question(prompt)

            question = clean_generated_question(raw_question)

            valid, reason = validate_int006_final_question(
                question,
                language,
                question_type,
                ground_truth
            )

            if not valid:
                print(
                    f"Attempt {attempt}: INVALID -> {reason}"
                )
                continue

            # Exact duplicate check
            if question in previous_questions:
                print(
                    f"Attempt {attempt}: DUPLICATE -> retry"
                )
                continue

            # Similarity check
            is_similar = False

            for old_q in previous_questions:

                try:
                    sim = similarity(
                        question,
                        old_q
                    )

                    if sim >= 0.90:
                        is_similar = True
                        break

                except Exception:
                    pass

            if is_similar:
                print(
                    f"Attempt {attempt}: TOO SIMILAR -> retry"
                )
                continue

            return question, True, "Valid"

        except Exception as e:

            print(
                f"Attempt {attempt}: ERROR -> {e}"
            )

    return "", False, "Could not generate unique question"


# ------------------------------------------------------------
# 8. GENERATE ALL QUESTIONS
# ------------------------------------------------------------

previous_questions = []

generated_questions = []

print("\n" + "=" * 80)
print("GENERATING INT006 QUESTIONS")
print("=" * 80)

for idx, row in int006_df.iterrows():

    faq_id = row["faq_id"]
    language = row["language"]
    question_type = row["question_type"]

    print("\n" + "-" * 80)
    print(
        f"{idx + 1}/{len(int006_df)} | "
        f"{faq_id} | {language} | {question_type}"
    )

    question, success, reason = generate_unique_int006_question(
        row,
        previous_questions,
        max_attempts=7
    )

    if success:

        print(f"GENERATED: {question}")

        previous_questions.append(question)
        generated_questions.append(question)

    else:

        print(f"FAILED: {reason}")

        generated_questions.append("")


# ------------------------------------------------------------
# 9. STORE QUESTIONS
# ------------------------------------------------------------

int006_df["question"] = generated_questions


# ------------------------------------------------------------
# 10. STORAGE VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("INT006 STORAGE VALIDATION")
print("=" * 80)

non_empty = (
    int006_df["question"]
    .fillna("")
    .astype(str)
    .str.strip()
    .ne("")
    .sum()
)

empty = len(int006_df) - non_empty

print(f"Rows: {len(int006_df)}")
print(f"Non-empty questions: {non_empty}")
print(f"Empty questions: {empty}")


# ------------------------------------------------------------
# 11. EXACT DUPLICATE CHECK
# ------------------------------------------------------------

duplicate_mask = (
    int006_df["question"]
    .duplicated(keep=False)
    &
    int006_df["question"].ne("")
)

duplicate_rows = int006_df[duplicate_mask]

print("\n" + "=" * 80)
print("EXACT DUPLICATE CHECK")
print("=" * 80)

print(f"Exact duplicate rows: {len(duplicate_rows)}")

if len(duplicate_rows) > 0:

    print("⚠️ DUPLICATES FOUND")

    print(
        duplicate_rows[
            ["faq_id", "language", "question_type", "question"]
        ].to_string(index=False)
    )

else:

    print("✅ NO EXACT DUPLICATES")


# ------------------------------------------------------------
# 12. SAVE
# ------------------------------------------------------------

os.makedirs(
    os.path.dirname(OUTPUT_PATH),
    exist_ok=True
)

int006_df.to_csv(
    OUTPUT_PATH,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 80)
print("INT006 SAVE")
print("=" * 80)

print(f"Path: {OUTPUT_PATH}")
print(f"Rows saved: {len(int006_df)}")

print("\n" + "=" * 80)
print("FINAL SAMPLE")
print("=" * 80)

print(
    int006_df[
        ["faq_id", "language", "question_type", "question"]
    ].to_string(index=False)
)

print("\n" + "=" * 80)
print("✅ INT006 PROCESS COMPLETE")
print("=" * 80)
````

````python
# ============================================================
# INT006 - REGENERATE EMPTY QUESTIONS ONLY
# ============================================================

import pandas as pd

print("=" * 80)
print("INT006 REGENERATE EMPTY QUESTIONS")
print("=" * 80)

# Load current dataset
int006_path = "03_Structured_Data/CSV/faq_INT006_final.csv"
int006_df = pd.read_csv(int006_path)

# Find empty questions
empty_mask = (
    int006_df["question"].isna() |
    (int006_df["question"].astype(str).str.strip() == "")
)

empty_rows = int006_df[empty_mask].copy()

print(f"Total rows: {len(int006_df)}")
print(f"Empty questions: {len(empty_rows)}")
print()

if len(empty_rows) == 0:
    print("✅ No empty questions found.")
else:

    print("Questions to regenerate:")
    print(empty_rows[["faq_id", "language", "question_type"]].to_string(index=False))
    print()

    # --------------------------------------------------------
    # Targeted generation instructions
    # --------------------------------------------------------

    regeneration_prompts = {

        "FAQ0164": """
Generate ONE English paraphrased question about EARLY CHECK-IN
at Nile Pearl Hotel.

Do NOT repeat the wording:
"Is early check-in available at Nile Pearl Hotel?"

Use different wording but keep the exact same intent.

Return ONLY the question.
""",

        "FAQ0166": """
Generate ONE English paraphrased question about EARLY CHECK-IN
at Nile Pearl Hotel.

Use natural English and clearly ask whether the guest can request
or receive early check-in.

Do NOT repeat existing wording.

Return ONLY the question.
""",

        "FAQ0167": """
Generate ONE English SCENARIO question about EARLY CHECK-IN
at Nile Pearl Hotel.

The question must describe a realistic guest situation, such as
arriving before the normal check-in time.

It must clearly ask about early check-in.

Return ONLY the question.
""",

        "FAQ0180": """
Generate ONE natural French paraphrased question about
EARLY CHECK-IN at Nile Pearl Hotel.

Use correct French grammar.

Do NOT use:
"L'hôtel autorise-t-il l'arrivée anticipée ?"

The question must clearly be about requesting or getting
an early check-in.

Return ONLY the question.
"""
    }

    for idx in empty_rows.index:

        faq_id = int006_df.loc[idx, "faq_id"]
        language = int006_df.loc[idx, "language"]
        question_type = int006_df.loc[idx, "question_type"]

        print("-" * 80)
        print(f"{faq_id} | {language} | {question_type}")

        prompt = regeneration_prompts.get(faq_id)

        if prompt is None:
            print("❌ No regeneration prompt found")
            continue

        generated_question = None

        for attempt in range(1, 8):

            try:

                response = generate_ollama_question(prompt)

                if response is None:
                    print(f"Attempt {attempt}: EMPTY RESPONSE")
                    continue

                question = str(response).strip()

                # Remove accidental quotes
                question = question.strip('"').strip("'").strip()

                # Basic cleanup
                question = question.replace("؟؟", "؟")

                if not question:
                    print(f"Attempt {attempt}: EMPTY")
                    continue

                # Make sure it is a question
                if not question.endswith(("?", "؟")):
                    question += "?"

                # Basic language/intent validation
                if faq_id == "FAQ0164":
                    valid = (
                        "early check-in" in question.lower()
                        and "nile pearl hotel" in question.lower()
                    )

                elif faq_id == "FAQ0166":
                    valid = (
                        "check-in" in question.lower()
                        and "early" in question.lower()
                        and "nile pearl hotel" in question.lower()
                    )

                elif faq_id == "FAQ0167":
                    q = question.lower()

                    valid = (
                        "nile pearl hotel" in q
                        and (
                            "early check-in" in q
                            or "check in early" in q
                            or "check-in early" in q
                        )
                    )

                elif faq_id == "FAQ0180":
                    q = question.lower()

                    valid = (
                        "nile pearl hotel" in q
                        and (
                            "anticip" in q
                            or "check-in" in q
                            or "enregistrement" in q
                        )
                    )

                else:
                    valid = True

                if not valid:
                    print(f"Attempt {attempt}: INVALID -> intent/hotel check failed")
                    continue

                # Check against existing questions
                existing_questions = (
                    int006_df["question"]
                    .fillna("")
                    .astype(str)
                    .str.strip()
                )

                if question in existing_questions.values:
                    print(f"Attempt {attempt}: DUPLICATE -> retry")
                    continue

                generated_question = question
                print(f"GENERATED: {generated_question}")
                break

            except Exception as e:
                print(f"Attempt {attempt}: ERROR -> {e}")

        if generated_question:
            int006_df.loc[idx, "question"] = generated_question
        else:
            print("❌ FAILED")

# ============================================================
# SAVE
# ============================================================

int006_df.to_csv(
    int006_path,
    index=False,
    encoding="utf-8-sig"
)

print()
print("=" * 80)
print("STORAGE VALIDATION")
print("=" * 80)

non_empty = (
    int006_df["question"].notna()
    & (int006_df["question"].astype(str).str.strip() != "")
).sum()

empty = len(int006_df) - non_empty

print(f"Rows: {len(int006_df)}")
print(f"Non-empty questions: {non_empty}")
print(f"Empty questions: {empty}")

if empty == 0:
    print("✅ ALL 32 QUESTIONS ARE STORED")
else:
    print("⚠️ SOME QUESTIONS ARE STILL EMPTY")

print("=" * 80)
````

````python
# ============================================================
# INT006 FINAL QUALITY AUDIT
# ============================================================

import pandas as pd
from difflib import SequenceMatcher
import re

PATH = "03_Structured_Data/CSV/faq_INT006_final.csv"

df = pd.read_csv(PATH)

print("=" * 80)
print("INT006 FINAL QUALITY AUDIT")
print("=" * 80)

# ------------------------------------------------------------
# 1. BASIC VALIDATION
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("1. BASIC VALIDATION")
print("-" * 80)

print(f"Total rows: {len(df)}")

empty_mask = (
    df["question"].isna()
    | (df["question"].astype(str).str.strip() == "")
)

print(f"Non-empty questions: {(~empty_mask).sum()}")
print(f"Empty questions: {empty_mask.sum()}")

print("\nLanguages:")
print(df["language"].value_counts().to_dict())

print("\nQuestion types:")
print(df["question_type"].value_counts().to_dict())


# ------------------------------------------------------------
# 2. EXACT DUPLICATES
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("2. EXACT DUPLICATE CHECK")
print("-" * 80)

questions = df["question"].fillna("").astype(str).str.strip()

duplicate_mask = questions.duplicated(keep=False)

exact_duplicates = df[duplicate_mask].copy()

print(f"Exact duplicate rows: {len(exact_duplicates)}")

if len(exact_duplicates) == 0:
    print("✅ NO EXACT DUPLICATES")
else:
    print("⚠️ EXACT DUPLICATES FOUND")
    print(
        exact_duplicates[
            ["faq_id", "language", "question_type", "question"]
        ].to_string(index=False)
    )


# ------------------------------------------------------------
# 3. SIMILARITY CHECK
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("3. SIMILARITY CHECK")
print("-" * 80)

def normalize_question(text):
    text = str(text).lower().strip()

    # Remove punctuation
    text = re.sub(r"[?!؟.,:;\"'«»()]", " ", text)

    # Normalize spaces
    text = re.sub(r"\s+", " ", text)

    return text


similar_pairs = []

for i in range(len(df)):
    q1 = normalize_question(df.iloc[i]["question"])

    if not q1:
        continue

    for j in range(i + 1, len(df)):
        q2 = normalize_question(df.iloc[j]["question"])

        if not q2:
            continue

        # Compare only same language
        if df.iloc[i]["language"] != df.iloc[j]["language"]:
            continue

        similarity = SequenceMatcher(None, q1, q2).ratio()

        if similarity >= 0.85:
            similar_pairs.append({
                "faq_1": df.iloc[i]["faq_id"],
                "faq_2": df.iloc[j]["faq_id"],
                "language": df.iloc[i]["language"],
                "type_1": df.iloc[i]["question_type"],
                "type_2": df.iloc[j]["question_type"],
                "similarity": round(similarity, 3),
                "question_1": df.iloc[i]["question"],
                "question_2": df.iloc[j]["question"],
            })

similar_df = pd.DataFrame(similar_pairs)

print(f"Potential similar pairs: {len(similar_df)}")

if len(similar_df) > 0:
    print("\nTop similar pairs:")

    print(
        similar_df
        .sort_values("similarity", ascending=False)
        .head(30)
        .to_string(index=False)
    )
else:
    print("✅ NO HIGH-SIMILARITY PAIRS")


# ------------------------------------------------------------
# 4. INTENT CHECK
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("4. EARLY CHECK-IN INTENT CHECK")
print("-" * 80)

def has_early_checkin_intent(question, language):

    q = str(question).lower()

    if language == "English":
        keywords = [
            "early check-in",
            "early check in",
            "check in early",
            "check-in early",
            "earlier check-in",
        ]

    elif language == "Arabic":
        keywords = [
            "تسجيل الوصول المبكر",
            "تسجيل الوصول مبكر",
            "تسجيل مبكر",
            "الوصول مبكر",
            "الدخول مبكر",
        ]

    elif language == "French":
        keywords = [
            "arrivée anticipée",
            "enregistrement anticipé",
            "check-in anticipé",
            "check in anticipé",
        ]

    elif language == "Russian":
        keywords = [
            "ранн",
            "заранее",
            "до обычного времени",
        ]

    else:
        keywords = []

    return any(keyword in q for keyword in keywords)


intent_failures = []

for idx, row in df.iterrows():

    question = str(row["question"]).strip()

    if not question:
        continue

    if not has_early_checkin_intent(
        question,
        row["language"]
    ):
        intent_failures.append({
            "faq_id": row["faq_id"],
            "language": row["language"],
            "question_type": row["question_type"],
            "question": question,
        })

print(f"Intent failures: {len(intent_failures)}")

if len(intent_failures) == 0:
    print("✅ ALL QUESTIONS CLEARLY RELATE TO EARLY CHECK-IN")
else:
    print("⚠️ QUESTIONS WITH POSSIBLE INTENT ISSUES")

    print(
        pd.DataFrame(intent_failures).to_string(index=False)
    )


# ------------------------------------------------------------
# 5. QUESTION MARK CHECK
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("5. QUESTION FORMAT CHECK")
print("-" * 80)

format_failures = []

for idx, row in df.iterrows():

    q = str(row["question"]).strip()

    if not q:
        continue

    if not q.endswith(("?", "؟")):
        format_failures.append({
            "faq_id": row["faq_id"],
            "question": q
        })

print(f"Questions without question mark: {len(format_failures)}")

if len(format_failures) == 0:
    print("✅ ALL QUESTIONS HAVE QUESTION MARKS")
else:
    print(
        pd.DataFrame(format_failures).to_string(index=False)
    )


# ------------------------------------------------------------
# 6. HOTEL NAME CHECK
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("6. HOTEL NAME CHECK")
print("-" * 80)

hotel_failures = []

for idx, row in df.iterrows():

    q = str(row["question"]).strip()

    if not q:
        continue

    # Require hotel name for scenario/paraphrased questions
    # and check direct questions that should identify the hotel.
    if "Nile Pearl Hotel" not in q:

        hotel_failures.append({
            "faq_id": row["faq_id"],
            "language": row["language"],
            "question_type": row["question_type"],
            "question": q
        })

print(f"Questions without 'Nile Pearl Hotel': {len(hotel_failures)}")

if len(hotel_failures) == 0:
    print("✅ HOTEL NAME PRESENT IN ALL QUESTIONS")
else:
    print("⚠️ QUESTIONS WITHOUT HOTEL NAME")
    print(
        pd.DataFrame(hotel_failures).to_string(index=False)
    )


# ------------------------------------------------------------
# 7. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("INT006 FINAL AUDIT SUMMARY")
print("=" * 80)

print(f"Total rows                  : {len(df)}")
print(f"Non-empty questions        : {(~empty_mask).sum()}")
print(f"Empty questions            : {empty_mask.sum()}")
print(f"Exact duplicate rows       : {len(exact_duplicates)}")
print(f"Similarity pairs >= 0.85   : {len(similar_df)}")
print(f"Intent failures             : {len(intent_failures)}")
print(f"Format failures             : {len(format_failures)}")
print(f"Hotel-name failures         : {len(hotel_failures)}")

print("\n" + "=" * 80)

if (
    len(df) == 32
    and empty_mask.sum() == 0
    and len(exact_duplicates) == 0
    and len(intent_failures) == 0
    and len(format_failures) == 0
):
    print("✅ INT006 PASSED CORE QUALITY CHECKS")
else:
    print("⚠️ INT006 NEEDS REVIEW")

print("=" * 80)
````

````python
import pandas as pd
from difflib import SequenceMatcher

PATH = "03_Structured_Data/CSV/faq_INT006_final.csv"

df = pd.read_csv(PATH)

print("=" * 80)
print("INT006 FINAL QUALITY AUDIT")
print("=" * 80)

# ============================================================
# 1. BASIC VALIDATION
# ============================================================

print("\n" + "-" * 80)
print("1. BASIC VALIDATION")
print("-" * 80)

total_rows = len(df)
non_empty = df["question"].fillna("").astype(str).str.strip().ne("").sum()
empty = total_rows - non_empty

print(f"Total rows: {total_rows}")
print(f"Non-empty questions: {non_empty}")
print(f"Empty questions: {empty}")

print("\nLanguages:")
print(df["language"].value_counts().to_dict())

print("\nQuestion types:")
print(df["question_type"].value_counts().to_dict())


# ============================================================
# 2. EXACT DUPLICATE CHECK
# ============================================================

print("\n" + "-" * 80)
print("2. EXACT DUPLICATE CHECK")
print("-" * 80)

questions = df["question"].fillna("").astype(str).str.strip()

duplicate_count = questions.duplicated(keep=False).sum()

print(f"Exact duplicate rows: {duplicate_count}")

if duplicate_count == 0:
    print("✅ NO EXACT DUPLICATES")
else:
    print("⚠️ EXACT DUPLICATES FOUND")
    print(
        df[questions.duplicated(keep=False)][
            ["faq_id", "language", "question_type", "question"]
        ].to_string(index=False)
    )


# ============================================================
# 3. SIMILARITY CHECK
# ============================================================

print("\n" + "-" * 80)
print("3. SIMILARITY CHECK")
print("-" * 80)

similar_pairs = []

for i in range(len(df)):
    q1 = str(df.iloc[i]["question"]).strip()

    if not q1:
        continue

    for j in range(i + 1, len(df)):
        q2 = str(df.iloc[j]["question"]).strip()

        if not q2:
            continue

        similarity = SequenceMatcher(None, q1.lower(), q2.lower()).ratio()

        if similarity >= 0.85:
            similar_pairs.append({
                "faq_1": df.iloc[i]["faq_id"],
                "faq_2": df.iloc[j]["faq_id"],
                "language": df.iloc[i]["language"],
                "type_1": df.iloc[i]["question_type"],
                "type_2": df.iloc[j]["question_type"],
                "similarity": round(similarity, 3),
                "question_1": q1,
                "question_2": q2
            })

print(f"Potential similar pairs: {len(similar_pairs)}")

if similar_pairs:
    similarity_df = pd.DataFrame(similar_pairs)
    similarity_df = similarity_df.sort_values(
        "similarity",
        ascending=False
    )

    print("\nTop similar pairs:")
    print(similarity_df.head(10).to_string(index=False))
else:
    print("✅ NO HIGH-SIMILARITY PAIRS")


# ============================================================
# 4. EARLY CHECK-IN INTENT CHECK
# ============================================================

print("\n" + "-" * 80)
print("4. EARLY CHECK-IN INTENT CHECK")
print("-" * 80)

intent_keywords = [
    "early check-in",
    "early check in",
    "earlier check-in",
    "earlier arrival",
    "arrive early",
    "check in early",
    "تسجيل الوصول المبكر",
    "تسجيل الوصول مبكر",
    "الوصول مبكر",
    "وصول مبكر",
    "arrivée anticipée",
    "enregistrement anticipé",
    "enregistrement avant",
    "ранний заезд",
    "ранняя регистрация",
    "зарегистрироваться раньше"
]

intent_failures = []

for _, row in df.iterrows():

    question = str(row["question"]).lower().strip()

    if not any(keyword.lower() in question for keyword in intent_keywords):
        intent_failures.append({
            "faq_id": row["faq_id"],
            "language": row["language"],
            "question_type": row["question_type"],
            "question": row["question"]
        })

print(f"Intent failures: {len(intent_failures)}")

if len(intent_failures) == 0:
    print("✅ ALL QUESTIONS CLEARLY RELATE TO EARLY CHECK-IN")
else:
    print("⚠️ INTENT FAILURES FOUND")
    print(pd.DataFrame(intent_failures).to_string(index=False))


# ============================================================
# 5. QUESTION FORMAT CHECK
# ============================================================

print("\n" + "-" * 80)
print("5. QUESTION FORMAT CHECK")
print("-" * 80)

format_failures = []

for _, row in df.iterrows():

    question = str(row["question"]).strip()

    if question and not question.endswith("?") and not question.endswith("؟"):
        format_failures.append({
            "faq_id": row["faq_id"],
            "language": row["language"],
            "question": row["question"]
        })

print(f"Questions without question mark: {len(format_failures)}")

if len(format_failures) == 0:
    print("✅ ALL QUESTIONS HAVE QUESTION MARKS")
else:
    print("⚠️ FORMAT FAILURES FOUND")
    print(pd.DataFrame(format_failures).to_string(index=False))


# ============================================================
# 6. HOTEL NAME CHECK
# ============================================================

print("\n" + "-" * 80)
print("6. HOTEL NAME CHECK")
print("-" * 80)

hotel_failures = []

for _, row in df.iterrows():

    question = str(row["question"])

    if "Nile Pearl Hotel" not in question:
        hotel_failures.append({
            "faq_id": row["faq_id"],
            "language": row["language"],
            "question_type": row["question_type"],
            "question": row["question"]
        })

print(f"Questions without 'Nile Pearl Hotel': {len(hotel_failures)}")

if len(hotel_failures) == 0:
    print("✅ ALL QUESTIONS CONTAIN HOTEL NAME")
else:
    print("⚠️ QUESTIONS WITHOUT HOTEL NAME")
    print(pd.DataFrame(hotel_failures).to_string(index=False))


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("INT006 FINAL AUDIT SUMMARY")
print("=" * 80)

print(f"Total rows                  : {total_rows}")
print(f"Non-empty questions        : {non_empty}")
print(f"Empty questions            : {empty}")
print(f"Exact duplicate rows       : {duplicate_count}")
print(f"Similarity pairs >= 0.85   : {len(similar_pairs)}")
print(f"Intent failures             : {len(intent_failures)}")
print(f"Format failures             : {len(format_failures)}")
print(f"Hotel-name failures         : {len(hotel_failures)}")

print("\n" + "=" * 80)

if (
    total_rows == 32
    and non_empty == 32
    and empty == 0
    and duplicate_count == 0
    and len(intent_failures) == 0
    and len(format_failures) == 0
    and len(hotel_failures) == 0
):
    print("✅ INT006 PASSED CORE QUALITY CHECKS")
    print("✅ INT006 READY")
else:
    print("⚠️ INT006 NEEDS REVIEW")

print("=" * 80)
````

````python
import pandas as pd
from difflib import SequenceMatcher

# ============================================================
# INT006 FINAL QUALITY AUDIT
# ============================================================

PATH = "03_Structured_Data/CSV/faq_INT006_final.csv"

df = pd.read_csv(PATH)

print("=" * 80)
print("INT006 FINAL QUALITY AUDIT")
print("=" * 80)


# ============================================================
# 1. BASIC VALIDATION
# ============================================================

print("\n" + "-" * 80)
print("1. BASIC VALIDATION")
print("-" * 80)

total_rows = len(df)

questions = (
    df["question"]
    .fillna("")
    .astype(str)
    .str.strip()
)

non_empty = questions.ne("").sum()
empty = total_rows - non_empty

print(f"Total rows: {total_rows}")
print(f"Non-empty questions: {non_empty}")
print(f"Empty questions: {empty}")

print("\nLanguages:")
print(df["language"].value_counts().to_dict())

print("\nQuestion types:")
print(df["question_type"].value_counts().to_dict())


# ============================================================
# 2. EXACT DUPLICATE CHECK
# ============================================================

print("\n" + "-" * 80)
print("2. EXACT DUPLICATE CHECK")
print("-" * 80)

duplicate_mask = questions.duplicated(keep=False)
duplicate_count = duplicate_mask.sum()

print(f"Exact duplicate rows: {duplicate_count}")

if duplicate_count == 0:
    print("✅ NO EXACT DUPLICATES")
else:
    print("⚠️ EXACT DUPLICATES FOUND")

    print(
        df.loc[
            duplicate_mask,
            ["faq_id", "language", "question_type", "question"]
        ].to_string(index=False)
    )


# ============================================================
# 3. SIMILARITY CHECK
# ============================================================

print("\n" + "-" * 80)
print("3. SIMILARITY CHECK")
print("-" * 80)

similar_pairs = []

for i in range(len(df)):

    q1 = str(df.iloc[i]["question"]).strip()

    if not q1:
        continue

    for j in range(i + 1, len(df)):

        q2 = str(df.iloc[j]["question"]).strip()

        if not q2:
            continue

        similarity = SequenceMatcher(
            None,
            q1.lower(),
            q2.lower()
        ).ratio()

        if similarity >= 0.85:

            similar_pairs.append({
                "faq_1": df.iloc[i]["faq_id"],
                "faq_2": df.iloc[j]["faq_id"],
                "language": df.iloc[i]["language"],
                "type_1": df.iloc[i]["question_type"],
                "type_2": df.iloc[j]["question_type"],
                "similarity": round(similarity, 3),
                "question_1": q1,
                "question_2": q2
            })


print(f"Potential similar pairs: {len(similar_pairs)}")

if similar_pairs:

    similarity_df = pd.DataFrame(similar_pairs)

    similarity_df = similarity_df.sort_values(
        "similarity",
        ascending=False
    )

    print("\nTop similar pairs:")

    print(
        similarity_df
        .head(10)
        .to_string(index=False)
    )

else:

    print("✅ NO HIGH-SIMILARITY PAIRS")


# ============================================================
# 4. EARLY CHECK-IN INTENT CHECK
# ============================================================

print("\n" + "-" * 80)
print("4. EARLY CHECK-IN INTENT CHECK")
print("-" * 80)


intent_patterns = {

    "English": [

        "early check-in",
        "early check in",

        "earlier check-in",
        "earlier check in",

        "earlier arrival",

        "arrive early",

        "check in early",
        "check-in early",

        "arrive before",

        "check in before",
        "check-in before",

        "before the standard check-in",
        "before the usual check-in",
        "before the normal check-in",

    ],


    "Arabic": [

        "تسجيل الوصول المبكر",
        "تسجيل الوصول مبكر",

        "الوصول مبكر",
        "وصول مبكر",

        "قبل موعد تسجيل الوصول",
        "قبل الوقت المعتاد",

        "دخول الغرفة مبكر",

    ],


    "French": [

        "check-in anticipé",
        "check in anticipé",

        "arrivée anticipée",

        "enregistrement anticipé",

        "enregistrement avant",

        "avant l'heure normale",

        "avant l'heure habituelle",

    ],


    "Russian": [

        "ранний заезд",

        "ранняя регистрация",
        "раннюю регистрацию",

        "зарегистрироваться раньше",

        "приехать раньше",

        "прибыть раньше",

        "заранее прибыть",
        "заранее приехать",

        "зарегистрироваться заезда заранее",

        "до обычного времени",

        "до времени обычного заезда",

        "раньше времени заезда",

    ]
}


intent_failures = []


for _, row in df.iterrows():

    question = str(row["question"]).strip().lower()

    language = row["language"]

    patterns = intent_patterns.get(language, [])

    matched = any(
        pattern.lower() in question
        for pattern in patterns
    )

    if not matched:

        intent_failures.append({

            "faq_id": row["faq_id"],
            "language": language,
            "question_type": row["question_type"],
            "question": row["question"]

        })


print(f"Intent failures: {len(intent_failures)}")


if len(intent_failures) == 0:

    print(
        "✅ ALL QUESTIONS CLEARLY RELATE TO EARLY CHECK-IN"
    )

else:

    print("⚠️ INTENT FAILURES FOUND")

    print(
        pd.DataFrame(intent_failures)
        .to_string(index=False)
    )


# ============================================================
# 5. QUESTION FORMAT CHECK
# ============================================================

print("\n" + "-" * 80)
print("5. QUESTION FORMAT CHECK")
print("-" * 80)

format_failures = []


for _, row in df.iterrows():

    question = str(row["question"]).strip()

    if not question:
        continue

    if not (
        question.endswith("?")
        or question.endswith("؟")
    ):

        format_failures.append({

            "faq_id": row["faq_id"],
            "language": row["language"],
            "question": row["question"]

        })


print(
    f"Questions without question mark: "
    f"{len(format_failures)}"
)


if len(format_failures) == 0:

    print(
        "✅ ALL QUESTIONS HAVE QUESTION MARKS"
    )

else:

    print("⚠️ FORMAT FAILURES FOUND")

    print(
        pd.DataFrame(format_failures)
        .to_string(index=False)
    )


# ============================================================
# 6. HOTEL NAME CHECK
# ============================================================

print("\n" + "-" * 80)
print("6. HOTEL NAME CHECK")
print("-" * 80)

hotel_failures = []


for _, row in df.iterrows():

    question = str(row["question"])

    if "Nile Pearl Hotel" not in question:

        hotel_failures.append({

            "faq_id": row["faq_id"],
            "language": row["language"],
            "question_type": row["question_type"],
            "question": row["question"]

        })


print(
    f"Questions without 'Nile Pearl Hotel': "
    f"{len(hotel_failures)}"
)


if len(hotel_failures) == 0:

    print(
        "✅ ALL QUESTIONS CONTAIN HOTEL NAME"
    )

else:

    print(
        "⚠️ QUESTIONS WITHOUT HOTEL NAME"
    )

    print(
        pd.DataFrame(hotel_failures)
        .to_string(index=False)
    )


# ============================================================
# 7. EXPECTED DISTRIBUTION CHECK
# ============================================================

print("\n" + "-" * 80)
print("7. EXPECTED DISTRIBUTION CHECK")
print("-" * 80)


expected_languages = {
    "English": 8,
    "Arabic": 8,
    "French": 8,
    "Russian": 8
}

expected_types = {
    "Direct": 12,
    "Paraphrased": 12,
    "Scenario": 8
}


actual_languages = df["language"].value_counts().to_dict()
actual_types = df["question_type"].value_counts().to_dict()


language_distribution_ok = (
    actual_languages == expected_languages
)

type_distribution_ok = (
    actual_types == expected_types
)


print(
    f"Language distribution correct: "
    f"{language_distribution_ok}"
)

print(
    f"Question type distribution correct: "
    f"{type_distribution_ok}"
)


if language_distribution_ok and type_distribution_ok:

    print("✅ DISTRIBUTION IS CORRECT")

else:

    print("⚠️ DISTRIBUTION NEEDS REVIEW")


# ============================================================
# 8. FINAL AUDIT SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("INT006 FINAL AUDIT SUMMARY")
print("=" * 80)

print(
    f"Total rows                  : {total_rows}"
)

print(
    f"Non-empty questions        : {non_empty}"
)

print(
    f"Empty questions            : {empty}"
)

print(
    f"Exact duplicate rows       : {duplicate_count}"
)

print(
    f"Similarity pairs >= 0.85   : {len(similar_pairs)}"
)

print(
    f"Intent failures             : {len(intent_failures)}"
)

print(
    f"Format failures             : {len(format_failures)}"
)

print(
    f"Hotel-name failures         : {len(hotel_failures)}"
)

print(
    f"Language distribution OK    : {language_distribution_ok}"
)

print(
    f"Question types distribution : {type_distribution_ok}"
)


# ============================================================
# FINAL STATUS
# ============================================================

core_checks_passed = (

    total_rows == 32

    and non_empty == 32

    and empty == 0

    and duplicate_count == 0

    and len(intent_failures) == 0

    and len(format_failures) == 0

    and len(hotel_failures) == 0

    and language_distribution_ok

    and type_distribution_ok
)


print("\n" + "=" * 80)


if core_checks_passed:

    print("✅ INT006 PASSED ALL CORE QUALITY CHECKS")
    print("✅ INT006 READY FOR FINAL DATASET")

else:

    print("⚠️ INT006 NEEDS REVIEW")


print("=" * 80)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 01 PREPARATION
# Add ONLY 32 Questions
# ============================================================

from pathlib import Path
import pandas as pd
import re
import shutil

print("=" * 100)
print("🏨 HOTEL AI - BATCH 01 | 32 QUESTIONS")
print("=" * 100)

# ------------------------------------------------------------
# 1. USE DATASET PATH FROM PREVIOUS CELL
# ------------------------------------------------------------

if "DATASET_PATH" not in globals():
    raise RuntimeError(
        "❌ DATASET_PATH is not initialized.\n"
        "Run the PROJECT AUTO DETECTION cell first."
    )

DATASET_PATH = Path(DATASET_PATH)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"❌ Dataset not found:\n{DATASET_PATH}"
    )

# ------------------------------------------------------------
# 2. LOAD CURRENT DATASET
# ------------------------------------------------------------

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print(f"📂 Dataset: {DATASET_PATH}")
print(f"📊 Current rows: {len(df)}")

# ------------------------------------------------------------
# 3. VALIDATE CURRENT DATASET
# ------------------------------------------------------------

required_columns = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer"
]

missing = [
    c for c in required_columns
    if c not in df.columns
]

if missing:
    raise ValueError(
        f"❌ Missing columns:\n{missing}"
    )

# ------------------------------------------------------------
# 4. FIND NEXT FAQ ID SAFELY
# ------------------------------------------------------------

faq_numbers = (
    df["faq_id"]
    .astype(str)
    .str.extract(r"(\d+)", expand=False)
    .astype(int)
)

max_faq_number = faq_numbers.max()
next_faq_number = max_faq_number + 1

batch_size = 32

batch_start = next_faq_number
batch_end = batch_start + batch_size - 1

print("\n" + "=" * 100)
print("🆔 FAQ ID RANGE FOR THIS BATCH")
print("=" * 100)

print(f"Existing maximum : FAQ{max_faq_number:04d}")
print(f"Batch start      : FAQ{batch_start:04d}")
print(f"Batch end        : FAQ{batch_end:04d}")
print(f"Batch size       : {batch_size}")

# ------------------------------------------------------------
# 5. BALANCED BATCH PLAN
# ------------------------------------------------------------

languages = [
    "Arabic",
    "English",
    "French",
    "Russian"
]

intents = (
    df[
        ["intent_id", "intent_name"]
    ]
    .drop_duplicates()
    .sort_values("intent_id")
)

print("\n" + "=" * 100)
print("🎯 BATCH PLAN")
print("=" * 100)

print(
    f"\n8 intents × 4 languages = {len(intents) * len(languages)} questions"
)

# ------------------------------------------------------------
# 6. BUILD EMPTY BATCH TEMPLATE
# ------------------------------------------------------------

batch_rows = []

current_id = batch_start

for _, intent_row in intents.iterrows():

    intent_id = intent_row["intent_id"]
    intent_name = intent_row["intent_name"]

    for language in languages:

        batch_rows.append({
            "faq_id": f"FAQ{current_id:04d}",
            "intent_id": intent_id,
            "language": language,
            "question_type": None,
            "question": None,
            "status": "generated",
            "intent_name": intent_name,
            "ground_truth": None,
            "category": None,
            "source": "batch_01",
            "attempts": 0,
            "difficulty": "Easy",
            "answer": None
        })

        current_id += 1

batch_df = pd.DataFrame(
    batch_rows,
    columns=required_columns
)

# ------------------------------------------------------------
# 7. VALIDATE BATCH SIZE
# ------------------------------------------------------------

assert len(batch_df) == 32

print("\n" + "=" * 100)
print("📊 BATCH STRUCTURE")
print("=" * 100)

print(f"Questions: {len(batch_df)}")

print("\nIntent distribution:")
print(
    batch_df["intent_name"].value_counts()
)

print("\nLanguage distribution:")
print(
    batch_df["language"].value_counts()
)

# ------------------------------------------------------------
# 8. CREATE BATCH DIRECTORY
# ------------------------------------------------------------

project_root = DATASET_PATH

# /final/hotel_faq_final.csv
# project root = .../CSV/final
final_dir = DATASET_PATH.parent

batch_dir = final_dir / "batches"

batch_dir.mkdir(
    parents=True,
    exist_ok=True
)

batch_path = (
    batch_dir
    / "hotel_faq_batch_01_32.csv"
)

# ------------------------------------------------------------
# 9. SAVE TEMPLATE
# ------------------------------------------------------------

batch_df.to_csv(
    batch_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 100)
print("💾 BATCH TEMPLATE CREATED")
print("=" * 100)

print(f"📄 {batch_path}")

# ------------------------------------------------------------
# 10. DISPLAY BATCH
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("📋 BATCH 01")
print("=" * 100)

display(batch_df)

print("\n" + "=" * 100)
print("✅ BATCH 01 READY")
print("=" * 100)

print("""
32 سؤال فقط.

❗ لم يتم تعديل hotel_faq_final.csv.
❗ لم يتم دمج الـ32 سؤال.
❗ الـbatch محفوظ منفصلًا للمراجعة.

Next step:
سنملأ الـ32 سؤال فعليًا مع الحفاظ على:
- Intent correctness
- Language
- Question diversity
- Ground truth
- Answer
- Question type
- No duplicates
""")
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 02
# CONTINUE FROM FAQ0256
# GENERATE EXACTLY 32 NEW QUESTIONS
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata

print("=" * 100)
print("🏨 HOTEL AI - BATCH 02 — CONTINUE")
print("=" * 100)

# ============================================================
# 1. PROJECT
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"❌ Dataset not found:\n{DATASET_PATH}"
    )

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)

print(f"📂 Dataset: {DATASET_PATH}")
print(f"📊 Existing questions: {len(df)}")

if len(df) != 256:
    print(
        f"⚠️ Warning: expected 256 existing rows, "
        f"found {len(df)}."
    )

# ============================================================
# 3. NORMALIZATION
# ============================================================

def normalize_text(text):

    if pd.isna(text):
        return ""

    text = str(text).strip().lower()

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    text = re.sub(
        r"[?!.,;:]+",
        "",
        text
    )

    return text


existing_questions = set(
    df["question"].map(normalize_text)
)

# ============================================================
# 4. CONTINUE FAQ IDs
# ============================================================

def faq_number(value):

    match = re.search(
        r"(\d+)",
        str(value)
    )

    return int(match.group(1)) if match else 0


last_id = max(
    df["faq_id"].map(faq_number)
)

start_id = last_id + 1
end_id = start_id + 31

print("\n" + "=" * 100)
print("🆔 FAQ ID CONTINUATION")
print("=" * 100)

print(f"Last existing : FAQ{last_id:04d}")
print(f"Next batch    : FAQ{start_id:04d} → FAQ{end_id:04d}")

# ============================================================
# 5. BATCH 02 QUESTIONS
#
# IMPORTANT:
# These are NEW phrasings.
# ============================================================

questions = [

# ------------------------------------------------------------
# INT001 - HOTEL LOCATION
# ------------------------------------------------------------

("INT001", "hotel_location", "Arabic", "Paraphrased",
 "في أي منطقة يقع فندق Nile Pearl Hotel؟"),

("INT001", "hotel_location", "English", "Paraphrased",
 "Which area is Nile Pearl Hotel situated in?"),

("INT001", "hotel_location", "French", "Paraphrased",
 "Dans quel quartier se trouve l’hôtel Nile Pearl Hotel ?"),

("INT001", "hotel_location", "Russian", "Paraphrased",
 "В каком районе расположен отель Nile Pearl Hotel?"),

# ------------------------------------------------------------
# INT002 - HOTEL CATEGORY
# ------------------------------------------------------------

("INT002", "hotel_category", "Arabic", "Paraphrased",
 "إلى أي فئة ينتمي فندق Nile Pearl Hotel؟"),

("INT002", "hotel_category", "English", "Paraphrased",
 "Which hotel category does Nile Pearl Hotel belong to?"),

("INT002", "hotel_category", "French", "Paraphrased",
 "À quelle catégorie appartient l’hôtel Nile Pearl Hotel ?"),

("INT002", "hotel_category", "Russian", "Paraphrased",
 "К какой категории относится отель Nile Pearl Hotel?"),

# ------------------------------------------------------------
# INT003 - RECEPTION AVAILABILITY
# ------------------------------------------------------------

("INT003", "reception_availability", "Arabic", "Direct",
 "هل يمكن التواصل مع مكتب الاستقبال في أي وقت؟"),

("INT003", "reception_availability", "English", "Direct",
 "Can I contact the hotel reception at any time?"),

("INT003", "reception_availability", "French", "Direct",
 "Puis-je contacter la réception de l’hôtel à tout moment ?"),

("INT003", "reception_availability", "Russian", "Direct",
 "Можно ли связаться со стойкой регистрации в любое время?"),

# ------------------------------------------------------------
# INT004 - CHECK IN TIME
# ------------------------------------------------------------

("INT004", "check_in_time", "Arabic", "Paraphrased",
 "متى يبدأ وقت الوصول وتسجيل الدخول للفندق؟"),

("INT004", "check_in_time", "English", "Paraphrased",
 "When does the hotel's check-in period begin?"),

("INT004", "check_in_time", "French", "Paraphrased",
 "Quand commence la période d’enregistrement à l’hôtel ?"),

("INT004", "check_in_time", "Russian", "Paraphrased",
 "Когда начинается время регистрации заезда в отель?"),

# ------------------------------------------------------------
# INT005 - CHECK OUT TIME
# ------------------------------------------------------------

("INT005", "check_out_time", "Direct",
 "ما الوقت المحدد لتسجيل الخروج من الفندق؟"),

("INT005", "check_out_time", "English", "Direct",
 "What is the designated check-out time at the hotel?"),

("INT005", "check_out_time", "French", "Direct",
 "Quelle est l’heure prévue pour le départ de l’hôtel ?"),

("INT005", "check_out_time", "Russian", "Direct",
 "Какое время установлено для выезда из отеля?"),

# ------------------------------------------------------------
# INT006 - EARLY CHECK IN
# ------------------------------------------------------------

("INT006", "early_check_in", "Arabic", "Scenario",
 "إذا وصلت قبل وقت تسجيل الدخول، هل يمكنني دخول غرفتي مبكرًا؟"),

("INT006", "early_check_in", "English", "Scenario",
 "If I arrive before check-in time, can I access my room early?"),

("INT006", "early_check_in", "French", "Scenario",
 "Si j’arrive avant l’heure d’enregistrement, puis-je accéder à ma chambre plus tôt ?"),

("INT006", "early_check_in", "Russian", "Scenario",
 "Если я приеду до времени заезда, смогу ли я попасть в номер раньше?"),

# ------------------------------------------------------------
# INT007 - LATE CHECK OUT
# ------------------------------------------------------------

("INT007", "late_check_out", "Arabic", "Scenario",
 "هل يمكنني الاحتفاظ بالغرفة حتى وقت متأخر من يوم المغادرة؟"),

("INT007", "late_check_out", "English", "Scenario",
 "Can I keep my room until later on the day of departure?"),

("INT007", "late_check_out", "French", "Scenario",
 "Puis-je garder ma chambre plus tard le jour de mon départ ?"),

("INT007", "late_check_out", "Russian", "Scenario",
 "Могу ли я оставить номер до более позднего времени в день выезда?"),

# ------------------------------------------------------------
# INT008 - IDENTIFICATION REQUIRED
# ------------------------------------------------------------

("INT008", "identification_required", "Direct",
 "ما المستند الذي يجب تقديمه لإثبات الهوية عند الوصول؟"),

("INT008", "identification_required", "English", "Direct",
 "Which identification document should I present when I arrive?"),

("INT008", "identification_required", "French", "Direct",
 "Quel document d’identité dois-je présenter à mon arrivée ?"),

("INT008", "identification_required", "Russian", "Direct",
 "Какой документ удостоверяющий личность нужно предъявить при прибытии?"),
]

# ============================================================
# 6. FIXED VALIDATION
# ============================================================

if len(questions) != 32:
    raise ValueError(
        f"❌ Expected 32 questions, got {len(questions)}"
    )

print("\n✅ Exactly 32 questions prepared.")

# ============================================================
# 7. CHECK DUPLICATES INSIDE BATCH
# ============================================================

batch_questions = [
    normalize_text(q[4])
    for q in questions
]

duplicate_inside_batch = (
    len(batch_questions)
    != len(set(batch_questions))
)

if duplicate_inside_batch:

    seen = set()

    print("\n❌ DUPLICATES INSIDE BATCH:")

    for q in questions:

        normalized = normalize_text(q[4])

        if normalized in seen:
            print("-", q[4])

        seen.add(normalized)

    raise ValueError(
        "❌ Batch contains duplicate questions."
    )

print("✅ No duplicates inside Batch 02.")

# ============================================================
# 8. CHECK AGAINST EXISTING 256
# ============================================================

conflicts = []

for q in questions:

    normalized = normalize_text(q[4])

    if normalized in existing_questions:
        conflicts.append(q[4])


if conflicts:

    print("\n❌ EXISTING QUESTION CONFLICTS:")

    for q in conflicts:
        print("-", q)

    raise ValueError(
        "❌ One or more questions already exist."
    )

print("✅ No conflicts with existing dataset.")

# ============================================================
# 9. LANGUAGE BALANCE
# ============================================================

language_counts = {}

for q in questions:

    language = q[2]

    language_counts[language] = (
        language_counts.get(language, 0) + 1
    )

print("\n" + "=" * 100)
print("🌍 LANGUAGE DISTRIBUTION")
print("=" * 100)

for language, count in language_counts.items():

    print(
        f"{language:<10} {count}"
    )

if set(language_counts.values()) != {8}:

    raise ValueError(
        "❌ Language balance is not 8 questions per language."
    )

# ============================================================
# 10. INTENT BALANCE
# ============================================================

intent_counts = {}

for q in questions:

    intent = q[0]

    intent_counts[intent] = (
        intent_counts.get(intent, 0) + 1
    )

print("\n" + "=" * 100)
print("🎯 INTENT DISTRIBUTION")
print("=" * 100)

for intent, count in sorted(intent_counts.items()):

    print(
        f"{intent:<8} {count}"
    )

if len(intent_counts) != 8:

    raise ValueError(
        "❌ Expected 8 intents."
    )

if set(intent_counts.values()) != {4}:

    raise ValueError(
        "❌ Every intent must have exactly 4 questions."
    )

# ============================================================
# 11. CREATE BATCH
# ============================================================

batch_rows = []

for i, q in enumerate(questions):

    intent_id = q[0]
    intent_name = q[1]
    language = q[2]
    question_type = q[3]
    question = q[4]

    batch_rows.append({

        "faq_id":
            f"FAQ{start_id + i:04d}",

        "intent_id":
            intent_id,

        "language":
            language,

        "question_type":
            question_type,

        "question":
            question,

        "status":
            "generated",

        "intent_name":
            intent_name,

        "ground_truth":
            "",

        "category":
            "Hotel FAQ",

        "source":
            "generated_batch_02",

        "attempts":
            0,

        "difficulty":
            "Easy",

        "answer":
            ""
    })


batch_df = pd.DataFrame(
    batch_rows,
    columns=[
        "faq_id",
        "intent_id",
        "language",
        "question_type",
        "question",
        "status",
        "intent_name",
        "ground_truth",
        "category",
        "source",
        "attempts",
        "difficulty",
        "answer"
    ]
)

# ============================================================
# 12. SAVE SEPARATE BATCH
# ============================================================

BATCH_DIR = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
)

BATCH_DIR.mkdir(
    parents=True,
    exist_ok=True
)

BATCH_PATH = (
    BATCH_DIR
    / "faq_batch_02.csv"
)

batch_df.to_csv(
    BATCH_PATH,
    index=False,
    encoding="utf-8-sig"
)

# ============================================================
# 13. FINAL VALIDATION
# ============================================================

assert len(batch_df) == 32

assert (
    batch_df["faq_id"].nunique()
    == 32
)

assert (
    batch_df["question"].map(normalize_text).nunique()
    == 32
)

# ============================================================
# 14. REPORT
# ============================================================

print("\n" + "=" * 100)
print("🎉 BATCH 02 READY")
print("=" * 100)

print(
    f"Questions : {len(batch_df)}"
)

print(
    f"IDs       : FAQ{start_id:04d} → FAQ{end_id:04d}"
)

print("\n🌍 Languages:")

print(
    batch_df["language"].value_counts()
)

print("\n🎯 Intents:")

print(
    batch_df.groupby(
        ["intent_id", "intent_name"]
    ).size()
)

print("\n📝 Question Types:")

print(
    batch_df["question_type"].value_counts()
)

print("\n📄 Saved:")

print(BATCH_PATH)

print("\n" + "=" * 100)
print("🛡️ ORIGINAL DATASET PROTECTED")
print("=" * 100)

print("❗ hotel_faq_final.csv was NOT modified.")
print("❗ Existing 256 questions remain untouched.")
print("❗ Only Batch 02 was created.")
print("❗ Next step: review + fill ground_truth/answer.")
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 02 VALIDATION
# ============================================================

print("=" * 100)
print("🔎 BATCH 02 VALIDATION")
print("=" * 100)

# ------------------------------------------------------------
# Basic checks
# ------------------------------------------------------------

assert len(batch_df) == 32, (
    f"❌ Expected 32 questions, found {len(batch_df)}"
)

# ------------------------------------------------------------
# FAQ IDs
# ------------------------------------------------------------

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(257, 289)
]

actual_ids = batch_df["faq_id"].tolist()

assert actual_ids == expected_ids, (
    "❌ FAQ ID sequence is incorrect."
)

print("✅ FAQ IDs: FAQ0257 → FAQ0288")

# ------------------------------------------------------------
# Duplicate IDs
# ------------------------------------------------------------

duplicate_ids = batch_df[
    batch_df["faq_id"].duplicated(keep=False)
]

assert duplicate_ids.empty, (
    f"❌ Duplicate FAQ IDs found:\n{duplicate_ids}"
)

print("✅ No duplicate FAQ IDs.")

# ------------------------------------------------------------
# Normalize questions
# ------------------------------------------------------------

def normalize_question(text):

    text = str(text).strip().lower()

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    text = re.sub(
        r"[?!.,;:]+",
        "",
        text
    )

    return text


batch_normalized = (
    batch_df["question"]
    .map(normalize_question)
)

# ------------------------------------------------------------
# Duplicates inside batch
# ------------------------------------------------------------

duplicate_questions = batch_df[
    batch_normalized.duplicated(keep=False)
]

assert duplicate_questions.empty, (
    f"❌ Duplicate questions inside Batch 02:\n"
    f"{duplicate_questions[['faq_id', 'question']]}"
)

print("✅ No duplicate questions inside Batch 02.")

# ------------------------------------------------------------
# Compare against existing 256
# ------------------------------------------------------------

existing_normalized = set(
    df["question"]
    .map(normalize_question)
)

conflicts = batch_df[
    batch_normalized.isin(existing_normalized)
]

assert conflicts.empty, (
    f"❌ Questions already exist in dataset:\n"
    f"{conflicts[['faq_id', 'question']]}"
)

print("✅ No Batch 02 question already exists.")

# ------------------------------------------------------------
# Empty questions
# ------------------------------------------------------------

empty_questions = batch_df[
    batch_df["question"].isna()
    |
    (
        batch_df["question"]
        .astype(str)
        .str.strip()
        .eq("")
    )
]

assert empty_questions.empty, (
    "❌ Empty questions found."
)

print("✅ No empty questions.")

# ------------------------------------------------------------
# Language balance
# ------------------------------------------------------------

language_counts = (
    batch_df["language"]
    .value_counts()
)

print("\n🌍 LANGUAGE DISTRIBUTION")
print(language_counts)

expected_languages = {
    "Arabic": 8,
    "English": 8,
    "French": 8,
    "Russian": 8
}

assert language_counts.to_dict() == expected_languages, (
    "❌ Language distribution is incorrect."
)

print("✅ Perfect language balance.")

# ------------------------------------------------------------
# Intent balance
# ------------------------------------------------------------

intent_counts = (
    batch_df["intent_id"]
    .value_counts()
    .sort_index()
)

print("\n🎯 INTENT DISTRIBUTION")
print(intent_counts)

assert len(intent_counts) == 8

assert all(
    count == 4
    for count in intent_counts
), (
    "❌ Every intent must contain exactly 4 questions."
)

print("✅ Perfect intent balance.")

# ------------------------------------------------------------
# Intent ↔ Intent Name
# ------------------------------------------------------------

intent_mapping = (
    batch_df
    .groupby("intent_id")["intent_name"]
    .nunique()
)

assert all(intent_mapping == 1), (
    "❌ Intent ID maps to multiple intent names."
)

print("✅ Intent ID ↔ Intent Name consistency.")

# ------------------------------------------------------------
# Question types
# ------------------------------------------------------------

print("\n📝 QUESTION TYPES")

print(
    batch_df["question_type"]
    .value_counts()
)

# ------------------------------------------------------------
# Final result
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("🎉 BATCH 02 VALIDATION PASSED")
print("=" * 100)

print("""
✅ 32 NEW QUESTIONS
✅ IDs FAQ0257 → FAQ0288
✅ No duplicate IDs
✅ No duplicate questions
✅ No overlap with existing 256
✅ No empty questions
✅ 8 questions × 4 languages
✅ 4 questions × 8 intents
✅ Intent mapping valid

🛡️ hotel_faq_final.csv is still untouched.
""")

# ------------------------------------------------------------
# Save validated batch
# ------------------------------------------------------------

BATCH_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_02_validated.csv"
)

BATCH_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)

batch_df.to_csv(
    BATCH_PATH,
    index=False,
    encoding="utf-8-sig"
)

print(f"📄 Batch saved:\n{BATCH_PATH}")
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 02
# CONTINUE FROM EXISTING 256 QUESTIONS
# Generate ONLY FAQ0257 → FAQ0288
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata


# ============================================================
# 1. PROJECT AUTO DETECTION
# ============================================================

print("=" * 100)
print("🏨 HOTEL AI - BATCH 02 GENERATION")
print("=" * 100)

possible_roots = [
    Path("/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"),
    Path.home() / "Desktop" / "HotelAI-Intelligent-Agent",
    Path.home() / "Desktop" / "HotelAI",
]

PROJECT_ROOT = None

for root in possible_roots:
    if root.exists():
        PROJECT_ROOT = root
        break

if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "❌ HotelAI project folder was not found.\n\n"
        "Checked locations:\n"
        + "\n".join(str(x) for x in possible_roots)
    )

print(f"📂 Project root:\n{PROJECT_ROOT}")


# ============================================================
# 2. FINAL DATASET PATH
# ============================================================

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"❌ Final dataset not found:\n{DATASET_PATH}"
    )

print(f"\n✅ Final dataset found:")
print(DATASET_PATH)


# ============================================================
# 3. LOAD EXISTING DATASET
# ============================================================

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print("\n" + "=" * 100)
print("📊 EXISTING DATASET")
print("=" * 100)

print(f"Rows    : {len(df)}")
print(f"Columns : {len(df.columns)}")

print("\nColumns:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")


# ============================================================
# 4. REQUIRED COLUMNS
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

missing_columns = [
    col
    for col in REQUIRED_COLUMNS
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        "❌ Missing required columns:\n"
        + "\n".join(missing_columns)
    )

print("\n✅ All required columns exist.")


# ============================================================
# 5. CURRENT DATASET VALIDATION
# ============================================================

if len(df) != 256:
    raise ValueError(
        f"❌ Expected exactly 256 existing questions.\n"
        f"Found: {len(df)}"
    )

print("\n✅ Existing dataset contains exactly 256 questions.")


# ============================================================
# 6. FAQ ID EXTRACTION
# ============================================================

def extract_faq_number(value):

    match = re.fullmatch(
        r"FAQ(\d+)",
        str(value).strip()
    )

    if not match:
        return None

    return int(match.group(1))


faq_numbers = df["faq_id"].apply(
    extract_faq_number
)

if faq_numbers.isna().any():

    invalid_ids = df.loc[
        faq_numbers.isna(),
        "faq_id"
    ].tolist()

    raise ValueError(
        f"❌ Invalid FAQ IDs:\n{invalid_ids}"
    )


# ============================================================
# 7. LAST FAQ ID
# ============================================================

last_faq_number = int(
    faq_numbers.max()
)

if last_faq_number != 256:
    raise ValueError(
        f"❌ Expected last FAQ ID = FAQ0256.\n"
        f"Found = FAQ{last_faq_number:04d}"
    )

print("\n" + "=" * 100)
print("🆔 FAQ ID CONTINUATION")
print("=" * 100)

print("Current rows : 256")
print("Last FAQ ID  : FAQ0256")


# ============================================================
# 8. BATCH SETTINGS
# ============================================================

BATCH_SIZE = 32

START_ID = last_faq_number + 1
END_ID = START_ID + BATCH_SIZE - 1

NEW_IDS = [
    f"FAQ{i:04d}"
    for i in range(
        START_ID,
        END_ID + 1
    )
]

if len(NEW_IDS) != BATCH_SIZE:
    raise AssertionError(
        f"❌ Internal error: expected 32 IDs, "
        f"got {len(NEW_IDS)}"
    )


print("\n" + "=" * 100)
print("🚀 NEXT BATCH")
print("=" * 100)

print(f"Batch size : {BATCH_SIZE}")
print(f"IDs        : {NEW_IDS[0]} → {NEW_IDS[-1]}")


# ============================================================
# 9. NORMALIZATION
# ============================================================

def normalize_question(text):

    if pd.isna(text):
        return ""

    text = str(text)

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = text.strip().lower()

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    text = re.sub(
        r"[؟?!.,;:]+",
        "",
        text
    )

    return text


existing_questions = set(
    df["question"]
    .map(normalize_question)
)


# ============================================================
# 10. EXISTING DATASET SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("📊 EXISTING DATASET SUMMARY")
print("=" * 100)

print("\n🌍 Languages:")
print(
    df["language"]
    .value_counts()
    .sort_index()
)

print("\n🎯 Intents:")
print(
    df.groupby(
        ["intent_id", "intent_name"]
    )
    .size()
    .reset_index(name="questions")
    .sort_values("intent_id")
)

print("\n📝 Question Types:")
print(
    df["question_type"]
    .value_counts()
)


# ============================================================
# 11. BATCH CONTENT
#
# 8 intents × 4 languages = 32 questions
# Each intent gets exactly 4 questions.
# ============================================================

batch_content = [

    # --------------------------------------------------------
    # INT001 - hotel_location
    # --------------------------------------------------------

    {
        "intent_id": "INT001",
        "intent_name": "hotel_location",
        "language": "Arabic",
        "question_type": "Paraphrased",
        "question": "في أي منطقة من المدينة يقع فندق Nile Pearl Hotel؟",
        "ground_truth": "The question asks about the location of Nile Pearl Hotel.",
        "answer": "Nile Pearl Hotel is located in the specified hotel area of the city."
    },

    {
        "intent_id": "INT001",
        "intent_name": "hotel_location",
        "language": "English",
        "question_type": "Paraphrased",
        "question": "Which part of the city is Nile Pearl Hotel located in?",
        "ground_truth": "The question asks about the location of Nile Pearl Hotel.",
        "answer": "Nile Pearl Hotel is located in the specified hotel area of the city."
    },

    {
        "intent_id": "INT001",
        "intent_name": "hotel_location",
        "language": "French",
        "question_type": "Paraphrased",
        "question": "Dans quel quartier de la ville se trouve le Nile Pearl Hotel ?",
        "ground_truth": "La question porte sur l'emplacement du Nile Pearl Hotel.",
        "answer": "Le Nile Pearl Hotel se trouve dans la zone indiquée de la ville."
    },

    {
        "intent_id": "INT001",
        "intent_name": "hotel_location",
        "language": "Russian",
        "question_type": "Paraphrased",
        "question": "В каком районе города находится отель Nile Pearl Hotel?",
        "ground_truth": "Вопрос касается расположения отеля Nile Pearl Hotel.",
        "answer": "Отель Nile Pearl Hotel расположен в указанной части города."
    },


    # --------------------------------------------------------
    # INT002 - hotel_category
    # --------------------------------------------------------

    {
        "intent_id": "INT002",
        "intent_name": "hotel_category",
        "language": "Arabic",
        "question_type": "Paraphrased",
        "question": "ما فئة فندق Nile Pearl Hotel؟",
        "ground_truth": "The question asks about the category of Nile Pearl Hotel.",
        "answer": "Nile Pearl Hotel belongs to the hotel category specified in the hotel information."
    },

    {
        "intent_id": "INT002",
        "intent_name": "hotel_category",
        "language": "English",
        "question_type": "Paraphrased",
        "question": "What hotel category is Nile Pearl Hotel?",
        "ground_truth": "The question asks about the category of Nile Pearl Hotel.",
        "answer": "Nile Pearl Hotel belongs to the hotel category specified in the hotel information."
    },

    {
        "intent_id": "INT002",
        "intent_name": "hotel_category",
        "language": "French",
        "question_type": "Paraphrased",
        "question": "Quelle est la catégorie du Nile Pearl Hotel ?",
        "ground_truth": "La question porte sur la catégorie du Nile Pearl Hotel.",
        "answer": "Le Nile Pearl Hotel appartient à la catégorie indiquée dans les informations de l'hôtel."
    },

    {
        "intent_id": "INT002",
        "intent_name": "hotel_category",
        "language": "Russian",
        "question_type": "Paraphrased",
        "question": "Какова категория отеля Nile Pearl Hotel?",
        "ground_truth": "Вопрос касается категории отеля Nile Pearl Hotel.",
        "answer": "Отель Nile Pearl Hotel относится к категории, указанной в информации об отеле."
    },


    # --------------------------------------------------------
    # INT003 - reception_availability
    # --------------------------------------------------------

    {
        "intent_id": "INT003",
        "intent_name": "reception_availability",
        "language": "Arabic",
        "question_type": "Paraphrased",
        "question": "هل مكتب استقبال الفندق يعمل على مدار الساعة؟",
        "ground_truth": "The question asks whether the hotel reception is available around the clock.",
        "answer": "The hotel's reception availability is provided in the hotel reception policy."
    },

    {
        "intent_id": "INT003",
        "intent_name": "reception_availability",
        "language": "English",
        "question_type": "Paraphrased",
        "question": "Does the hotel reception operate 24 hours a day?",
        "ground_truth": "The question asks whether the hotel reception is available around the clock.",
        "answer": "The hotel's reception availability is provided in the hotel reception policy."
    },

    {
        "intent_id": "INT003",
        "intent_name": "reception_availability",
        "language": "French",
        "question_type": "Paraphrased",
        "question": "La réception de l'hôtel est-elle ouverte 24 heures sur 24 ?",
        "ground_truth": "La question porte sur la disponibilité de la réception de l'hôtel.",
        "answer": "Les horaires de disponibilité de la réception sont indiqués dans la politique de l'hôtel."
    },

    {
        "intent_id": "INT003",
        "intent_name": "reception_availability",
        "language": "Russian",
        "question_type": "Paraphrased",
        "question": "Работает ли стойка регистрации отеля круглосуточно?",
        "ground_truth": "Вопрос касается круглосуточной работы стойки регистрации.",
        "answer": "Время работы стойки регистрации указано в правилах отеля."
    },


    # --------------------------------------------------------
    # INT004 - check_in_time
    # --------------------------------------------------------

    {
        "intent_id": "INT004",
        "intent_name": "check_in_time",
        "language": "Arabic",
        "question_type": "Paraphrased",
        "question": "ما هو الوقت المحدد لتسجيل الوصول إلى الفندق؟",
        "ground_truth": "The question asks for the standard hotel check-in time.",
        "answer": "The standard check-in time is provided in the hotel's check-in policy."
    },

    {
        "intent_id": "INT004",
        "intent_name": "check_in_time",
        "language": "English",
        "question_type": "Paraphrased",
        "question": "What is the hotel's standard check-in time?",
        "ground_truth": "The question asks for the standard hotel check-in time.",
        "answer": "The standard check-in time is provided in the hotel's check-in policy."
    },

    {
        "intent_id": "INT004",
        "intent_name": "check_in_time",
        "language": "French",
        "question_type": "Paraphrased",
        "question": "Quelle est l'heure habituelle d'arrivée à l'hôtel ?",
        "ground_truth": "La question demande l'heure standard d'enregistrement à l'hôtel.",
        "answer": "L'heure standard d'enregistrement est indiquée dans la politique d'arrivée de l'hôtel."
    },

    {
        "intent_id": "INT004",
        "intent_name": "check_in_time",
        "language": "Russian",
        "question_type": "Paraphrased",
        "question": "Какое стандартное время регистрации заезда в отеле?",
        "ground_truth": "Вопрос касается стандартного времени регистрации заезда.",
        "answer": "Стандартное время заезда указано в правилах регистрации отеля."
    },


    # --------------------------------------------------------
    # INT005 - check_out_time
    # --------------------------------------------------------

    {
        "intent_id": "INT005",
        "intent_name": "check_out_time",
        "language": "Arabic",
        "question_type": "Paraphrased",
        "question": "ما هو الوقت المعتاد لتسجيل المغادرة من الفندق؟",
        "ground_truth": "The question asks for the standard hotel check-out time.",
        "answer": "The standard check-out time is provided in the hotel's check-out policy."
    },

    {
        "intent_id": "INT005",
        "intent_name": "check_out_time",
        "language": "English",
        "question_type": "Paraphrased",
        "question": "What is the hotel's standard check-out time?",
        "ground_truth": "The question asks for the standard hotel check-out time.",
        "answer": "The standard check-out time is provided in the hotel's check-out policy."
    },

    {
        "intent_id": "INT005",
        "intent_name": "check_out_time",
        "language": "French",
        "question_type": "Paraphrased",
        "question": "Quelle est l'heure habituelle de départ de l'hôtel ?",
        "ground_truth": "La question demande l'heure standard de départ de l'hôtel.",
        "answer": "L'heure standard de départ est indiquée dans la politique de départ de l'hôtel."
    },

    {
        "intent_id": "INT005",
        "intent_name": "check_out_time",
        "language": "Russian",
        "question_type": "Paraphrased",
        "question": "Какое стандартное время регистрации выезда из отеля?",
        "ground_truth": "Вопрос касается стандартного времени выезда.",
        "answer": "Стандартное время выезда указано в правилах отеля."
    },


    # --------------------------------------------------------
    # INT006 - early_check_in
    # --------------------------------------------------------

    {
        "intent_id": "INT006",
        "intent_name": "early_check_in",
        "language": "Arabic",
        "question_type": "Paraphrased",
        "question": "هل يسمح الفندق بتسجيل الوصول في وقت مبكر؟",
        "ground_truth": "The question asks whether early check-in is allowed.",
        "answer": "Early check-in is subject to the hotel's early check-in policy and availability."
    },

    {
        "intent_id": "INT006",
        "intent_name": "early_check_in",
        "language": "English",
        "question_type": "Paraphrased",
        "question": "Does the hotel allow early check-in?",
        "ground_truth": "The question asks whether early check-in is allowed.",
        "answer": "Early check-in is subject to the hotel's early check-in policy and availability."
    },

    {
        "intent_id": "INT006",
        "intent_name": "early_check_in",
        "language": "French",
        "question_type": "Paraphrased",
        "question": "L'hôtel autorise-t-il l'enregistrement anticipé ?",
        "ground_truth": "La question porte sur la possibilité d'un enregistrement anticipé.",
        "answer": "L'enregistrement anticipé dépend de la politique de l'hôtel et des disponibilités."
    },

    {
        "intent_id": "INT006",
        "intent_name": "early_check_in",
        "language": "Russian",
        "question_type": "Paraphrased",
        "question": "Разрешена ли в отеле ранняя регистрация заезда?",
        "ground_truth": "Вопрос касается возможности раннего заезда.",
        "answer": "Ранний заезд зависит от правил отеля и наличия свободных номеров."
    },


    # --------------------------------------------------------
    # INT007 - late_check_out
    # --------------------------------------------------------

    {
        "intent_id": "INT007",
        "intent_name": "late_check_out",
        "language": "Arabic",
        "question_type": "Paraphrased",
        "question": "هل يمكنني طلب تسجيل المغادرة في وقت متأخر؟",
        "ground_truth": "The question asks whether late check-out can be requested.",
        "answer": "Late check-out is subject to the hotel's policy and availability."
    },

    {
        "intent_id": "INT007",
        "intent_name": "late_check_out",
        "language": "English",
        "question_type": "Paraphrased",
        "question": "Can I request a late check-out?",
        "ground_truth": "The question asks whether late check-out can be requested.",
        "answer": "Late check-out is subject to the hotel's policy and availability."
    },

    {
        "intent_id": "INT007",
        "intent_name": "late_check_out",
        "language": "French",
        "question_type": "Paraphrased",
        "question": "Puis-je demander un départ tardif de l'hôtel ?",
        "ground_truth": "La question porte sur la possibilité de demander un départ tardif.",
        "answer": "Le départ tardif dépend de la politique de l'hôtel et des disponibilités."
    },

    {
        "intent_id": "INT007",
        "intent_name": "late_check_out",
        "language": "Russian",
        "question_type": "Paraphrased",
        "question": "Можно ли запросить поздний выезд из отеля?",
        "ground_truth": "Вопрос касается возможности позднего выезда.",
        "answer": "Поздний выезд зависит от правил отеля и наличия свободных номеров."
    },


    # --------------------------------------------------------
    # INT008 - identification_required
    # --------------------------------------------------------

    {
        "intent_id": "INT008",
        "intent_name": "identification_required",
        "language": "Arabic",
        "question_type": "Paraphrased",
        "question": "ما هي وثيقة إثبات الهوية المطلوبة عند تسجيل الوصول؟",
        "ground_truth": "The question asks which identification document is required at check-in.",
        "answer": "The required identification document is specified in the hotel's identification policy."
    },

    {
        "intent_id": "INT008",
        "intent_name": "identification_required",
        "language": "English",
        "question_type": "Paraphrased",
        "question": "What identification document do I need when checking in?",
        "ground_truth": "The question asks which identification document is required at check-in.",
        "answer": "The required identification document is specified in the hotel's identification policy."
    },

    {
        "intent_id": "INT008",
        "intent_name": "identification_required",
        "language": "French",
        "question_type": "Paraphrased",
        "question": "Quelle pièce d'identité dois-je présenter lors de l'enregistrement ?",
        "ground_truth": "La question demande quelle pièce d'identité est nécessaire lors de l'enregistrement.",
        "answer": "La pièce d'identité requise est indiquée dans la politique d'identification de l'hôtel."
    },

    {
        "intent_id": "INT008",
        "intent_name": "identification_required",
        "language": "Russian",
        "question_type": "Paraphrased",
        "question": "Какой документ удостоверяющий личность нужно предъявить при регистрации?",
        "ground_truth": "Вопрос касается документа удостоверяющего личность, необходимого при регистрации.",
        "answer": "Требуемый документ удостоверяющий личность указан в правилах отеля."
    },
]


# ============================================================
# 12. CREATE DATAFRAME
# ============================================================

batch_df = pd.DataFrame(
    batch_content
)

# Add IDs
batch_df.insert(
    0,
    "faq_id",
    NEW_IDS
)

# Add remaining columns
batch_df["status"] = "generated"
batch_df["category"] = batch_df["intent_name"]
batch_df["source"] = "batch_02"
batch_df["attempts"] = 1
batch_df["difficulty"] = "Easy"


# ============================================================
# 13. REORDER COLUMNS
# ============================================================

batch_df = batch_df[
    REQUIRED_COLUMNS
]


# ============================================================
# 14. HARD VALIDATION
# ============================================================

print("\n" + "=" * 100)
print("🔎 BATCH 02 VALIDATION")
print("=" * 100)


# ------------------------------------------------------------
# Exact row count
# ------------------------------------------------------------

assert len(batch_df) == 32, (
    f"❌ Expected 32 rows, found {len(batch_df)}"
)

print("✅ Exactly 32 questions.")


# ------------------------------------------------------------
# Exact IDs
# ------------------------------------------------------------

actual_ids = batch_df["faq_id"].tolist()

assert actual_ids == NEW_IDS, (
    "❌ FAQ IDs are incorrect.\n"
    f"Expected: {NEW_IDS}\n"
    f"Actual:   {actual_ids}"
)

print(
    f"✅ IDs: {NEW_IDS[0]} → {NEW_IDS[-1]}"
)


# ------------------------------------------------------------
# Duplicate IDs
# ------------------------------------------------------------

assert not batch_df["faq_id"].duplicated().any()

print("✅ No duplicate FAQ IDs.")


# ------------------------------------------------------------
# Empty questions
# ------------------------------------------------------------

empty_questions = (
    batch_df["question"]
    .isna()
    |
    batch_df["question"]
    .astype(str)
    .str.strip()
    .eq("")
)

assert not empty_questions.any(), (
    "❌ Empty questions detected."
)

print("✅ No empty questions.")


# ------------------------------------------------------------
# Empty ground truth
# ------------------------------------------------------------

assert (
    batch_df["ground_truth"]
    .notna()
    .all()
)

assert (
    batch_df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

print("✅ Ground truth complete.")


# ------------------------------------------------------------
# Empty answers
# ------------------------------------------------------------

assert (
    batch_df["answer"]
    .notna()
    .all()
)

assert (
    batch_df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

print("✅ Answers complete.")


# ------------------------------------------------------------
# Duplicate questions inside batch
# ------------------------------------------------------------

batch_normalized = (
    batch_df["question"]
    .map(normalize_question)
)

duplicate_mask = (
    batch_normalized
    .duplicated(keep=False)
)

if duplicate_mask.any():

    print("\n❌ DUPLICATES INSIDE BATCH:")

    print(
        batch_df.loc[
            duplicate_mask,
            [
                "faq_id",
                "language",
                "intent_id",
                "question"
            ]
        ].to_string(index=False)
    )

    raise AssertionError(
        "❌ Duplicate questions inside Batch 02."
    )

print("✅ No duplicate questions inside batch.")


# ------------------------------------------------------------
# Compare with existing dataset
# ------------------------------------------------------------

existing_conflicts = batch_df[
    batch_normalized.isin(
        existing_questions
    )
]

if not existing_conflicts.empty:

    print("\n❌ QUESTIONS ALREADY EXIST:")

    print(
        existing_conflicts[
            [
                "faq_id",
                "language",
                "intent_id",
                "question"
            ]
        ].to_string(index=False)
    )

    raise AssertionError(
        "❌ Batch contains questions already "
        "present in the existing dataset."
    )

print(
    "✅ No questions overlap with existing 256."
)


# ============================================================
# 15. LANGUAGE BALANCE
# ============================================================

language_counts = (
    batch_df["language"]
    .value_counts()
    .to_dict()
)

expected_language_counts = {
    "Arabic": 8,
    "English": 8,
    "French": 8,
    "Russian": 8,
}

assert (
    language_counts
    == expected_language_counts
)

print(
    "✅ Language balance: "
    "Arabic 8 / English 8 / French 8 / Russian 8"
)


# ============================================================
# 16. INTENT BALANCE
# ============================================================

intent_counts = (
    batch_df["intent_id"]
    .value_counts()
    .to_dict()
)

for intent_id in [
    "INT001",
    "INT002",
    "INT003",
    "INT004",
    "INT005",
    "INT006",
    "INT007",
    "INT008",
]:

    assert intent_counts.get(
        intent_id,
        0
    ) == 4

print(
    "✅ Intent balance: 4 questions × 8 intents"
)


# ============================================================
# 17. INTENT NAME CONSISTENCY
# ============================================================

intent_mapping = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

for intent_id, intent_name in intent_mapping.items():

    subset = batch_df[
        batch_df["intent_id"]
        == intent_id
    ]

    assert (
        subset["intent_name"]
        == intent_name
    ).all()

print("✅ Intent ID ↔ Intent Name consistency.")


# ============================================================
# 18. SHOW BATCH
# ============================================================

print("\n" + "=" * 100)
print("📝 BATCH 02 QUESTIONS")
print("=" * 100)

print(
    batch_df[
        [
            "faq_id",
            "intent_id",
            "language",
            "question_type",
            "question"
        ]
    ].to_string(index=False)
)


# ============================================================
# 19. SAVE BATCH SEPARATELY
# ============================================================

BATCH_DIR = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
)

BATCH_DIR.mkdir(
    parents=True,
    exist_ok=True
)

BATCH_PATH = (
    BATCH_DIR
    / "faq_batch_02.csv"
)

batch_df.to_csv(
    BATCH_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 20. FINAL CONFIRMATION
# ============================================================

print("\n" + "=" * 100)
print("🎉 BATCH 02 READY")
print("=" * 100)

print(
    f"""
Existing dataset : 256 questions
New batch        : 32 questions

FAQ IDs:
{NEW_IDS[0]} → {NEW_IDS[-1]}

Languages:
Arabic   : 8
English  : 8
French   : 8
Russian  : 8

Intents:
INT001 : 4
INT002 : 4
INT003 : 4
INT004 : 4
INT005 : 4
INT006 : 4
INT007 : 4
INT008 : 4

Validation:
✅ Exactly 32 rows
✅ IDs correct
✅ No duplicate IDs
✅ No empty questions
✅ No duplicate questions
✅ No overlap with existing 256
✅ Ground truth complete
✅ Answers complete
✅ Language balance correct
✅ Intent balance correct

⚠️ IMPORTANT:
hotel_faq_final.csv was NOT modified.

📄 Batch file:
{BATCH_PATH}
"""
)

print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 02 vs ORIGINAL 256 QUESTIONS
# READ-ONLY DUPLICATE / OVERLAP CHECK
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata

print("=" * 100)
print("🏨 HOTEL AI - BATCH 02 vs ORIGINAL 256")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_02.csv"
)

# ============================================================
# 2. CHECK FILES
# ============================================================

assert FINAL_DATASET.exists(), (
    f"❌ Final dataset not found:\n{FINAL_DATASET}"
)

assert BATCH_FILE.exists(), (
    f"❌ Batch 02 not found:\n{BATCH_FILE}"
)

print(f"✅ Final dataset found.")
print(f"✅ Batch 02 found.")

# ============================================================
# 3. LOAD
# ============================================================

df = pd.read_csv(FINAL_DATASET)
batch = pd.read_csv(BATCH_FILE)

print("\n" + "=" * 100)
print("📊 FILE INFORMATION")
print("=" * 100)

print(f"Final dataset rows : {len(df)}")
print(f"Batch 02 rows      : {len(batch)}")

# ============================================================
# 4. MAKE SURE BATCH IS THE EXPECTED 32
# ============================================================

assert len(batch) == 32, (
    f"❌ Batch 02 should contain 32 questions, "
    f"found {len(batch)}."
)

print("✅ Batch 02 contains exactly 32 questions.")

# ============================================================
# 5. ORIGINAL 256
# ============================================================
# IMPORTANT:
# Original questions = FAQ0009 → FAQ0256
#
# We intentionally exclude FAQ0001 → FAQ0008 because
# those are outside the original 256 being compared here.

original_256 = df[
    df["faq_id"].between("FAQ0009", "FAQ0256")
].copy()

print("\n" + "=" * 100)
print("📦 ORIGINAL DATASET")
print("=" * 100)

print(f"Original questions selected: {len(original_256)}")

assert len(original_256) == 248, (
    "⚠️ Note: FAQ0009 → FAQ0256 contains 248 questions, "
    "not 256."
)

print(
    "⚠️ IMPORTANT: Your current ID structure has FAQ0001 → FAQ0288."
)
print(
    "FAQ0009 → FAQ0256 = 248 questions."
)
print(
    "Therefore, this check cannot represent the old 256 exactly "
    "using IDs alone."
)

# ============================================================
# 6. BETTER METHOD:
#    TAKE THE FIRST 256 ROWS OF THE ORIGINAL DATASET
# ============================================================

original_256 = df.iloc[:256].copy()

print("\n" + "=" * 100)
print("📌 ORIGINAL 256 USED FOR COMPARISON")
print("=" * 100)

print(
    f"First ID : {original_256.iloc[0]['faq_id']}"
)

print(
    f"Last ID  : {original_256.iloc[-1]['faq_id']}"
)

print(
    f"Rows     : {len(original_256)}"
)

assert len(original_256) == 256

# ============================================================
# 7. NORMALIZATION FUNCTION
# ============================================================

def normalize_question(text):

    text = str(text)

    # Unicode normalization
    text = unicodedata.normalize(
        "NFKC",
        text
    )

    # Lowercase
    text = text.lower()

    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    # Remove punctuation
    text = re.sub(
        r"[^\w\s]",
        "",
        text,
        flags=re.UNICODE
    )

    # Final trim
    text = text.strip()

    return text


original_256["_normalized_question"] = (
    original_256["question"]
    .map(normalize_question)
)

batch["_normalized_question"] = (
    batch["question"]
    .map(normalize_question)
)

# ============================================================
# 8. EXACT DUPLICATE CHECK
# ============================================================

print("\n" + "=" * 100)
print("🔎 EXACT QUESTION OVERLAP")
print("=" * 100)

old_questions = set(
    original_256["question"]
    .astype(str)
    .str.strip()
)

new_questions = set(
    batch["question"]
    .astype(str)
    .str.strip()
)

exact_overlap = (
    new_questions
    .intersection(old_questions)
)

if exact_overlap:

    print(
        f"❌ {len(exact_overlap)} exact duplicate(s) found:"
    )

    for q in sorted(exact_overlap):
        print(f"- {q}")

else:

    print(
        "✅ No exact duplicate questions."
    )

# ============================================================
# 9. NORMALIZED OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("🔎 NORMALIZED QUESTION OVERLAP")
print("=" * 100)

old_normalized = set(
    original_256["_normalized_question"]
)

new_normalized = set(
    batch["_normalized_question"]
)

normalized_overlap = (
    new_normalized
    .intersection(old_normalized)
)

if normalized_overlap:

    print(
        f"❌ {len(normalized_overlap)} normalized "
        f"duplicate(s) found."
    )

    for normalized in sorted(normalized_overlap):

        old_rows = original_256[
            original_256["_normalized_question"]
            == normalized
        ]

        new_rows = batch[
            batch["_normalized_question"]
            == normalized
        ]

        print("\nOLD:")
        print(
            old_rows[
                ["faq_id", "question"]
            ].to_string(index=False)
        )

        print("\nNEW:")
        print(
            new_rows[
                ["faq_id", "question"]
            ].to_string(index=False)
        )

else:

    print(
        "✅ No normalized duplicate questions."
    )

# ============================================================
# 10. DUPLICATES INSIDE BATCH
# ============================================================

print("\n" + "=" * 100)
print("📝 DUPLICATES INSIDE BATCH 02")
print("=" * 100)

batch_duplicates = batch[
    batch["_normalized_question"]
    .duplicated(keep=False)
]

if not batch_duplicates.empty:

    print(
        "❌ Duplicate questions inside Batch 02:"
    )

    print(
        batch_duplicates[
            ["faq_id", "question"]
        ].to_string(index=False)
    )

else:

    print(
        "✅ No duplicates inside Batch 02."
    )

# ============================================================
# 11. FINAL RESULT
# ============================================================

print("\n" + "=" * 100)
print("🏆 BATCH 02 vs ORIGINAL 256 RESULT")
print("=" * 100)

if (
    not exact_overlap
    and not normalized_overlap
    and batch_duplicates.empty
):

    print("""
🟢 PASS

✅ 32 new questions checked
✅ Compared against original 256
✅ No exact duplicates
✅ No normalized duplicates
✅ No duplicates inside Batch 02
✅ Dataset was NOT modified
""")

else:

    print("""
🔴 FAIL

⚠️ There are duplicate/overlapping questions.

DO NOT MERGE BATCH 02 yet.
Review the questions listed above.
""")

print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - COMPLETE DATASET QUALITY CHECK
# READ-ONLY — DOES NOT MODIFY THE DATASET
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata

# ============================================================
# 1. PROJECT AUTO DETECTION
# ============================================================

print("=" * 100)
print("🏨 HOTEL AI - COMPLETE DATASET QUALITY CHECK")
print("=" * 100)

possible_paths = [
    Path("/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"),
    Path.home() / "Desktop" / "HotelAI-Intelligent-Agent",
    Path.home() / "OneDrive" / "سطح المكتب" / "HotelAI",
    Path.home() / "OneDrive" / "Desktop" / "HotelAI",
]

PROJECT_ROOT = None

for path in possible_paths:
    if path.exists():
        PROJECT_ROOT = path
        break

if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "❌ Could not automatically detect HotelAI project root.\n"
        "Please update PROJECT_ROOT manually."
    )

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

print(f"\n📂 Project root:")
print(PROJECT_ROOT)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"\n❌ Dataset not found:\n{DATASET_PATH}"
    )

print(f"\n📄 Dataset:")
print(DATASET_PATH)

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)

print("\n" + "=" * 100)
print("📊 DATASET INFORMATION")
print("=" * 100)

print(f"Rows    : {len(df)}")
print(f"Columns : {len(df.columns)}")

# ============================================================
# 3. REQUIRED COLUMNS
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

print("\n" + "=" * 100)
print("🧪 COLUMN VALIDATION")
print("=" * 100)

missing_columns = [
    col for col in REQUIRED_COLUMNS
    if col not in df.columns
]

extra_columns = [
    col for col in df.columns
    if col not in REQUIRED_COLUMNS
]

if missing_columns:
    print("❌ Missing columns:")
    print(missing_columns)
else:
    print("✅ All 13 required columns exist.")

if list(df.columns) == REQUIRED_COLUMNS:
    print("✅ Column order is correct.")
else:
    print("⚠️ Column order differs from expected.")
    print("\nActual:")
    print(list(df.columns))

# ============================================================
# 4. ROW COUNT
# ============================================================

print("\n" + "=" * 100)
print("📦 ROW COUNT")
print("=" * 100)

print(f"Total questions: {len(df)}")

if len(df) == 288:
    print("✅ Dataset contains exactly 288 questions.")
else:
    print(
        f"⚠️ Expected 288 questions, found {len(df)}."
    )

# ============================================================
# 5. FAQ ID FORMAT
# ============================================================

print("\n" + "=" * 100)
print("🆔 FAQ ID VALIDATION")
print("=" * 100)

id_pattern = re.compile(r"^FAQ\d{4}$")

invalid_ids = df[
    ~df["faq_id"].astype(str).str.match(id_pattern)
]

if invalid_ids.empty:
    print("✅ All FAQ IDs have valid format.")
else:
    print("❌ Invalid FAQ IDs:")
    print(
        invalid_ids[
            ["faq_id", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 6. DUPLICATE FAQ IDs
# ============================================================

duplicate_ids = df[
    df["faq_id"].duplicated(keep=False)
]

if duplicate_ids.empty:
    print("✅ No duplicate FAQ IDs.")
else:
    print("❌ Duplicate FAQ IDs:")
    print(
        duplicate_ids[
            ["faq_id", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 7. FAQ ID CONTINUITY
# ============================================================

print("\n" + "=" * 100)
print("🔢 FAQ ID CONTINUITY")
print("=" * 100)

numeric_ids = (
    df["faq_id"]
    .astype(str)
    .str.extract(r"FAQ(\d{4})")[0]
)

numeric_ids = pd.to_numeric(
    numeric_ids,
    errors="coerce"
)

expected_ids = list(
    range(
        int(numeric_ids.min()),
        int(numeric_ids.max()) + 1
    )
)

actual_ids_sorted = sorted(
    numeric_ids.dropna().astype(int).tolist()
)

missing_ids = sorted(
    set(expected_ids) - set(actual_ids_sorted)
)

extra_count = len(
    actual_ids_sorted
) - len(
    set(actual_ids_sorted)
)

if not missing_ids and extra_count == 0:
    print(
        f"✅ IDs are continuous: "
        f"FAQ{int(numeric_ids.min()):04d} → "
        f"FAQ{int(numeric_ids.max()):04d}"
    )
else:
    print("❌ ID continuity problem.")

    if missing_ids:
        print("Missing IDs:")
        print(
            [
                f"FAQ{x:04d}"
                for x in missing_ids
            ]
        )

# ============================================================
# 8. ID ORDER
# ============================================================

ordered_ids = numeric_ids.tolist()

expected_order = list(
    range(
        int(numeric_ids.min()),
        int(numeric_ids.max()) + 1
    )
)

if ordered_ids == expected_order:
    print("✅ FAQ IDs are correctly ordered.")
else:
    print("❌ FAQ IDs are NOT correctly ordered.")

    problems = []

    for i, (actual, expected) in enumerate(
        zip(ordered_ids, expected_order),
        start=1
    ):
        if actual != expected:
            problems.append(
                (
                    i,
                    f"FAQ{expected:04d}",
                    f"FAQ{int(actual):04d}"
                )
            )

    print("\nFirst order problems:")
    for row, expected, actual in problems[:20]:
        print(
            f"Row {row}: "
            f"Expected {expected} "
            f"but found {actual}"
        )

# ============================================================
# 9. REQUIRED FIELD VALIDATION
# ============================================================

print("\n" + "=" * 100)
print("🧪 REQUIRED FIELD VALIDATION")
print("=" * 100)

field_errors = False

for col in REQUIRED_COLUMNS:

    missing = df[col].isna().sum()

    empty = (
        df[col]
        .astype(str)
        .str.strip()
        .isin(["", "nan", "None"])
        .sum()
    )

    print(
        f"{'✅' if missing == 0 and empty == 0 else '❌'} "
        f"{col:<15} "
        f"Missing: {missing:<4} "
        f"Empty: {empty}"
    )

    if missing > 0 or empty > 0:
        field_errors = True

# ============================================================
# 10. NORMALIZATION FUNCTION
# ============================================================

def normalize_question(text):

    if pd.isna(text):
        return ""

    text = str(text)

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = text.strip().lower()

    # Remove punctuation
    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    # Normalize spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# 11. EXACT DUPLICATES
# ============================================================

print("\n" + "=" * 100)
print("📝 EXACT QUESTION DUPLICATE CHECK")
print("=" * 100)

exact_duplicates = df[
    df["question"]
    .astype(str)
    .str.strip()
    .duplicated(keep=False)
]

if exact_duplicates.empty:
    print("✅ No exact duplicate questions.")
else:
    print(
        f"❌ Found "
        f"{len(exact_duplicates)} rows "
        f"in duplicate groups."
    )

    print(
        exact_duplicates[
            ["faq_id", "language", "intent_id", "question"]
        ]
        .sort_values("question")
        .to_string(index=False)
    )

# ============================================================
# 12. NORMALIZED DUPLICATES
# ============================================================

print("\n" + "=" * 100)
print("🔎 NORMALIZED QUESTION DUPLICATE CHECK")
print("=" * 100)

df["_normalized_question"] = (
    df["question"].apply(normalize_question)
)

normalized_duplicates = df[
    df["_normalized_question"]
    .duplicated(keep=False)
]

if normalized_duplicates.empty:
    print("✅ No normalized duplicate questions.")
else:
    print(
        f"❌ Found "
        f"{len(normalized_duplicates)} rows "
        f"with normalized duplicates."
    )

    print(
        normalized_duplicates[
            [
                "faq_id",
                "language",
                "intent_id",
                "question",
            ]
        ]
        .sort_values("_normalized_question")
        .to_string(index=False)
    )

# ============================================================
# 13. LANGUAGE DISTRIBUTION
# ============================================================

print("\n" + "=" * 100)
print("🌍 LANGUAGE DISTRIBUTION")
print("=" * 100)

language_counts = (
    df["language"]
    .value_counts()
)

print(language_counts)

expected_languages = {
    "Arabic": 72,
    "English": 72,
    "French": 72,
    "Russian": 72,
}

language_ok = True

for lang, expected in expected_languages.items():

    actual = language_counts.get(lang, 0)

    if actual != expected:
        language_ok = False
        print(
            f"❌ {lang}: expected "
            f"{expected}, found {actual}"
        )

if language_ok:
    print(
        "✅ Language balance: "
        "72 × 4 languages."
    )

# ============================================================
# 14. INTENT DISTRIBUTION
# ============================================================

print("\n" + "=" * 100)
print("🎯 INTENT DISTRIBUTION")
print("=" * 100)

intent_summary = (
    df.groupby(
        ["intent_id", "intent_name"]
    )
    .size()
    .reset_index(name="questions")
)

print(
    intent_summary.to_string(index=False)
)

intent_ok = True

for intent_id in [
    "INT001",
    "INT002",
    "INT003",
    "INT004",
    "INT005",
    "INT006",
    "INT007",
    "INT008",
]:

    count = (
        df["intent_id"] == intent_id
    ).sum()

    if count != 36:
        intent_ok = False
        print(
            f"❌ {intent_id}: "
            f"expected 36, found {count}"
        )

if intent_ok:
    print(
        "✅ Intent balance: "
        "36 × 8 intents."
    )

# ============================================================
# 15. INTENT ID ↔ NAME CONSISTENCY
# ============================================================

print("\n" + "=" * 100)
print("🔗 INTENT ID ↔ INTENT NAME")
print("=" * 100)

intent_mapping = (
    df.groupby("intent_id")["intent_name"]
    .nunique()
)

bad_mappings = (
    intent_mapping[
        intent_mapping > 1
    ]
)

if bad_mappings.empty:
    print(
        "✅ Every intent_id maps "
        "to exactly one intent_name."
    )
else:
    print(
        "❌ Intent ID ↔ name inconsistency:"
    )

    for intent_id in bad_mappings.index:

        names = (
            df.loc[
                df["intent_id"] == intent_id,
                "intent_name"
            ]
            .unique()
        )

        print(
            intent_id,
            "→",
            list(names)
        )

# ============================================================
# 16. QUESTION TYPE
# ============================================================

print("\n" + "=" * 100)
print("📝 QUESTION TYPE DISTRIBUTION")
print("=" * 100)

question_type_counts = (
    df["question_type"]
    .value_counts()
)

print(question_type_counts)

allowed_question_types = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

invalid_types = set(
    question_type_counts.index
) - allowed_question_types

if not invalid_types:
    print("✅ All question types are valid.")
else:
    print(
        "❌ Invalid question types:"
    )
    print(invalid_types)

# ============================================================
# 17. STATUS
# ============================================================

print("\n" + "=" * 100)
print("📌 STATUS DISTRIBUTION")
print("=" * 100)

print(
    df["status"].value_counts()
)

# ============================================================
# 18. DIFFICULTY
# ============================================================

print("\n" + "=" * 100)
print("📌 DIFFICULTY DISTRIBUTION")
print("=" * 100)

print(
    df["difficulty"].value_counts()
)

# ============================================================
# 19. GROUND TRUTH / ANSWER
# ============================================================

print("\n" + "=" * 100)
print("🎯 GROUND TRUTH / ANSWER")
print("=" * 100)

ground_truth_empty = (
    df["ground_truth"]
    .astype(str)
    .str.strip()
    .isin(["", "nan", "None"])
)

answer_empty = (
    df["answer"]
    .astype(str)
    .str.strip()
    .isin(["", "nan", "None"])
)

print(
    "✅ Ground truth complete."
    if ground_truth_empty.sum() == 0
    else
    f"❌ Ground truth missing: "
    f"{ground_truth_empty.sum()}"
)

print(
    "✅ Answers complete."
    if answer_empty.sum() == 0
    else
    f"❌ Answers missing: "
    f"{answer_empty.sum()}"
)

# ============================================================
# 20. QUESTIONS PER LANGUAGE × INTENT
# ============================================================

print("\n" + "=" * 100)
print("📊 LANGUAGE × INTENT MATRIX")
print("=" * 100)

matrix = pd.crosstab(
    df["intent_id"],
    df["language"]
)

print(matrix)

# ============================================================
# 21. QUESTION TYPE × LANGUAGE
# ============================================================

print("\n" + "=" * 100)
print("📊 QUESTION TYPE × LANGUAGE")
print("=" * 100)

type_language = pd.crosstab(
    df["language"],
    df["question_type"]
)

print(type_language)

# ============================================================
# 22. POSSIBLE SAME-INTENT DUPLICATES
# ============================================================

print("\n" + "=" * 100)
print("🔍 DUPLICATES BY INTENT")
print("=" * 100)

same_intent_duplicates = (
    df.groupby(
        ["intent_id", "_normalized_question"]
    )
    .size()
    .reset_index(name="count")
)

same_intent_duplicates = (
    same_intent_duplicates[
        same_intent_duplicates["count"] > 1
    ]
)

if same_intent_duplicates.empty:
    print(
        "✅ No normalized duplicates "
        "inside the same intent."
    )
else:
    print(
        "⚠️ Possible repeated questions "
        "inside the same intent:"
    )

    for _, row in same_intent_duplicates.iterrows():

        subset = df[
            (df["intent_id"] == row["intent_id"])
            &
            (
                df["_normalized_question"]
                == row["_normalized_question"]
            )
        ]

        print("\nIntent:", row["intent_id"])

        print(
            subset[
                [
                    "faq_id",
                    "language",
                    "question",
                ]
            ].to_string(index=False)
        )

# ============================================================
# 23. FINAL RESULT
# ============================================================

print("\n" + "=" * 100)
print("🏆 FINAL DATASET CHECK RESULT")
print("=" * 100)

errors = []

if missing_columns:
    errors.append("Missing columns")

if not invalid_ids.empty:
    errors.append("Invalid FAQ IDs")

if not duplicate_ids.empty:
    errors.append("Duplicate FAQ IDs")

if missing_ids:
    errors.append("Missing FAQ IDs")

if ordered_ids != expected_order:
    errors.append("FAQ ID order")

if field_errors:
    errors.append("Required fields")

if not exact_duplicates.empty:
    errors.append("Exact duplicate questions")

if not normalized_duplicates.empty:
    errors.append("Normalized duplicate questions")

if not language_ok:
    errors.append("Language balance")

if not intent_ok:
    errors.append("Intent balance")

if not bad_mappings.empty:
    errors.append("Intent mapping")

if invalid_types:
    errors.append("Invalid question types")

if ground_truth_empty.sum() > 0:
    errors.append("Missing ground truth")

if answer_empty.sum() > 0:
    errors.append("Missing answers")

if not errors:

    print("""
🟢 DATASET IS VALID

✅ All IDs valid
✅ IDs continuous and ordered
✅ No duplicate IDs
✅ No missing IDs
✅ No duplicate questions
✅ No normalized duplicates
✅ Required fields complete
✅ Languages validated
✅ Intents validated
✅ Intent names consistent
✅ Question types valid
✅ Ground truth complete
✅ Answers complete
""")

else:

    print("🔴 DATASET HAS ISSUES")
    print("\nProblems found:")

    for error in errors:
        print(f"❌ {error}")

print("\n" + "=" * 100)
print("🔒 READ-ONLY CHECK")
print("=" * 100)

print(
    "✅ hotel_faq_final.csv was NOT modified."
)

# ============================================================
# 24. CLEAN TEMP COLUMN
# ============================================================

df.drop(
    columns=["_normalized_question"],
    inplace=True
)

print("\n" + "=" * 100)
print("🏁 CHECK COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - DUPLICATE + SIMILARITY CHECK
# ============================================================
# READ-ONLY
# Checks:
# 1. Exact duplicates
# 2. Normalized duplicates
# 3. Very similar questions
# 4. Similar questions inside the SAME intent
# 5. Similar questions across DIFFERENT intents
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata
from difflib import SequenceMatcher

print("=" * 100)
print("🏨 HOTEL AI - DUPLICATE + SIMILARITY CHECK")
print("=" * 100)

# ============================================================
# 1. FIND DATASET
# ============================================================

possible_roots = [
    Path("/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"),
    Path.home() / "Desktop" / "HotelAI-Intelligent-Agent",
    Path.home() / "Desktop" / "HotelAI",
]

PROJECT_ROOT = next(
    (p for p in possible_roots if p.exists()),
    None
)

if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "❌ HotelAI project root not found."
    )

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"❌ Dataset not found:\n{DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

print(f"\n📄 Dataset:\n{DATASET_PATH}")
print(f"📊 Total questions: {len(df)}")

# ============================================================
# 2. NORMALIZATION
# ============================================================

def normalize_question(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.lower().strip()

    # Remove punctuation
    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    # Normalize spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


df["_normalized"] = df["question"].apply(
    normalize_question
)

# ============================================================
# 3. EXACT DUPLICATES
# ============================================================

print("\n" + "=" * 100)
print("📝 1. EXACT DUPLICATE CHECK")
print("=" * 100)

exact_duplicates = df[
    df["question"].duplicated(keep=False)
].copy()

if exact_duplicates.empty:

    print("✅ No exact duplicate questions.")

else:

    print(
        f"🔴 Found {len(exact_duplicates)} rows "
        "in exact duplicate groups.\n"
    )

    print(
        exact_duplicates[
            [
                "faq_id",
                "intent_id",
                "language",
                "question"
            ]
        ].to_string(index=False)
    )

# ============================================================
# 4. NORMALIZED DUPLICATES
# ============================================================

print("\n" + "=" * 100)
print("🔎 2. NORMALIZED DUPLICATE CHECK")
print("=" * 100)

normalized_duplicates = df[
    df["_normalized"].duplicated(keep=False)
].copy()

if normalized_duplicates.empty:

    print("✅ No normalized duplicate questions.")

else:

    print(
        f"🔴 Found {len(normalized_duplicates)} rows "
        "with normalized duplicates.\n"
    )

    print(
        normalized_duplicates[
            [
                "faq_id",
                "intent_id",
                "language",
                "question"
            ]
        ].to_string(index=False)
    )

# ============================================================
# 5. SIMILARITY FUNCTION
# ============================================================

def similarity(a, b):

    return SequenceMatcher(
        None,
        normalize_question(a),
        normalize_question(b)
    ).ratio()

# ============================================================
# 6. PAIRWISE SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🔍 3. QUESTION SIMILARITY CHECK")
print("=" * 100)

print("""
Similarity levels:

🟢 < 0.70  = Usually different
🟡 0.70–0.84 = Similar wording
🟠 0.85–0.94 = Very similar
🔴 >= 0.95 = Almost duplicate
""")

THRESHOLD = 0.85

similar_pairs = []

questions = df[
    [
        "faq_id",
        "intent_id",
        "language",
        "question"
    ]
].to_dict("records")

total_comparisons = (
    len(questions) * (len(questions) - 1) // 2
)

print(
    f"🔄 Comparing {total_comparisons:,} question pairs..."
)

for i in range(len(questions)):

    q1 = questions[i]

    for j in range(i + 1, len(questions)):

        q2 = questions[j]

        # Skip exact same normalized text
        if normalize_question(q1["question"]) == normalize_question(
            q2["question"]
        ):
            continue

        score = similarity(
            q1["question"],
            q2["question"]
        )

        if score >= THRESHOLD:

            similar_pairs.append({
                "faq_id_1": q1["faq_id"],
                "intent_1": q1["intent_id"],
                "language_1": q1["language"],
                "question_1": q1["question"],

                "faq_id_2": q2["faq_id"],
                "intent_2": q2["intent_id"],
                "language_2": q2["language"],
                "question_2": q2["question"],

                "similarity": round(score, 3)
            })

print(
    f"\n🔎 Similar pairs found: {len(similar_pairs)}"
)

# ============================================================
# 7. DISPLAY SIMILAR QUESTIONS
# ============================================================

if similar_pairs:

    similarity_df = pd.DataFrame(
        similar_pairs
    ).sort_values(
        "similarity",
        ascending=False
    )

    print("\n" + "=" * 100)
    print("📋 MOST SIMILAR QUESTIONS")
    print("=" * 100)

    pd.set_option(
        "display.max_colwidth",
        120
    )

    print(
        similarity_df[
            [
                "similarity",
                "faq_id_1",
                "intent_1",
                "question_1",
                "faq_id_2",
                "intent_2",
                "question_2"
            ]
        ].head(50).to_string(index=False)
    )

else:

    print(
        "✅ No highly similar questions found "
        f"at threshold {THRESHOLD}."
    )

# ============================================================
# 8. SAME INTENT SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🎯 4. SIMILARITY INSIDE SAME INTENT")
print("=" * 100)

same_intent_pairs = [
    x for x in similar_pairs
    if x["intent_1"] == x["intent_2"]
]

print(
    f"Same-intent similar pairs: "
    f"{len(same_intent_pairs)}"
)

if same_intent_pairs:

    same_intent_df = pd.DataFrame(
        same_intent_pairs
    ).sort_values(
        "similarity",
        ascending=False
    )

    print(
        same_intent_df[
            [
                "similarity",
                "faq_id_1",
                "intent_1",
                "question_1",
                "faq_id_2",
                "question_2"
            ]
        ].head(50).to_string(index=False)
    )

else:

    print(
        "✅ No highly similar questions "
        "inside the same intent."
    )

# ============================================================
# 9. DIFFERENT INTENT SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("⚠️ 5. SIMILARITY ACROSS DIFFERENT INTENTS")
print("=" * 100)

different_intent_pairs = [
    x for x in similar_pairs
    if x["intent_1"] != x["intent_2"]
]

print(
    f"Different-intent similar pairs: "
    f"{len(different_intent_pairs)}"
)

if different_intent_pairs:

    different_intent_df = pd.DataFrame(
        different_intent_pairs
    ).sort_values(
        "similarity",
        ascending=False
    )

    print(
        different_intent_df[
            [
                "similarity",
                "faq_id_1",
                "intent_1",
                "question_1",
                "faq_id_2",
                "intent_2",
                "question_2"
            ]
        ].head(50).to_string(index=False)
    )

else:

    print(
        "✅ No highly similar questions "
        "across different intents."
    )

# ============================================================
# 10. SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("🏆 SIMILARITY CHECK SUMMARY")
print("=" * 100)

print(
    f"""
Total questions                 : {len(df)}
Exact duplicate rows            : {len(exact_duplicates)}
Normalized duplicate rows       : {len(normalized_duplicates)}
Similarity threshold            : {THRESHOLD}
Highly similar pairs             : {len(similar_pairs)}
Same-intent similar pairs        : {len(same_intent_pairs)}
Different-intent similar pairs   : {len(different_intent_pairs)}
"""
)

# ============================================================
# 11. FINAL INTERPRETATION
# ============================================================

print("=" * 100)
print("📌 INTERPRETATION")
print("=" * 100)

if exact_duplicates.empty:
    print("✅ Exact duplicates: PASS")
else:
    print("🔴 Exact duplicates: REVIEW")

if normalized_duplicates.empty:
    print("✅ Normalized duplicates: PASS")
else:
    print("🔴 Normalized duplicates: REVIEW")

if len(same_intent_pairs) == 0:
    print("✅ Same-intent similarity: PASS")
else:
    print(
        f"🟡 Same-intent similarity: "
        f"{len(same_intent_pairs)} pairs need review"
    )

if len(different_intent_pairs) == 0:
    print("✅ Cross-intent similarity: PASS")
else:
    print(
        f"🟠 Cross-intent similarity: "
        f"{len(different_intent_pairs)} pairs need review"
    )

print("\n" + "=" * 100)
print("🔒 READ-ONLY")
print("=" * 100)

print("✅ Dataset was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

print("\n" + "=" * 100)
print("🎉 DUPLICATE + SIMILARITY CHECK COMPLETE")
print("=" * 100)
````

````python
import pandas as pd
import os
import re
from datetime import datetime
from difflib import SequenceMatcher

# ============================================================
# PATHS
# ============================================================

PROJECT_ROOT = "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"

DATASET_PATH = os.path.join(
    PROJECT_ROOT,
    "03_Structured_Data",
    "CSV",
    "final",
    "hotel_faq_final.csv"
)

BACKUP_DIR = os.path.join(
    PROJECT_ROOT,
    "03_Structured_Data",
    "CSV",
    "final",
    "backups"
)

os.makedirs(BACKUP_DIR, exist_ok=True)

print("=" * 100)
print("🏨 HOTEL AI - FIX FINAL CROSS-INTENT SIMILARITY")
print("=" * 100)

# ============================================================
# LOAD
# ============================================================

assert os.path.exists(DATASET_PATH), "❌ Dataset not found."

df = pd.read_csv(DATASET_PATH)

print(f"📊 Rows before: {len(df)}")

# ============================================================
# BACKUP
# ============================================================

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

backup_path = os.path.join(
    BACKUP_DIR,
    f"hotel_faq_before_final_similarity_fix_{timestamp}.csv"
)

df.to_csv(backup_path, index=False)

print(f"💾 Backup created:")
print(backup_path)

# ============================================================
# ORIGINAL VALUES
# ============================================================

original_questions = {
    "FAQ0105": "ما هو وقت تسجيل الوصول في Nile Pearl Hotel؟",
    "FAQ0141": "ما هو وقت تسجيل الخروج المعتاد في Nile Pearl Hotel؟",
    "FAQ0011": "ما عنوان فندق Nile Pearl Hotel؟",
    "FAQ0041": "كم عدد نجوم فندق Nile Pearl Hotel؟",
}

# ============================================================
# NEW QUESTIONS
# ============================================================

new_questions = {

    # INT004 - Check-in
    "FAQ0105":
        "عند الوصول إلى Nile Pearl Hotel، متى يُسمح للضيف ببدء إجراءات تسجيل الوصول؟",

    # INT005 - Check-out
    "FAQ0141":
        "عند انتهاء الإقامة في Nile Pearl Hotel، ما الموعد النهائي لمغادرة الغرفة؟",

    # INT001 - Hotel location/address
    "FAQ0011":
        "ما الموقع أو العنوان الذي يمكنني استخدامه للوصول إلى Nile Pearl Hotel؟",

    # INT002 - Hotel category/stars
    "FAQ0041":
        "كم نجمة مُصنَّف بها Nile Pearl Hotel؟",
}

# ============================================================
# VALIDATE IDs
# ============================================================

target_ids = list(new_questions.keys())

assert df["faq_id"].isin(target_ids).sum() == 4, \
    "❌ One or more target FAQ IDs were not found."

# ============================================================
# APPLY CHANGES
# ============================================================

print()
print("=" * 100)
print("🔄 APPLYING QUESTION CHANGES")
print("=" * 100)

for faq_id, new_question in new_questions.items():

    idx = df.index[df["faq_id"] == faq_id][0]

    old_question = df.at[idx, "question"]

    print()
    print(f"🔄 {faq_id}")
    print(f"OLD: {old_question}")
    print(f"NEW: {new_question}")

    df.at[idx, "question"] = new_question

# ============================================================
# BASIC STRUCTURE CHECK
# ============================================================

print()
print("=" * 100)
print("🧪 BASIC VALIDATION")
print("=" * 100)

assert len(df) == 320, \
    "❌ Row count changed."

assert df["faq_id"].nunique() == 320, \
    "❌ FAQ IDs are no longer unique."

assert df["question"].notna().all(), \
    "❌ Missing questions found."

assert df["question"].astype(str).str.strip().ne("").all(), \
    "❌ Empty questions found."

print("✅ Row count preserved: 320")
print("✅ FAQ IDs preserved.")
print("✅ No missing questions.")
print("✅ No empty questions.")

# ============================================================
# EXACT DUPLICATES
# ============================================================

print()
print("=" * 100)
print("📝 EXACT DUPLICATE CHECK")
print("=" * 100)

exact_duplicates = df[
    df["question"].duplicated(keep=False)
]

if exact_duplicates.empty:
    print("✅ No exact duplicate questions.")
else:
    print("❌ Exact duplicates found:")
    print(
        exact_duplicates[
            ["faq_id", "intent_id", "question"]
        ].to_string(index=False)
    )
    raise AssertionError("Exact duplicates detected.")

# ============================================================
# NORMALIZATION
# ============================================================

def normalize(text):
    text = str(text).lower().strip()

    text = re.sub(r"\s+", " ", text)

    text = re.sub(r"[؟?!.,;:،؛\"'«»()\-]", "", text)

    return text

df["_normalized"] = df["question"].apply(normalize)

# ============================================================
# NORMALIZED DUPLICATES
# ============================================================

print()
print("=" * 100)
print("🔎 NORMALIZED DUPLICATE CHECK")
print("=" * 100)

normalized_duplicates = df[
    df["_normalized"].duplicated(keep=False)
]

if normalized_duplicates.empty:
    print("✅ No normalized duplicate questions.")
else:
    print("❌ Normalized duplicates found:")
    print(
        normalized_duplicates[
            ["faq_id", "intent_id", "question"]
        ].sort_values("_normalized").to_string(index=False)
    )
    raise AssertionError("Normalized duplicates detected.")

# ============================================================
# SIMILARITY CHECK
# ============================================================

print()
print("=" * 100)
print("🔍 QUESTION SIMILARITY CHECK")
print("=" * 100)

questions = df[
    ["faq_id", "intent_id", "question"]
].to_dict("records")

similar_pairs = []

for i in range(len(questions)):
    for j in range(i + 1, len(questions)):

        q1 = questions[i]
        q2 = questions[j]

        score = SequenceMatcher(
            None,
            normalize(q1["question"]),
            normalize(q2["question"])
        ).ratio()

        if score >= 0.85:

            similar_pairs.append({
                "similarity": round(score, 3),
                "faq_id_1": q1["faq_id"],
                "intent_1": q1["intent_id"],
                "question_1": q1["question"],
                "faq_id_2": q2["faq_id"],
                "intent_2": q2["intent_id"],
                "question_2": q2["question"],
            })

similar_df = pd.DataFrame(similar_pairs)

if similar_df.empty:

    print("🎉 No highly similar questions found.")
    print("✅ Similarity < 0.85 for all question pairs.")

else:

    similar_df = similar_df.sort_values(
        "similarity",
        ascending=False
    )

    print(
        f"⚠️ Highly similar pairs remaining: "
        f"{len(similar_df)}"
    )

    print()
    print(similar_df.to_string(index=False))

# ============================================================
# SAME INTENT SIMILARITY
# ============================================================

print()
print("=" * 100)
print("🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

same_intent = []

if not similar_df.empty:

    for _, row in similar_df.iterrows():

        if row["intent_1"] == row["intent_2"]:
            same_intent.append(row)

if not same_intent:

    print("✅ No highly similar questions inside the same intent.")

else:

    same_intent_df = pd.DataFrame(same_intent)

    print(
        f"⚠️ Same-intent similar pairs: "
        f"{len(same_intent_df)}"
    )

    print(
        same_intent_df.to_string(index=False)
    )

# ============================================================
# DIFFERENT INTENT SIMILARITY
# ============================================================

print()
print("=" * 100)
print("⚠️ DIFFERENT-INTENT SIMILARITY")
print("=" * 100)

different_intent = []

if not similar_df.empty:

    for _, row in similar_df.iterrows():

        if row["intent_1"] != row["intent_2"]:
            different_intent.append(row)

if not different_intent:

    print("✅ No highly similar questions across different intents.")

else:

    different_intent_df = pd.DataFrame(
        different_intent
    )

    print(
        f"⚠️ Different-intent similar pairs: "
        f"{len(different_intent_df)}"
    )

    print(
        different_intent_df.to_string(index=False)
    )

# ============================================================
# REMOVE TEMP COLUMN
# ============================================================

df.drop(columns=["_normalized"], inplace=True)

# ============================================================
# FINAL SAFETY CHECK
# ============================================================

assert len(df) == 320
assert df["faq_id"].tolist() == [
    f"FAQ{i:04d}" for i in range(1, 321)
]

# ============================================================
# SAVE
# ============================================================

df.to_csv(DATASET_PATH, index=False)

print()
print("=" * 100)
print("💾 DATASET SAVED")
print("=" * 100)

print(DATASET_PATH)

# ============================================================
# FINAL RESULT
# ============================================================

print()
print("=" * 100)
print("🏆 FINAL RESULT")
print("=" * 100)

print("🟢 FINAL SIMILARITY FIX COMPLETED")
print()
print("✅ 320 questions preserved")
print("✅ FAQ0001 → FAQ0320")
print("✅ No rows added")
print("✅ No rows deleted")
print("✅ FAQ IDs preserved")
print("✅ Backup created")
print("✅ Only 4 selected questions modified")
print("✅ No exact duplicates")
print("✅ No normalized duplicates")

if similar_df.empty:
    print("✅ No similarity >= 0.85")
else:
    print(
        f"⚠️ Similarity pairs >= 0.85 remaining: "
        f"{len(similar_df)}"
    )

print()
print("🔒 Backup:")
print(backup_path)

print("=" * 100)
print("🎉 CHECK COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - FULL DATASET QUALITY + DUPLICATE + SIMILARITY CHECK
# READ-ONLY — DOES NOT MODIFY THE DATASET
# ============================================================

from pathlib import Path
from itertools import combinations
import re
import unicodedata
import pandas as pd
import numpy as np
from difflib import SequenceMatcher

# ============================================================
# 1. CONFIGURATION
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

EXPECTED_ROWS = 320

EXPECTED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

EXPECTED_LANGUAGES = {
    "Arabic",
    "English",
    "French",
    "Russian",
}

EXPECTED_INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

EXPECTED_QUESTION_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

EXPECTED_STATUS = {
    "valid",
    "generated",
}

EXPECTED_DIFFICULTY = {
    "Easy",
}

SIMILARITY_THRESHOLD = 0.85

print("=" * 100)
print("🏨 HOTEL AI - FULL DATASET QUALITY + DUPLICATE + SIMILARITY CHECK")
print("=" * 100)

# ============================================================
# 2. PATH CHECK
# ============================================================

print("\n📂 PATH VALIDATION")
print("=" * 100)

assert PROJECT_ROOT.exists(), f"❌ Project root not found: {PROJECT_ROOT}"
print(f"✅ Project root found:\n{PROJECT_ROOT}")

assert DATASET_PATH.exists(), f"❌ Dataset not found: {DATASET_PATH}"
print(f"✅ Dataset found:\n{DATASET_PATH}")

# ============================================================
# 3. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)

print("\n📊 DATASET INFORMATION")
print("=" * 100)

print(f"Rows    : {len(df)}")
print(f"Columns : {len(df.columns)}")

# ============================================================
# 4. COLUMN CHECK
# ============================================================

print("\n🧪 COLUMN VALIDATION")
print("=" * 100)

missing_columns = [
    col for col in EXPECTED_COLUMNS
    if col not in df.columns
]

extra_columns = [
    col for col in df.columns
    if col not in EXPECTED_COLUMNS
]

assert not missing_columns, (
    f"❌ Missing columns: {missing_columns}"
)

assert not extra_columns, (
    f"❌ Unexpected columns: {extra_columns}"
)

assert list(df.columns) == EXPECTED_COLUMNS, (
    "❌ Column order is incorrect."
)

print("✅ All 13 required columns exist.")
print("✅ Column order is correct.")

# ============================================================
# 5. ROW COUNT
# ============================================================

print("\n📦 ROW COUNT")
print("=" * 100)

assert len(df) == EXPECTED_ROWS, (
    f"❌ Expected {EXPECTED_ROWS} rows, found {len(df)}"
)

print(f"Total questions: {len(df)}")
print(f"✅ Exactly {EXPECTED_ROWS} questions.")

# ============================================================
# 6. REQUIRED FIELD CHECK
# ============================================================

print("\n🧪 REQUIRED FIELD VALIDATION")
print("=" * 100)

required_fields_ok = True

for col in EXPECTED_COLUMNS:

    missing = int(df[col].isna().sum())

    empty = int(
        df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        f"✅ {col:<15} "
        f"Missing: {missing:<3} "
        f"Empty: {empty:<3}"
    )

    if missing > 0 or empty > 0:
        required_fields_ok = False

assert required_fields_ok, (
    "❌ Required fields contain missing/empty values."
)

print("\n✅ All required fields are complete.")

# ============================================================
# 7. FAQ ID VALIDATION
# ============================================================

print("\n🆔 FAQ ID VALIDATION")
print("=" * 100)

id_pattern = re.compile(r"^FAQ\d{4}$")

invalid_ids = df[
    ~df["faq_id"]
    .astype(str)
    .str.match(id_pattern)
]

assert invalid_ids.empty, (
    f"❌ Invalid FAQ IDs:\n{invalid_ids[['faq_id']]}"
)

print("✅ All FAQ IDs have valid format.")

duplicate_ids = df[df["faq_id"].duplicated(keep=False)]

assert duplicate_ids.empty, (
    f"❌ Duplicate FAQ IDs:\n{duplicate_ids[['faq_id']]}"
)

print("✅ No duplicate FAQ IDs.")

# ============================================================
# 8. FAQ ID CONTINUITY + ORDER
# ============================================================

print("\n🔢 FAQ ID CONTINUITY")
print("=" * 100)

numeric_ids = (
    df["faq_id"]
    .str.extract(r"FAQ(\d{4})")[0]
    .astype(int)
)

expected_ids = np.arange(1, EXPECTED_ROWS + 1)

missing_ids = sorted(
    set(expected_ids) - set(numeric_ids)
)

extra_ids = sorted(
    set(numeric_ids) - set(expected_ids)
)

assert not missing_ids, (
    f"❌ Missing IDs: {missing_ids}"
)

assert not extra_ids, (
    f"❌ Unexpected IDs: {extra_ids}"
)

assert numeric_ids.tolist() == expected_ids.tolist(), (
    "❌ FAQ IDs are not correctly ordered."
)

print(
    f"✅ IDs continuous and ordered: "
    f"FAQ0001 → FAQ{EXPECTED_ROWS:04d}"
)

# ============================================================
# 9. NORMALIZATION FUNCTION
# ============================================================

def normalize_text(text):

    if pd.isna(text):
        return ""

    text = str(text)

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Lowercase
    text = text.lower()

    # Remove punctuation
    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


df["_normalized_question"] = (
    df["question"]
    .map(normalize_text)
)

# ============================================================
# 10. EXACT DUPLICATES
# ============================================================

print("\n📝 EXACT QUESTION DUPLICATE CHECK")
print("=" * 100)

exact_duplicates = df[
    df["question"]
    .duplicated(keep=False)
]

if exact_duplicates.empty:

    print("✅ No exact duplicate questions.")

else:

    print(
        f"🔴 Exact duplicate rows: "
        f"{len(exact_duplicates)}"
    )

    print(
        exact_duplicates[
            ["faq_id", "intent_id", "language", "question"]
        ]
        .sort_values("question")
        .to_string(index=False)
    )

# ============================================================
# 11. NORMALIZED DUPLICATES
# ============================================================

print("\n🔎 NORMALIZED DUPLICATE CHECK")
print("=" * 100)

normalized_duplicates = df[
    df["_normalized_question"]
    .duplicated(keep=False)
]

if normalized_duplicates.empty:

    print("✅ No normalized duplicate questions.")

else:

    print(
        f"🔴 Normalized duplicate rows: "
        f"{len(normalized_duplicates)}"
    )

    print(
        normalized_duplicates[
            [
                "faq_id",
                "intent_id",
                "language",
                "question",
                "_normalized_question",
            ]
        ]
        .sort_values("_normalized_question")
        .to_string(index=False)
    )

# ============================================================
# 12. DUPLICATES INSIDE SAME INTENT
# ============================================================

print("\n🔍 DUPLICATES INSIDE SAME INTENT")
print("=" * 100)

same_intent_duplicates = (
    df[
        df.duplicated(
            subset=["intent_id", "_normalized_question"],
            keep=False
        )
    ]
)

if same_intent_duplicates.empty:

    print(
        "✅ No normalized duplicates "
        "inside the same intent."
    )

else:

    print(
        f"🔴 Same-intent duplicate rows: "
        f"{len(same_intent_duplicates)}"
    )

    print(
        same_intent_duplicates[
            [
                "faq_id",
                "intent_id",
                "language",
                "question",
            ]
        ]
        .sort_values(
            ["intent_id", "question"]
        )
        .to_string(index=False)
    )

# ============================================================
# 13. INTENT ID ↔ INTENT NAME
# ============================================================

print("\n🔗 INTENT ID ↔ INTENT NAME")
print("=" * 100)

intent_errors = []

for intent_id, expected_name in EXPECTED_INTENTS.items():

    rows = df[df["intent_id"] == intent_id]

    if rows.empty:
        intent_errors.append(
            f"{intent_id}: missing"
        )
        continue

    actual_names = set(
        rows["intent_name"]
        .dropna()
        .astype(str)
    )

    if actual_names != {expected_name}:

        intent_errors.append(
            f"{intent_id}: "
            f"expected {expected_name}, "
            f"found {actual_names}"
        )

unexpected_intents = set(df["intent_id"]) - set(
    EXPECTED_INTENTS
)

if unexpected_intents:
    intent_errors.append(
        f"Unexpected intent IDs: {unexpected_intents}"
    )

assert not intent_errors, (
    "❌ Intent mapping errors:\n"
    + "\n".join(intent_errors)
)

print("✅ All intent mappings are correct.")

# ============================================================
# 14. LANGUAGE VALIDATION
# ============================================================

print("\n🌍 LANGUAGE DISTRIBUTION")
print("=" * 100)

language_counts = df["language"].value_counts()

print(language_counts.to_string())

unexpected_languages = (
    set(df["language"]) - EXPECTED_LANGUAGES
)

assert not unexpected_languages, (
    f"❌ Unexpected languages: {unexpected_languages}"
)

assert set(language_counts.index) == EXPECTED_LANGUAGES, (
    "❌ Language set is incorrect."
)

for language in EXPECTED_LANGUAGES:

    count = int(language_counts.get(language, 0))

    assert count == 80, (
        f"❌ {language}: expected 80, found {count}"
    )

print("\n✅ 80 questions × 4 languages.")

# ============================================================
# 15. INTENT DISTRIBUTION
# ============================================================

print("\n🎯 INTENT DISTRIBUTION")
print("=" * 100)

intent_distribution = (
    df.groupby(
        ["intent_id", "intent_name"]
    )
    .size()
    .reset_index(name="questions")
)

print(
    intent_distribution.to_string(index=False)
)

for intent_id in EXPECTED_INTENTS:

    count = int(
        (df["intent_id"] == intent_id).sum()
    )

    assert count == 40, (
        f"❌ {intent_id}: "
        f"expected 40, found {count}"
    )

print("\n✅ 40 questions × 8 intents.")

# ============================================================
# 16. QUESTION TYPE VALIDATION
# ============================================================

print("\n📝 QUESTION TYPE DISTRIBUTION")
print("=" * 100)

question_type_counts = df["question_type"].value_counts()

print(question_type_counts.to_string())

unexpected_types = (
    set(df["question_type"])
    - EXPECTED_QUESTION_TYPES
)

assert not unexpected_types, (
    f"❌ Unexpected question types: {unexpected_types}"
)

print("\n✅ All question types are valid.")

# ============================================================
# 17. STATUS VALIDATION
# ============================================================

print("\n📌 STATUS DISTRIBUTION")
print("=" * 100)

status_counts = df["status"].value_counts()

print(status_counts.to_string())

unexpected_status = (
    set(df["status"])
    - EXPECTED_STATUS
)

assert not unexpected_status, (
    f"❌ Unexpected status values: {unexpected_status}"
)

print("\n✅ Status values are valid.")

# ============================================================
# 18. DIFFICULTY VALIDATION
# ============================================================

print("\n📌 DIFFICULTY DISTRIBUTION")
print("=" * 100)

difficulty_counts = df["difficulty"].value_counts()

print(difficulty_counts.to_string())

unexpected_difficulty = (
    set(df["difficulty"])
    - EXPECTED_DIFFICULTY
)

assert not unexpected_difficulty, (
    f"❌ Unexpected difficulty values: "
    f"{unexpected_difficulty}"
)

print("\n✅ Difficulty values are valid.")

# ============================================================
# 19. GROUND TRUTH / ANSWER CHECK
# ============================================================

print("\n🎯 GROUND TRUTH / ANSWER")
print("=" * 100)

gt_empty = (
    df["ground_truth"]
    .fillna("")
    .astype(str)
    .str.strip()
    .eq("")
)

answer_empty = (
    df["answer"]
    .fillna("")
    .astype(str)
    .str.strip()
    .eq("")
)

assert not gt_empty.any(), (
    "❌ Empty ground_truth values found."
)

assert not answer_empty.any(), (
    "❌ Empty answer values found."
)

print("✅ Ground truth complete.")
print("✅ Answers complete.")

# ============================================================
# 20. GROUND TRUTH ↔ ANSWER CONSISTENCY
# ============================================================

print("\n🔗 GROUND TRUTH ↔ ANSWER CONSISTENCY")
print("=" * 100)

df["_normalized_ground_truth"] = (
    df["ground_truth"]
    .map(normalize_text)
)

df["_normalized_answer"] = (
    df["answer"]
    .map(normalize_text)
)

answer_mismatches = df[
    df["_normalized_ground_truth"]
    != df["_normalized_answer"]
]

if answer_mismatches.empty:

    print(
        "✅ Ground truth and answer are identical "
        "after normalization."
    )

else:

    print(
        f"🟡 {len(answer_mismatches)} rows have "
        "different ground_truth and answer text."
    )

    print(
        answer_mismatches[
            [
                "faq_id",
                "ground_truth",
                "answer",
            ]
        ]
        .to_string(index=False)
    )

# ============================================================
# 21. LANGUAGE × INTENT
# ============================================================

print("\n📊 LANGUAGE × INTENT MATRIX")
print("=" * 100)

language_intent = pd.crosstab(
    df["intent_id"],
    df["language"]
)

print(language_intent.to_string())

for intent_id in EXPECTED_INTENTS:

    for language in EXPECTED_LANGUAGES:

        count = int(
            language_intent
            .get(language, pd.Series())
            .get(intent_id, 0)
        )

        assert count == 10, (
            f"❌ {intent_id} × {language}: "
            f"expected 10, found {count}"
        )

print(
    "\n✅ Every intent contains "
    "10 questions in every language."
)

# ============================================================
# 22. QUESTION TYPE × LANGUAGE
# ============================================================

print("\n📊 QUESTION TYPE × LANGUAGE")
print("=" * 100)

type_language = pd.crosstab(
    df["language"],
    df["question_type"]
)

print(type_language.to_string())

# ============================================================
# 23. QUESTION LENGTH CHECK
# ============================================================

print("\n📏 QUESTION LENGTH CHECK")
print("=" * 100)

df["_question_length"] = (
    df["question"]
    .astype(str)
    .str.len()
)

too_short = df[
    df["_question_length"] < 10
]

if too_short.empty:

    print(
        "✅ No suspiciously short questions "
        "(< 10 characters)."
    )

else:

    print(
        f"⚠️ {len(too_short)} suspiciously short questions:"
    )

    print(
        too_short[
            ["faq_id", "language", "question"]
        ]
        .to_string(index=False)
    )

# ============================================================
# 24. FULL PAIRWISE SIMILARITY
# ============================================================

print("\n🔎 FULL QUESTION SIMILARITY CHECK")
print("=" * 100)

questions = df["question"].astype(str).tolist()
faq_ids = df["faq_id"].tolist()
intents = df["intent_id"].tolist()
languages = df["language"].tolist()

pairs = []

total_pairs = len(questions) * (len(questions) - 1) // 2

print(
    f"Comparing {total_pairs:,} question pairs..."
)

for i, j in combinations(range(len(questions)), 2):

    score = SequenceMatcher(
        None,
        normalize_text(questions[i]),
        normalize_text(questions[j])
    ).ratio()

    if score >= 0.80:

        pairs.append({
            "similarity": round(score, 3),
            "faq_id_1": faq_ids[i],
            "intent_1": intents[i],
            "language_1": languages[i],
            "question_1": questions[i],
            "faq_id_2": faq_ids[j],
            "intent_2": intents[j],
            "language_2": languages[j],
            "question_2": questions[j],
        })

similarity_df = pd.DataFrame(pairs)

if similarity_df.empty:

    print(
        "🎉 No pairs with similarity >= 0.80."
    )

else:

    similarity_df = similarity_df.sort_values(
        "similarity",
        ascending=False
    ).reset_index(drop=True)

    print(
        f"🔎 Pairs with similarity >= 0.80: "
        f"{len(similarity_df)}"
    )

# ============================================================
# 25. SIMILARITY THRESHOLD COUNTS
# ============================================================

print("\n📊 SIMILARITY THRESHOLD SUMMARY")
print("=" * 100)

if similarity_df.empty:

    for threshold in [0.80, 0.85, 0.90, 0.95]:

        print(
            f"{threshold:.2f}+ : 0 pairs"
        )

else:

    for threshold in [0.80, 0.85, 0.90, 0.95]:

        count = int(
            (similarity_df["similarity"] >= threshold)
            .sum()
        )

        print(
            f"{threshold:.2f}+ : {count} pairs"
        )

# ============================================================
# 26. TOP 30 MOST SIMILAR
# ============================================================

print("\n📋 TOP 30 MOST SIMILAR QUESTIONS")
print("=" * 100)

if similarity_df.empty:

    print("None.")

else:

    print(
        similarity_df
        .head(30)
        .to_string(index=False)
    )

# ============================================================
# 27. SAME-INTENT SIMILARITY
# ============================================================

print("\n🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    same_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["intent_1"]
        ==
        similarity_df["intent_2"]
    ]

if same_intent.empty:

    print(
        "✅ No similar pairs inside the same intent "
        "at >= 0.80."
    )

else:

    print(
        f"⚠️ Same-intent similar pairs: "
        f"{len(same_intent)}"
    )

    print(
        same_intent
        .head(30)
        .to_string(index=False)
    )

# ============================================================
# 28. CROSS-INTENT SIMILARITY
# ============================================================

print("\n⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    cross_intent = pd.DataFrame()

else:

    cross_intent = similarity_df[
        similarity_df["intent_1"]
        !=
        similarity_df["intent_2"]
    ]

if cross_intent.empty:

    print(
        "✅ No cross-intent similar pairs "
        "at >= 0.80."
    )

else:

    print(
        f"⚠️ Cross-intent similar pairs: "
        f"{len(cross_intent)}"
    )

    print(
        cross_intent
        .head(30)
        .to_string(index=False)
    )

# ============================================================
# 29. SAME-LANGUAGE SIMILARITY
# ============================================================

print("\n🌍 SAME-LANGUAGE SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    same_language = pd.DataFrame()

else:

    same_language = similarity_df[
        similarity_df["language_1"]
        ==
        similarity_df["language_2"]
    ]

if same_language.empty:

    print(
        "✅ No same-language similar pairs "
        "at >= 0.80."
    )

else:

    print(
        f"Same-language similar pairs: "
        f"{len(same_language)}"
    )

# ============================================================
# 30. CROSS-LANGUAGE SIMILARITY
# ============================================================

print("\n🌐 CROSS-LANGUAGE SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    cross_language = pd.DataFrame()

else:

    cross_language = similarity_df[
        similarity_df["language_1"]
        !=
        similarity_df["language_2"]
    ]

if cross_language.empty:

    print(
        "✅ No cross-language similar pairs "
        "at >= 0.80."
    )

else:

    print(
        f"Cross-language similar pairs: "
        f"{len(cross_language)}"
    )

# ============================================================
# 31. HIGH-SIMILARITY REVIEW
# ============================================================

print("\n🚨 HIGH-SIMILARITY REVIEW")
print("=" * 100)

if similarity_df.empty:

    high_similarity = pd.DataFrame()

else:

    high_similarity = similarity_df[
        similarity_df["similarity"]
        >= SIMILARITY_THRESHOLD
    ]

if high_similarity.empty:

    print(
        f"🎉 No similarity >= "
        f"{SIMILARITY_THRESHOLD:.2f}"
    )

else:

    print(
        f"⚠️ {len(high_similarity)} pairs "
        f"have similarity >= "
        f"{SIMILARITY_THRESHOLD:.2f}"
    )

    print(
        high_similarity
        .to_string(index=False)
    )

# ============================================================
# 32. VERY HIGH SIMILARITY
# ============================================================

print("\n🔴 VERY HIGH SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    very_high = pd.DataFrame()

else:

    very_high = similarity_df[
        similarity_df["similarity"]
        >= 0.90
    ]

if very_high.empty:

    print(
        "✅ No pairs with similarity >= 0.90."
    )

else:

    print(
        f"🔴 {len(very_high)} pairs "
        "with similarity >= 0.90:"
    )

    print(
        very_high.to_string(index=False)
    )

# ============================================================
# 33. EXTREME SIMILARITY
# ============================================================

print("\n🔴 EXTREME SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    extreme = pd.DataFrame()

else:

    extreme = similarity_df[
        similarity_df["similarity"]
        >= 0.95
    ]

if extreme.empty:

    print(
        "✅ No pairs with similarity >= 0.95."
    )

else:

    print(
        f"🚨 {len(extreme)} pairs "
        "with similarity >= 0.95:"
    )

    print(
        extreme.to_string(index=False)
    )

# ============================================================
# 34. SOURCE / CATEGORY CONSISTENCY
# ============================================================

print("\n📌 CATEGORY / SOURCE CHECK")
print("=" * 100)

for col in ["category", "source"]:

    unique_values = (
        df[col]
        .dropna()
        .astype(str)
        .str.strip()
        .unique()
    )

    print(
        f"{col}: {len(unique_values)} unique value(s)"
    )

    print(
        list(unique_values)
    )

# ============================================================
# 35. ATTEMPTS VALIDATION
# ============================================================

print("\n🔢 ATTEMPTS VALIDATION")
print("=" * 100)

attempts_numeric = pd.to_numeric(
    df["attempts"],
    errors="coerce"
)

invalid_attempts = df[
    attempts_numeric.isna()
]

negative_attempts = df[
    attempts_numeric < 0
]

assert invalid_attempts.empty, (
    "❌ Non-numeric attempts found."
)

assert negative_attempts.empty, (
    "❌ Negative attempts found."
)

print("✅ Attempts values are numeric and non-negative.")

print(
    f"Attempts min : {attempts_numeric.min()}"
)

print(
    f"Attempts max : {attempts_numeric.max()}"
)

print(
    f"Attempts mean: {attempts_numeric.mean():.2f}"
)

# ============================================================
# 36. NULL / WHITESPACE CHECK ACROSS ALL TEXT
# ============================================================

print("\n🧹 WHITESPACE / EMPTY TEXT CHECK")
print("=" * 100)

whitespace_issues = []

for col in EXPECTED_COLUMNS:

    series = df[col].astype(str)

    leading = series[
        series.str.match(r"^\s")
    ]

    trailing = series[
        series.str.match(r".*\s$")
    ]

    if len(leading) > 0:

        whitespace_issues.append(
            (col, "leading", len(leading))
        )

    if len(trailing) > 0:

        whitespace_issues.append(
            (col, "trailing", len(trailing))
        )

if not whitespace_issues:

    print(
        "✅ No obvious leading/trailing whitespace issues."
    )

else:

    print(
        "⚠️ Whitespace issues detected:"
    )

    for issue in whitespace_issues:

        print(issue)

# ============================================================
# 37. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("🏆 FINAL FULL DATASET CHECK RESULT")
print("=" * 100)

structural_pass = True

if len(df) != EXPECTED_ROWS:
    structural_pass = False

if missing_columns or extra_columns:
    structural_pass = False

if not duplicate_ids.empty:
    structural_pass = False

if not missing_ids == []:
    structural_pass = False

if not extra_ids == []:
    structural_pass = False

if not required_fields_ok:
    structural_pass = False

if not intent_errors == []:
    structural_pass = False

if not exact_duplicates.empty:
    structural_pass = False

if not normalized_duplicates.empty:
    structural_pass = False

if not gt_empty.any() == False:
    structural_pass = False

if not answer_empty.any() == False:
    structural_pass = False

if structural_pass:

    print("🟢 STRUCTURAL CHECK: PASS")

else:

    print("🔴 STRUCTURAL CHECK: FAIL")

# Similarity result

if similarity_df.empty:

    similarity_pass = True

else:

    similarity_pass = (
        (similarity_df["similarity"] >= 0.85)
        .sum()
        == 0
    )

if similarity_pass:

    print("🟢 SIMILARITY CHECK: PASS")
    print("✅ No similarity >= 0.85.")

else:

    print("🟡 SIMILARITY CHECK: REVIEW REQUIRED")
    print(
        f"⚠️ {len(high_similarity)} "
        "pairs >= 0.85."
    )

# ============================================================
# 38. FINAL CHECKLIST
# ============================================================

print("\n" + "=" * 100)
print("📋 FINAL CHECKLIST")
print("=" * 100)

checks = [
    ("320 questions", len(df) == 320),
    ("13 columns", len(df.columns) == 13),
    ("Valid FAQ IDs", invalid_ids.empty),
    ("No duplicate IDs", duplicate_ids.empty),
    ("Continuous IDs", numeric_ids.tolist() == expected_ids.tolist()),
    ("Required fields complete", required_fields_ok),
    ("No exact duplicates", exact_duplicates.empty),
    ("No normalized duplicates", normalized_duplicates.empty),
    ("No same-intent duplicates", same_intent_duplicates.empty),
    ("Intent mappings correct", not intent_errors),
    ("4 languages", set(df["language"]) == EXPECTED_LANGUAGES),
    ("40 questions per intent", all(
        (df["intent_id"] == x).sum() == 40
        for x in EXPECTED_INTENTS
    )),
    ("10 per language per intent", all(
        language_intent.loc[x, y] == 10
        for x in EXPECTED_INTENTS
        for y in EXPECTED_LANGUAGES
    )),
    ("Question types valid", not bool(unexpected_types)),
    ("Status valid", not bool(unexpected_status)),
    ("Difficulty valid", not bool(unexpected_difficulty)),
    ("Ground truth complete", not gt_empty.any()),
    ("Answers complete", not answer_empty.any()),
    ("No similarity >= 0.85", similarity_pass),
]

for name, result in checks:

    print(
        f"{'✅' if result else '❌'} {name}"
    )

# ============================================================
# 39. READ-ONLY CONFIRMATION
# ============================================================

print("\n" + "=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ Dataset was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

# ============================================================
# 40. FINAL STATUS
# ============================================================

print("\n" + "=" * 100)

all_pass = all(
    result for _, result in checks
)

if all_pass:

    print("🎉🟢 FULL DATASET CHECK PASSED")
    print("=" * 100)
    print("✅ ALL 320 QUESTIONS PASSED ALL CHECKS.")
    print("✅ Dataset is clean and ready.")

else:

    print("⚠️🟡 FULL DATASET CHECK REQUIRES REVIEW")
    print("=" * 100)
    print("⚠️ One or more checks require attention.")

print("=" * 100)

# ============================================================
# 41. CLEAN TEMPORARY COLUMNS
# ============================================================

# مهم:
# ده فقط في الذاكرة، ولن يتم حفظ أي تعديل على CSV.

df.drop(
    columns=[
        "_normalized_question",
        "_normalized_ground_truth",
        "_normalized_answer",
        "_question_length",
    ],
    inplace=True,
    errors="ignore"
)

print("\n🏁 CHECK COMPLETE")
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 04 vs EXISTING 320
# DUPLICATE + SIMILARITY CHECK
# READ-ONLY
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata
from difflib import SequenceMatcher

# ============================================================
# PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_04.csv"
)

print("=" * 100)
print("🏨 HOTEL AI - BATCH 04 vs EXISTING 320")
print("=" * 100)

assert FINAL_DATASET.exists()
assert BATCH_FILE.exists()

df = pd.read_csv(FINAL_DATASET)
batch_df = pd.read_csv(BATCH_FILE)

assert len(df) == 320
assert len(batch_df) == 32

print(f"Existing questions : {len(df)}")
print(f"Batch 04 questions : {len(batch_df)}")

# ============================================================
# NORMALIZATION
# ============================================================

def normalize(text):
    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# EXACT OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("📝 EXACT OVERLAP")
print("=" * 100)

existing_exact = set(
    df["question"]
    .astype(str)
    .str.strip()
)

batch_exact = set(
    batch_df["question"]
    .astype(str)
    .str.strip()
)

exact_overlap = existing_exact.intersection(
    batch_exact
)

if exact_overlap:
    print("❌ Exact overlap found:")
    for q in sorted(exact_overlap):
        print(q)
else:
    print("✅ No exact overlap.")

# ============================================================
# NORMALIZED OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("🔎 NORMALIZED OVERLAP")
print("=" * 100)

existing_norm = set(
    df["question"].map(normalize)
)

batch_norm = set(
    batch_df["question"].map(normalize)
)

normalized_overlap = (
    existing_norm.intersection(batch_norm)
)

if normalized_overlap:
    print("❌ Normalized overlap found:")
    for q in sorted(normalized_overlap):
        print(q)
else:
    print("✅ No normalized overlap.")

# ============================================================
# DUPLICATES INSIDE BATCH
# ============================================================

print("\n" + "=" * 100)
print("📝 DUPLICATES INSIDE BATCH 04")
print("=" * 100)

batch_duplicate = batch_df[
    batch_df["question"].duplicated(keep=False)
]

if batch_duplicate.empty:
    print("✅ No duplicate questions inside Batch 04.")
else:
    print(
        batch_duplicate[
            ["faq_id", "question"]
        ].to_string(index=False)
    )

# ============================================================
# SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🔍 SEMANTIC / TEXT SIMILARITY")
print("=" * 100)

pairs = []

for _, new_row in batch_df.iterrows():

    for _, old_row in df.iterrows():

        score = SequenceMatcher(
            None,
            normalize(new_row["question"]),
            normalize(old_row["question"])
        ).ratio()

        if score >= 0.80:

            pairs.append({
                "similarity": round(score, 3),

                "new_faq_id": new_row["faq_id"],
                "new_intent": new_row["intent_id"],
                "new_language": new_row["language"],
                "new_question": new_row["question"],

                "old_faq_id": old_row["faq_id"],
                "old_intent": old_row["intent_id"],
                "old_language": old_row["language"],
                "old_question": old_row["question"],
            })

similarity_df = pd.DataFrame(pairs)

if similarity_df.empty:

    print("🎉 No pairs with similarity >= 0.80.")

else:

    similarity_df = similarity_df.sort_values(
        "similarity",
        ascending=False
    )

    print(
        f"Pairs with similarity >= 0.80: "
        f"{len(similarity_df)}"
    )

    print("\nTOP 30:")
    print(
        similarity_df.head(30).to_string(
            index=False
        )
    )

# ============================================================
# SAME INTENT
# ============================================================

print("\n" + "=" * 100)
print("🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    same_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["new_intent"]
        ==
        similarity_df["old_intent"]
    ]

print(
    f"Same-intent pairs >= 0.80: "
    f"{len(same_intent)}"
)

if not same_intent.empty:

    print(
        same_intent.head(30).to_string(
            index=False
        )
    )

# ============================================================
# CROSS INTENT
# ============================================================

print("\n" + "=" * 100)
print("⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    cross_intent = pd.DataFrame()

else:

    cross_intent = similarity_df[
        similarity_df["new_intent"]
        !=
        similarity_df["old_intent"]
    ]

print(
    f"Cross-intent pairs >= 0.80: "
    f"{len(cross_intent)}"
)

if not cross_intent.empty:

    print(
        cross_intent.head(30).to_string(
            index=False
        )
    )

# ============================================================
# HIGH SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🚨 HIGH SIMILARITY REVIEW")
print("=" * 100)

if similarity_df.empty:

    high = pd.DataFrame()

else:

    high = similarity_df[
        similarity_df["similarity"] >= 0.85
    ]

print(
    f"Similarity >= 0.85: {len(high)}"
)

if not high.empty:

    print(
        high.to_string(index=False)
    )
else:
    print("✅ No similarity >= 0.85.")

# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("🏆 BATCH 04 SIMILARITY SUMMARY")
print("=" * 100)

print(
    f"""
Existing questions          : 320
Batch 04 questions           : 32
Exact overlaps               : {len(exact_overlap)}
Normalized overlaps          : {len(normalized_overlap)}
Pairs >= 0.80                : {len(similarity_df)}
Same-intent pairs >= 0.80    : {len(same_intent)}
Cross-intent pairs >= 0.80   : {len(cross_intent)}
Pairs >= 0.85                : {len(high)}
"""
)

# ============================================================
# FINAL
# ============================================================

print("=" * 100)
print("🔒 READ-ONLY")
print("=" * 100)

print("✅ Existing dataset was NOT modified.")
print("✅ Batch 04 was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")

print("\n" + "=" * 100)
print("🎉 BATCH 04 CHECK COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 04 SIMILARITY REPAIR
# READ-ONLY AGAINST FINAL DATASET
# MODIFIES BATCH 04 ONLY
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata

print("=" * 100)
print("🏨 HOTEL AI - BATCH 04 SIMILARITY REPAIR")
print("=" * 100)

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_04.csv"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

assert BATCH_FILE.exists(), "❌ Batch 04 not found."
assert FINAL_DATASET.exists(), "❌ Final dataset not found."

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

df = pd.read_csv(
    FINAL_DATASET,
    encoding="utf-8-sig"
)

assert len(batch_df) == 32
assert len(df) == 320

# ============================================================
# REWRITES
# ============================================================

rewrites = {

    # INT002 Arabic
    "FAQ0325":
        "ما فئة Nile Pearl Hotel ضمن تصنيفات الفنادق؟",

    # INT004 Arabic
    "FAQ0333":
        "في أي ساعة يمكن للنزلاء بدء إجراءات الوصول إلى Nile Pearl Hotel؟",

    # INT004 Russian
    "FAQ0336":
        "С какого часа гостям разрешается заселение в Nile Pearl Hotel?",

    # INT004 English
    "FAQ0334":
        "What time are guests first permitted to check in at Nile Pearl Hotel?",

    # INT005 English
    "FAQ0338":
        "By what time should I have completed my departure from Nile Pearl Hotel?",

    # Cross-intent protection / stronger distinction
    "FAQ0326":
        "What hotel classification does Nile Pearl Hotel have?",
}

# ============================================================
# APPLY
# ============================================================

print("\n" + "=" * 100)
print("✍️ APPLYING REWRITES")
print("=" * 100)

for faq_id, new_question in rewrites.items():

    idx = batch_df.index[
        batch_df["faq_id"] == faq_id
    ].tolist()

    assert len(idx) == 1, (
        f"❌ {faq_id} not found exactly once."
    )

    idx = idx[0]

    old_question = batch_df.at[
        idx,
        "question"
    ]

    batch_df.at[
        idx,
        "question"
    ] = new_question

    print(f"\n🔄 {faq_id}")
    print(f"OLD: {old_question}")
    print(f"NEW: {new_question}")

# ============================================================
# NORMALIZATION
# ============================================================

def normalize(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# DUPLICATES INSIDE BATCH
# ============================================================

batch_normalized = batch_df["question"].map(
    normalize
)

assert not batch_df["question"].duplicated().any(), (
    "❌ Exact duplicate inside Batch 04."
)

assert not batch_normalized.duplicated().any(), (
    "❌ Normalized duplicate inside Batch 04."
)

# ============================================================
# OVERLAP WITH EXISTING DATASET
# ============================================================

existing_normalized = set(
    df["question"].map(normalize)
)

overlap = [
    q for q in batch_normalized
    if q in existing_normalized
]

assert not overlap, (
    f"❌ Batch 04 overlaps with existing 320:\n{overlap}"
)

print("\n✅ No duplicate/normalized overlap with existing 320.")

# ============================================================
# SAVE BATCH
# ============================================================

batch_df.to_csv(
    BATCH_FILE,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 100)
print("💾 BATCH 04 UPDATED")
print("=" * 100)

print(f"📄 {BATCH_FILE}")
print(f"✅ Rows preserved: {len(batch_df)}")
print("✅ hotel_faq_final.csv was NOT modified.")

print("\n" + "=" * 100)
print("🎉 BATCH 04 SIMILARITY REPAIR COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 04 FINAL SIMILARITY CHECK
# ============================================================
# READ-ONLY
# Existing dataset : 320 questions
# Batch 04         : 32 questions
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata
from difflib import SequenceMatcher

print("=" * 100)
print("🏨 HOTEL AI - BATCH 04 FINAL SIMILARITY CHECK")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_04.csv"
)

assert PROJECT_ROOT.exists(), (
    f"❌ Project root not found:\n{PROJECT_ROOT}"
)

assert FINAL_DATASET.exists(), (
    f"❌ Final dataset not found:\n{FINAL_DATASET}"
)

assert BATCH_FILE.exists(), (
    f"❌ Batch 04 not found:\n{BATCH_FILE}"
)

print(f"📂 Project root:\n{PROJECT_ROOT}")
print(f"📄 Existing dataset:\n{FINAL_DATASET}")
print(f"📄 Batch 04:\n{BATCH_FILE}")

# ============================================================
# 2. LOAD
# ============================================================

df = pd.read_csv(
    FINAL_DATASET,
    encoding="utf-8-sig"
)

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

print("\n" + "=" * 100)
print("📊 FILE INFORMATION")
print("=" * 100)

print(f"Existing dataset rows : {len(df)}")
print(f"Batch 04 rows         : {len(batch_df)}")

assert len(df) == 320, (
    f"❌ Existing dataset must contain 320 rows, "
    f"found {len(df)}"
)

assert len(batch_df) == 32, (
    f"❌ Batch 04 must contain 32 rows, "
    f"found {len(batch_df)}"
)

print("✅ Existing dataset = 320")
print("✅ Batch 04 = 32")

# ============================================================
# 3. REQUIRED COLUMNS
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

assert list(df.columns) == REQUIRED_COLUMNS, (
    "❌ Existing dataset columns are incorrect."
)

assert list(batch_df.columns) == REQUIRED_COLUMNS, (
    "❌ Batch 04 columns are incorrect."
)

print("\n✅ All 13 columns exist in both files.")
print("✅ Column order is correct.")

# ============================================================
# 4. FAQ ID CHECK
# ============================================================

print("\n" + "=" * 100)
print("🆔 BATCH 04 FAQ ID CHECK")
print("=" * 100)

expected_batch_ids = [
    f"FAQ{i:04d}"
    for i in range(321, 353)
]

actual_batch_ids = batch_df["faq_id"].tolist()

assert actual_batch_ids == expected_batch_ids, (
    "❌ Batch 04 FAQ IDs are incorrect.\n"
    f"Expected first: {expected_batch_ids[:5]}\n"
    f"Actual first  : {actual_batch_ids[:5]}\n"
    f"Expected last : {expected_batch_ids[-5:]}\n"
    f"Actual last  : {actual_batch_ids[-5:]}"
)

assert batch_df["faq_id"].is_unique

print("✅ IDs are exactly FAQ0321 → FAQ0352.")
print("✅ No duplicate IDs inside Batch 04.")

# ============================================================
# 5. NORMALIZATION
# ============================================================

def normalize_question(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# 6. EXACT OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("📝 EXACT OVERLAP")
print("=" * 100)

existing_exact = set(
    df["question"]
    .astype(str)
    .str.strip()
)

batch_exact = set(
    batch_df["question"]
    .astype(str)
    .str.strip()
)

exact_overlap = sorted(
    existing_exact.intersection(
        batch_exact
    )
)

if exact_overlap:

    print("❌ Exact overlaps found:")

    for question in exact_overlap:
        print(f"- {question}")

else:

    print("✅ No exact overlap.")

# ============================================================
# 7. NORMALIZED OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("🔎 NORMALIZED OVERLAP")
print("=" * 100)

existing_normalized = set(
    df["question"].map(
        normalize_question
    )
)

batch_normalized = set(
    batch_df["question"].map(
        normalize_question
    )
)

normalized_overlap = sorted(
    existing_normalized.intersection(
        batch_normalized
    )
)

if normalized_overlap:

    print("❌ Normalized overlaps found:")

    for question in normalized_overlap:
        print(f"- {question}")

else:

    print("✅ No normalized overlap.")

# ============================================================
# 8. DUPLICATES INSIDE BATCH
# ============================================================

print("\n" + "=" * 100)
print("📝 DUPLICATES INSIDE BATCH 04")
print("=" * 100)

batch_duplicate_exact = batch_df[
    batch_df["question"].duplicated(
        keep=False
    )
]

batch_normalized_series = (
    batch_df["question"]
    .map(normalize_question)
)

batch_duplicate_normalized = batch_df[
    batch_normalized_series.duplicated(
        keep=False
    )
]

if batch_duplicate_exact.empty:

    print("✅ No exact duplicates inside Batch 04.")

else:

    print("❌ Exact duplicates inside Batch 04:")
    print(
        batch_duplicate_exact[
            ["faq_id", "question"]
        ].to_string(index=False)
    )

if batch_duplicate_normalized.empty:

    print(
        "✅ No normalized duplicates inside Batch 04."
    )

else:

    print(
        "❌ Normalized duplicates inside Batch 04:"
    )

    print(
        batch_duplicate_normalized[
            ["faq_id", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 9. PAIRWISE SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🔍 SEMANTIC / TEXT SIMILARITY")
print("=" * 100)

similar_pairs = []

for _, new_row in batch_df.iterrows():

    for _, old_row in df.iterrows():

        score = SequenceMatcher(
            None,
            normalize_question(
                new_row["question"]
            ),
            normalize_question(
                old_row["question"]
            )
        ).ratio()

        # Collect >= 0.80
        if score >= 0.80:

            similar_pairs.append({
                "similarity": round(score, 3),

                "new_faq_id": new_row["faq_id"],
                "new_intent": new_row["intent_id"],
                "new_language": new_row["language"],
                "new_question": new_row["question"],

                "old_faq_id": old_row["faq_id"],
                "old_intent": old_row["intent_id"],
                "old_language": old_row["language"],
                "old_question": old_row["question"],
            })

similarity_df = pd.DataFrame(
    similar_pairs
)

if similarity_df.empty:

    print(
        "🎉 No pairs with similarity >= 0.80."
    )

else:

    similarity_df = similarity_df.sort_values(
        "similarity",
        ascending=False
    ).reset_index(drop=True)

    print(
        f"Pairs with similarity >= 0.80: "
        f"{len(similarity_df)}"
    )

    print("\nTOP SIMILAR PAIRS:")
    print(
        similarity_df.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 10. SIMILARITY THRESHOLDS
# ============================================================

print("\n" + "=" * 100)
print("📊 SIMILARITY THRESHOLD SUMMARY")
print("=" * 100)

for threshold in [0.80, 0.85, 0.90, 0.95]:

    if similarity_df.empty:

        count = 0

    else:

        count = int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )

    print(
        f"{threshold:.2f}+ : {count} pairs"
    )

# ============================================================
# 11. SAME-INTENT
# ============================================================

print("\n" + "=" * 100)
print("🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    same_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["new_intent"]
        ==
        similarity_df["old_intent"]
    ].copy()

print(
    f"Same-intent pairs >= 0.80: "
    f"{len(same_intent)}"
)

if not same_intent.empty:

    print(
        same_intent.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 12. CROSS-INTENT
# ============================================================

print("\n" + "=" * 100)
print("⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    cross_intent = pd.DataFrame()

else:

    cross_intent = similarity_df[
        similarity_df["new_intent"]
        !=
        similarity_df["old_intent"]
    ].copy()

print(
    f"Cross-intent pairs >= 0.80: "
    f"{len(cross_intent)}"
)

if not cross_intent.empty:

    print(
        cross_intent.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 13. SAME-LANGUAGE
# ============================================================

print("\n" + "=" * 100)
print("🌍 SAME-LANGUAGE SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    same_language = pd.DataFrame()

else:

    same_language = similarity_df[
        similarity_df["new_language"]
        ==
        similarity_df["old_language"]
    ].copy()

print(
    f"Same-language pairs >= 0.80: "
    f"{len(same_language)}"
)

# ============================================================
# 14. CROSS-LANGUAGE
# ============================================================

print("\n" + "=" * 100)
print("🌐 CROSS-LANGUAGE SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    cross_language = pd.DataFrame()

else:

    cross_language = similarity_df[
        similarity_df["new_language"]
        !=
        similarity_df["old_language"]
    ].copy()

print(
    f"Cross-language pairs >= 0.80: "
    f"{len(cross_language)}"
)

# ============================================================
# 15. HIGH SIMILARITY REVIEW
# ============================================================

print("\n" + "=" * 100)
print("🚨 HIGH SIMILARITY REVIEW")
print("=" * 100)

if similarity_df.empty:

    high_similarity = pd.DataFrame()

else:

    high_similarity = similarity_df[
        similarity_df["similarity"] >= 0.85
    ].copy()

print(
    f"Similarity >= 0.85: "
    f"{len(high_similarity)}"
)

if high_similarity.empty:

    print("✅ No similarity >= 0.85.")

else:

    print(
        high_similarity.to_string(
            index=False
        )
    )

# ============================================================
# 16. VERY HIGH SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🔴 VERY HIGH SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    very_high = pd.DataFrame()

else:

    very_high = similarity_df[
        similarity_df["similarity"] >= 0.90
    ].copy()

print(
    f"Similarity >= 0.90: "
    f"{len(very_high)}"
)

if not very_high.empty:

    print(
        very_high.to_string(
            index=False
        )
    )
else:

    print("✅ No similarity >= 0.90.")

# ============================================================
# 17. EXTREME SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🚨 EXTREME SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    extreme = pd.DataFrame()

else:

    extreme = similarity_df[
        similarity_df["similarity"] >= 0.95
    ].copy()

print(
    f"Similarity >= 0.95: "
    f"{len(extreme)}"
)

if not extreme.empty:

    print(
        extreme.to_string(
            index=False
        )
    )

else:

    print("✅ No similarity >= 0.95.")

# ============================================================
# 18. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("🏆 BATCH 04 FINAL SIMILARITY SUMMARY")
print("=" * 100)

print(
    f"""
Existing questions            : 320
Batch 04 questions             : 32

Exact overlaps                 : {len(exact_overlap)}
Normalized overlaps            : {len(normalized_overlap)}

Pairs >= 0.80                  : {len(similarity_df)}
Same-intent >= 0.80            : {len(same_intent)}
Cross-intent >= 0.80           : {len(cross_intent)}

Pairs >= 0.85                  : {len(high_similarity)}
Pairs >= 0.90                  : {len(very_high)}
Pairs >= 0.95                  : {len(extreme)}
"""
)

# ============================================================
# 19. DECISION
# ============================================================

print("=" * 100)
print("🏁 BATCH 04 DECISION")
print("=" * 100)

if (
    len(exact_overlap) == 0
    and len(normalized_overlap) == 0
    and batch_duplicate_exact.empty
    and batch_duplicate_normalized.empty
    and len(high_similarity) == 0
    and len(cross_intent) == 0
):

    print("""
🟢 BATCH 04 IS READY FOR THE NEXT STEP

✅ No exact overlap
✅ No normalized overlap
✅ No duplicates inside Batch 04
✅ No similarity >= 0.85
✅ No cross-intent similarity >= 0.80
""")

else:

    print("""
🟡 BATCH 04 STILL NEEDS REVIEW

⚠️ One or more similarity/duplicate checks
still require attention.
""")

# ============================================================
# 20. READ-ONLY CONFIRMATION
# ============================================================

print("=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ faq_batch_04.csv was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

print("\n" + "=" * 100)
print("🎉 BATCH 04 FINAL CHECK COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 04 REPAIR
# GROUND TRUTH + ANSWER + FULL BATCH VALIDATION
# ============================================================
# IMPORTANT:
# - Modifies BATCH 04 only
# - Does NOT modify hotel_faq_final.csv
# - Uses existing dataset knowledge as the source of truth
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata

print("=" * 100)
print("🏨 HOTEL AI - BATCH 04 REPAIR + FINAL VALIDATION")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_04.csv"
)

assert PROJECT_ROOT.exists(), (
    f"❌ Project root not found:\n{PROJECT_ROOT}"
)

assert FINAL_DATASET.exists(), (
    f"❌ Final dataset not found:\n{FINAL_DATASET}"
)

assert BATCH_FILE.exists(), (
    f"❌ Batch 04 not found:\n{BATCH_FILE}"
)

print("✅ Project root found.")
print("✅ Final dataset found.")
print("✅ Batch 04 found.")

# ============================================================
# 2. LOAD
# ============================================================

df = pd.read_csv(
    FINAL_DATASET,
    encoding="utf-8-sig"
)

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

print("\n" + "=" * 100)
print("📊 FILE INFORMATION")
print("=" * 100)

print(f"Existing dataset : {len(df)} rows")
print(f"Batch 04         : {len(batch_df)} rows")

assert len(df) == 320
assert len(batch_df) == 32

# ============================================================
# 3. REQUIRED COLUMNS
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

assert list(df.columns) == REQUIRED_COLUMNS
assert list(batch_df.columns) == REQUIRED_COLUMNS

print("✅ All 13 columns exist.")
print("✅ Column order is correct.")

# ============================================================
# 4. INTENT MAP
# ============================================================

INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

VALID_LANGUAGES = {
    "Arabic",
    "English",
    "French",
    "Russian",
}

VALID_QUESTION_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

# ============================================================
# 5. FAQ ID VALIDATION
# ============================================================

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(321, 353)
]

assert batch_df["faq_id"].tolist() == expected_ids
assert batch_df["faq_id"].is_unique

print("\n" + "=" * 100)
print("🆔 FAQ ID VALIDATION")
print("=" * 100)

print("✅ IDs are exactly FAQ0321 → FAQ0352.")
print("✅ No duplicate IDs.")

# ============================================================
# 6. NORMALIZATION
# ============================================================

def normalize_text(text):
    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


def normalize_question(text):
    text = normalize_text(text)

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# 7. CHECK BATCH QUESTION OVERLAP
# ============================================================

existing_questions = set(
    df["question"]
    .map(normalize_question)
)

batch_questions = batch_df["question"].map(
    normalize_question
)

assert not batch_questions.duplicated().any()

overlap = [
    q for q in batch_questions
    if q in existing_questions
]

assert not overlap, (
    f"❌ Batch 04 contains questions already in "
    f"the existing dataset:\n{overlap}"
)

print("\n" + "=" * 100)
print("📝 QUESTION OVERLAP")
print("=" * 100)

print("✅ No duplicate questions inside Batch 04.")
print("✅ No normalized overlap with existing 320.")

# ============================================================
# 8. BUILD CANONICAL KNOWLEDGE MAP
# ============================================================
# We extract a canonical answer for every
# Intent × Language from the existing dataset.
#
# Priority:
# 1. Existing non-empty ground_truth
# 2. Existing non-empty answer
#
# We choose the most frequent normalized ground_truth
# within each Intent × Language.

print("\n" + "=" * 100)
print("🧠 BUILDING CANONICAL KNOWLEDGE MAP")
print("=" * 100)

knowledge_map = {}

for intent_id, intent_name in INTENTS.items():

    for language in VALID_LANGUAGES:

        subset = df[
            (df["intent_id"] == intent_id)
            &
            (df["language"] == language)
        ].copy()

        assert len(subset) == 10, (
            f"❌ Expected 10 rows for "
            f"{intent_id} × {language}, "
            f"found {len(subset)}"
        )

        # Ground truth candidates
        gt_values = (
            subset["ground_truth"]
            .fillna("")
            .astype(str)
            .str.strip()
        )

        gt_values = gt_values[
            gt_values.ne("")
        ]

        assert not gt_values.empty, (
            f"❌ No ground truth available for "
            f"{intent_id} × {language}"
        )

        # Most common exact ground truth
        canonical_gt = (
            gt_values
            .value_counts()
            .index[0]
        )

        # Answer fallback
        answer_values = (
            subset["answer"]
            .fillna("")
            .astype(str)
            .str.strip()
        )

        answer_values = answer_values[
            answer_values.ne("")
        ]

        assert not answer_values.empty, (
            f"❌ No answer available for "
            f"{intent_id} × {language}"
        )

        canonical_answer = (
            answer_values
            .value_counts()
            .index[0]
        )

        knowledge_map[
            (intent_id, language)
        ] = {
            "ground_truth": canonical_gt,
            "answer": canonical_answer,
        }

        print(
            f"✅ {intent_id:<6} "
            f"{language:<8}"
        )

print("✅ Canonical knowledge map created.")

# ============================================================
# 9. APPLY GROUND TRUTH + ANSWER
# ============================================================

print("\n" + "=" * 100)
print("✍️ APPLYING GROUND TRUTH + ANSWERS")
print("=" * 100)

for idx, row in batch_df.iterrows():

    intent_id = str(
        row["intent_id"]
    ).strip()

    language = str(
        row["language"]
    ).strip()

    assert intent_id in INTENTS, (
        f"❌ Unknown intent: {intent_id}"
    )

    assert language in VALID_LANGUAGES, (
        f"❌ Unknown language: {language}"
    )

    source_knowledge = knowledge_map[
        (intent_id, language)
    ]

    batch_df.at[
        idx,
        "ground_truth"
    ] = source_knowledge["ground_truth"]

    batch_df.at[
        idx,
        "answer"
    ] = source_knowledge["answer"]

    batch_df.at[
        idx,
        "status"
    ] = "generated"

    batch_df.at[
        idx,
        "category"
    ] = "Hotel FAQ"

    batch_df.at[
        idx,
        "source"
    ] = "Generated Batch 04"

    batch_df.at[
        idx,
        "attempts"
    ] = 0

    batch_df.at[
        idx,
        "difficulty"
    ] = "Easy"

print("✅ Ground truth filled for all 32.")
print("✅ Answers filled for all 32.")

# ============================================================
# 10. REQUIRED FIELD CHECK
# ============================================================

print("\n" + "=" * 100)
print("🧪 REQUIRED FIELD VALIDATION")
print("=" * 100)

for col in REQUIRED_COLUMNS:

    missing = int(
        batch_df[col].isna().sum()
    )

    empty = int(
        batch_df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        f"✅ {col:<15}"
        f"Missing: {missing:<3}"
        f"Empty: {empty:<3}"
    )

    assert missing == 0
    assert empty == 0

print("✅ All required fields complete.")

# ============================================================
# 11. LANGUAGE BALANCE
# ============================================================

print("\n" + "=" * 100)
print("🌍 LANGUAGE DISTRIBUTION")
print("=" * 100)

language_counts = (
    batch_df["language"]
    .value_counts()
    .sort_index()
)

print(language_counts)

expected_language_counts = {
    "Arabic": 8,
    "English": 8,
    "French": 8,
    "Russian": 8,
}

assert (
    language_counts.to_dict()
    == expected_language_counts
)

print("✅ 8 × 4 languages.")

# ============================================================
# 12. INTENT BALANCE
# ============================================================

print("\n" + "=" * 100)
print("🎯 INTENT DISTRIBUTION")
print("=" * 100)

intent_counts = (
    batch_df["intent_id"]
    .value_counts()
    .sort_index()
)

print(intent_counts)

expected_intent_counts = {
    "INT001": 4,
    "INT002": 4,
    "INT003": 4,
    "INT004": 4,
    "INT005": 4,
    "INT006": 4,
    "INT007": 4,
    "INT008": 4,
}

assert (
    intent_counts.to_dict()
    == expected_intent_counts
)

print("✅ 4 × 8 intents.")

# ============================================================
# 13. INTENT NAME
# ============================================================

for _, row in batch_df.iterrows():

    expected_name = INTENTS[
        row["intent_id"]
    ]

    assert row["intent_name"] == expected_name

print("✅ Intent ID ↔ Intent Name consistency.")

# ============================================================
# 14. QUESTION TYPES
# ============================================================

invalid_types = (
    set(batch_df["question_type"])
    - VALID_QUESTION_TYPES
)

assert not invalid_types

print("\n" + "=" * 100)
print("📝 QUESTION TYPES")
print("=" * 100)

print(
    batch_df["question_type"]
    .value_counts()
)

print("✅ All question types are valid.")

# ============================================================
# 15. GROUND TRUTH / ANSWER
# ============================================================

print("\n" + "=" * 100)
print("🎯 GROUND TRUTH / ANSWER")
print("=" * 100)

assert (
    batch_df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

assert (
    batch_df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

print("✅ Ground truth complete.")
print("✅ Answers complete.")

# ============================================================
# 16. SAVE BATCH ONLY
# ============================================================

batch_df.to_csv(
    BATCH_FILE,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 100)
print("💾 BATCH 04 SAVED")
print("=" * 100)

print(BATCH_FILE)
print("✅ hotel_faq_final.csv was NOT modified.")

# ============================================================
# 17. FINAL BATCH RESULT
# ============================================================

print("\n" + "=" * 100)
print("🏆 BATCH 04 REPAIR RESULT")
print("=" * 100)

print("""
🟢 BATCH 04 IS COMPLETE

✅ 32 questions
✅ FAQ0321 → FAQ0352
✅ Ground truth complete
✅ Answers complete
✅ 8 Arabic
✅ 8 English
✅ 8 French
✅ 8 Russian
✅ 4 questions per intent
✅ Required fields complete
✅ Intent mapping correct
✅ Question types valid
✅ No duplicate questions
✅ No overlap with existing 320

⚠️ hotel_faq_final.csv was NOT modified.

🎉 SAFE FOR FINAL BATCH 04 VALIDATION
""")

print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 04 FINAL CHECK
# GROUND TRUTH + ANSWER + DUPLICATE + SIMILARITY
# READ-ONLY
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata
from difflib import SequenceMatcher

print("=" * 100)
print("🏨 HOTEL AI - BATCH 04 FINAL CHECK")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_04.csv"
)

assert FINAL_DATASET.exists(), "❌ Final dataset not found."
assert BATCH_FILE.exists(), "❌ Batch 04 not found."

df = pd.read_csv(FINAL_DATASET, encoding="utf-8-sig")
batch_df = pd.read_csv(BATCH_FILE, encoding="utf-8-sig")

# ============================================================
# 2. BASIC FILE CHECK
# ============================================================

print("\n" + "=" * 100)
print("📊 FILE INFORMATION")
print("=" * 100)

print(f"Existing dataset : {len(df)}")
print(f"Batch 04         : {len(batch_df)}")

assert len(df) == 320
assert len(batch_df) == 32

print("✅ Existing dataset = 320")
print("✅ Batch 04 = 32")

# ============================================================
# 3. COLUMN CHECK
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

assert list(df.columns) == REQUIRED_COLUMNS
assert list(batch_df.columns) == REQUIRED_COLUMNS

print("✅ All 13 columns exist.")
print("✅ Column order correct.")

# ============================================================
# 4. ID CHECK
# ============================================================

print("\n" + "=" * 100)
print("🆔 FAQ ID CHECK")
print("=" * 100)

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(321, 353)
]

assert batch_df["faq_id"].tolist() == expected_ids
assert batch_df["faq_id"].is_unique

print("✅ IDs: FAQ0321 → FAQ0352")
print("✅ No duplicate IDs.")

# ============================================================
# 5. REQUIRED FIELD CHECK
# ============================================================

print("\n" + "=" * 100)
print("🧪 REQUIRED FIELD CHECK")
print("=" * 100)

for col in REQUIRED_COLUMNS:

    missing = int(batch_df[col].isna().sum())

    empty = int(
        batch_df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        f"✅ {col:<15}"
        f" Missing: {missing:<3}"
        f" Empty: {empty:<3}"
    )

    assert missing == 0
    assert empty == 0

print("✅ All required fields complete.")

# ============================================================
# 6. NORMALIZATION
# ============================================================

def normalize(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# 7. DUPLICATE CHECK INSIDE BATCH
# ============================================================

print("\n" + "=" * 100)
print("📝 BATCH DUPLICATE CHECK")
print("=" * 100)

assert not batch_df["question"].duplicated().any()

batch_norm = batch_df["question"].map(normalize)

assert not batch_norm.duplicated().any()

print("✅ No exact duplicates inside Batch 04.")
print("✅ No normalized duplicates inside Batch 04.")

# ============================================================
# 8. OVERLAP AGAINST 320
# ============================================================

print("\n" + "=" * 100)
print("🔎 OVERLAP WITH EXISTING 320")
print("=" * 100)

existing_norm = set(
    df["question"].map(normalize)
)

overlap = [
    q for q in batch_norm
    if q in existing_norm
]

assert not overlap, (
    f"❌ Overlap found:\n{overlap}"
)

print("✅ No exact/normalized overlap with existing 320.")

# ============================================================
# 9. LANGUAGE
# ============================================================

print("\n" + "=" * 100)
print("🌍 LANGUAGE CHECK")
print("=" * 100)

language_counts = batch_df["language"].value_counts()

print(language_counts)

assert language_counts.to_dict() == {
    "Arabic": 8,
    "English": 8,
    "French": 8,
    "Russian": 8,
}

print("✅ 8 × 4 languages.")

# ============================================================
# 10. INTENT
# ============================================================

print("\n" + "=" * 100)
print("🎯 INTENT CHECK")
print("=" * 100)

intent_counts = batch_df["intent_id"].value_counts()

print(intent_counts)

assert intent_counts.to_dict() == {
    "INT001": 4,
    "INT002": 4,
    "INT003": 4,
    "INT004": 4,
    "INT005": 4,
    "INT006": 4,
    "INT007": 4,
    "INT008": 4,
}

print("✅ 4 × 8 intents.")

# ============================================================
# 11. INTENT NAME
# ============================================================

INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

for _, row in batch_df.iterrows():

    assert row["intent_name"] == INTENTS[row["intent_id"]]

print("✅ Intent ID ↔ Intent Name consistency.")

# ============================================================
# 12. QUESTION TYPES
# ============================================================

VALID_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

assert (
    set(batch_df["question_type"])
    <= VALID_TYPES
)

print("\n" + "=" * 100)
print("📝 QUESTION TYPE CHECK")
print("=" * 100)

print(batch_df["question_type"].value_counts())
print("✅ All question types valid.")

# ============================================================
# 13. GROUND TRUTH + ANSWER
# ============================================================

print("\n" + "=" * 100)
print("🎯 GROUND TRUTH / ANSWER CHECK")
print("=" * 100)

assert (
    batch_df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

assert (
    batch_df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

print("✅ Ground truth complete.")
print("✅ Answers complete.")

# ============================================================
# 14. FULL SIMILARITY AGAINST 320
# ============================================================

print("\n" + "=" * 100)
print("🔍 BATCH 04 vs EXISTING 320 SIMILARITY")
print("=" * 100)

similar_pairs = []

for _, new_row in batch_df.iterrows():

    for _, old_row in df.iterrows():

        score = SequenceMatcher(
            None,
            normalize(new_row["question"]),
            normalize(old_row["question"])
        ).ratio()

        if score >= 0.80:

            similar_pairs.append({
                "similarity": round(score, 3),
                "new_faq_id": new_row["faq_id"],
                "new_intent": new_row["intent_id"],
                "new_language": new_row["language"],
                "new_question": new_row["question"],
                "old_faq_id": old_row["faq_id"],
                "old_intent": old_row["intent_id"],
                "old_language": old_row["language"],
                "old_question": old_row["question"],
            })

similarity_df = pd.DataFrame(similar_pairs)

if similarity_df.empty:

    print("✅ No pairs >= 0.80.")

else:

    similarity_df = similarity_df.sort_values(
        "similarity",
        ascending=False
    )

    print(
        f"Pairs >= 0.80: {len(similarity_df)}"
    )

    print(
        similarity_df.to_string(index=False)
    )

# ============================================================
# 15. THRESHOLDS
# ============================================================

print("\n" + "=" * 100)
print("📊 SIMILARITY THRESHOLDS")
print("=" * 100)

for threshold in [0.80, 0.85, 0.90, 0.95]:

    if similarity_df.empty:
        count = 0
    else:
        count = int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )

    print(
        f"{threshold:.2f}+ : {count}"
    )

# ============================================================
# 16. SAME INTENT
# ============================================================

if similarity_df.empty:
    same_intent = pd.DataFrame()
else:
    same_intent = similarity_df[
        similarity_df["new_intent"]
        ==
        similarity_df["old_intent"]
    ]

print("\n" + "=" * 100)
print("🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

print(
    f">= 0.80: {len(same_intent)}"
)

if not same_intent.empty:
    print(
        same_intent.to_string(index=False)
    )

# ============================================================
# 17. CROSS INTENT
# ============================================================

if similarity_df.empty:
    cross_intent = pd.DataFrame()
else:
    cross_intent = similarity_df[
        similarity_df["new_intent"]
        !=
        similarity_df["old_intent"]
    ]

print("\n" + "=" * 100)
print("⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

print(
    f">= 0.80: {len(cross_intent)}"
)

if not cross_intent.empty:
    print(
        cross_intent.to_string(index=False)
    )

# ============================================================
# 18. FINAL DECISION
# ============================================================

high_similarity = (
    similarity_df[
        similarity_df["similarity"] >= 0.85
    ]
    if not similarity_df.empty
    else pd.DataFrame()
)

print("\n" + "=" * 100)
print("🏆 BATCH 04 FINAL RESULT")
print("=" * 100)

if (
    len(overlap) == 0
    and high_similarity.empty
    and cross_intent.empty
    and not batch_df["question"].duplicated().any()
    and not batch_norm.duplicated().any()
):

    print("""
🟢 BATCH 04 IS FULLY VALID

✅ 32 questions
✅ FAQ0321 → FAQ0352
✅ Required fields complete
✅ Ground truth complete
✅ Answers complete
✅ No duplicate questions
✅ No normalized overlap
✅ No similarity >= 0.85
✅ No cross-intent similarity >= 0.80
✅ Language balanced
✅ Intent balanced
✅ Intent mapping correct
✅ Question types valid

🎉 SAFE TO MERGE BATCH 04
""")

else:

    print("""
🟡 BATCH 04 STILL NEEDS REVIEW
""")

# ============================================================
# 19. READ-ONLY
# ============================================================

print("=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ faq_batch_04.csv was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

print("\n" + "=" * 100)
print("🎉 BATCH 04 FINAL CHECK COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - FULL DATASET QUALITY + SIMILARITY CHECK
# VERSION: 352 QUESTIONS
# READ-ONLY — DOES NOT MODIFY THE DATASET
# ============================================================

from pathlib import Path
from itertools import combinations
from difflib import SequenceMatcher
import pandas as pd
import re
import unicodedata

# ============================================================
# 1. CONFIG
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

EXPECTED_ROWS = 352

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

LANGUAGES = {
    "Arabic",
    "English",
    "French",
    "Russian",
}

QUESTION_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

STATUS_VALUES = {
    "valid",
    "generated",
}

DIFFICULTY_VALUES = {
    "Easy",
}

print("=" * 100)
print("🏨 HOTEL AI - FULL DATASET QUALITY + SIMILARITY CHECK")
print("=" * 100)

# ============================================================
# 2. PATH
# ============================================================

assert PROJECT_ROOT.exists(), (
    f"❌ Project root not found:\n{PROJECT_ROOT}"
)

assert DATASET_PATH.exists(), (
    f"❌ Dataset not found:\n{DATASET_PATH}"
)

print("\n📂 Project root:")
print(PROJECT_ROOT)

print("\n📄 Dataset:")
print(DATASET_PATH)

print("\n✅ Project root found.")
print("✅ Dataset found.")

# ============================================================
# 3. LOAD
# ============================================================

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print("\n" + "=" * 100)
print("📊 DATASET INFORMATION")
print("=" * 100)

print(f"Rows    : {len(df)}")
print(f"Columns : {len(df.columns)}")

# ============================================================
# 4. COLUMN CHECK
# ============================================================

print("\n" + "=" * 100)
print("🧪 COLUMN VALIDATION")
print("=" * 100)

missing_columns = [
    c for c in REQUIRED_COLUMNS
    if c not in df.columns
]

extra_columns = [
    c for c in df.columns
    if c not in REQUIRED_COLUMNS
]

assert not missing_columns, (
    f"❌ Missing columns: {missing_columns}"
)

assert not extra_columns, (
    f"❌ Unexpected columns: {extra_columns}"
)

assert list(df.columns) == REQUIRED_COLUMNS, (
    "❌ Column order is incorrect."
)

print("✅ All 13 required columns exist.")
print("✅ Column order is correct.")

# ============================================================
# 5. ROW COUNT
# ============================================================

print("\n" + "=" * 100)
print("📦 ROW COUNT")
print("=" * 100)

assert len(df) == EXPECTED_ROWS, (
    f"❌ Expected {EXPECTED_ROWS}, found {len(df)}"
)

print(f"✅ Exactly {EXPECTED_ROWS} questions.")

# ============================================================
# 6. REQUIRED FIELD CHECK
# ============================================================

print("\n" + "=" * 100)
print("🧪 REQUIRED FIELD VALIDATION")
print("=" * 100)

required_fields_ok = True

for col in REQUIRED_COLUMNS:

    missing = int(df[col].isna().sum())

    empty = int(
        df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        f"✅ {col:<15}"
        f"Missing: {missing:<3}"
        f"Empty: {empty:<3}"
    )

    if missing > 0 or empty > 0:
        required_fields_ok = False

assert required_fields_ok, (
    "❌ Required fields contain missing or empty values."
)

print("\n✅ All required fields complete.")

# ============================================================
# 7. FAQ ID FORMAT
# ============================================================

print("\n" + "=" * 100)
print("🆔 FAQ ID VALIDATION")
print("=" * 100)

valid_id_mask = (
    df["faq_id"]
    .astype(str)
    .str.fullmatch(r"FAQ\d{4}")
)

invalid_ids = df[~valid_id_mask]

assert invalid_ids.empty, (
    f"❌ Invalid FAQ IDs:\n"
    f"{invalid_ids[['faq_id']]}"
)

print("✅ All FAQ IDs have valid format.")

# ============================================================
# 8. DUPLICATE IDS
# ============================================================

duplicate_ids = df[
    df["faq_id"].duplicated(keep=False)
]

assert duplicate_ids.empty, (
    f"❌ Duplicate FAQ IDs:\n"
    f"{duplicate_ids[['faq_id']]}"
)

print("✅ No duplicate FAQ IDs.")

# ============================================================
# 9. ID CONTINUITY
# ============================================================

print("\n" + "=" * 100)
print("🔢 FAQ ID CONTINUITY")
print("=" * 100)

numeric_ids = (
    df["faq_id"]
    .astype(str)
    .str.extract(r"FAQ(\d{4})")[0]
    .astype(int)
)

expected_numeric_ids = list(
    range(1, EXPECTED_ROWS + 1)
)

assert numeric_ids.tolist() == expected_numeric_ids, (
    "❌ FAQ IDs are not continuous or ordered."
)

print(
    f"✅ FAQ0001 → FAQ{EXPECTED_ROWS:04d}"
)
print("✅ IDs are continuous and correctly ordered.")

# ============================================================
# 10. NORMALIZATION
# ============================================================

def normalize_text(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


df["_normalized_question"] = (
    df["question"]
    .map(normalize_text)
)

# ============================================================
# 11. EXACT DUPLICATES
# ============================================================

print("\n" + "=" * 100)
print("📝 EXACT QUESTION DUPLICATE CHECK")
print("=" * 100)

exact_duplicates = df[
    df["question"].duplicated(keep=False)
]

if exact_duplicates.empty:

    print("✅ No exact duplicate questions.")

else:

    print(
        exact_duplicates[
            ["faq_id", "intent_id", "language", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 12. NORMALIZED DUPLICATES
# ============================================================

print("\n" + "=" * 100)
print("🔎 NORMALIZED DUPLICATE CHECK")
print("=" * 100)

normalized_duplicates = df[
    df["_normalized_question"].duplicated(keep=False)
]

if normalized_duplicates.empty:

    print("✅ No normalized duplicate questions.")

else:

    print(
        normalized_duplicates[
            [
                "faq_id",
                "intent_id",
                "language",
                "question"
            ]
        ].to_string(index=False)
    )

# ============================================================
# 13. SAME INTENT DUPLICATES
# ============================================================

print("\n" + "=" * 100)
print("🔍 DUPLICATES INSIDE SAME INTENT")
print("=" * 100)

same_intent_duplicates = df[
    df.duplicated(
        subset=[
            "intent_id",
            "_normalized_question"
        ],
        keep=False
    )
]

if same_intent_duplicates.empty:

    print(
        "✅ No normalized duplicates "
        "inside the same intent."
    )

else:

    print(
        same_intent_duplicates[
            [
                "faq_id",
                "intent_id",
                "language",
                "question"
            ]
        ].to_string(index=False)
    )

# ============================================================
# 14. INTENT MAPPING
# ============================================================

print("\n" + "=" * 100)
print("🔗 INTENT ID ↔ INTENT NAME")
print("=" * 100)

intent_errors = []

for intent_id, intent_name in INTENTS.items():

    subset = df[
        df["intent_id"] == intent_id
    ]

    if len(subset) != 44:
        intent_errors.append(
            f"{intent_id}: expected 44, found {len(subset)}"
        )

    names = set(
        subset["intent_name"]
        .astype(str)
    )

    if names != {intent_name}:
        intent_errors.append(
            f"{intent_id}: expected {intent_name}, found {names}"
        )

unexpected_intents = (
    set(df["intent_id"])
    - set(INTENTS.keys())
)

if unexpected_intents:
    intent_errors.append(
        f"Unexpected intents: {unexpected_intents}"
    )

assert not intent_errors, (
    "❌ Intent validation errors:\n"
    + "\n".join(intent_errors)
)

print("✅ All intent mappings are correct.")
print("✅ 44 questions × 8 intents.")

# ============================================================
# 15. LANGUAGE
# ============================================================

print("\n" + "=" * 100)
print("🌍 LANGUAGE DISTRIBUTION")
print("=" * 100)

language_counts = df["language"].value_counts()

print(language_counts.to_string())

assert set(language_counts.index) == LANGUAGES

for language in LANGUAGES:

    count = int(
        language_counts.get(language, 0)
    )

    assert count == 88, (
        f"❌ {language}: expected 88, found {count}"
    )

print("✅ 88 questions × 4 languages.")

# ============================================================
# 16. LANGUAGE × INTENT
# ============================================================

print("\n" + "=" * 100)
print("📊 LANGUAGE × INTENT MATRIX")
print("=" * 100)

language_intent = pd.crosstab(
    df["intent_id"],
    df["language"]
)

print(language_intent.to_string())

for intent_id in INTENTS:

    for language in LANGUAGES:

        count = int(
            language_intent.loc[
                intent_id,
                language
            ]
        )

        assert count == 11, (
            f"❌ {intent_id} × {language}: "
            f"expected 11, found {count}"
        )

print(
    "\n✅ Every intent has "
    "11 questions per language."
)

# ============================================================
# 17. QUESTION TYPES
# ============================================================

print("\n" + "=" * 100)
print("📝 QUESTION TYPE DISTRIBUTION")
print("=" * 100)

question_type_counts = (
    df["question_type"]
    .value_counts()
)

print(
    question_type_counts.to_string()
)

unexpected_types = (
    set(df["question_type"])
    - QUESTION_TYPES
)

assert not unexpected_types

print("✅ All question types are valid.")

# ============================================================
# 18. STATUS
# ============================================================

print("\n" + "=" * 100)
print("📌 STATUS DISTRIBUTION")
print("=" * 100)

status_counts = df["status"].value_counts()

print(
    status_counts.to_string()
)

unexpected_status = (
    set(df["status"])
    - STATUS_VALUES
)

assert not unexpected_status

print("✅ Status values are valid.")

# ============================================================
# 19. DIFFICULTY
# ============================================================

print("\n" + "=" * 100)
print("📌 DIFFICULTY DISTRIBUTION")
print("=" * 100)

difficulty_counts = df["difficulty"].value_counts()

print(
    difficulty_counts.to_string()
)

unexpected_difficulty = (
    set(df["difficulty"])
    - DIFFICULTY_VALUES
)

assert not unexpected_difficulty

print("✅ Difficulty values are valid.")

# ============================================================
# 20. GROUND TRUTH / ANSWER
# ============================================================

print("\n" + "=" * 100)
print("🎯 GROUND TRUTH / ANSWER")
print("=" * 100)

gt_empty = (
    df["ground_truth"]
    .fillna("")
    .astype(str)
    .str.strip()
    .eq("")
)

answer_empty = (
    df["answer"]
    .fillna("")
    .astype(str)
    .str.strip()
    .eq("")
)

assert not gt_empty.any(), (
    "❌ Empty ground_truth found."
)

assert not answer_empty.any(), (
    "❌ Empty answer found."
)

print("✅ Ground truth complete.")
print("✅ Answers complete.")

# ============================================================
# 21. GROUND TRUTH vs ANSWER DIFFERENCE REPORT
# ============================================================

print("\n" + "=" * 100)
print("🔗 GROUND TRUTH ↔ ANSWER")
print("=" * 100)

gt_norm = (
    df["ground_truth"]
    .map(normalize_text)
)

answer_norm = (
    df["answer"]
    .map(normalize_text)
)

answer_difference = df[
    gt_norm != answer_norm
]

print(
    f"Rows with different wording: "
    f"{len(answer_difference)}"
)

print(
    "ℹ️ Different wording is not automatically an error."
)

# ============================================================
# 22. ATTEMPTS
# ============================================================

print("\n" + "=" * 100)
print("🔢 ATTEMPTS VALIDATION")
print("=" * 100)

attempts = pd.to_numeric(
    df["attempts"],
    errors="coerce"
)

assert not attempts.isna().any()
assert (attempts >= 0).all()

print("✅ Attempts are numeric and non-negative.")
print(f"Min    : {attempts.min()}")
print(f"Max    : {attempts.max()}")
print(f"Average: {attempts.mean():.2f}")

# ============================================================
# 23. QUESTION LENGTH
# ============================================================

print("\n" + "=" * 100)
print("📏 QUESTION LENGTH")
print("=" * 100)

question_lengths = (
    df["question"]
    .astype(str)
    .str.len()
)

print(f"Shortest : {question_lengths.min()}")
print(f"Longest  : {question_lengths.max()}")
print(f"Average  : {question_lengths.mean():.2f}")

very_short = df[
    question_lengths < 10
]

if very_short.empty:

    print("✅ No very short questions.")

else:

    print(
        f"⚠️ Very short questions: "
        f"{len(very_short)}"
    )

# ============================================================
# 24. FULL PAIRWISE SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🔍 FULL QUESTION SIMILARITY CHECK")
print("=" * 100)

questions = df["question"].astype(str).tolist()
faq_ids = df["faq_id"].tolist()
intents = df["intent_id"].tolist()
languages = df["language"].tolist()

total_comparisons = (
    EXPECTED_ROWS
    * (EXPECTED_ROWS - 1)
    // 2
)

print(
    f"Comparisons performed: "
    f"{total_comparisons:,}"
)

similar_pairs = []

for i, j in combinations(
    range(EXPECTED_ROWS),
    2
):

    score = SequenceMatcher(
        None,
        normalize_text(questions[i]),
        normalize_text(questions[j])
    ).ratio()

    if score >= 0.80:

        similar_pairs.append({
            "similarity": round(score, 3),

            "faq_id_1": faq_ids[i],
            "intent_1": intents[i],
            "language_1": languages[i],
            "question_1": questions[i],

            "faq_id_2": faq_ids[j],
            "intent_2": intents[j],
            "language_2": languages[j],
            "question_2": questions[j],
        })

similarity_df = pd.DataFrame(
    similar_pairs
)

if similarity_df.empty:

    print(
        "✅ No pairs with similarity >= 0.80."
    )

else:

    similarity_df = similarity_df.sort_values(
        "similarity",
        ascending=False
    ).reset_index(drop=True)

    print(
        f"Pairs >= 0.80: "
        f"{len(similarity_df)}"
    )

# ============================================================
# 25. THRESHOLD SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("📊 SIMILARITY THRESHOLD SUMMARY")
print("=" * 100)

for threshold in [
    0.80,
    0.85,
    0.90,
    0.95
]:

    if similarity_df.empty:

        count = 0

    else:

        count = int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )

    print(
        f"{threshold:.2f}+ : {count} pairs"
    )

# ============================================================
# 26. TOP SIMILAR QUESTIONS
# ============================================================

print("\n" + "=" * 100)
print("📋 TOP 30 MOST SIMILAR QUESTIONS")
print("=" * 100)

if similarity_df.empty:

    print("None.")

else:

    print(
        similarity_df.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 27. SAME-INTENT
# ============================================================

print("\n" + "=" * 100)
print("🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    same_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["intent_1"]
        ==
        similarity_df["intent_2"]
    ]

print(
    f"Same-intent pairs >= 0.80: "
    f"{len(same_intent)}"
)

if not same_intent.empty:

    print(
        same_intent.head(30)
        .to_string(index=False)
    )

# ============================================================
# 28. CROSS-INTENT
# ============================================================

print("\n" + "=" * 100)
print("⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    cross_intent = pd.DataFrame()

else:

    cross_intent = similarity_df[
        similarity_df["intent_1"]
        !=
        similarity_df["intent_2"]
    ]

print(
    f"Cross-intent pairs >= 0.80: "
    f"{len(cross_intent)}"
)

if not cross_intent.empty:

    print(
        cross_intent.head(30)
        .to_string(index=False)
    )

# ============================================================
# 29. SAME LANGUAGE
# ============================================================

print("\n" + "=" * 100)
print("🌍 SAME-LANGUAGE SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    same_language = pd.DataFrame()

else:

    same_language = similarity_df[
        similarity_df["language_1"]
        ==
        similarity_df["language_2"]
    ]

print(
    f"Same-language pairs >= 0.80: "
    f"{len(same_language)}"
)

# ============================================================
# 30. CROSS LANGUAGE
# ============================================================

print("\n" + "=" * 100)
print("🌐 CROSS-LANGUAGE SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    cross_language = pd.DataFrame()

else:

    cross_language = similarity_df[
        similarity_df["language_1"]
        !=
        similarity_df["language_2"]
    ]

print(
    f"Cross-language pairs >= 0.80: "
    f"{len(cross_language)}"
)

# ============================================================
# 31. HIGH SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🚨 HIGH-SIMILARITY REVIEW")
print("=" * 100)

if similarity_df.empty:

    high_similarity = pd.DataFrame()

else:

    high_similarity = similarity_df[
        similarity_df["similarity"] >= 0.85
    ]

if high_similarity.empty:

    print(
        "✅ No similarity >= 0.85."
    )

else:

    print(
        f"⚠️ {len(high_similarity)} "
        f"pairs >= 0.85"
    )

    print(
        high_similarity.to_string(
            index=False
        )
    )

# ============================================================
# 32. VERY HIGH
# ============================================================

print("\n" + "=" * 100)
print("🔴 VERY HIGH SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    very_high = pd.DataFrame()

else:

    very_high = similarity_df[
        similarity_df["similarity"] >= 0.90
    ]

print(
    f">= 0.90 : {len(very_high)}"
)

if not very_high.empty:

    print(
        very_high.to_string(
            index=False
        )
    )

# ============================================================
# 33. EXTREME
# ============================================================

print("\n" + "=" * 100)
print("🚨 EXTREME SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    extreme = pd.DataFrame()

else:

    extreme = similarity_df[
        similarity_df["similarity"] >= 0.95
    ]

print(
    f">= 0.95 : {len(extreme)}"
)

if not extreme.empty:

    print(
        extreme.to_string(
            index=False
        )
    )

# ============================================================
# 34. FINAL CHECKS
# ============================================================

structural_pass = all([
    len(df) == EXPECTED_ROWS,
    list(df.columns) == REQUIRED_COLUMNS,
    invalid_ids.empty,
    duplicate_ids.empty,
    numeric_ids.tolist() == expected_numeric_ids,
    required_fields_ok,
    exact_duplicates.empty,
    normalized_duplicates.empty,
    same_intent_duplicates.empty,
    not intent_errors,
    set(language_counts.index) == LANGUAGES,
    all(
        (df["intent_id"] == x).sum() == 44
        for x in INTENTS
    ),
    all(
        language_intent.loc[x, y] == 11
        for x in INTENTS
        for y in LANGUAGES
    ),
    not unexpected_types,
    not unexpected_status,
    not unexpected_difficulty,
    not gt_empty.any(),
    not answer_empty.any(),
    very_short.empty,
])

similarity_pass = (
    similarity_df.empty
    or not (
        similarity_df["similarity"] >= 0.85
    ).any()
)

# ============================================================
# 35. FINAL SCORE
# ============================================================

checks = {
    "352 questions": len(df) == 352,
    "13 columns": len(df.columns) == 13,
    "Valid FAQ IDs": invalid_ids.empty,
    "No duplicate IDs": duplicate_ids.empty,
    "Continuous IDs": numeric_ids.tolist() == expected_numeric_ids,
    "Required fields": required_fields_ok,
    "No exact duplicates": exact_duplicates.empty,
    "No normalized duplicates": normalized_duplicates.empty,
    "No same-intent duplicates": same_intent_duplicates.empty,
    "Intent mappings": not intent_errors,
    "Language balance": all(
        language_counts.get(x, 0) == 88
        for x in LANGUAGES
    ),
    "Intent balance": all(
        (df["intent_id"] == x).sum() == 44
        for x in INTENTS
    ),
    "Language × Intent": all(
        language_intent.loc[x, y] == 11
        for x in INTENTS
        for y in LANGUAGES
    ),
    "Question types": not unexpected_types,
    "Status": not unexpected_status,
    "Difficulty": not unexpected_difficulty,
    "Ground truth": not gt_empty.any(),
    "Answers": not answer_empty.any(),
    "Similarity < 0.85": similarity_pass,
}

passed = sum(checks.values())
total = len(checks)
score = passed / total * 100

print("\n" + "=" * 100)
print("🏆 FINAL DATASET VALIDATION SCORE")
print("=" * 100)

for name, result in checks.items():

    print(
        f"{'✅' if result else '❌'} {name}"
    )

print("-" * 100)
print(f"Passed checks : {passed}/{total}")
print(f"Validation Score : {score:.1f}%")

if score == 100:
    print("🟢 DATASET IS FULLY VALIDATED")
else:
    print("🟡 DATASET NEEDS REVIEW")

# ============================================================
# 36. READ ONLY
# ============================================================

print("\n" + "=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

# ============================================================
# 37. CLEAN MEMORY ONLY
# ============================================================

df.drop(
    columns=[
        "_normalized_question"
    ],
    inplace=True,
    errors="ignore"
)

print("\n" + "=" * 100)
print("🎉 FULL 352-QUESTION CHECK COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - FINAL VALIDATION SCORE CHECK
# ALL 352 QUESTIONS
# READ-ONLY
# ============================================================

from pathlib import Path
from itertools import combinations
from difflib import SequenceMatcher
import pandas as pd
import re
import unicodedata

# ============================================================
# 1. PATH
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

assert PROJECT_ROOT.exists(), (
    f"❌ Project root not found:\n{PROJECT_ROOT}"
)

assert DATASET_PATH.exists(), (
    f"❌ Dataset not found:\n{DATASET_PATH}"
)

# ============================================================
# 2. LOAD
# ============================================================

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print("=" * 100)
print("🏨 HOTEL AI - FINAL DATASET VALIDATION")
print("=" * 100)

print(f"\n📄 Dataset:\n{DATASET_PATH}")
print(f"📊 Rows: {len(df)}")
print(f"📊 Columns: {len(df.columns)}")

# ============================================================
# 3. CONFIG
# ============================================================

EXPECTED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

EXPECTED_INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

EXPECTED_LANGUAGES = {
    "Arabic",
    "English",
    "French",
    "Russian",
}

VALID_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

VALID_STATUS = {
    "valid",
    "generated",
}

VALID_DIFFICULTY = {
    "Easy",
}

EXPECTED_ROWS = 352

# ============================================================
# 4. NORMALIZATION
# ============================================================

def normalize_question(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# 5. CHECKS
# ============================================================

checks = {}

# ------------------------------------------------------------
# Check 1 - Row count
# ------------------------------------------------------------

checks["352 questions"] = (
    len(df) == EXPECTED_ROWS
)

# ------------------------------------------------------------
# Check 2 - Columns
# ------------------------------------------------------------

checks["13 columns"] = (
    len(df.columns) == 13
    and df.columns.tolist() == EXPECTED_COLUMNS
)

# ------------------------------------------------------------
# Check 3 - FAQ ID format
# ------------------------------------------------------------

valid_id = (
    df["faq_id"]
    .astype(str)
    .str.fullmatch(r"FAQ\d{4}")
    .all()
)

checks["Valid FAQ IDs"] = bool(valid_id)

# ------------------------------------------------------------
# Check 4 - Duplicate IDs
# ------------------------------------------------------------

checks["No duplicate IDs"] = (
    not df["faq_id"].duplicated().any()
)

# ------------------------------------------------------------
# Check 5 - Continuous IDs
# ------------------------------------------------------------

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(1, 353)
]

checks["Continuous IDs"] = (
    df["faq_id"].tolist() == expected_ids
)

# ------------------------------------------------------------
# Check 6 - Required fields
# ------------------------------------------------------------

required_fields_ok = True

for col in EXPECTED_COLUMNS:

    if df[col].isna().any():
        required_fields_ok = False
        break

    if (
        df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .any()
    ):
        required_fields_ok = False
        break

checks["Required fields"] = required_fields_ok

# ------------------------------------------------------------
# Normalized questions
# ------------------------------------------------------------

df["_normalized"] = (
    df["question"]
    .map(normalize_question)
)

# ------------------------------------------------------------
# Check 7 - Exact duplicates
# ------------------------------------------------------------

checks["No exact duplicates"] = (
    not df["question"]
    .duplicated()
    .any()
)

# ------------------------------------------------------------
# Check 8 - Normalized duplicates
# ------------------------------------------------------------

checks["No normalized duplicates"] = (
    not df["_normalized"]
    .duplicated()
    .any()
)

# ------------------------------------------------------------
# Check 9 - Same-intent duplicates
# ------------------------------------------------------------

same_intent_duplicates = df[
    df.duplicated(
        subset=[
            "intent_id",
            "_normalized"
        ],
        keep=False
    )
]

checks["No same-intent duplicates"] = (
    same_intent_duplicates.empty
)

# ------------------------------------------------------------
# Check 10 - Intent mappings
# ------------------------------------------------------------

intent_mapping_ok = True

for intent_id, intent_name in EXPECTED_INTENTS.items():

    subset = df[
        df["intent_id"] == intent_id
    ]

    if len(subset) != 44:
        intent_mapping_ok = False
        break

    if set(subset["intent_name"]) != {intent_name}:
        intent_mapping_ok = False
        break

checks["Intent mappings"] = intent_mapping_ok

# ------------------------------------------------------------
# Check 11 - Language balance
# ------------------------------------------------------------

language_counts = df["language"].value_counts()

language_balance_ok = (
    set(language_counts.index)
    == EXPECTED_LANGUAGES
    and all(
        language_counts.get(lang, 0) == 88
        for lang in EXPECTED_LANGUAGES
    )
)

checks["Language balance"] = language_balance_ok

# ------------------------------------------------------------
# Check 12 - Intent balance
# ------------------------------------------------------------

intent_balance_ok = all(
    (
        df["intent_id"] == intent_id
    ).sum() == 44
    for intent_id in EXPECTED_INTENTS
)

checks["Intent balance"] = intent_balance_ok

# ------------------------------------------------------------
# Check 13 - Language × Intent
# ------------------------------------------------------------

matrix = pd.crosstab(
    df["intent_id"],
    df["language"]
)

language_intent_ok = True

for intent_id in EXPECTED_INTENTS:

    for language in EXPECTED_LANGUAGES:

        if (
            matrix.loc[
                intent_id,
                language
            ] != 11
        ):
            language_intent_ok = False

checks["Language × Intent"] = language_intent_ok

# ------------------------------------------------------------
# Check 14 - Question types
# ------------------------------------------------------------

checks["Question types"] = (
    set(df["question_type"])
    <= VALID_TYPES
)

# ------------------------------------------------------------
# Check 15 - Status
# ------------------------------------------------------------

checks["Status"] = (
    set(df["status"])
    <= VALID_STATUS
)

# ------------------------------------------------------------
# Check 16 - Difficulty
# ------------------------------------------------------------

checks["Difficulty"] = (
    set(df["difficulty"])
    <= VALID_DIFFICULTY
)

# ------------------------------------------------------------
# Check 17 - Ground truth
# ------------------------------------------------------------

checks["Ground truth"] = (
    df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

# ------------------------------------------------------------
# Check 18 - Answers
# ------------------------------------------------------------

checks["Answers"] = (
    df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

# ============================================================
# 6. SIMILARITY CHECK
# ============================================================

print("\n" + "=" * 100)
print("🔍 SIMILARITY CHECK")
print("=" * 100)

questions = df["question"].astype(str).tolist()

total_pairs = (
    len(questions)
    * (len(questions) - 1)
    // 2
)

print(
    f"Comparing {total_pairs:,} question pairs..."
)

high_similarity_pairs = []

for i, j in combinations(
    range(len(questions)),
    2
):

    score = SequenceMatcher(
        None,
        questions[i].casefold(),
        questions[j].casefold()
    ).ratio()

    if score >= 0.85:

        high_similarity_pairs.append({
            "faq_id_1": df.iloc[i]["faq_id"],
            "faq_id_2": df.iloc[j]["faq_id"],
            "intent_1": df.iloc[i]["intent_id"],
            "intent_2": df.iloc[j]["intent_id"],
            "similarity": round(score, 3),
            "question_1": questions[i],
            "question_2": questions[j],
        })

checks["Similarity < 0.85"] = (
    len(high_similarity_pairs) == 0
)

print(
    f"Pairs >= 0.85: "
    f"{len(high_similarity_pairs)}"
)

if high_similarity_pairs:

    similarity_df = pd.DataFrame(
        high_similarity_pairs
    ).sort_values(
        "similarity",
        ascending=False
    )

    print(
        similarity_df.to_string(
            index=False
        )
    )

else:

    print("✅ No similarity >= 0.85.")

# ============================================================
# 7. SCORE
# ============================================================

print("\n" + "=" * 100)
print("🏆 FINAL VALIDATION SCORE")
print("=" * 100)

for name, result in checks.items():

    print(
        f"{'✅' if result else '❌'} {name}"
    )

passed = sum(checks.values())
total = len(checks)
score = (
    passed / total * 100
)

print("-" * 100)
print(f"Passed checks      : {passed}/{total}")
print(
    f"Validation Score  : {score:.1f}%"
)

# ============================================================
# 8. FINAL RESULT
# ============================================================

if score == 100:

    print(
        "🟢 DATASET IS FULLY VALIDATED"
    )

else:

    print(
        "🟡 DATASET NEEDS REVIEW"
    )

# ============================================================
# 9. QUICK SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("📊 QUICK DATASET SUMMARY")
print("=" * 100)

print(f"Questions : {len(df)}")

print("\nLanguages:")
print(
    language_counts.to_string()
)

print("\nIntents:")
print(
    df.groupby(
        ["intent_id", "intent_name"]
    ).size()
    .to_string()
)

print("\nQuestion types:")
print(
    df["question_type"]
    .value_counts()
    .to_string()
)

# ============================================================
# 10. READ-ONLY
# ============================================================

print("\n" + "=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ Dataset was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

# Remove temp column from memory only
df.drop(
    columns=["_normalized"],
    inplace=True,
    errors="ignore"
)

print("\n" + "=" * 100)
print("🎉 VALIDATION COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 05 vs EXISTING 352
# DUPLICATE + SIMILARITY CHECK
# READ-ONLY
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata
from difflib import SequenceMatcher

print("=" * 100)
print("🏨 HOTEL AI - BATCH 05 vs EXISTING 352")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/User/Desktop/HotelAI-Intelligent-Agent"
)

# Fallback to the actual path used in the notebook
if not PROJECT_ROOT.exists():
    PROJECT_ROOT = Path(
        "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
    )

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_05.csv"
)

assert FINAL_DATASET.exists(), (
    f"❌ Final dataset not found:\n{FINAL_DATASET}"
)

assert BATCH_FILE.exists(), (
    f"❌ Batch 05 not found:\n{BATCH_FILE}"
)

print(f"📂 Project root:\n{PROJECT_ROOT}")
print(f"📄 Existing dataset:\n{FINAL_DATASET}")
print(f"📄 Batch 05:\n{BATCH_FILE}")

# ============================================================
# 2. LOAD
# ============================================================

df = pd.read_csv(
    FINAL_DATASET,
    encoding="utf-8-sig"
)

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

print("\n" + "=" * 100)
print("📊 FILE INFORMATION")
print("=" * 100)

print(f"Existing questions : {len(df)}")
print(f"Batch 05 questions : {len(batch_df)}")

assert len(df) == 352
assert len(batch_df) == 32

print("✅ Existing dataset = 352")
print("✅ Batch 05 = 32")

# ============================================================
# 3. REQUIRED COLUMNS
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

assert list(df.columns) == REQUIRED_COLUMNS
assert list(batch_df.columns) == REQUIRED_COLUMNS

print("✅ All 13 columns exist.")
print("✅ Column order is correct.")

# ============================================================
# 4. BATCH ID CHECK
# ============================================================

print("\n" + "=" * 100)
print("🆔 BATCH 05 FAQ ID CHECK")
print("=" * 100)

expected_batch_ids = [
    f"FAQ{i:04d}"
    for i in range(353, 385)
]

assert batch_df["faq_id"].tolist() == expected_batch_ids
assert batch_df["faq_id"].is_unique

print("✅ IDs are exactly FAQ0353 → FAQ0384.")
print("✅ No duplicate IDs.")

# ============================================================
# 5. NORMALIZATION
# ============================================================

def normalize_question(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# 6. EXACT OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("📝 EXACT OVERLAP")
print("=" * 100)

existing_exact = set(
    df["question"]
    .astype(str)
    .str.strip()
)

batch_exact = set(
    batch_df["question"]
    .astype(str)
    .str.strip()
)

exact_overlap = sorted(
    existing_exact.intersection(
        batch_exact
    )
)

if exact_overlap:

    print("❌ Exact overlaps found:")

    for q in exact_overlap:
        print(f"- {q}")

else:

    print("✅ No exact overlap.")

# ============================================================
# 7. NORMALIZED OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("🔎 NORMALIZED OVERLAP")
print("=" * 100)

existing_normalized = set(
    df["question"].map(normalize_question)
)

batch_normalized = set(
    batch_df["question"].map(normalize_question)
)

normalized_overlap = sorted(
    existing_normalized.intersection(
        batch_normalized
    )
)

if normalized_overlap:

    print("❌ Normalized overlaps found:")

    for q in normalized_overlap:
        print(f"- {q}")

else:

    print("✅ No normalized overlap.")

# ============================================================
# 8. DUPLICATES INSIDE BATCH
# ============================================================

print("\n" + "=" * 100)
print("📝 DUPLICATES INSIDE BATCH 05")
print("=" * 100)

batch_exact_duplicates = batch_df[
    batch_df["question"].duplicated(
        keep=False
    )
]

batch_norm_series = batch_df[
    "question"
].map(normalize_question)

batch_normalized_duplicates = batch_df[
    batch_norm_series.duplicated(
        keep=False
    )
]

if batch_exact_duplicates.empty:
    print("✅ No exact duplicates inside Batch 05.")
else:
    print(
        batch_exact_duplicates[
            ["faq_id", "question"]
        ].to_string(index=False)
    )

if batch_normalized_duplicates.empty:
    print("✅ No normalized duplicates inside Batch 05.")
else:
    print(
        batch_normalized_duplicates[
            ["faq_id", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 9. SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🔍 SEMANTIC / TEXT SIMILARITY")
print("=" * 100)

similar_pairs = []

for _, new_row in batch_df.iterrows():

    for _, old_row in df.iterrows():

        score = SequenceMatcher(
            None,
            normalize_question(
                new_row["question"]
            ),
            normalize_question(
                old_row["question"]
            )
        ).ratio()

        if score >= 0.80:

            similar_pairs.append({
                "similarity": round(score, 3),

                "new_faq_id": new_row["faq_id"],
                "new_intent": new_row["intent_id"],
                "new_language": new_row["language"],
                "new_question": new_row["question"],

                "old_faq_id": old_row["faq_id"],
                "old_intent": old_row["intent_id"],
                "old_language": old_row["language"],
                "old_question": old_row["question"],
            })

similarity_df = pd.DataFrame(similar_pairs)

if similarity_df.empty:

    print("🎉 No pairs with similarity >= 0.80.")

else:

    similarity_df = similarity_df.sort_values(
        "similarity",
        ascending=False
    ).reset_index(drop=True)

    print(
        f"Pairs with similarity >= 0.80: "
        f"{len(similarity_df)}"
    )

    print("\nTOP 30 SIMILAR PAIRS:")
    print(
        similarity_df.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 10. THRESHOLDS
# ============================================================

print("\n" + "=" * 100)
print("📊 SIMILARITY THRESHOLD SUMMARY")
print("=" * 100)

for threshold in [0.80, 0.85, 0.90, 0.95]:

    if similarity_df.empty:
        count = 0
    else:
        count = int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )

    print(
        f"{threshold:.2f}+ : {count} pairs"
    )

# ============================================================
# 11. SAME INTENT
# ============================================================

if similarity_df.empty:
    same_intent = pd.DataFrame()
else:
    same_intent = similarity_df[
        similarity_df["new_intent"]
        ==
        similarity_df["old_intent"]
    ]

print("\n" + "=" * 100)
print("🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

print(
    f"Same-intent pairs >= 0.80: "
    f"{len(same_intent)}"
)

if not same_intent.empty:

    print(
        same_intent.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 12. CROSS INTENT
# ============================================================

if similarity_df.empty:
    cross_intent = pd.DataFrame()
else:
    cross_intent = similarity_df[
        similarity_df["new_intent"]
        !=
        similarity_df["old_intent"]
    ]

print("\n" + "=" * 100)
print("⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

print(
    f"Cross-intent pairs >= 0.80: "
    f"{len(cross_intent)}"
)

if not cross_intent.empty:

    print(
        cross_intent.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 13. HIGH SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🚨 HIGH SIMILARITY REVIEW")
print("=" * 100)

if similarity_df.empty:

    high_similarity = pd.DataFrame()

else:

    high_similarity = similarity_df[
        similarity_df["similarity"] >= 0.85
    ]

print(
    f"Similarity >= 0.85: "
    f"{len(high_similarity)}"
)

if high_similarity.empty:

    print("✅ No similarity >= 0.85.")

else:

    print(
        high_similarity.to_string(
            index=False
        )
    )

# ============================================================
# 14. VERY HIGH
# ============================================================

print("\n" + "=" * 100)
print("🔴 VERY HIGH SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    very_high = pd.DataFrame()

else:

    very_high = similarity_df[
        similarity_df["similarity"] >= 0.90
    ]

print(
    f"Similarity >= 0.90: "
    f"{len(very_high)}"
)

if not very_high.empty:

    print(
        very_high.to_string(
            index=False
        )
    )

# ============================================================
# 15. EXTREME
# ============================================================

print("\n" + "=" * 100)
print("🚨 EXTREME SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    extreme = pd.DataFrame()

else:

    extreme = similarity_df[
        similarity_df["similarity"] >= 0.95
    ]

print(
    f"Similarity >= 0.95: "
    f"{len(extreme)}"
)

if not extreme.empty:

    print(
        extreme.to_string(
            index=False
        )
    )

# ============================================================
# 16. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("🏆 BATCH 05 SIMILARITY SUMMARY")
print("=" * 100)

print(
    f"""
Existing questions          : 352
Batch 05 questions           : 32

Exact overlaps               : {len(exact_overlap)}
Normalized overlaps          : {len(normalized_overlap)}

Pairs >= 0.80                : {len(similarity_df)}
Same-intent pairs >= 0.80    : {len(same_intent)}
Cross-intent pairs >= 0.80   : {len(cross_intent)}

Pairs >= 0.85                : {len(high_similarity)}
Pairs >= 0.90                : {len(very_high)}
Pairs >= 0.95                : {len(extreme)}
"""
)

# ============================================================
# 17. DECISION
# ============================================================

print("=" * 100)
print("🏁 BATCH 05 DECISION")
print("=" * 100)

if (
    len(exact_overlap) == 0
    and len(normalized_overlap) == 0
    and batch_exact_duplicates.empty
    and batch_normalized_duplicates.empty
    and len(high_similarity) == 0
    and len(cross_intent) == 0
):

    print("""
🟢 BATCH 05 IS READY FOR THE NEXT STEP

✅ No exact overlap
✅ No normalized overlap
✅ No duplicates inside Batch 05
✅ No similarity >= 0.85
✅ No cross-intent similarity >= 0.80
""")

else:

    print("""
🟡 BATCH 05 STILL NEEDS REVIEW

⚠️ One or more duplicate/similarity checks
require attention.
""")

# ============================================================
# 18. READ-ONLY
# ============================================================

print("=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ faq_batch_05.csv was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

print("\n" + "=" * 100)
print("🎉 BATCH 05 SIMILARITY CHECK COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 05 FINAL SIMILARITY CHECK
# Existing dataset : 352
# Batch 05         : 32
# READ-ONLY
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata
from difflib import SequenceMatcher

print("=" * 100)
print("🏨 HOTEL AI - BATCH 05 FINAL SIMILARITY CHECK")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_05.csv"
)

assert PROJECT_ROOT.exists(), "❌ Project root not found."
assert FINAL_DATASET.exists(), "❌ Final dataset not found."
assert BATCH_FILE.exists(), "❌ Batch 05 not found."

# ============================================================
# 2. LOAD
# ============================================================

df = pd.read_csv(
    FINAL_DATASET,
    encoding="utf-8-sig"
)

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

print(f"📂 Project root:\n{PROJECT_ROOT}")
print(f"📄 Existing dataset:\n{FINAL_DATASET}")
print(f"📄 Batch 05:\n{BATCH_FILE}")

print("\n" + "=" * 100)
print("📊 FILE INFORMATION")
print("=" * 100)

print(f"Existing questions : {len(df)}")
print(f"Batch 05 questions : {len(batch_df)}")

assert len(df) == 352, (
    f"❌ Existing dataset must contain 352 rows, found {len(df)}"
)

assert len(batch_df) == 32, (
    f"❌ Batch 05 must contain 32 rows, found {len(batch_df)}"
)

print("✅ Existing dataset = 352")
print("✅ Batch 05 = 32")

# ============================================================
# 3. COLUMNS
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

assert list(df.columns) == REQUIRED_COLUMNS
assert list(batch_df.columns) == REQUIRED_COLUMNS

print("✅ All 13 columns exist.")
print("✅ Column order is correct.")

# ============================================================
# 4. BATCH ID CHECK
# ============================================================

print("\n" + "=" * 100)
print("🆔 BATCH 05 FAQ ID CHECK")
print("=" * 100)

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(353, 385)
]

assert batch_df["faq_id"].tolist() == expected_ids
assert batch_df["faq_id"].is_unique

print("✅ IDs are exactly FAQ0353 → FAQ0384.")
print("✅ No duplicate IDs.")

# ============================================================
# 5. NORMALIZATION
# ============================================================

def normalize_question(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# 6. EXACT OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("📝 EXACT OVERLAP")
print("=" * 100)

existing_exact = set(
    df["question"]
    .astype(str)
    .str.strip()
)

batch_exact = set(
    batch_df["question"]
    .astype(str)
    .str.strip()
)

exact_overlap = sorted(
    existing_exact.intersection(
        batch_exact
    )
)

if exact_overlap:

    print("❌ Exact overlaps found:")

    for q in exact_overlap:
        print(f"- {q}")

else:

    print("✅ No exact overlap.")

# ============================================================
# 7. NORMALIZED OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("🔎 NORMALIZED OVERLAP")
print("=" * 100)

existing_normalized = set(
    df["question"].map(normalize_question)
)

batch_normalized = set(
    batch_df["question"].map(normalize_question)
)

normalized_overlap = sorted(
    existing_normalized.intersection(
        batch_normalized
    )
)

if normalized_overlap:

    print("❌ Normalized overlaps found:")

    for q in normalized_overlap:
        print(f"- {q}")

else:

    print("✅ No normalized overlap.")

# ============================================================
# 8. DUPLICATES INSIDE BATCH
# ============================================================

print("\n" + "=" * 100)
print("📝 DUPLICATES INSIDE BATCH 05")
print("=" * 100)

exact_dupes = batch_df[
    batch_df["question"].duplicated(keep=False)
]

batch_norm_series = (
    batch_df["question"]
    .map(normalize_question)
)

normalized_dupes = batch_df[
    batch_norm_series.duplicated(keep=False)
]

if exact_dupes.empty:
    print("✅ No exact duplicates inside Batch 05.")
else:
    print(
        exact_dupes[
            ["faq_id", "question"]
        ].to_string(index=False)
    )

if normalized_dupes.empty:
    print("✅ No normalized duplicates inside Batch 05.")
else:
    print(
        normalized_dupes[
            ["faq_id", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 9. SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🔍 SEMANTIC / TEXT SIMILARITY")
print("=" * 100)

similar_pairs = []

for _, new_row in batch_df.iterrows():

    for _, old_row in df.iterrows():

        score = SequenceMatcher(
            None,
            normalize_question(new_row["question"]),
            normalize_question(old_row["question"])
        ).ratio()

        if score >= 0.80:

            similar_pairs.append({
                "similarity": round(score, 3),

                "new_faq_id": new_row["faq_id"],
                "new_intent": new_row["intent_id"],
                "new_language": new_row["language"],
                "new_question": new_row["question"],

                "old_faq_id": old_row["faq_id"],
                "old_intent": old_row["intent_id"],
                "old_language": old_row["language"],
                "old_question": old_row["question"],
            })

similarity_df = pd.DataFrame(similar_pairs)

if similarity_df.empty:

    print("🎉 No pairs with similarity >= 0.80.")

else:

    similarity_df = similarity_df.sort_values(
        "similarity",
        ascending=False
    ).reset_index(drop=True)

    print(
        f"Pairs with similarity >= 0.80: "
        f"{len(similarity_df)}"
    )

    print("\nTOP SIMILAR PAIRS:")

    print(
        similarity_df.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 10. THRESHOLD SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("📊 SIMILARITY THRESHOLD SUMMARY")
print("=" * 100)

for threshold in [0.80, 0.85, 0.90, 0.95]:

    if similarity_df.empty:
        count = 0
    else:
        count = int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )

    print(f"{threshold:.2f}+ : {count} pairs")

# ============================================================
# 11. SAME INTENT
# ============================================================

print("\n" + "=" * 100)
print("🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    same_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["new_intent"]
        ==
        similarity_df["old_intent"]
    ]

print(
    f"Same-intent pairs >= 0.80: "
    f"{len(same_intent)}"
)

if not same_intent.empty:

    print(
        same_intent.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 12. CROSS INTENT
# ============================================================

print("\n" + "=" * 100)
print("⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    cross_intent = pd.DataFrame()

else:

    cross_intent = similarity_df[
        similarity_df["new_intent"]
        !=
        similarity_df["old_intent"]
    ]

print(
    f"Cross-intent pairs >= 0.80: "
    f"{len(cross_intent)}"
)

if not cross_intent.empty:

    print(
        cross_intent.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 13. HIGH SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🚨 HIGH SIMILARITY REVIEW")
print("=" * 100)

if similarity_df.empty:

    high_similarity = pd.DataFrame()

else:

    high_similarity = similarity_df[
        similarity_df["similarity"] >= 0.85
    ]

print(
    f"Similarity >= 0.85: "
    f"{len(high_similarity)}"
)

if high_similarity.empty:

    print("✅ No similarity >= 0.85.")

else:

    print(
        high_similarity.to_string(
            index=False
        )
    )

# ============================================================
# 14. VERY HIGH
# ============================================================

print("\n" + "=" * 100)
print("🔴 VERY HIGH SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    very_high = pd.DataFrame()

else:

    very_high = similarity_df[
        similarity_df["similarity"] >= 0.90
    ]

print(
    f"Similarity >= 0.90: "
    f"{len(very_high)}"
)

if not very_high.empty:

    print(
        very_high.to_string(
            index=False
        )
    )

# ============================================================
# 15. EXTREME
# ============================================================

print("\n" + "=" * 100)
print("🚨 EXTREME SIMILARITY")
print("=" * 100)

if similarity_df.empty:

    extreme = pd.DataFrame()

else:

    extreme = similarity_df[
        similarity_df["similarity"] >= 0.95
    ]

print(
    f"Similarity >= 0.95: "
    f"{len(extreme)}"
)

if not extreme.empty:

    print(
        extreme.to_string(
            index=False
        )
    )

# ============================================================
# 16. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("🏆 BATCH 05 FINAL SIMILARITY SUMMARY")
print("=" * 100)

print(
    f"""
Existing questions          : 352
Batch 05 questions           : 32

Exact overlaps               : {len(exact_overlap)}
Normalized overlaps          : {len(normalized_overlap)}

Pairs >= 0.80                : {len(similarity_df)}
Same-intent >= 0.80          : {len(same_intent)}
Cross-intent >= 0.80         : {len(cross_intent)}

Pairs >= 0.85                : {len(high_similarity)}
Pairs >= 0.90                : {len(very_high)}
Pairs >= 0.95                : {len(extreme)}
"""
)

# ============================================================
# 17. DECISION
# ============================================================

print("=" * 100)
print("🏁 BATCH 05 DECISION")
print("=" * 100)

if (
    len(exact_overlap) == 0
    and len(normalized_overlap) == 0
    and exact_dupes.empty
    and normalized_dupes.empty
    and len(high_similarity) == 0
    and len(cross_intent) == 0
):

    print("""
🟢 BATCH 05 IS READY FOR THE NEXT STEP

✅ No exact overlap
✅ No normalized overlap
✅ No duplicates inside Batch 05
✅ No similarity >= 0.85
✅ No cross-intent similarity >= 0.80
""")

else:

    print("""
🟡 BATCH 05 STILL NEEDS REVIEW

⚠️ One or more duplicate/similarity checks
require attention.
""")

# ============================================================
# 18. READ-ONLY
# ============================================================

print("=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ faq_batch_05.csv was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

print("\n" + "=" * 100)
print("🎉 BATCH 05 FINAL SIMILARITY CHECK COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 05 REPAIR
# GROUND TRUTH + ANSWER + VALIDATION
# ============================================================
# Modifies:
#   03_Structured_Data/CSV/batches/faq_batch_05.csv
#
# Does NOT modify:
#   03_Structured_Data/CSV/final/hotel_faq_final.csv
# ============================================================

from pathlib import Path
import pandas as pd
import re
import unicodedata

print("=" * 100)
print("🏨 HOTEL AI - BATCH 05 REPAIR + VALIDATION")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_05.csv"
)

assert PROJECT_ROOT.exists(), (
    f"❌ Project root not found:\n{PROJECT_ROOT}"
)

assert FINAL_DATASET.exists(), (
    f"❌ Final dataset not found:\n{FINAL_DATASET}"
)

assert BATCH_FILE.exists(), (
    f"❌ Batch 05 not found:\n{BATCH_FILE}"
)

print("✅ Project root found.")
print("✅ Final dataset found.")
print("✅ Batch 05 found.")

# ============================================================
# 2. LOAD
# ============================================================

df = pd.read_csv(
    FINAL_DATASET,
    encoding="utf-8-sig"
)

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

assert len(df) == 352
assert len(batch_df) == 32

print("\n" + "=" * 100)
print("📊 FILE INFORMATION")
print("=" * 100)

print(f"Existing dataset : {len(df)} rows")
print(f"Batch 05         : {len(batch_df)} rows")

# ============================================================
# 3. REQUIRED COLUMNS
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

assert list(df.columns) == REQUIRED_COLUMNS
assert list(batch_df.columns) == REQUIRED_COLUMNS

print("✅ All 13 columns exist.")
print("✅ Column order is correct.")

# ============================================================
# 4. FORCE CORRECT DTYPES
# ============================================================
# Prevents pandas FutureWarning when writing text into columns
# that may have been inferred as float64 because they were empty.

TEXT_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "difficulty",
    "answer",
]

for col in TEXT_COLUMNS:
    batch_df[col] = (
        batch_df[col]
        .fillna("")
        .astype("string")
    )

batch_df["attempts"] = pd.to_numeric(
    batch_df["attempts"],
    errors="raise"
)

# ============================================================
# 5. INTENTS
# ============================================================

INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

LANGUAGES = {
    "Arabic",
    "English",
    "French",
    "Russian",
}

# ============================================================
# 6. BATCH ID VALIDATION
# ============================================================

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(353, 385)
]

assert batch_df["faq_id"].tolist() == expected_ids
assert batch_df["faq_id"].is_unique

print("\n" + "=" * 100)
print("🆔 FAQ ID VALIDATION")
print("=" * 100)

print("✅ IDs are exactly FAQ0353 → FAQ0384.")
print("✅ No duplicate IDs.")

# ============================================================
# 7. NORMALIZATION
# ============================================================

def normalize_text(text):
    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# 8. DUPLICATE / OVERLAP CHECK
# ============================================================

batch_norm = batch_df["question"].map(
    normalize_text
)

existing_norm = set(
    df["question"].map(normalize_text)
)

assert not batch_df["question"].duplicated().any()
assert not batch_norm.duplicated().any()

overlap = [
    q for q in batch_norm
    if q in existing_norm
]

assert not overlap, (
    f"❌ Batch 05 overlaps with existing 352:\n{overlap}"
)

print("\n" + "=" * 100)
print("📝 QUESTION OVERLAP")
print("=" * 100)

print("✅ No exact duplicates inside Batch 05.")
print("✅ No normalized duplicates inside Batch 05.")
print("✅ No overlap with existing 352.")

# ============================================================
# 9. BUILD CANONICAL KNOWLEDGE MAP
# ============================================================
# Use the existing 352 as the source of truth.
#
# We pick the most common ground_truth and answer
# for each Intent × Language.

print("\n" + "=" * 100)
print("🧠 BUILDING KNOWLEDGE MAP")
print("=" * 100)

knowledge_map = {}

for intent_id, intent_name in INTENTS.items():

    for language in LANGUAGES:

        subset = df[
            (df["intent_id"] == intent_id)
            &
            (df["language"] == language)
        ].copy()

        assert len(subset) == 11, (
            f"❌ Expected 11 rows for "
            f"{intent_id} × {language}, "
            f"found {len(subset)}"
        )

        # -----------------------------
        # Ground truth candidates
        # -----------------------------

        gt = (
            subset["ground_truth"]
            .fillna("")
            .astype(str)
            .str.strip()
        )

        gt = gt[gt.ne("")]

        assert not gt.empty, (
            f"❌ No ground truth for "
            f"{intent_id} × {language}"
        )

        canonical_gt = (
            gt.value_counts()
            .index[0]
        )

        # -----------------------------
        # Answer candidates
        # -----------------------------

        answers = (
            subset["answer"]
            .fillna("")
            .astype(str)
            .str.strip()
        )

        answers = answers[
            answers.ne("")
        ]

        assert not answers.empty, (
            f"❌ No answer for "
            f"{intent_id} × {language}"
        )

        canonical_answer = (
            answers.value_counts()
            .index[0]
        )

        knowledge_map[
            (intent_id, language)
        ] = {
            "ground_truth": canonical_gt,
            "answer": canonical_answer,
        }

        print(
            f"✅ {intent_id:<6} "
            f"{language:<8}"
        )

print("✅ Knowledge map created.")

# ============================================================
# 10. APPLY GROUND TRUTH + ANSWERS
# ============================================================

print("\n" + "=" * 100)
print("✍️ APPLYING GROUND TRUTH + ANSWERS")
print("=" * 100)

for idx, row in batch_df.iterrows():

    intent_id = str(
        row["intent_id"]
    ).strip()

    language = str(
        row["language"]
    ).strip()

    assert intent_id in INTENTS
    assert language in LANGUAGES

    knowledge = knowledge_map[
        (intent_id, language)
    ]

    batch_df.at[
        idx,
        "ground_truth"
    ] = knowledge["ground_truth"]

    batch_df.at[
        idx,
        "answer"
    ] = knowledge["answer"]

    batch_df.at[
        idx,
        "status"
    ] = "generated"

    batch_df.at[
        idx,
        "category"
    ] = "Hotel FAQ"

    batch_df.at[
        idx,
        "source"
    ] = "Generated Batch 05"

    batch_df.at[
        idx,
        "attempts"
    ] = 0

    batch_df.at[
        idx,
        "difficulty"
    ] = "Easy"

print("✅ Ground truth filled for all 32.")
print("✅ Answers filled for all 32.")

# ============================================================
# 11. RECAST TEXT COLUMNS AFTER ASSIGNMENT
# ============================================================

for col in TEXT_COLUMNS:
    batch_df[col] = (
        batch_df[col]
        .fillna("")
        .astype("string")
    )

# ============================================================
# 12. REQUIRED FIELD VALIDATION
# ============================================================

print("\n" + "=" * 100)
print("🧪 REQUIRED FIELD VALIDATION")
print("=" * 100)

for col in REQUIRED_COLUMNS:

    missing = int(
        batch_df[col].isna().sum()
    )

    empty = int(
        batch_df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        f"✅ {col:<15}"
        f"Missing: {missing:<3}"
        f"Empty: {empty:<3}"
    )

    assert missing == 0
    assert empty == 0

print("✅ All required fields complete.")

# ============================================================
# 13. LANGUAGE CHECK
# ============================================================

print("\n" + "=" * 100)
print("🌍 LANGUAGE DISTRIBUTION")
print("=" * 100)

language_counts = (
    batch_df["language"]
    .value_counts()
    .sort_index()
)

print(language_counts)

assert language_counts.to_dict() == {
    "Arabic": 8,
    "English": 8,
    "French": 8,
    "Russian": 8,
}

print("✅ 8 questions × 4 languages.")

# ============================================================
# 14. INTENT CHECK
# ============================================================

print("\n" + "=" * 100)
print("🎯 INTENT DISTRIBUTION")
print("=" * 100)

intent_counts = (
    batch_df["intent_id"]
    .value_counts()
    .sort_index()
)

print(intent_counts)

assert intent_counts.to_dict() == {
    "INT001": 4,
    "INT002": 4,
    "INT003": 4,
    "INT004": 4,
    "INT005": 4,
    "INT006": 4,
    "INT007": 4,
    "INT008": 4,
}

print("✅ 4 questions × 8 intents.")

# ============================================================
# 15. INTENT NAME CHECK
# ============================================================

for _, row in batch_df.iterrows():

    assert (
        row["intent_name"]
        == INTENTS[row["intent_id"]]
    )

print("✅ Intent ID ↔ Intent Name consistency.")

# ============================================================
# 16. QUESTION TYPE CHECK
# ============================================================

VALID_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

assert (
    set(batch_df["question_type"])
    <= VALID_TYPES
)

print("\n" + "=" * 100)
print("📝 QUESTION TYPES")
print("=" * 100)

print(
    batch_df["question_type"]
    .value_counts()
)

print("✅ All question types are valid.")

# ============================================================
# 17. GROUND TRUTH / ANSWER CHECK
# ============================================================

print("\n" + "=" * 100)
print("🎯 GROUND TRUTH / ANSWER")
print("=" * 100)

assert (
    batch_df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

assert (
    batch_df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

print("✅ Ground truth complete.")
print("✅ Answers complete.")

# ============================================================
# 18. SAVE BATCH ONLY
# ============================================================

batch_df.to_csv(
    BATCH_FILE,
    index=False,
    encoding="utf-8-sig"
)

# ============================================================
# 19. RELOAD CHECK
# ============================================================

reloaded = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

assert len(reloaded) == 32

assert reloaded["faq_id"].tolist() == expected_ids

assert (
    reloaded["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

assert (
    reloaded["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

# ============================================================
# 20. FINAL RESULT
# ============================================================

print("\n" + "=" * 100)
print("🏆 BATCH 05 REPAIR RESULT")
print("=" * 100)

print("""
🟢 BATCH 05 IS COMPLETE

✅ 32 questions
✅ FAQ0353 → FAQ0384
✅ Ground truth complete
✅ Answers complete
✅ Required fields complete
✅ 8 Arabic
✅ 8 English
✅ 8 French
✅ 8 Russian
✅ 4 questions per intent
✅ Intent names consistent
✅ Question types valid
✅ No duplicate questions
✅ No normalized overlap with existing 352

⚠️ hotel_faq_final.csv was NOT modified.

🎉 SAFE FOR FINAL BATCH 05 VALIDATION
""")

print(f"📄 Batch saved to:")
print(BATCH_FILE)

print("=" * 100)
print("🎉 BATCH 05 REPAIR COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - FULL VALIDATION SCORE
# ALL 384 QUESTIONS
# READ-ONLY
# ============================================================

from pathlib import Path
from itertools import combinations
from difflib import SequenceMatcher
import pandas as pd
import re
import unicodedata

# ============================================================
# 1. CONFIG
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

EXPECTED_ROWS = 384

EXPECTED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

EXPECTED_INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

EXPECTED_LANGUAGES = {
    "Arabic",
    "English",
    "French",
    "Russian",
}

VALID_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

VALID_STATUS = {
    "valid",
    "generated",
}

VALID_DIFFICULTY = {
    "Easy",
}

print("=" * 100)
print("🏨 HOTEL AI - FULL DATASET VALIDATION")
print("=" * 100)

# ============================================================
# 2. LOAD
# ============================================================

assert PROJECT_ROOT.exists(), (
    f"❌ Project root not found:\n{PROJECT_ROOT}"
)

assert DATASET_PATH.exists(), (
    f"❌ Dataset not found:\n{DATASET_PATH}"
)

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print(f"\n📄 Dataset:\n{DATASET_PATH}")
print(f"📊 Rows    : {len(df)}")
print(f"📊 Columns : {len(df.columns)}")

# ============================================================
# 3. SCORE DICTIONARY
# ============================================================

checks = {}

# ============================================================
# 4. ROW COUNT
# ============================================================

checks["384 questions"] = (
    len(df) == EXPECTED_ROWS
)

# ============================================================
# 5. COLUMNS
# ============================================================

checks["13 columns"] = (
    len(df.columns) == 13
    and df.columns.tolist() == EXPECTED_COLUMNS
)

# ============================================================
# 6. REQUIRED FIELDS
# ============================================================

required_fields_ok = True

for col in EXPECTED_COLUMNS:

    if df[col].isna().any():
        required_fields_ok = False
        break

    if (
        df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .any()
    ):
        required_fields_ok = False
        break

checks["Required fields"] = required_fields_ok

# ============================================================
# 7. FAQ ID FORMAT
# ============================================================

valid_id_format = (
    df["faq_id"]
    .astype(str)
    .str.fullmatch(r"FAQ\d{4}")
    .all()
)

checks["Valid FAQ IDs"] = bool(valid_id_format)

# ============================================================
# 8. DUPLICATE IDs
# ============================================================

checks["No duplicate IDs"] = (
    not df["faq_id"].duplicated().any()
)

# ============================================================
# 9. CONTINUOUS IDs
# ============================================================

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(1, EXPECTED_ROWS + 1)
]

checks["Continuous IDs"] = (
    df["faq_id"].tolist()
    == expected_ids
)

# ============================================================
# 10. NORMALIZATION
# ============================================================

def normalize_question(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


df["_normalized"] = (
    df["question"]
    .map(normalize_question)
)

# ============================================================
# 11. EXACT DUPLICATES
# ============================================================

checks["No exact duplicates"] = (
    not df["question"].duplicated().any()
)

# ============================================================
# 12. NORMALIZED DUPLICATES
# ============================================================

checks["No normalized duplicates"] = (
    not df["_normalized"].duplicated().any()
)

# ============================================================
# 13. SAME-INTENT DUPLICATES
# ============================================================

same_intent_dupes = df[
    df.duplicated(
        subset=[
            "intent_id",
            "_normalized"
        ],
        keep=False
    )
]

checks["No same-intent duplicates"] = (
    same_intent_dupes.empty
)

# ============================================================
# 14. INTENT MAPPING
# ============================================================

intent_mapping_ok = True

for intent_id, intent_name in EXPECTED_INTENTS.items():

    subset = df[
        df["intent_id"] == intent_id
    ]

    if len(subset) != 48:
        intent_mapping_ok = False
        break

    if set(subset["intent_name"]) != {intent_name}:
        intent_mapping_ok = False
        break

checks["Intent mappings"] = intent_mapping_ok

# ============================================================
# 15. LANGUAGE BALANCE
# ============================================================

language_counts = df["language"].value_counts()

checks["Language balance"] = (
    set(language_counts.index) == EXPECTED_LANGUAGES
    and all(
        language_counts.get(lang, 0) == 96
        for lang in EXPECTED_LANGUAGES
    )
)

# ============================================================
# 16. INTENT BALANCE
# ============================================================

checks["Intent balance"] = all(
    (
        df["intent_id"] == intent_id
    ).sum() == 48
    for intent_id in EXPECTED_INTENTS
)

# ============================================================
# 17. LANGUAGE × INTENT
# ============================================================

matrix = pd.crosstab(
    df["intent_id"],
    df["language"]
)

checks["Language × Intent"] = all(
    matrix.loc[intent_id, language] == 12
    for intent_id in EXPECTED_INTENTS
    for language in EXPECTED_LANGUAGES
)

# ============================================================
# 18. QUESTION TYPES
# ============================================================

checks["Question types"] = (
    set(df["question_type"])
    <= VALID_TYPES
)

# ============================================================
# 19. STATUS
# ============================================================

checks["Status"] = (
    set(df["status"])
    <= VALID_STATUS
)

# ============================================================
# 20. DIFFICULTY
# ============================================================

checks["Difficulty"] = (
    set(df["difficulty"])
    <= VALID_DIFFICULTY
)

# ============================================================
# 21. GROUND TRUTH
# ============================================================

checks["Ground truth"] = (
    df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

# ============================================================
# 22. ANSWERS
# ============================================================

checks["Answers"] = (
    df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

# ============================================================
# 23. SIMILARITY CHECK
# ============================================================

print("\n" + "=" * 100)
print("🔍 FULL QUESTION SIMILARITY CHECK")
print("=" * 100)

questions = (
    df["question"]
    .astype(str)
    .tolist()
)

total_pairs = (
    EXPECTED_ROWS
    * (EXPECTED_ROWS - 1)
    // 2
)

print(
    f"Comparing {total_pairs:,} question pairs..."
)

high_similarity_pairs = []

for i, j in combinations(
    range(EXPECTED_ROWS),
    2
):

    score = SequenceMatcher(
        None,
        normalize_question(questions[i]),
        normalize_question(questions[j])
    ).ratio()

    if score >= 0.85:

        high_similarity_pairs.append({
            "similarity": round(score, 3),
            "faq_id_1": df.iloc[i]["faq_id"],
            "faq_id_2": df.iloc[j]["faq_id"],
            "intent_1": df.iloc[i]["intent_id"],
            "intent_2": df.iloc[j]["intent_id"],
            "question_1": questions[i],
            "question_2": questions[j],
        })

checks["Similarity < 0.85"] = (
    len(high_similarity_pairs) == 0
)

print(
    f"Pairs >= 0.85: {len(high_similarity_pairs)}"
)

if high_similarity_pairs:

    similarity_df = pd.DataFrame(
        high_similarity_pairs
    ).sort_values(
        "similarity",
        ascending=False
    )

    print(
        similarity_df.head(30).to_string(
            index=False
        )
    )

else:

    print("✅ No similarity >= 0.85.")

# ============================================================
# 24. FINAL SCORE
# ============================================================

print("\n" + "=" * 100)
print("🏆 FINAL VALIDATION SCORE")
print("=" * 100)

for name, result in checks.items():

    print(
        f"{'✅' if result else '❌'} {name}"
    )

passed = sum(checks.values())
total = len(checks)
score = (
    passed / total * 100
)

print("-" * 100)
print(f"Passed checks      : {passed}/{total}")
print(f"Validation Score   : {score:.1f}%")

if score == 100:

    print("🟢 DATASET IS FULLY VALIDATED")

else:

    print("🟡 DATASET NEEDS REVIEW")

# ============================================================
# 25. QUICK SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("📊 QUICK DATASET SUMMARY")
print("=" * 100)

print(f"Questions : {len(df)}")

print("\nLanguages:")
print(
    language_counts.to_string()
)

print("\nIntents:")
print(
    df.groupby(
        ["intent_id", "intent_name"]
    ).size()
    .to_string()
)

print("\nLanguage × Intent:")
print(
    matrix.to_string()
)

print("\nQuestion types:")
print(
    df["question_type"]
    .value_counts()
    .to_string()
)

print("\nStatus:")
print(
    df["status"]
    .value_counts()
    .to_string()
)

print("\nDifficulty:")
print(
    df["difficulty"]
    .value_counts()
    .to_string()
)

# ============================================================
# 26. READ-ONLY
# ============================================================

print("\n" + "=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ Dataset was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

# Remove temporary column from memory only
df.drop(
    columns=["_normalized"],
    inplace=True,
    errors="ignore"
)

print("\n" + "=" * 100)
print("🎉 VALIDATION COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - COMPLETE 384 DATASET AUDIT
# FULL QUALITY + DUPLICATE + SIMILARITY + BALANCE CHECK
# READ-ONLY — DOES NOT MODIFY CSV
# ============================================================

from pathlib import Path
from itertools import combinations
from difflib import SequenceMatcher
import pandas as pd
import numpy as np
import re
import unicodedata

# ============================================================
# 1. PATHS / CONFIG
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

EXPECTED_ROWS = 384

EXPECTED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

EXPECTED_INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

EXPECTED_LANGUAGES = {
    "Arabic",
    "English",
    "French",
    "Russian",
}

VALID_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

VALID_STATUS = {
    "valid",
    "generated",
}

VALID_DIFFICULTY = {
    "Easy",
}

SIMILARITY_THRESHOLD = 0.85

print("=" * 100)
print("🏨 HOTEL AI - COMPLETE 384 DATASET AUDIT")
print("=" * 100)

# ============================================================
# 2. PATH VALIDATION
# ============================================================

assert PROJECT_ROOT.exists(), (
    f"❌ Project root not found:\n{PROJECT_ROOT}"
)

assert DATASET_PATH.exists(), (
    f"❌ Dataset not found:\n{DATASET_PATH}"
)

print("\n📂 PROJECT")
print("=" * 100)
print(PROJECT_ROOT)

print("\n📄 DATASET")
print("=" * 100)
print(DATASET_PATH)

print("\n✅ Project root found.")
print("✅ Dataset found.")

# ============================================================
# 3. LOAD
# ============================================================

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print("\n📊 DATASET INFORMATION")
print("=" * 100)
print(f"Rows    : {len(df)}")
print(f"Columns : {len(df.columns)}")

# ============================================================
# 4. CHECK REGISTRY
# ============================================================

checks = {}

# ============================================================
# 5. COLUMN CHECK
# ============================================================

missing_columns = [
    col
    for col in EXPECTED_COLUMNS
    if col not in df.columns
]

extra_columns = [
    col
    for col in df.columns
    if col not in EXPECTED_COLUMNS
]

columns_ok = (
    not missing_columns
    and not extra_columns
    and df.columns.tolist() == EXPECTED_COLUMNS
)

checks["13 columns"] = columns_ok

print("\n🧪 COLUMN VALIDATION")
print("=" * 100)

if columns_ok:
    print("✅ All 13 required columns exist.")
    print("✅ Column order is correct.")
else:
    print("❌ Column validation failed.")
    print("Missing:", missing_columns)
    print("Extra:", extra_columns)

# ============================================================
# 6. ROW COUNT
# ============================================================

row_count_ok = (
    len(df) == EXPECTED_ROWS
)

checks["384 questions"] = row_count_ok

print("\n📦 ROW COUNT")
print("=" * 100)
print(f"Total questions: {len(df)}")

if row_count_ok:
    print("✅ Exactly 384 questions.")
else:
    print("❌ Expected 384 questions.")

# ============================================================
# 7. REQUIRED FIELD CHECK
# ============================================================

required_fields_ok = True

print("\n🧪 REQUIRED FIELD VALIDATION")
print("=" * 100)

if columns_ok:

    for col in EXPECTED_COLUMNS:

        missing = int(df[col].isna().sum())

        empty = int(
            df[col]
            .fillna("")
            .astype(str)
            .str.strip()
            .eq("")
            .sum()
        )

        print(
            f"{'✅' if missing == 0 and empty == 0 else '❌'} "
            f"{col:<15}"
            f"Missing: {missing:<3}"
            f"Empty: {empty:<3}"
        )

        if missing > 0 or empty > 0:
            required_fields_ok = False

else:
    required_fields_ok = False

checks["Required fields"] = required_fields_ok

# ============================================================
# 8. FAQ ID CHECK
# ============================================================

print("\n🆔 FAQ ID VALIDATION")
print("=" * 100)

if "faq_id" in df.columns:

    valid_id_format = (
        df["faq_id"]
        .astype(str)
        .str.fullmatch(r"FAQ\d{4}")
        .all()
    )

    duplicate_ids = (
        df[df["faq_id"].duplicated(keep=False)]
        .copy()
    )

    numeric_ids = pd.to_numeric(
        df["faq_id"]
        .astype(str)
        .str.extract(r"FAQ(\d{4})")[0],
        errors="coerce"
    )

    expected_numeric_ids = np.arange(
        1,
        EXPECTED_ROWS + 1
    )

    continuous_ids = (
        numeric_ids.notna().all()
        and numeric_ids.astype(int).tolist()
        == expected_numeric_ids.tolist()
    )

else:

    valid_id_format = False
    duplicate_ids = pd.DataFrame()
    continuous_ids = False

checks["Valid FAQ IDs"] = bool(valid_id_format)
checks["No duplicate IDs"] = duplicate_ids.empty
checks["Continuous IDs"] = bool(continuous_ids)

print(
    "✅ All FAQ IDs have valid format."
    if valid_id_format
    else "❌ Invalid FAQ ID format."
)

print(
    "✅ No duplicate FAQ IDs."
    if duplicate_ids.empty
    else f"❌ Duplicate FAQ IDs: {len(duplicate_ids)}"
)

print(
    "✅ IDs continuous: FAQ0001 → FAQ0384"
    if continuous_ids
    else "❌ FAQ IDs are not continuous/ordered."
)

# ============================================================
# 9. NORMALIZATION
# ============================================================

def normalize_text(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


df["_normalized_question"] = (
    df["question"]
    .map(normalize_text)
)

# ============================================================
# 10. EXACT DUPLICATES
# ============================================================

exact_duplicates = df[
    df["question"]
    .duplicated(keep=False)
].copy()

checks["No exact duplicates"] = (
    exact_duplicates.empty
)

print("\n📝 EXACT DUPLICATE CHECK")
print("=" * 100)

if exact_duplicates.empty:
    print("✅ No exact duplicate questions.")
else:
    print(
        exact_duplicates[
            ["faq_id", "intent_id", "language", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 11. NORMALIZED DUPLICATES
# ============================================================

normalized_duplicates = df[
    df["_normalized_question"]
    .duplicated(keep=False)
].copy()

checks["No normalized duplicates"] = (
    normalized_duplicates.empty
)

print("\n🔎 NORMALIZED DUPLICATE CHECK")
print("=" * 100)

if normalized_duplicates.empty:
    print("✅ No normalized duplicate questions.")
else:
    print(
        normalized_duplicates[
            ["faq_id", "intent_id", "language", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 12. SAME-INTENT DUPLICATES
# ============================================================

same_intent_duplicates = df[
    df.duplicated(
        subset=[
            "intent_id",
            "_normalized_question"
        ],
        keep=False
    )
].copy()

checks["No same-intent duplicates"] = (
    same_intent_duplicates.empty
)

print("\n🔍 DUPLICATES INSIDE SAME INTENT")
print("=" * 100)

if same_intent_duplicates.empty:
    print("✅ No normalized duplicates inside the same intent.")
else:
    print(
        same_intent_duplicates[
            [
                "faq_id",
                "intent_id",
                "language",
                "question"
            ]
        ].to_string(index=False)
    )

# ============================================================
# 13. INTENT MAPPING
# ============================================================

intent_mapping_ok = True

print("\n🔗 INTENT ID ↔ INTENT NAME")
print("=" * 100)

for intent_id, intent_name in EXPECTED_INTENTS.items():

    subset = df[
        df["intent_id"] == intent_id
    ]

    actual_names = set(
        subset["intent_name"]
        .astype(str)
    )

    if len(subset) != 48:
        intent_mapping_ok = False

    if actual_names != {intent_name}:
        intent_mapping_ok = False

    print(
        f"{'✅' if len(subset) == 48 and actual_names == {intent_name} else '❌'} "
        f"{intent_id} → {intent_name} | rows={len(subset)}"
    )

unexpected_intents = (
    set(df["intent_id"])
    - set(EXPECTED_INTENTS.keys())
)

if unexpected_intents:
    intent_mapping_ok = False
    print(
        f"❌ Unexpected intent IDs: {unexpected_intents}"
    )

checks["Intent mappings"] = intent_mapping_ok

# ============================================================
# 14. LANGUAGE DISTRIBUTION
# ============================================================

language_counts = (
    df["language"]
    .value_counts()
)

language_balance_ok = (
    set(language_counts.index)
    == EXPECTED_LANGUAGES
    and all(
        int(language_counts.get(language, 0)) == 96
        for language in EXPECTED_LANGUAGES
    )
)

checks["Language balance"] = language_balance_ok

print("\n🌍 LANGUAGE DISTRIBUTION")
print("=" * 100)
print(language_counts.to_string())

print(
    "✅ 96 questions × 4 languages."
    if language_balance_ok
    else "❌ Language balance failed."
)

# ============================================================
# 15. INTENT DISTRIBUTION
# ============================================================

intent_counts = (
    df["intent_id"]
    .value_counts()
)

intent_balance_ok = all(
    int(intent_counts.get(intent_id, 0)) == 48
    for intent_id in EXPECTED_INTENTS
)

checks["Intent balance"] = intent_balance_ok

print("\n🎯 INTENT DISTRIBUTION")
print("=" * 100)

print(
    df.groupby(
        ["intent_id", "intent_name"]
    )
    .size()
    .rename("questions")
    .reset_index()
    .to_string(index=False)
)

print(
    "✅ 48 questions × 8 intents."
    if intent_balance_ok
    else "❌ Intent balance failed."
)

# ============================================================
# 16. LANGUAGE × INTENT
# ============================================================

language_intent = pd.crosstab(
    df["intent_id"],
    df["language"]
)

language_intent_ok = True

for intent_id in EXPECTED_INTENTS:
    for language in EXPECTED_LANGUAGES:

        count = int(
            language_intent.loc[
                intent_id,
                language
            ]
        )

        if count != 12:
            language_intent_ok = False

checks["Language × Intent"] = language_intent_ok

print("\n📊 LANGUAGE × INTENT MATRIX")
print("=" * 100)
print(language_intent.to_string())

print(
    "✅ Every intent has 12 questions per language."
    if language_intent_ok
    else "❌ Language × Intent balance failed."
)

# ============================================================
# 17. QUESTION TYPES
# ============================================================

question_type_counts = (
    df["question_type"]
    .value_counts()
)

unexpected_types = (
    set(df["question_type"])
    - VALID_TYPES
)

question_types_ok = (
    not unexpected_types
    and question_type_counts.sum() == EXPECTED_ROWS
)

checks["Question types"] = question_types_ok

print("\n📝 QUESTION TYPE DISTRIBUTION")
print("=" * 100)
print(question_type_counts.to_string())

print(
    "✅ All question types are valid."
    if question_types_ok
    else f"❌ Invalid question types: {unexpected_types}"
)

# ============================================================
# 18. STATUS
# ============================================================

status_counts = (
    df["status"]
    .value_counts()
)

unexpected_status = (
    set(df["status"])
    - VALID_STATUS
)

status_ok = (
    not unexpected_status
)

checks["Status"] = status_ok

print("\n📌 STATUS DISTRIBUTION")
print("=" * 100)
print(status_counts.to_string())

print(
    "✅ Status values are valid."
    if status_ok
    else f"❌ Invalid status values: {unexpected_status}"
)

# ============================================================
# 19. DIFFICULTY
# ============================================================

difficulty_counts = (
    df["difficulty"]
    .value_counts()
)

unexpected_difficulty = (
    set(df["difficulty"])
    - VALID_DIFFICULTY
)

difficulty_ok = (
    not unexpected_difficulty
)

checks["Difficulty"] = difficulty_ok

print("\n📌 DIFFICULTY DISTRIBUTION")
print("=" * 100)
print(difficulty_counts.to_string())

print(
    "✅ Difficulty values are valid."
    if difficulty_ok
    else f"❌ Invalid difficulty: {unexpected_difficulty}"
)

# ============================================================
# 20. GROUND TRUTH / ANSWERS
# ============================================================

ground_truth_ok = (
    df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

answers_ok = (
    df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

checks["Ground truth"] = bool(ground_truth_ok)
checks["Answers"] = bool(answers_ok)

print("\n🎯 GROUND TRUTH / ANSWER")
print("=" * 100)

print(
    "✅ Ground truth complete."
    if ground_truth_ok
    else "❌ Ground truth contains empty values."
)

print(
    "✅ Answers complete."
    if answers_ok
    else "❌ Answers contain empty values."
)

# ============================================================
# 21. GROUND TRUTH / ANSWER WORDING DIFFERENCE
# ============================================================

gt_norm = (
    df["ground_truth"]
    .map(normalize_text)
)

answer_norm = (
    df["answer"]
    .map(normalize_text)
)

answer_wording_difference = (
    gt_norm != answer_norm
)

print("\n🔗 GROUND TRUTH ↔ ANSWER")
print("=" * 100)

print(
    "Rows with different wording:",
    int(answer_wording_difference.sum())
)

print(
    "ℹ️ Different wording alone is not treated as an error."
)

# ============================================================
# 22. ATTEMPTS
# ============================================================

attempts_numeric = pd.to_numeric(
    df["attempts"],
    errors="coerce"
)

attempts_ok = (
    attempts_numeric.notna().all()
    and (attempts_numeric >= 0).all()
)

checks["Attempts"] = attempts_ok

print("\n🔢 ATTEMPTS")
print("=" * 100)

print(
    "✅ Attempts are numeric and non-negative."
    if attempts_ok
    else "❌ Invalid attempts values."
)

print(
    f"Min    : {attempts_numeric.min()}"
)
print(
    f"Max    : {attempts_numeric.max()}"
)
print(
    f"Average: {attempts_numeric.mean():.2f}"
)

# ============================================================
# 23. QUESTION LENGTH
# ============================================================

question_lengths = (
    df["question"]
    .astype(str)
    .str.len()
)

very_short_questions = df[
    question_lengths < 10
].copy()

question_length_ok = (
    very_short_questions.empty
)

checks["Question length"] = question_length_ok

print("\n📏 QUESTION LENGTH")
print("=" * 100)

print(
    f"Shortest : {question_lengths.min()}"
)
print(
    f"Longest  : {question_lengths.max()}"
)
print(
    f"Average  : {question_lengths.mean():.2f}"
)

print(
    "✅ No suspiciously short questions."
    if question_length_ok
    else "⚠️ Suspiciously short questions found."
)

# ============================================================
# 24. FULL PAIRWISE SIMILARITY
# ============================================================

print("\n🔍 FULL QUESTION SIMILARITY")
print("=" * 100)

questions = (
    df["question"]
    .astype(str)
    .tolist()
)

total_pairs = (
    EXPECTED_ROWS
    * (EXPECTED_ROWS - 1)
    // 2
)

print(
    f"Comparing {total_pairs:,} question pairs..."
)

similarity_pairs = []

for i, j in combinations(
    range(EXPECTED_ROWS),
    2
):

    score = SequenceMatcher(
        None,
        normalize_text(questions[i]),
        normalize_text(questions[j])
    ).ratio()

    if score >= 0.80:

        similarity_pairs.append({
            "similarity": round(score, 3),
            "faq_id_1": df.iloc[i]["faq_id"],
            "intent_1": df.iloc[i]["intent_id"],
            "language_1": df.iloc[i]["language"],
            "question_1": questions[i],
            "faq_id_2": df.iloc[j]["faq_id"],
            "intent_2": df.iloc[j]["intent_id"],
            "language_2": df.iloc[j]["language"],
            "question_2": questions[j],
        })

similarity_df = pd.DataFrame(
    similarity_pairs
)

# ============================================================
# 25. SIMILARITY THRESHOLDS
# ============================================================

threshold_counts = {}

for threshold in [
    0.80,
    0.85,
    0.90,
    0.95,
]:

    if similarity_df.empty:
        count = 0
    else:
        count = int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )

    threshold_counts[threshold] = count

print("\n📊 SIMILARITY THRESHOLD SUMMARY")
print("=" * 100)

for threshold, count in threshold_counts.items():
    print(
        f"{threshold:.2f}+ : {count} pairs"
    )

# ============================================================
# 26. SAME / CROSS INTENT
# ============================================================

if similarity_df.empty:

    same_intent = pd.DataFrame()
    cross_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["intent_1"]
        ==
        similarity_df["intent_2"]
    ]

    cross_intent = similarity_df[
        similarity_df["intent_1"]
        !=
        similarity_df["intent_2"]
    ]

print("\n🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

print(
    f"Pairs >= 0.80: {len(same_intent)}"
)

print("\n⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

print(
    f"Pairs >= 0.80: {len(cross_intent)}"
)

# ============================================================
# 27. TOP SIMILAR PAIRS
# ============================================================

print("\n📋 TOP 30 MOST SIMILAR QUESTIONS")
print("=" * 100)

if similarity_df.empty:
    print("None.")
else:
    print(
        similarity_df
        .sort_values(
            "similarity",
            ascending=False
        )
        .head(30)
        .to_string(index=False)
    )

# ============================================================
# 28. HIGH SIMILARITY DECISION
# ============================================================

high_similarity = (
    similarity_df[
        similarity_df["similarity"]
        >= SIMILARITY_THRESHOLD
    ]
    if not similarity_df.empty
    else pd.DataFrame()
)

similarity_ok = (
    high_similarity.empty
)

checks["Similarity < 0.85"] = similarity_ok

print("\n🚨 HIGH-SIMILARITY REVIEW")
print("=" * 100)

if high_similarity.empty:

    print("✅ No similarity >= 0.85.")

else:

    print(
        f"❌ {len(high_similarity)} pairs >= 0.85"
    )

    print(
        high_similarity
        .sort_values(
            "similarity",
            ascending=False
        )
        .to_string(index=False)
    )

# ============================================================
# 29. SCORE
# ============================================================

print("\n" + "=" * 100)
print("🏆 FINAL VALIDATION SCORE")
print("=" * 100)

for name, result in checks.items():

    print(
        f"{'✅' if result else '❌'} {name}"
    )

passed = sum(checks.values())
total = len(checks)

score = (
    passed / total * 100
)

print("-" * 100)
print(
    f"Passed checks      : "
    f"{passed}/{total}"
)

print(
    f"Validation Score   : "
    f"{score:.1f}%"
)

if score == 100:

    print(
        "🟢 DATASET IS FULLY VALIDATED"
    )

else:

    print(
        "🟡 DATASET NEEDS REVIEW"
    )

# ============================================================
# 30. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("📊 FINAL SUMMARY")
print("=" * 100)

print(f"Questions: {len(df)}")

print("\nLanguages:")
print(
    language_counts.to_string()
)

print("\nIntents:")
print(
    intent_counts.to_string()
)

print("\nQuestion types:")
print(
    question_type_counts.to_string()
)

print("\nStatus:")
print(
    status_counts.to_string()
)

print("\nDifficulty:")
print(
    difficulty_counts.to_string()
)

print(
    "\nSimilarity >= 0.85:",
    len(high_similarity)
)

# ============================================================
# 31. READ-ONLY CONFIRMATION
# ============================================================

print("\n" + "=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ Dataset was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

# ============================================================
# 32. CLEAN MEMORY ONLY
# ============================================================

df.drop(
    columns=[
        "_normalized_question"
    ],
    inplace=True,
    errors="ignore"
)

print("\n" + "=" * 100)
print("🎉 COMPLETE 384-QUESTION AUDIT FINISHED")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - VALIDATION SCORE FOR ALL 384 QUESTIONS
# READ-ONLY
# ============================================================

from pathlib import Path
from itertools import combinations
from difflib import SequenceMatcher
import pandas as pd
import re
import unicodedata

print("=" * 100)
print("🏨 HOTEL AI - 384 QUESTIONS VALIDATION SCORE")
print("=" * 100)

# ============================================================
# 1. PATH
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

assert PROJECT_ROOT.exists(), "❌ Project root not found."
assert DATASET_PATH.exists(), "❌ Dataset not found."

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print(f"\n📄 Dataset:\n{DATASET_PATH}")
print(f"📊 Questions: {len(df)}")

# ============================================================
# 2. EXPECTED STRUCTURE
# ============================================================

EXPECTED_ROWS = 384

EXPECTED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

EXPECTED_INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

EXPECTED_LANGUAGES = {
    "Arabic",
    "English",
    "French",
    "Russian",
}

VALID_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

VALID_STATUS = {
    "valid",
    "generated",
}

VALID_DIFFICULTY = {
    "Easy",
}

# ============================================================
# 3. NORMALIZATION
# ============================================================

def normalize_question(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


# ============================================================
# 4. INDIVIDUAL SCORE CALCULATIONS
# ============================================================

score_checks = {}

# ------------------------------------------------------------
# CHECK 1 - ROW COUNT
# ------------------------------------------------------------

score_checks["Row Count"] = (
    len(df) == EXPECTED_ROWS
)

# ------------------------------------------------------------
# CHECK 2 - COLUMNS
# ------------------------------------------------------------

score_checks["Columns"] = (
    len(df.columns) == 13
    and df.columns.tolist() == EXPECTED_COLUMNS
)

# ------------------------------------------------------------
# CHECK 3 - REQUIRED FIELDS
# ------------------------------------------------------------

required_fields_ok = True

for col in EXPECTED_COLUMNS:

    if df[col].isna().any():
        required_fields_ok = False
        break

    if (
        df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .any()
    ):
        required_fields_ok = False
        break

score_checks["Required Fields"] = required_fields_ok

# ------------------------------------------------------------
# CHECK 4 - FAQ ID FORMAT
# ------------------------------------------------------------

score_checks["FAQ ID Format"] = (
    df["faq_id"]
    .astype(str)
    .str.fullmatch(r"FAQ\d{4}")
    .all()
)

# ------------------------------------------------------------
# CHECK 5 - DUPLICATE IDS
# ------------------------------------------------------------

score_checks["Duplicate IDs"] = (
    not df["faq_id"].duplicated().any()
)

# ------------------------------------------------------------
# CHECK 6 - CONTINUOUS IDS
# ------------------------------------------------------------

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(1, 385)
]

score_checks["Continuous IDs"] = (
    df["faq_id"].tolist()
    == expected_ids
)

# ------------------------------------------------------------
# NORMALIZED QUESTIONS
# ------------------------------------------------------------

df["_normalized"] = (
    df["question"]
    .map(normalize_question)
)

# ------------------------------------------------------------
# CHECK 7 - EXACT DUPLICATES
# ------------------------------------------------------------

score_checks["Exact Duplicates"] = (
    not df["question"].duplicated().any()
)

# ------------------------------------------------------------
# CHECK 8 - NORMALIZED DUPLICATES
# ------------------------------------------------------------

score_checks["Normalized Duplicates"] = (
    not df["_normalized"].duplicated().any()
)

# ------------------------------------------------------------
# CHECK 9 - SAME INTENT DUPLICATES
# ------------------------------------------------------------

same_intent_dupes = df[
    df.duplicated(
        subset=[
            "intent_id",
            "_normalized"
        ],
        keep=False
    )
]

score_checks["Same-Intent Duplicates"] = (
    same_intent_dupes.empty
)

# ------------------------------------------------------------
# CHECK 10 - INTENT MAPPING
# ------------------------------------------------------------

intent_mapping_ok = True

for intent_id, intent_name in EXPECTED_INTENTS.items():

    subset = df[
        df["intent_id"] == intent_id
    ]

    if len(subset) != 48:
        intent_mapping_ok = False
        break

    if set(subset["intent_name"]) != {intent_name}:
        intent_mapping_ok = False
        break

score_checks["Intent Mapping"] = (
    intent_mapping_ok
)

# ------------------------------------------------------------
# CHECK 11 - LANGUAGE BALANCE
# ------------------------------------------------------------

language_counts = (
    df["language"]
    .value_counts()
)

score_checks["Language Balance"] = (
    set(language_counts.index)
    == EXPECTED_LANGUAGES
    and all(
        language_counts.get(
            language,
            0
        ) == 96
        for language in EXPECTED_LANGUAGES
    )
)

# ------------------------------------------------------------
# CHECK 12 - INTENT BALANCE
# ------------------------------------------------------------

score_checks["Intent Balance"] = all(
    (
        df["intent_id"] == intent_id
    ).sum() == 48
    for intent_id in EXPECTED_INTENTS
)

# ------------------------------------------------------------
# CHECK 13 - LANGUAGE × INTENT
# ------------------------------------------------------------

matrix = pd.crosstab(
    df["intent_id"],
    df["language"]
)

score_checks["Language × Intent"] = all(
    matrix.loc[
        intent_id,
        language
    ] == 12

    for intent_id in EXPECTED_INTENTS

    for language in EXPECTED_LANGUAGES
)

# ------------------------------------------------------------
# CHECK 14 - QUESTION TYPES
# ------------------------------------------------------------

score_checks["Question Types"] = (
    set(df["question_type"])
    <= VALID_TYPES
)

# ------------------------------------------------------------
# CHECK 15 - STATUS
# ------------------------------------------------------------

score_checks["Status"] = (
    set(df["status"])
    <= VALID_STATUS
)

# ------------------------------------------------------------
# CHECK 16 - DIFFICULTY
# ------------------------------------------------------------

score_checks["Difficulty"] = (
    set(df["difficulty"])
    <= VALID_DIFFICULTY
)

# ------------------------------------------------------------
# CHECK 17 - GROUND TRUTH
# ------------------------------------------------------------

score_checks["Ground Truth"] = (
    df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

# ------------------------------------------------------------
# CHECK 18 - ANSWERS
# ------------------------------------------------------------

score_checks["Answers"] = (
    df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

# ------------------------------------------------------------
# CHECK 19 - ATTEMPTS
# ------------------------------------------------------------

attempts_numeric = pd.to_numeric(
    df["attempts"],
    errors="coerce"
)

score_checks["Attempts"] = (
    attempts_numeric.notna().all()
    and (attempts_numeric >= 0).all()
)

# ------------------------------------------------------------
# CHECK 20 - QUESTION LENGTH
# ------------------------------------------------------------

question_lengths = (
    df["question"]
    .astype(str)
    .str.len()
)

score_checks["Question Length"] = (
    not (question_lengths < 10).any()
)

# ============================================================
# 5. SIMILARITY CHECK
# ============================================================

print("\n" + "=" * 100)
print("🔍 FULL SIMILARITY CHECK")
print("=" * 100)

questions = (
    df["question"]
    .astype(str)
    .tolist()
)

total_pairs = (
    EXPECTED_ROWS
    * (EXPECTED_ROWS - 1)
    // 2
)

print(
    f"Comparing {total_pairs:,} question pairs..."
)

high_similarity_count = 0

for i, j in combinations(
    range(EXPECTED_ROWS),
    2
):

    score = SequenceMatcher(
        None,
        normalize_question(questions[i]),
        normalize_question(questions[j])
    ).ratio()

    if score >= 0.85:
        high_similarity_count += 1

score_checks["Similarity < 0.85"] = (
    high_similarity_count == 0
)

print(
    f"Pairs >= 0.85: {high_similarity_count}"
)

if high_similarity_count == 0:
    print("✅ No similarity >= 0.85.")
else:
    print(
        f"⚠️ {high_similarity_count} "
        f"high-similarity pairs found."
    )

# ============================================================
# 6. CALCULATE SCORE
# ============================================================

passed = sum(
    score_checks.values()
)

total = len(score_checks)

validation_score = (
    passed / total * 100
)

# ============================================================
# 7. DISPLAY EACH CHECK
# ============================================================

print("\n" + "=" * 100)
print("🏆 VALIDATION CHECKLIST")
print("=" * 100)

for check_name, result in score_checks.items():

    print(
        f"{'✅' if result else '❌'} "
        f"{check_name}"
    )

# ============================================================
# 8. FINAL SCORE
# ============================================================

print("\n" + "=" * 100)
print("🏆 FINAL VALIDATION SCORE")
print("=" * 100)

print(
    f"Passed Checks      : {passed}/{total}"
)

print(
    f"Validation Score   : {validation_score:.1f}%"
)

if validation_score == 100:

    print(
        "🟢 DATASET IS FULLY VALIDATED"
    )

else:

    print(
        "🟡 DATASET NEEDS REVIEW"
    )

# ============================================================
# 9. DATASET SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("📊 DATASET SUMMARY")
print("=" * 100)

print(
    f"Total Questions : {len(df)}"
)

print("\nLanguages:")
print(
    language_counts.to_string()
)

print("\nIntents:")
print(
    df.groupby(
        ["intent_id", "intent_name"]
    )
    .size()
    .to_string()
)

print("\nQuestion Types:")
print(
    df["question_type"]
    .value_counts()
    .to_string()
)

print("\nStatus:")
print(
    df["status"]
    .value_counts()
    .to_string()
)

print("\nDifficulty:")
print(
    df["difficulty"]
    .value_counts()
    .to_string()
)

# ============================================================
# 10. READ-ONLY CONFIRMATION
# ============================================================

print("\n" + "=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

# ============================================================
# 11. CLEAN MEMORY ONLY
# ============================================================

df.drop(
    columns=["_normalized"],
    inplace=True,
    errors="ignore"
)

print("\n" + "=" * 100)
print("🎉 VALIDATION SCORE CHECK COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 06 FINAL SIMILARITY CHECK
# Existing dataset : 384
# Batch 06         : 32
# READ-ONLY
# ============================================================

from pathlib import Path
from difflib import SequenceMatcher
import pandas as pd
import re
import unicodedata

print("=" * 100)
print("🏨 HOTEL AI - BATCH 06 FINAL SIMILARITY CHECK")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_06.csv"
)

assert FINAL_DATASET.exists(), "❌ Final dataset not found."
assert BATCH_FILE.exists(), "❌ Batch 06 not found."

# ============================================================
# 2. LOAD
# ============================================================

df = pd.read_csv(
    FINAL_DATASET,
    encoding="utf-8-sig"
)

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

assert len(df) == 384
assert len(batch_df) == 32

print(f"Existing questions : {len(df)}")
print(f"Batch 06 questions : {len(batch_df)}")

# ============================================================
# 3. COLUMNS
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

assert list(df.columns) == REQUIRED_COLUMNS
assert list(batch_df.columns) == REQUIRED_COLUMNS

print("✅ All 13 columns exist.")
print("✅ Column order is correct.")

# ============================================================
# 4. ID CHECK
# ============================================================

print("\n" + "=" * 100)
print("🆔 BATCH 06 FAQ ID CHECK")
print("=" * 100)

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(385, 417)
]

assert batch_df["faq_id"].tolist() == expected_ids
assert batch_df["faq_id"].is_unique

print("✅ IDs are exactly FAQ0385 → FAQ0416.")
print("✅ No duplicate IDs.")

# ============================================================
# 5. NORMALIZATION
# ============================================================

def normalize_question(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# 6. EXACT OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("📝 EXACT OVERLAP")
print("=" * 100)

existing_exact = set(
    df["question"]
    .astype(str)
    .str.strip()
)

batch_exact = set(
    batch_df["question"]
    .astype(str)
    .str.strip()
)

exact_overlap = sorted(
    existing_exact.intersection(
        batch_exact
    )
)

if exact_overlap:

    print("❌ Exact overlaps found:")
    for q in exact_overlap:
        print("-", q)

else:

    print("✅ No exact overlap.")

# ============================================================
# 7. NORMALIZED OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("🔎 NORMALIZED OVERLAP")
print("=" * 100)

existing_normalized = set(
    df["question"].map(normalize_question)
)

batch_normalized = set(
    batch_df["question"].map(normalize_question)
)

normalized_overlap = sorted(
    existing_normalized.intersection(
        batch_normalized
    )
)

if normalized_overlap:

    print("❌ Normalized overlaps found:")

    for q in normalized_overlap:
        print("-", q)

else:

    print("✅ No normalized overlap.")

# ============================================================
# 8. DUPLICATES INSIDE BATCH
# ============================================================

print("\n" + "=" * 100)
print("📝 DUPLICATES INSIDE BATCH 06")
print("=" * 100)

exact_duplicates = batch_df[
    batch_df["question"].duplicated(keep=False)
]

normalized_series = batch_df[
    "question"
].map(normalize_question)

normalized_duplicates = batch_df[
    normalized_series.duplicated(keep=False)
]

assert exact_duplicates.empty
assert normalized_duplicates.empty

print("✅ No exact duplicates inside Batch 06.")
print("✅ No normalized duplicates inside Batch 06.")

# ============================================================
# 9. SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🔍 SEMANTIC / TEXT SIMILARITY")
print("=" * 100)

similar_pairs = []

for _, new_row in batch_df.iterrows():

    for _, old_row in df.iterrows():

        score = SequenceMatcher(
            None,
            normalize_question(
                new_row["question"]
            ),
            normalize_question(
                old_row["question"]
            )
        ).ratio()

        if score >= 0.80:

            similar_pairs.append({
                "similarity": round(score, 3),

                "new_faq_id": new_row["faq_id"],
                "new_intent": new_row["intent_id"],
                "new_language": new_row["language"],
                "new_question": new_row["question"],

                "old_faq_id": old_row["faq_id"],
                "old_intent": old_row["intent_id"],
                "old_language": old_row["language"],
                "old_question": old_row["question"],
            })

similarity_df = pd.DataFrame(
    similar_pairs
)

if similarity_df.empty:

    print("🎉 No pairs with similarity >= 0.80.")

else:

    similarity_df = similarity_df.sort_values(
        "similarity",
        ascending=False
    )

    print(
        f"Pairs >= 0.80: {len(similarity_df)}"
    )

    print(
        similarity_df.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 10. THRESHOLDS
# ============================================================

print("\n" + "=" * 100)
print("📊 SIMILARITY THRESHOLDS")
print("=" * 100)

for threshold in [
    0.80,
    0.85,
    0.90,
    0.95
]:

    if similarity_df.empty:
        count = 0
    else:
        count = int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )

    print(
        f"{threshold:.2f}+ : {count} pairs"
    )

# ============================================================
# 11. SAME INTENT
# ============================================================

if similarity_df.empty:

    same_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["new_intent"]
        ==
        similarity_df["old_intent"]
    ]

print("\n" + "=" * 100)
print("🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

print(
    f">= 0.80 : {len(same_intent)}"
)

if not same_intent.empty:

    print(
        same_intent.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 12. CROSS INTENT
# ============================================================

if similarity_df.empty:

    cross_intent = pd.DataFrame()

else:

    cross_intent = similarity_df[
        similarity_df["new_intent"]
        !=
        similarity_df["old_intent"]
    ]

print("\n" + "=" * 100)
print("⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

print(
    f">= 0.80 : {len(cross_intent)}"
)

if not cross_intent.empty:

    print(
        cross_intent.head(30).to_string(
            index=False
        )
    )

# ============================================================
# 13. HIGH SIMILARITY
# ============================================================

high_similarity = (
    similarity_df[
        similarity_df["similarity"] >= 0.85
    ]
    if not similarity_df.empty
    else pd.DataFrame()
)

print("\n" + "=" * 100)
print("🚨 HIGH SIMILARITY REVIEW")
print("=" * 100)

print(
    f"Similarity >= 0.85: "
    f"{len(high_similarity)}"
)

if high_similarity.empty:

    print("✅ No similarity >= 0.85.")

else:

    print(
        high_similarity.to_string(
            index=False
        )
    )

# ============================================================
# 14. FINAL DECISION
# ============================================================

print("\n" + "=" * 100)
print("🏆 BATCH 06 FINAL RESULT")
print("=" * 100)

if (
    len(exact_overlap) == 0
    and len(normalized_overlap) == 0
    and exact_duplicates.empty
    and normalized_duplicates.empty
    and high_similarity.empty
    and cross_intent.empty
):

    print("""
🟢 BATCH 06 IS READY FOR THE NEXT STEP

✅ No exact overlap
✅ No normalized overlap
✅ No duplicates inside Batch 06
✅ No similarity >= 0.85
✅ No cross-intent similarity >= 0.80

🎉 SAFE TO FILL GROUND TRUTH + ANSWERS
""")

else:

    print("""
🟡 BATCH 06 STILL NEEDS REVIEW

⚠️ One or more duplicate/similarity checks require attention.
""")

# ============================================================
# 15. READ-ONLY
# ============================================================

print("=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ faq_batch_06.csv was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

print("\n" + "=" * 100)
print("🎉 BATCH 06 SIMILARITY CHECK COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 06 REPAIR
# FILL GROUND TRUTH + ANSWER
# DOES NOT MODIFY FINAL DATASET
# ============================================================

from pathlib import Path
import pandas as pd

print("=" * 100)
print("🏨 HOTEL AI - BATCH 06 GROUND TRUTH + ANSWER REPAIR")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_06.csv"
)

assert FINAL_DATASET.exists()
assert BATCH_FILE.exists()

# ============================================================
# 2. LOAD
# ============================================================

df = pd.read_csv(
    FINAL_DATASET,
    encoding="utf-8-sig"
)

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

assert len(df) == 384
assert len(batch_df) == 32

print(f"Existing dataset : {len(df)}")
print(f"Batch 06         : {len(batch_df)}")

# ============================================================
# 3. CONVERT TEXT COLUMNS
# ============================================================

TEXT_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "difficulty",
    "answer",
]

for col in TEXT_COLUMNS:

    batch_df[col] = (
        batch_df[col]
        .fillna("")
        .astype("string")
    )

batch_df["attempts"] = pd.to_numeric(
    batch_df["attempts"],
    errors="raise"
)

# ============================================================
# 4. INTENTS
# ============================================================

INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

# ============================================================
# 5. BUILD KNOWLEDGE MAP
# ============================================================
# Use the existing 384 as source of truth.
# One canonical answer per Intent × Language.

print("\n" + "=" * 100)
print("🧠 BUILDING KNOWLEDGE MAP")
print("=" * 100)

knowledge_map = {}

for intent_id, intent_name in INTENTS.items():

    for language in [
        "Arabic",
        "English",
        "French",
        "Russian",
    ]:

        subset = df[
            (df["intent_id"] == intent_id)
            &
            (df["language"] == language)
        ].copy()

        assert len(subset) == 12, (
            f"❌ Expected 12 rows for "
            f"{intent_id} × {language}, "
            f"found {len(subset)}"
        )

        # ----------------------------------------------------
        # Ground Truth
        # ----------------------------------------------------

        gt = (
            subset["ground_truth"]
            .fillna("")
            .astype(str)
            .str.strip()
        )

        gt = gt[gt.ne("")]

        assert not gt.empty, (
            f"❌ No ground truth for "
            f"{intent_id} × {language}"
        )

        canonical_gt = (
            gt.value_counts()
            .index[0]
        )

        # ----------------------------------------------------
        # Answer
        # ----------------------------------------------------

        answers = (
            subset["answer"]
            .fillna("")
            .astype(str)
            .str.strip()
        )

        answers = answers[
            answers.ne("")
        ]

        assert not answers.empty, (
            f"❌ No answer for "
            f"{intent_id} × {language}"
        )

        canonical_answer = (
            answers.value_counts()
            .index[0]
        )

        knowledge_map[
            (intent_id, language)
        ] = {
            "ground_truth": canonical_gt,
            "answer": canonical_answer,
        }

        print(
            f"✅ {intent_id:<6} {language:<8}"
        )

print("✅ Knowledge map created.")

# ============================================================
# 6. APPLY
# ============================================================

print("\n" + "=" * 100)
print("✍️ FILLING GROUND TRUTH + ANSWERS")
print("=" * 100)

for idx, row in batch_df.iterrows():

    intent_id = str(
        row["intent_id"]
    ).strip()

    language = str(
        row["language"]
    ).strip()

    knowledge = knowledge_map[
        (intent_id, language)
    ]

    batch_df.at[
        idx,
        "ground_truth"
    ] = knowledge["ground_truth"]

    batch_df.at[
        idx,
        "answer"
    ] = knowledge["answer"]

    batch_df.at[
        idx,
        "status"
    ] = "generated"

    batch_df.at[
        idx,
        "category"
    ] = "Hotel FAQ"

    batch_df.at[
        idx,
        "source"
    ] = "Generated Batch 06"

    batch_df.at[
        idx,
        "attempts"
    ] = 0

    batch_df.at[
        idx,
        "difficulty"
    ] = "Easy"

# Restore text dtype
for col in TEXT_COLUMNS:

    batch_df[col] = (
        batch_df[col]
        .fillna("")
        .astype("string")
    )

print("✅ Ground truth filled for all 32.")
print("✅ Answers filled for all 32.")

# ============================================================
# 7. VALIDATION
# ============================================================

print("\n" + "=" * 100)
print("🧪 REQUIRED FIELD VALIDATION")
print("=" * 100)

for col in TEXT_COLUMNS + ["attempts"]:

    missing = int(
        batch_df[col].isna().sum()
    )

    empty = int(
        batch_df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        f"{'✅' if missing == 0 and empty == 0 else '❌'} "
        f"{col:<15}"
        f"Missing: {missing:<3}"
        f"Empty: {empty:<3}"
    )

    assert missing == 0
    assert empty == 0

print("✅ All required fields complete.")

# ============================================================
# 8. SAVE BATCH ONLY
# ============================================================

batch_df.to_csv(
    BATCH_FILE,
    index=False,
    encoding="utf-8-sig"
)

# ============================================================
# 9. RELOAD
# ============================================================

check = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

assert len(check) == 32

assert (
    check["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

assert (
    check["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

print("\n" + "=" * 100)
print("🏆 BATCH 06 REPAIR RESULT")
print("=" * 100)

print("""
🟢 BATCH 06 GROUND TRUTH + ANSWER COMPLETE

✅ 32 questions
✅ FAQ0385 → FAQ0416
✅ Ground truth complete
✅ Answers complete
✅ Required fields complete
✅ Existing 384 dataset NOT modified

🎉 READY FOR FINAL BATCH 06 VALIDATION
""")

print("📄 Batch:")
print(BATCH_FILE)

print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 06 COMPLETE AUDIT
# Existing dataset : 384
# Batch 06         : 32
# READ-ONLY
# ============================================================

from pathlib import Path
from difflib import SequenceMatcher
import pandas as pd
import re
import unicodedata

print("=" * 100)
print("🏨 HOTEL AI - BATCH 06 COMPLETE AUDIT")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_06.csv"
)

assert PROJECT_ROOT.exists(), "❌ Project root not found."
assert FINAL_DATASET.exists(), "❌ Final dataset not found."
assert BATCH_FILE.exists(), "❌ Batch 06 not found."

# ============================================================
# 2. LOAD
# ============================================================

df = pd.read_csv(
    FINAL_DATASET,
    encoding="utf-8-sig"
)

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

assert len(df) == 384
assert len(batch_df) == 32

print("\n" + "=" * 100)
print("📊 FILE INFORMATION")
print("=" * 100)

print(f"Existing dataset : {len(df)} rows")
print(f"Batch 06         : {len(batch_df)} rows")

# ============================================================
# 3. REQUIRED COLUMNS
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

assert list(df.columns) == REQUIRED_COLUMNS
assert list(batch_df.columns) == REQUIRED_COLUMNS

print("✅ All 13 columns exist.")
print("✅ Column order is correct.")

# ============================================================
# 4. NORMALIZATION
# ============================================================

def normalize(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# 5. BATCH ID VALIDATION
# ============================================================

print("\n" + "=" * 100)
print("🆔 FAQ ID VALIDATION")
print("=" * 100)

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(385, 417)
]

assert batch_df["faq_id"].tolist() == expected_ids
assert batch_df["faq_id"].is_unique

print("✅ IDs are exactly FAQ0385 → FAQ0416.")
print("✅ No duplicate IDs.")

# ============================================================
# 6. CHECK ID OVERLAP
# ============================================================

id_overlap = set(
    df["faq_id"]
).intersection(
    set(batch_df["faq_id"])
)

assert not id_overlap

print("✅ No ID overlap with existing 384.")

# ============================================================
# 7. REQUIRED FIELDS
# ============================================================

print("\n" + "=" * 100)
print("🧪 REQUIRED FIELD VALIDATION")
print("=" * 100)

required_ok = True

for col in REQUIRED_COLUMNS:

    missing = int(
        batch_df[col].isna().sum()
    )

    empty = int(
        batch_df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        f"{'✅' if missing == 0 and empty == 0 else '❌'} "
        f"{col:<15}"
        f"Missing: {missing:<3}"
        f"Empty: {empty:<3}"
    )

    if missing or empty:
        required_ok = False

assert required_ok

print("✅ All required fields complete.")

# ============================================================
# 8. DUPLICATES INSIDE BATCH
# ============================================================

print("\n" + "=" * 100)
print("📝 DUPLICATE CHECK INSIDE BATCH 06")
print("=" * 100)

exact_dupes = batch_df[
    batch_df["question"].duplicated(keep=False)
]

batch_norm = batch_df["question"].map(normalize)

normalized_dupes = batch_df[
    batch_norm.duplicated(keep=False)
]

assert exact_dupes.empty
assert normalized_dupes.empty

print("✅ No exact duplicates.")
print("✅ No normalized duplicates.")

# ============================================================
# 9. OVERLAP WITH EXISTING 384
# ============================================================

print("\n" + "=" * 100)
print("🔎 OVERLAP WITH EXISTING 384")
print("=" * 100)

existing_exact = set(
    df["question"]
    .astype(str)
    .str.strip()
)

batch_exact = set(
    batch_df["question"]
    .astype(str)
    .str.strip()
)

exact_overlap = sorted(
    existing_exact.intersection(
        batch_exact
    )
)

existing_normalized = set(
    df["question"].map(normalize)
)

normalized_overlap = sorted(
    set(batch_norm).intersection(
        existing_normalized
    )
)

assert not exact_overlap
assert not normalized_overlap

print("✅ No exact overlap.")
print("✅ No normalized overlap.")

# ============================================================
# 10. INTENT VALIDATION
# ============================================================

INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

VALID_LANGUAGES = {
    "Arabic",
    "English",
    "French",
    "Russian",
}

VALID_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

print("\n" + "=" * 100)
print("🎯 INTENT VALIDATION")
print("=" * 100)

intent_counts = batch_df["intent_id"].value_counts()

assert intent_counts.to_dict() == {
    "INT001": 4,
    "INT002": 4,
    "INT003": 4,
    "INT004": 4,
    "INT005": 4,
    "INT006": 4,
    "INT007": 4,
    "INT008": 4,
}

for _, row in batch_df.iterrows():

    assert row["intent_id"] in INTENTS

    assert (
        row["intent_name"]
        == INTENTS[row["intent_id"]]
    )

print(intent_counts.to_string())
print("✅ 4 questions × 8 intents.")
print("✅ Intent ID ↔ Intent Name consistency.")

# ============================================================
# 11. LANGUAGE VALIDATION
# ============================================================

print("\n" + "=" * 100)
print("🌍 LANGUAGE VALIDATION")
print("=" * 100)

language_counts = batch_df["language"].value_counts()

assert set(language_counts.index) == VALID_LANGUAGES

assert language_counts.to_dict() == {
    "Arabic": 8,
    "English": 8,
    "French": 8,
    "Russian": 8,
}

print(language_counts.to_string())
print("✅ 8 questions × 4 languages.")

# ============================================================
# 12. QUESTION TYPES
# ============================================================

print("\n" + "=" * 100)
print("📝 QUESTION TYPE VALIDATION")
print("=" * 100)

assert set(
    batch_df["question_type"]
) <= VALID_TYPES

print(
    batch_df["question_type"]
    .value_counts()
    .to_string()
)

print("✅ All question types are valid.")

# ============================================================
# 13. STATUS / DIFFICULTY
# ============================================================

print("\n" + "=" * 100)
print("📌 STATUS / DIFFICULTY")
print("=" * 100)

assert set(batch_df["status"]) <= {
    "valid",
    "generated"
}

assert set(batch_df["difficulty"]) <= {
    "Easy"
}

print("Status:")
print(
    batch_df["status"]
    .value_counts()
    .to_string()
)

print("\nDifficulty:")
print(
    batch_df["difficulty"]
    .value_counts()
    .to_string()
)

print("✅ Status and difficulty values valid.")

# ============================================================
# 14. GROUND TRUTH / ANSWER
# ============================================================

print("\n" + "=" * 100)
print("🎯 GROUND TRUTH / ANSWER")
print("=" * 100)

ground_truth_complete = (
    batch_df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

answers_complete = (
    batch_df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

print(
    "✅ Ground truth complete."
    if ground_truth_complete
    else "⚠️ Ground truth not filled yet."
)

print(
    "✅ Answers complete."
    if answers_complete
    else "⚠️ Answers not filled yet."
)

# ============================================================
# 15. QUESTION LENGTH
# ============================================================

print("\n" + "=" * 100)
print("📏 QUESTION LENGTH")
print("=" * 100)

lengths = (
    batch_df["question"]
    .astype(str)
    .str.len()
)

print(f"Shortest : {lengths.min()}")
print(f"Longest  : {lengths.max()}")
print(f"Average  : {lengths.mean():.2f}")

short_questions = batch_df[
    lengths < 10
]

assert short_questions.empty

print("✅ No suspiciously short questions.")

# ============================================================
# 16. FULL SIMILARITY CHECK
# ============================================================

print("\n" + "=" * 100)
print("🔍 FULL SIMILARITY CHECK")
print("=" * 100)

similar_pairs = []

for _, new_row in batch_df.iterrows():

    for _, old_row in df.iterrows():

        score = SequenceMatcher(
            None,
            normalize(new_row["question"]),
            normalize(old_row["question"])
        ).ratio()

        if score >= 0.80:

            similar_pairs.append({
                "similarity": round(score, 3),

                "new_faq_id": new_row["faq_id"],
                "new_intent": new_row["intent_id"],
                "new_language": new_row["language"],
                "new_question": new_row["question"],

                "old_faq_id": old_row["faq_id"],
                "old_intent": old_row["intent_id"],
                "old_language": old_row["language"],
                "old_question": old_row["question"],
            })

similarity_df = pd.DataFrame(similar_pairs)

if similarity_df.empty:

    print("🎉 No pairs >= 0.80.")

else:

    similarity_df = similarity_df.sort_values(
        "similarity",
        ascending=False
    )

    print(
        f"Pairs >= 0.80: {len(similarity_df)}"
    )

    print(
        similarity_df.head(30)
        .to_string(index=False)
    )

# ============================================================
# 17. THRESHOLDS
# ============================================================

print("\n" + "=" * 100)
print("📊 SIMILARITY THRESHOLDS")
print("=" * 100)

for threshold in [
    0.80,
    0.85,
    0.90,
    0.95,
]:

    count = (
        0
        if similarity_df.empty
        else int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )
    )

    print(
        f"{threshold:.2f}+ : {count} pairs"
    )

# ============================================================
# 18. SAME / CROSS INTENT
# ============================================================

if similarity_df.empty:

    same_intent = pd.DataFrame()
    cross_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["new_intent"]
        ==
        similarity_df["old_intent"]
    ]

    cross_intent = similarity_df[
        similarity_df["new_intent"]
        !=
        similarity_df["old_intent"]
    ]

print("\n" + "=" * 100)
print("🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

print(
    f">= 0.80 : {len(same_intent)}"
)

print("\n" + "=" * 100)
print("⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

print(
    f">= 0.80 : {len(cross_intent)}"
)

# ============================================================
# 19. HIGH SIMILARITY
# ============================================================

if similarity_df.empty:

    high_similarity = pd.DataFrame()

else:

    high_similarity = similarity_df[
        similarity_df["similarity"] >= 0.85
    ]

print("\n" + "=" * 100)
print("🚨 HIGH SIMILARITY REVIEW")
print("=" * 100)

if high_similarity.empty:

    print("✅ No similarity >= 0.85.")

else:

    print(
        high_similarity.to_string(index=False)
    )

# ============================================================
# 20. FINAL DECISION
# ============================================================

batch_structure_ok = (
    len(batch_df) == 32
    and batch_df["faq_id"].is_unique
    and exact_dupes.empty
    and normalized_dupes.empty
    and not exact_overlap
    and not normalized_overlap
)

similarity_ok = (
    high_similarity.empty
    and cross_intent.empty
)

print("\n" + "=" * 100)
print("🏆 BATCH 06 COMPLETE AUDIT RESULT")
print("=" * 100)

if batch_structure_ok and similarity_ok:

    print("""
🟢 BATCH 06 PASSED STRUCTURAL + SIMILARITY CHECK

✅ 32 questions
✅ FAQ0385 → FAQ0416
✅ No duplicate IDs
✅ No duplicate questions
✅ No normalized duplicates
✅ No overlap with existing 384
✅ Language balance correct
✅ Intent balance correct
✅ Intent names correct
✅ Question types valid
✅ No similarity >= 0.85
✅ No cross-intent similarity >= 0.80
""")

else:

    print("""
🟡 BATCH 06 NEEDS REVIEW

⚠️ One or more checks require attention.
""")

# ============================================================
# 21. READ-ONLY
# ============================================================

print("=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ faq_batch_06.csv was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

print("\n" + "=" * 100)
print("🎉 BATCH 06 COMPLETE AUDIT FINISHED")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - COMPLETE 416 DATASET AUDIT
# FULL QUALITY + DUPLICATE + SIMILARITY CHECK
# READ-ONLY
# ============================================================

from pathlib import Path
from itertools import combinations
from difflib import SequenceMatcher
import pandas as pd
import re
import unicodedata
import numpy as np

# ============================================================
# 1. CONFIG
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

EXPECTED_ROWS = 416

EXPECTED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

EXPECTED_INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

EXPECTED_LANGUAGES = {
    "Arabic",
    "English",
    "French",
    "Russian",
}

VALID_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

VALID_STATUS = {
    "valid",
    "generated",
}

VALID_DIFFICULTY = {
    "Easy",
}

SIMILARITY_THRESHOLD = 0.85

print("=" * 100)
print("🏨 HOTEL AI - COMPLETE 416 DATASET AUDIT")
print("=" * 100)

# ============================================================
# 2. LOAD
# ============================================================

assert PROJECT_ROOT.exists(), (
    f"❌ Project root not found:\n{PROJECT_ROOT}"
)

assert DATASET_PATH.exists(), (
    f"❌ Dataset not found:\n{DATASET_PATH}"
)

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print("\n📂 PROJECT")
print("=" * 100)
print(PROJECT_ROOT)

print("\n📄 DATASET")
print("=" * 100)
print(DATASET_PATH)

print("\n📊 DATASET INFORMATION")
print("=" * 100)
print(f"Rows    : {len(df)}")
print(f"Columns : {len(df.columns)}")

# ============================================================
# 3. CHECK REGISTRY
# ============================================================

checks = {}

# ============================================================
# 4. COLUMN CHECK
# ============================================================

columns_ok = (
    len(df.columns) == 13
    and df.columns.tolist() == EXPECTED_COLUMNS
)

checks["13 columns"] = columns_ok

print("\n🧪 COLUMN VALIDATION")
print("=" * 100)

print(
    "✅ All 13 required columns exist."
    if columns_ok
    else "❌ Column validation failed."
)

if columns_ok:
    print("✅ Column order is correct.")

# ============================================================
# 5. ROW COUNT
# ============================================================

row_count_ok = (
    len(df) == EXPECTED_ROWS
)

checks["416 questions"] = row_count_ok

print("\n📦 ROW COUNT")
print("=" * 100)
print(f"Total questions: {len(df)}")

print(
    "✅ Exactly 416 questions."
    if row_count_ok
    else "❌ Expected 416 questions."
)

# ============================================================
# 6. REQUIRED FIELD VALIDATION
# ============================================================

required_fields_ok = True

print("\n🧪 REQUIRED FIELD VALIDATION")
print("=" * 100)

for col in EXPECTED_COLUMNS:

    if col not in df.columns:
        required_fields_ok = False
        print(f"❌ Missing column: {col}")
        continue

    missing = int(
        df[col].isna().sum()
    )

    empty = int(
        df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    ok = (
        missing == 0
        and empty == 0
    )

    print(
        f"{'✅' if ok else '❌'} "
        f"{col:<15}"
        f"Missing: {missing:<3}"
        f"Empty: {empty:<3}"
    )

    if not ok:
        required_fields_ok = False

checks["Required fields"] = required_fields_ok

# ============================================================
# 7. FAQ ID CHECK
# ============================================================

print("\n🆔 FAQ ID VALIDATION")
print("=" * 100)

valid_id_format = (
    df["faq_id"]
    .astype(str)
    .str.fullmatch(r"FAQ\d{4}")
    .all()
)

duplicate_ids = df[
    df["faq_id"]
    .duplicated(keep=False)
]

numeric_ids = pd.to_numeric(
    df["faq_id"]
    .astype(str)
    .str.extract(r"FAQ(\d{4})")[0],
    errors="coerce"
)

expected_numeric_ids = np.arange(
    1,
    EXPECTED_ROWS + 1
)

continuous_ids = (
    numeric_ids.notna().all()
    and numeric_ids.astype(int).tolist()
    == expected_numeric_ids.tolist()
)

checks["Valid FAQ IDs"] = bool(valid_id_format)
checks["No duplicate IDs"] = duplicate_ids.empty
checks["Continuous IDs"] = bool(continuous_ids)

print(
    "✅ All FAQ IDs have valid format."
    if valid_id_format
    else "❌ Invalid FAQ IDs."
)

print(
    "✅ No duplicate FAQ IDs."
    if duplicate_ids.empty
    else f"❌ Duplicate IDs found: {len(duplicate_ids)}"
)

print(
    "✅ IDs continuous: FAQ0001 → FAQ0416"
    if continuous_ids
    else "❌ IDs are not continuous."
)

# ============================================================
# 8. NORMALIZATION
# ============================================================

def normalize_text(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


df["_normalized_question"] = (
    df["question"]
    .map(normalize_text)
)

# ============================================================
# 9. EXACT DUPLICATES
# ============================================================

exact_duplicates = df[
    df["question"]
    .duplicated(keep=False)
]

checks["No exact duplicates"] = (
    exact_duplicates.empty
)

print("\n📝 EXACT DUPLICATE CHECK")
print("=" * 100)

if exact_duplicates.empty:
    print("✅ No exact duplicate questions.")
else:
    print(
        exact_duplicates[
            ["faq_id", "intent_id", "language", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 10. NORMALIZED DUPLICATES
# ============================================================

normalized_duplicates = df[
    df["_normalized_question"]
    .duplicated(keep=False)
]

checks["No normalized duplicates"] = (
    normalized_duplicates.empty
)

print("\n🔎 NORMALIZED DUPLICATE CHECK")
print("=" * 100)

if normalized_duplicates.empty:
    print("✅ No normalized duplicate questions.")
else:
    print(
        normalized_duplicates[
            ["faq_id", "intent_id", "language", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 11. SAME-INTENT DUPLICATES
# ============================================================

same_intent_duplicates = df[
    df.duplicated(
        subset=[
            "intent_id",
            "_normalized_question"
        ],
        keep=False
    )
]

checks["No same-intent duplicates"] = (
    same_intent_duplicates.empty
)

print("\n🔍 DUPLICATES INSIDE SAME INTENT")
print("=" * 100)

if same_intent_duplicates.empty:
    print("✅ No normalized duplicates inside the same intent.")
else:
    print(
        same_intent_duplicates[
            [
                "faq_id",
                "intent_id",
                "language",
                "question"
            ]
        ].to_string(index=False)
    )

# ============================================================
# 12. INTENT MAPPING
# ============================================================

intent_mapping_ok = True

print("\n🔗 INTENT ID ↔ INTENT NAME")
print("=" * 100)

for intent_id, intent_name in EXPECTED_INTENTS.items():

    subset = df[
        df["intent_id"] == intent_id
    ]

    actual_names = set(
        subset["intent_name"]
        .astype(str)
    )

    ok = (
        len(subset) == 52
        and actual_names == {intent_name}
    )

    if not ok:
        intent_mapping_ok = False

    print(
        f"{'✅' if ok else '❌'} "
        f"{intent_id} → {intent_name} "
        f"| rows={len(subset)}"
    )

checks["Intent mappings"] = intent_mapping_ok

# ============================================================
# 13. LANGUAGE DISTRIBUTION
# ============================================================

language_counts = (
    df["language"]
    .value_counts()
)

language_balance_ok = (
    set(language_counts.index)
    == EXPECTED_LANGUAGES
    and all(
        language_counts.get(
            lang,
            0
        ) == 104
        for lang in EXPECTED_LANGUAGES
    )
)

checks["Language balance"] = language_balance_ok

print("\n🌍 LANGUAGE DISTRIBUTION")
print("=" * 100)
print(language_counts.to_string())

print(
    "✅ 104 questions × 4 languages."
    if language_balance_ok
    else "❌ Language balance failed."
)

# ============================================================
# 14. INTENT DISTRIBUTION
# ============================================================

intent_counts = (
    df["intent_id"]
    .value_counts()
)

intent_balance_ok = all(
    int(
        intent_counts.get(
            intent_id,
            0
        )
    ) == 52
    for intent_id in EXPECTED_INTENTS
)

checks["Intent balance"] = intent_balance_ok

print("\n🎯 INTENT DISTRIBUTION")
print("=" * 100)

print(
    df.groupby(
        ["intent_id", "intent_name"]
    )
    .size()
    .rename("questions")
    .reset_index()
    .to_string(index=False)
)

print(
    "✅ 52 questions × 8 intents."
    if intent_balance_ok
    else "❌ Intent balance failed."
)

# ============================================================
# 15. LANGUAGE × INTENT
# ============================================================

language_intent = pd.crosstab(
    df["intent_id"],
    df["language"]
)

language_intent_ok = True

for intent_id in EXPECTED_INTENTS:

    for language in EXPECTED_LANGUAGES:

        count = int(
            language_intent.loc[
                intent_id,
                language
            ]
        )

        if count != 13:
            language_intent_ok = False

checks["Language × Intent"] = language_intent_ok

print("\n📊 LANGUAGE × INTENT MATRIX")
print("=" * 100)
print(language_intent.to_string())

print(
    "✅ Every intent has 13 questions per language."
    if language_intent_ok
    else "❌ Language × Intent balance failed."
)

# ============================================================
# 16. QUESTION TYPES
# ============================================================

question_type_counts = (
    df["question_type"]
    .value_counts()
)

unexpected_types = (
    set(df["question_type"])
    - VALID_TYPES
)

question_types_ok = (
    not unexpected_types
)

checks["Question types"] = question_types_ok

print("\n📝 QUESTION TYPE DISTRIBUTION")
print("=" * 100)
print(question_type_counts.to_string())

print(
    "✅ All question types are valid."
    if question_types_ok
    else f"❌ Invalid question types: {unexpected_types}"
)

# ============================================================
# 17. STATUS
# ============================================================

status_counts = (
    df["status"]
    .value_counts()
)

unexpected_status = (
    set(df["status"])
    - VALID_STATUS
)

status_ok = (
    not unexpected_status
)

checks["Status"] = status_ok

print("\n📌 STATUS DISTRIBUTION")
print("=" * 100)
print(status_counts.to_string())

print(
    "✅ Status values are valid."
    if status_ok
    else f"❌ Invalid statuses: {unexpected_status}"
)

# ============================================================
# 18. DIFFICULTY
# ============================================================

difficulty_counts = (
    df["difficulty"]
    .value_counts()
)

unexpected_difficulty = (
    set(df["difficulty"])
    - VALID_DIFFICULTY
)

difficulty_ok = (
    not unexpected_difficulty
)

checks["Difficulty"] = difficulty_ok

print("\n📌 DIFFICULTY DISTRIBUTION")
print("=" * 100)
print(difficulty_counts.to_string())

print(
    "✅ Difficulty values are valid."
    if difficulty_ok
    else f"❌ Invalid difficulty: {unexpected_difficulty}"
)

# ============================================================
# 19. GROUND TRUTH / ANSWERS
# ============================================================

ground_truth_ok = (
    df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

answers_ok = (
    df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

checks["Ground truth"] = bool(ground_truth_ok)
checks["Answers"] = bool(answers_ok)

print("\n🎯 GROUND TRUTH / ANSWER")
print("=" * 100)

print(
    "✅ Ground truth complete."
    if ground_truth_ok
    else "❌ Ground truth incomplete."
)

print(
    "✅ Answers complete."
    if answers_ok
    else "❌ Answers incomplete."
)

# ============================================================
# 20. GROUND TRUTH / ANSWER WORDING
# ============================================================

gt_norm = (
    df["ground_truth"]
    .map(normalize_text)
)

answer_norm = (
    df["answer"]
    .map(normalize_text)
)

wording_difference = (
    gt_norm != answer_norm
)

print("\n🔗 GROUND TRUTH ↔ ANSWER")
print("=" * 100)

print(
    "Rows with different wording:",
    int(wording_difference.sum())
)

print(
    "ℹ️ Different wording is not treated as a failure."
)

# ============================================================
# 21. ATTEMPTS
# ============================================================

attempts_numeric = pd.to_numeric(
    df["attempts"],
    errors="coerce"
)

attempts_ok = (
    attempts_numeric.notna().all()
    and (attempts_numeric >= 0).all()
)

checks["Attempts"] = attempts_ok

print("\n🔢 ATTEMPTS VALIDATION")
print("=" * 100)

print(
    "✅ Attempts are numeric and non-negative."
    if attempts_ok
    else "❌ Invalid attempts values."
)

print(
    f"Min    : {attempts_numeric.min()}"
)
print(
    f"Max    : {attempts_numeric.max()}"
)
print(
    f"Average: {attempts_numeric.mean():.2f}"
)

# ============================================================
# 22. QUESTION LENGTH
# ============================================================

question_lengths = (
    df["question"]
    .astype(str)
    .str.len()
)

very_short = df[
    question_lengths < 10
]

question_length_ok = (
    very_short.empty
)

checks["Question length"] = question_length_ok

print("\n📏 QUESTION LENGTH")
print("=" * 100)

print(
    f"Shortest : {question_lengths.min()}"
)
print(
    f"Longest  : {question_lengths.max()}"
)
print(
    f"Average  : {question_lengths.mean():.2f}"
)

print(
    "✅ No suspiciously short questions."
    if question_length_ok
    else "⚠️ Suspiciously short questions found."
)

# ============================================================
# 23. FULL PAIRWISE SIMILARITY
# ============================================================

print("\n🔍 FULL QUESTION SIMILARITY")
print("=" * 100)

questions = (
    df["question"]
    .astype(str)
    .tolist()
)

total_pairs = (
    EXPECTED_ROWS
    * (EXPECTED_ROWS - 1)
    // 2
)

print(
    f"Comparing {total_pairs:,} question pairs..."
)

similarity_pairs = []

for i, j in combinations(
    range(EXPECTED_ROWS),
    2
):

    score = SequenceMatcher(
        None,
        normalize_text(questions[i]),
        normalize_text(questions[j])
    ).ratio()

    if score >= 0.80:

        similarity_pairs.append({
            "similarity": round(score, 3),
            "faq_id_1": df.iloc[i]["faq_id"],
            "intent_1": df.iloc[i]["intent_id"],
            "language_1": df.iloc[i]["language"],
            "question_1": questions[i],
            "faq_id_2": df.iloc[j]["faq_id"],
            "intent_2": df.iloc[j]["intent_id"],
            "language_2": df.iloc[j]["language"],
            "question_2": questions[j],
        })

similarity_df = pd.DataFrame(
    similarity_pairs
)

# ============================================================
# 24. THRESHOLD SUMMARY
# ============================================================

print("\n📊 SIMILARITY THRESHOLD SUMMARY")
print("=" * 100)

for threshold in [
    0.80,
    0.85,
    0.90,
    0.95,
]:

    if similarity_df.empty:

        count = 0

    else:

        count = int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )

    print(
        f"{threshold:.2f}+ : {count} pairs"
    )

# ============================================================
# 25. SAME / CROSS INTENT
# ============================================================

if similarity_df.empty:

    same_intent = pd.DataFrame()
    cross_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["intent_1"]
        ==
        similarity_df["intent_2"]
    ]

    cross_intent = similarity_df[
        similarity_df["intent_1"]
        !=
        similarity_df["intent_2"]
    ]

print("\n🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

print(
    f"Pairs >= 0.80: {len(same_intent)}"
)

print("\n⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

print(
    f"Pairs >= 0.80: {len(cross_intent)}"
)

# ============================================================
# 26. TOP SIMILAR PAIRS
# ============================================================

print("\n📋 TOP 30 MOST SIMILAR QUESTIONS")
print("=" * 100)

if similarity_df.empty:

    print("None.")

else:

    print(
        similarity_df
        .sort_values(
            "similarity",
            ascending=False
        )
        .head(30)
        .to_string(index=False)
    )

# ============================================================
# 27. HIGH SIMILARITY DECISION
# ============================================================

high_similarity = (
    similarity_df[
        similarity_df["similarity"]
        >= SIMILARITY_THRESHOLD
    ]
    if not similarity_df.empty
    else pd.DataFrame()
)

checks["Similarity < 0.85"] = (
    high_similarity.empty
)

print("\n🚨 HIGH-SIMILARITY REVIEW")
print("=" * 100)

if high_similarity.empty:

    print("✅ No similarity >= 0.85.")

else:

    print(
        f"❌ {len(high_similarity)} pairs >= 0.85"
    )

    print(
        high_similarity
        .sort_values(
            "similarity",
            ascending=False
        )
        .to_string(index=False)
    )

# ============================================================
# 28. FINAL SCORE
# ============================================================

print("\n" + "=" * 100)
print("🏆 FINAL VALIDATION SCORE")
print("=" * 100)

for name, result in checks.items():

    print(
        f"{'✅' if result else '❌'} {name}"
    )

passed = sum(checks.values())
total = len(checks)

score = (
    passed / total * 100
)

print("-" * 100)
print(
    f"Passed checks      : {passed}/{total}"
)

print(
    f"Validation Score   : {score:.1f}%"
)

if score == 100:

    print(
        "🟢 DATASET IS FULLY VALIDATED"
    )

else:

    print(
        "🟡 DATASET NEEDS REVIEW"
    )

# ============================================================
# 29. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("📊 FINAL SUMMARY")
print("=" * 100)

print(
    f"Questions : {len(df)}"
)

print("\nLanguages:")
print(
    language_counts.to_string()
)

print("\nIntents:")
print(
    intent_counts.to_string()
)

print("\nLanguage × Intent:")
print(
    language_intent.to_string()
)

print("\nQuestion Types:")
print(
    question_type_counts.to_string()
)

print("\nStatus:")
print(
    status_counts.to_string()
)

print("\nDifficulty:")
print(
    difficulty_counts.to_string()
)

print(
    f"\nSimilarity >= 0.85: "
    f"{len(high_similarity)}"
)

# ============================================================
# 30. READ-ONLY CONFIRMATION
# ============================================================

print("\n" + "=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

# ============================================================
# 31. CLEAN MEMORY ONLY
# ============================================================

df.drop(
    columns=[
        "_normalized_question"
    ],
    inplace=True,
    errors="ignore"
)

print("\n" + "=" * 100)
print("🎉 COMPLETE 416-QUESTION AUDIT FINISHED")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 07 vs EXISTING 416
# DUPLICATE + SIMILARITY CHECK
# READ-ONLY
# ============================================================

from pathlib import Path
from difflib import SequenceMatcher
import pandas as pd
import re
import unicodedata

print("=" * 100)
print("🏨 HOTEL AI - BATCH 07 vs EXISTING 416")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_07.csv"
)

assert FINAL_DATASET.exists()
assert BATCH_FILE.exists()

# ============================================================
# 2. LOAD
# ============================================================

df = pd.read_csv(
    FINAL_DATASET,
    encoding="utf-8-sig"
)

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

assert len(df) == 416
assert len(batch_df) == 32

print(f"Existing questions : {len(df)}")
print(f"Batch 07 questions : {len(batch_df)}")

# ============================================================
# 3. COLUMNS
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

assert list(df.columns) == REQUIRED_COLUMNS
assert list(batch_df.columns) == REQUIRED_COLUMNS

print("✅ All 13 columns exist.")
print("✅ Column order is correct.")

# ============================================================
# 4. FAQ ID CHECK
# ============================================================

print("\n" + "=" * 100)
print("🆔 BATCH 07 FAQ ID CHECK")
print("=" * 100)

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(417, 449)
]

assert batch_df["faq_id"].tolist() == expected_ids
assert batch_df["faq_id"].is_unique

print("✅ IDs are exactly FAQ0417 → FAQ0448.")
print("✅ No duplicate IDs.")

id_overlap = set(
    df["faq_id"]
).intersection(
    set(batch_df["faq_id"])
)

assert not id_overlap

print("✅ No ID overlap with existing 416.")

# ============================================================
# 5. NORMALIZATION
# ============================================================

def normalize(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# 6. EXACT OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("📝 EXACT OVERLAP")
print("=" * 100)

existing_exact = set(
    df["question"]
    .astype(str)
    .str.strip()
)

batch_exact = set(
    batch_df["question"]
    .astype(str)
    .str.strip()
)

exact_overlap = sorted(
    existing_exact.intersection(
        batch_exact
    )
)

if exact_overlap:

    print("❌ Exact overlaps found:")

    for q in exact_overlap:
        print("-", q)

else:

    print("✅ No exact overlap.")

# ============================================================
# 7. NORMALIZED OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("🔎 NORMALIZED OVERLAP")
print("=" * 100)

existing_normalized = set(
    df["question"].map(normalize)
)

batch_normalized = set(
    batch_df["question"].map(normalize)
)

normalized_overlap = sorted(
    existing_normalized.intersection(
        batch_normalized
    )
)

if normalized_overlap:

    print("❌ Normalized overlaps found:")

    for q in normalized_overlap:
        print("-", q)

else:

    print("✅ No normalized overlap.")

# ============================================================
# 8. DUPLICATES INSIDE BATCH
# ============================================================

print("\n" + "=" * 100)
print("📝 DUPLICATES INSIDE BATCH 07")
print("=" * 100)

exact_duplicates = batch_df[
    batch_df["question"].duplicated(
        keep=False
    )
]

normalized_series = batch_df[
    "question"
].map(normalize)

normalized_duplicates = batch_df[
    normalized_series.duplicated(
        keep=False
    )
]

assert exact_duplicates.empty
assert normalized_duplicates.empty

print("✅ No exact duplicates inside Batch 07.")
print("✅ No normalized duplicates inside Batch 07.")

# ============================================================
# 9. SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🔍 SEMANTIC / TEXT SIMILARITY")
print("=" * 100)

similar_pairs = []

for _, new_row in batch_df.iterrows():

    for _, old_row in df.iterrows():

        score = SequenceMatcher(
            None,
            normalize(new_row["question"]),
            normalize(old_row["question"])
        ).ratio()

        if score >= 0.80:

            similar_pairs.append({
                "similarity": round(score, 3),

                "new_faq_id": new_row["faq_id"],
                "new_intent": new_row["intent_id"],
                "new_language": new_row["language"],
                "new_question": new_row["question"],

                "old_faq_id": old_row["faq_id"],
                "old_intent": old_row["intent_id"],
                "old_language": old_row["language"],
                "old_question": old_row["question"],
            })

similarity_df = pd.DataFrame(
    similar_pairs
)

if similarity_df.empty:

    print("🎉 No pairs >= 0.80.")

else:

    similarity_df = similarity_df.sort_values(
        "similarity",
        ascending=False
    ).reset_index(drop=True)

    print(
        f"Pairs >= 0.80: {len(similarity_df)}"
    )

    print(
        similarity_df.head(30)
        .to_string(index=False)
    )

# ============================================================
# 10. THRESHOLD SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("📊 SIMILARITY THRESHOLDS")
print("=" * 100)

for threshold in [
    0.80,
    0.85,
    0.90,
    0.95,
]:

    if similarity_df.empty:
        count = 0
    else:
        count = int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )

    print(
        f"{threshold:.2f}+ : {count} pairs"
    )

# ============================================================
# 11. SAME INTENT
# ============================================================

if similarity_df.empty:

    same_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["new_intent"]
        ==
        similarity_df["old_intent"]
    ]

print("\n" + "=" * 100)
print("🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

print(
    f">= 0.80 : {len(same_intent)}"
)

if not same_intent.empty:

    print(
        same_intent.head(30)
        .to_string(index=False)
    )

# ============================================================
# 12. CROSS INTENT
# ============================================================

if similarity_df.empty:

    cross_intent = pd.DataFrame()

else:

    cross_intent = similarity_df[
        similarity_df["new_intent"]
        !=
        similarity_df["old_intent"]
    ]

print("\n" + "=" * 100)
print("⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

print(
    f">= 0.80 : {len(cross_intent)}"
)

if not cross_intent.empty:

    print(
        cross_intent.head(30)
        .to_string(index=False)
    )

# ============================================================
# 13. HIGH SIMILARITY
# ============================================================

high_similarity = (
    similarity_df[
        similarity_df["similarity"] >= 0.85
    ]
    if not similarity_df.empty
    else pd.DataFrame()
)

print("\n" + "=" * 100)
print("🚨 HIGH SIMILARITY REVIEW")
print("=" * 100)

print(
    f"Similarity >= 0.85: "
    f"{len(high_similarity)}"
)

if high_similarity.empty:

    print("✅ No similarity >= 0.85.")

else:

    print(
        high_similarity.to_string(index=False)
    )

# ============================================================
# 14. FINAL DECISION
# ============================================================

print("\n" + "=" * 100)
print("🏆 BATCH 07 DECISION")
print("=" * 100)

if (
    len(exact_overlap) == 0
    and len(normalized_overlap) == 0
    and exact_duplicates.empty
    and normalized_duplicates.empty
    and high_similarity.empty
    and cross_intent.empty
):

    print("""
🟢 BATCH 07 IS READY FOR THE NEXT STEP

✅ No exact overlap
✅ No normalized overlap
✅ No duplicates inside Batch 07
✅ No similarity >= 0.85
✅ No cross-intent similarity >= 0.80

🎉 SAFE TO FILL GROUND TRUTH + ANSWERS
""")

else:

    print("""
🟡 BATCH 07 STILL NEEDS REVIEW

⚠️ One or more duplicate/similarity checks require attention.
""")

# ============================================================
# 15. READ ONLY
# ============================================================

print("=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ faq_batch_07.csv was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

print("\n" + "=" * 100)
print("🎉 BATCH 07 SIMILARITY CHECK COMPLETE")
print("=" * 100)
````

````python
from pathlib import Path
import pandas as pd
import unicodedata
import re

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_07.csv"
)

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

rewrites = {

    # INT002 - English
    "FAQ0422":
        "How many stars is Nile Pearl Hotel rated?",

    # INT006 - French
    "FAQ0439":
        "L'arrivée anticipée est-elle disponible sur demande au Nile Pearl Hotel ?",

    # INT002 - Arabic
    "FAQ0421":
        "إلى أي فئة من فئات الفنادق يُنسب Nile Pearl Hotel؟",

    # INT005 - English
    "FAQ0434":
        "By what time are guests expected to vacate their room?",

    # INT008 - French
    "FAQ0447":
        "Quel document d'identité faut-il avoir avec soi lors de l'arrivée à l'hôtel ?",
}

for faq_id, new_question in rewrites.items():

    matches = batch_df.index[
        batch_df["faq_id"] == faq_id
    ].tolist()

    assert len(matches) == 1, (
        f"❌ {faq_id} not found exactly once."
    )

    idx = matches[0]

    old_question = batch_df.at[
        idx,
        "question"
    ]

    batch_df.at[
        idx,
        "question"
    ] = new_question

    print(f"\n🔄 {faq_id}")
    print(f"OLD: {old_question}")
    print(f"NEW: {new_question}")


def normalize(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    return re.sub(
        r"\s+",
        " ",
        text
    ).strip()


batch_normalized = batch_df["question"].map(normalize)

assert not batch_df["question"].duplicated().any()
assert not batch_normalized.duplicated().any()

batch_df.to_csv(
    BATCH_FILE,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 100)
print("💾 BATCH 07 UPDATED")
print("=" * 100)
print(BATCH_FILE)
print("✅ 32 rows preserved.")
print("✅ hotel_faq_final.csv was NOT modified.")
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 07 GROUND TRUTH + ANSWER REPAIR
# FILL ONLY BATCH 07
# FINAL DATASET REMAINS UNCHANGED
# ============================================================

from pathlib import Path
import pandas as pd

print("=" * 100)
print("🏨 HOTEL AI - BATCH 07 GROUND TRUTH + ANSWER REPAIR")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_07.csv"
)

assert FINAL_DATASET.exists(), "❌ Final dataset not found."
assert BATCH_FILE.exists(), "❌ Batch 07 not found."

# ============================================================
# 2. LOAD
# ============================================================

df = pd.read_csv(
    FINAL_DATASET,
    encoding="utf-8-sig"
)

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

assert len(df) == 416
assert len(batch_df) == 32

print(f"Existing dataset : {len(df)}")
print(f"Batch 07         : {len(batch_df)}")

# ============================================================
# 3. REQUIRED COLUMNS
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

assert list(df.columns) == REQUIRED_COLUMNS
assert list(batch_df.columns) == REQUIRED_COLUMNS

print("✅ All 13 columns exist.")
print("✅ Column order is correct.")

# ============================================================
# 4. TEXT DTYPES
# ============================================================

TEXT_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "difficulty",
    "answer",
]

for col in TEXT_COLUMNS:
    batch_df[col] = (
        batch_df[col]
        .fillna("")
        .astype("string")
    )

batch_df["attempts"] = pd.to_numeric(
    batch_df["attempts"],
    errors="raise"
)

# ============================================================
# 5. INTENT MAP
# ============================================================

INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

LANGUAGES = [
    "Arabic",
    "English",
    "French",
    "Russian",
]

# ============================================================
# 6. BUILD CANONICAL KNOWLEDGE MAP
# ============================================================

print("\n" + "=" * 100)
print("🧠 BUILDING KNOWLEDGE MAP")
print("=" * 100)

knowledge_map = {}

for intent_id, intent_name in INTENTS.items():

    for language in LANGUAGES:

        subset = df[
            (df["intent_id"] == intent_id)
            &
            (df["language"] == language)
        ].copy()

        assert len(subset) == 13, (
            f"❌ Expected 13 rows for "
            f"{intent_id} × {language}, "
            f"found {len(subset)}"
        )

        # Ground truth candidates
        gt = (
            subset["ground_truth"]
            .fillna("")
            .astype(str)
            .str.strip()
        )

        gt = gt[gt.ne("")]

        assert not gt.empty, (
            f"❌ Ground truth missing for "
            f"{intent_id} × {language}"
        )

        canonical_gt = (
            gt.value_counts()
            .index[0]
        )

        # Answer candidates
        answers = (
            subset["answer"]
            .fillna("")
            .astype(str)
            .str.strip()
        )

        answers = answers[answers.ne("")]

        assert not answers.empty, (
            f"❌ Answer missing for "
            f"{intent_id} × {language}"
        )

        canonical_answer = (
            answers.value_counts()
            .index[0]
        )

        knowledge_map[
            (intent_id, language)
        ] = {
            "ground_truth": canonical_gt,
            "answer": canonical_answer,
        }

        print(
            f"✅ {intent_id:<6} {language:<8}"
        )

print("✅ Knowledge map created.")

# ============================================================
# 7. APPLY GROUND TRUTH + ANSWER
# ============================================================

print("\n" + "=" * 100)
print("✍️ FILLING GROUND TRUTH + ANSWERS")
print("=" * 100)

for idx, row in batch_df.iterrows():

    intent_id = str(
        row["intent_id"]
    ).strip()

    language = str(
        row["language"]
    ).strip()

    key = (
        intent_id,
        language
    )

    assert key in knowledge_map

    batch_df.at[
        idx,
        "ground_truth"
    ] = knowledge_map[key]["ground_truth"]

    batch_df.at[
        idx,
        "answer"
    ] = knowledge_map[key]["answer"]

    batch_df.at[
        idx,
        "status"
    ] = "generated"

    batch_df.at[
        idx,
        "category"
    ] = "Hotel FAQ"

    batch_df.at[
        idx,
        "source"
    ] = "Generated Batch 07"

    batch_df.at[
        idx,
        "attempts"
    ] = 0

    batch_df.at[
        idx,
        "difficulty"
    ] = "Easy"

# Re-cast text columns safely
for col in TEXT_COLUMNS:
    batch_df[col] = (
        batch_df[col]
        .fillna("")
        .astype("string")
    )

print("✅ Ground truth filled for all 32.")
print("✅ Answers filled for all 32.")

# ============================================================
# 8. REQUIRED FIELD VALIDATION
# ============================================================

print("\n" + "=" * 100)
print("🧪 REQUIRED FIELD VALIDATION")
print("=" * 100)

for col in REQUIRED_COLUMNS:

    missing = int(
        batch_df[col].isna().sum()
    )

    empty = int(
        batch_df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    ok = (
        missing == 0
        and empty == 0
    )

    print(
        f"{'✅' if ok else '❌'} "
        f"{col:<15}"
        f"Missing: {missing:<3}"
        f"Empty: {empty:<3}"
    )

    assert ok, (
        f"❌ Invalid values in {col}"
    )

print("✅ All required fields complete.")

# ============================================================
# 9. BASIC BATCH VALIDATION
# ============================================================

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(417, 449)
]

assert batch_df["faq_id"].tolist() == expected_ids
assert batch_df["faq_id"].is_unique

assert batch_df["intent_id"].value_counts().to_dict() == {
    "INT001": 4,
    "INT002": 4,
    "INT003": 4,
    "INT004": 4,
    "INT005": 4,
    "INT006": 4,
    "INT007": 4,
    "INT008": 4,
}

assert batch_df["language"].value_counts().to_dict() == {
    "Arabic": 8,
    "English": 8,
    "French": 8,
    "Russian": 8,
}

# ============================================================
# 10. SAVE BATCH ONLY
# ============================================================

batch_df.to_csv(
    BATCH_FILE,
    index=False,
    encoding="utf-8-sig"
)

# ============================================================
# 11. RELOAD + FINAL CHECK
# ============================================================

check_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

assert len(check_df) == 32

assert (
    check_df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

assert (
    check_df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

print("\n" + "=" * 100)
print("🏆 BATCH 07 REPAIR RESULT")
print("=" * 100)

print("""
🟢 BATCH 07 GROUND TRUTH + ANSWER COMPLETE

✅ 32 questions
✅ FAQ0417 → FAQ0448
✅ Ground truth complete
✅ Answers complete
✅ Required fields complete
✅ 8 Arabic
✅ 8 English
✅ 8 French
✅ 8 Russian
✅ 4 questions × 8 intents
✅ Existing 416 dataset NOT modified

🎉 READY FOR FINAL BATCH 07 AUDIT
""")

print("📄 Batch saved to:")
print(BATCH_FILE)

print("\n" + "=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)
print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ No rows added to final dataset.")
print("✅ No rows deleted from final dataset.")
print("✅ No final-dataset values changed.")

print("\n" + "=" * 100)
print("🎉 BATCH 07 REPAIR COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - BATCH 07 FINAL COMPLETE AUDIT
# Existing dataset : 416
# Batch 07         : 32
# READ-ONLY
# ============================================================

from pathlib import Path
from difflib import SequenceMatcher
import pandas as pd
import re
import unicodedata

print("=" * 100)
print("🏨 HOTEL AI - BATCH 07 FINAL COMPLETE AUDIT")
print("=" * 100)

# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

FINAL_DATASET = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

BATCH_FILE = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "batches"
    / "faq_batch_07.csv"
)

assert PROJECT_ROOT.exists()
assert FINAL_DATASET.exists()
assert BATCH_FILE.exists()

# ============================================================
# 2. LOAD
# ============================================================

df = pd.read_csv(
    FINAL_DATASET,
    encoding="utf-8-sig"
)

batch_df = pd.read_csv(
    BATCH_FILE,
    encoding="utf-8-sig"
)

assert len(df) == 416, (
    f"❌ Expected 416 existing rows, found {len(df)}"
)

assert len(batch_df) == 32, (
    f"❌ Expected 32 batch rows, found {len(batch_df)}"
)

print("\n" + "=" * 100)
print("📊 FILE INFORMATION")
print("=" * 100)

print(f"Existing dataset : {len(df)} rows")
print(f"Batch 07         : {len(batch_df)} rows")

# ============================================================
# 3. REQUIRED COLUMNS
# ============================================================

REQUIRED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

assert list(df.columns) == REQUIRED_COLUMNS
assert list(batch_df.columns) == REQUIRED_COLUMNS

print("✅ All 13 columns exist.")
print("✅ Column order is correct.")

# ============================================================
# 4. FAQ ID VALIDATION
# ============================================================

print("\n" + "=" * 100)
print("🆔 FAQ ID VALIDATION")
print("=" * 100)

expected_ids = [
    f"FAQ{i:04d}"
    for i in range(417, 449)
]

assert batch_df["faq_id"].tolist() == expected_ids
assert batch_df["faq_id"].is_unique

id_overlap = (
    set(df["faq_id"])
    .intersection(set(batch_df["faq_id"]))
)

assert not id_overlap

print("✅ IDs are exactly FAQ0417 → FAQ0448.")
print("✅ No duplicate IDs.")
print("✅ No ID overlap with existing 416.")

# ============================================================
# 5. REQUIRED FIELD CHECK
# ============================================================

print("\n" + "=" * 100)
print("🧪 REQUIRED FIELD VALIDATION")
print("=" * 100)

required_ok = True

for col in REQUIRED_COLUMNS:

    missing = int(
        batch_df[col].isna().sum()
    )

    empty = int(
        batch_df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    ok = (
        missing == 0
        and empty == 0
    )

    print(
        f"{'✅' if ok else '❌'} "
        f"{col:<15}"
        f"Missing: {missing:<3}"
        f"Empty: {empty:<3}"
    )

    if not ok:
        required_ok = False

assert required_ok

print("✅ All required fields complete.")

# ============================================================
# 6. NORMALIZATION
# ============================================================

def normalize(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

# ============================================================
# 7. DUPLICATE CHECK INSIDE BATCH
# ============================================================

print("\n" + "=" * 100)
print("📝 DUPLICATE CHECK INSIDE BATCH 07")
print("=" * 100)

exact_duplicates = batch_df[
    batch_df["question"].duplicated(
        keep=False
    )
]

batch_normalized = (
    batch_df["question"]
    .map(normalize)
)

normalized_duplicates = batch_df[
    batch_normalized.duplicated(
        keep=False
    )
]

assert exact_duplicates.empty
assert normalized_duplicates.empty

print("✅ No exact duplicates.")
print("✅ No normalized duplicates.")

# ============================================================
# 8. OVERLAP WITH EXISTING 416
# ============================================================

print("\n" + "=" * 100)
print("🔎 OVERLAP WITH EXISTING 416")
print("=" * 100)

existing_exact = set(
    df["question"]
    .astype(str)
    .str.strip()
)

batch_exact = set(
    batch_df["question"]
    .astype(str)
    .str.strip()
)

exact_overlap = (
    existing_exact
    .intersection(batch_exact)
)

existing_normalized = set(
    df["question"]
    .map(normalize)
)

normalized_overlap = (
    existing_normalized
    .intersection(set(batch_normalized))
)

assert not exact_overlap
assert not normalized_overlap

print("✅ No exact overlap.")
print("✅ No normalized overlap.")

# ============================================================
# 9. LANGUAGE CHECK
# ============================================================

print("\n" + "=" * 100)
print("🌍 LANGUAGE DISTRIBUTION")
print("=" * 100)

language_counts = (
    batch_df["language"]
    .value_counts()
    .sort_index()
)

print(language_counts.to_string())

assert language_counts.to_dict() == {
    "Arabic": 8,
    "English": 8,
    "French": 8,
    "Russian": 8,
}

print("✅ 8 questions × 4 languages.")

# ============================================================
# 10. INTENT CHECK
# ============================================================

print("\n" + "=" * 100)
print("🎯 INTENT DISTRIBUTION")
print("=" * 100)

intent_counts = (
    batch_df["intent_id"]
    .value_counts()
    .sort_index()
)

print(intent_counts.to_string())

assert intent_counts.to_dict() == {
    "INT001": 4,
    "INT002": 4,
    "INT003": 4,
    "INT004": 4,
    "INT005": 4,
    "INT006": 4,
    "INT007": 4,
    "INT008": 4,
}

print("✅ 4 questions × 8 intents.")

# ============================================================
# 11. INTENT NAME CONSISTENCY
# ============================================================

INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

for _, row in batch_df.iterrows():

    assert (
        row["intent_name"]
        == INTENTS[row["intent_id"]]
    )

print("✅ Intent ID ↔ Intent Name consistency.")

# ============================================================
# 12. QUESTION TYPES
# ============================================================

VALID_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

assert (
    set(batch_df["question_type"])
    <= VALID_TYPES
)

print("\n" + "=" * 100)
print("📝 QUESTION TYPE DISTRIBUTION")
print("=" * 100)

print(
    batch_df["question_type"]
    .value_counts()
    .to_string()
)

print("✅ All question types are valid.")

# ============================================================
# 13. STATUS / DIFFICULTY
# ============================================================

print("\n" + "=" * 100)
print("📌 STATUS / DIFFICULTY")
print("=" * 100)

assert set(batch_df["status"]) <= {
    "valid",
    "generated"
}

assert set(batch_df["difficulty"]) <= {
    "Easy"
}

print("Status:")
print(
    batch_df["status"]
    .value_counts()
    .to_string()
)

print("\nDifficulty:")
print(
    batch_df["difficulty"]
    .value_counts()
    .to_string()
)

print("✅ Status valid.")
print("✅ Difficulty valid.")

# ============================================================
# 14. GROUND TRUTH / ANSWER
# ============================================================

print("\n" + "=" * 100)
print("🎯 GROUND TRUTH / ANSWER")
print("=" * 100)

ground_truth_ok = (
    batch_df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

answers_ok = (
    batch_df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

assert ground_truth_ok
assert answers_ok

print("✅ Ground truth complete.")
print("✅ Answers complete.")

# ============================================================
# 15. ATTEMPTS
# ============================================================

attempts_numeric = pd.to_numeric(
    batch_df["attempts"],
    errors="coerce"
)

assert attempts_numeric.notna().all()
assert (attempts_numeric >= 0).all()

print("\n" + "=" * 100)
print("🔢 ATTEMPTS")
print("=" * 100)

print("✅ Attempts are numeric and non-negative.")
print(f"Min    : {attempts_numeric.min()}")
print(f"Max    : {attempts_numeric.max()}")
print(f"Average: {attempts_numeric.mean():.2f}")

# ============================================================
# 16. QUESTION LENGTH
# ============================================================

question_lengths = (
    batch_df["question"]
    .astype(str)
    .str.len()
)

short_questions = batch_df[
    question_lengths < 10
]

assert short_questions.empty

print("\n" + "=" * 100)
print("📏 QUESTION LENGTH")
print("=" * 100)

print(f"Shortest : {question_lengths.min()}")
print(f"Longest  : {question_lengths.max()}")
print(f"Average  : {question_lengths.mean():.2f}")
print("✅ No suspiciously short questions.")

# ============================================================
# 17. FULL SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🔍 FULL SIMILARITY CHECK")
print("=" * 100)

similar_pairs = []

for _, new_row in batch_df.iterrows():

    for _, old_row in df.iterrows():

        score = SequenceMatcher(
            None,
            normalize(new_row["question"]),
            normalize(old_row["question"])
        ).ratio()

        if score >= 0.80:

            similar_pairs.append({
                "similarity": round(score, 3),

                "new_faq_id": new_row["faq_id"],
                "new_intent": new_row["intent_id"],
                "new_language": new_row["language"],
                "new_question": new_row["question"],

                "old_faq_id": old_row["faq_id"],
                "old_intent": old_row["intent_id"],
                "old_language": old_row["language"],
                "old_question": old_row["question"],
            })

similarity_df = pd.DataFrame(similar_pairs)

if similarity_df.empty:

    print("🎉 No pairs >= 0.80.")

else:

    similarity_df = (
        similarity_df
        .sort_values(
            "similarity",
            ascending=False
        )
        .reset_index(drop=True)
    )

    print(
        f"Pairs >= 0.80: {len(similarity_df)}"
    )

    print(
        similarity_df
        .head(30)
        .to_string(index=False)
    )

# ============================================================
# 18. THRESHOLDS
# ============================================================

print("\n" + "=" * 100)
print("📊 SIMILARITY THRESHOLDS")
print("=" * 100)

for threshold in [
    0.80,
    0.85,
    0.90,
    0.95,
]:

    if similarity_df.empty:
        count = 0
    else:
        count = int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )

    print(
        f"{threshold:.2f}+ : {count} pairs"
    )

# ============================================================
# 19. SAME / CROSS INTENT
# ============================================================

if similarity_df.empty:

    same_intent = pd.DataFrame()
    cross_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["new_intent"]
        ==
        similarity_df["old_intent"]
    ]

    cross_intent = similarity_df[
        similarity_df["new_intent"]
        !=
        similarity_df["old_intent"]
    ]

print("\n" + "=" * 100)
print("🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

print(
    f">= 0.80 : {len(same_intent)}"
)

print("\n" + "=" * 100)
print("⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

print(
    f">= 0.80 : {len(cross_intent)}"
)

# ============================================================
# 20. HIGH SIMILARITY
# ============================================================

high_similarity = (
    similarity_df[
        similarity_df["similarity"] >= 0.85
    ]
    if not similarity_df.empty
    else pd.DataFrame()
)

print("\n" + "=" * 100)
print("🚨 HIGH SIMILARITY")
print("=" * 100)

print(
    f"Similarity >= 0.85 : {len(high_similarity)}"
)

if not high_similarity.empty:

    print(
        high_similarity
        .to_string(index=False)
    )

else:

    print("✅ No similarity >= 0.85.")

# ============================================================
# 21. FINAL DECISION
# ============================================================

print("\n" + "=" * 100)
print("🏆 BATCH 07 FINAL RESULT")
print("=" * 100)

passed = (
    len(exact_overlap) == 0
    and len(normalized_overlap) == 0
    and exact_duplicates.empty
    and normalized_duplicates.empty
    and ground_truth_ok
    and answers_ok
    and high_similarity.empty
    and cross_intent.empty
)

if passed:

    print("""
🟢 BATCH 07 IS FULLY VALID

✅ 32 questions
✅ FAQ0417 → FAQ0448
✅ Required fields complete
✅ Ground truth complete
✅ Answers complete
✅ No duplicate questions
✅ No normalized overlap
✅ No similarity >= 0.85
✅ No cross-intent similarity >= 0.80
✅ Language balanced
✅ Intent balanced
✅ Intent names consistent
✅ Question types valid

🎉 SAFE TO MERGE BATCH 07
""")

else:

    print("""
🟡 BATCH 07 STILL NEEDS REVIEW

⚠️ One or more checks require attention.
""")

# ============================================================
# 22. READ-ONLY
# ============================================================

print("=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ faq_batch_07.csv was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

print("\n" + "=" * 100)
print("🎉 BATCH 07 FINAL AUDIT COMPLETE")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - COMPLETE 448 DATASET AUDIT
# FULL QUALITY + DUPLICATE + SIMILARITY
# READ-ONLY
# ============================================================

from pathlib import Path
from itertools import combinations
from difflib import SequenceMatcher
import pandas as pd
import re
import unicodedata
import numpy as np

print("=" * 100)
print("🏨 HOTEL AI - COMPLETE 448 DATASET AUDIT")
print("=" * 100)

# ============================================================
# 1. CONFIG
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

EXPECTED_ROWS = 448

EXPECTED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

EXPECTED_INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

EXPECTED_LANGUAGES = {
    "Arabic",
    "English",
    "French",
    "Russian",
}

VALID_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

VALID_STATUS = {
    "valid",
    "generated",
}

VALID_DIFFICULTY = {
    "Easy",
}

SIMILARITY_THRESHOLD = 0.85

# ============================================================
# 2. LOAD
# ============================================================

assert PROJECT_ROOT.exists(), "❌ Project root not found."
assert DATASET_PATH.exists(), "❌ Dataset not found."

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print("\n📂 PROJECT")
print("=" * 100)
print(PROJECT_ROOT)

print("\n📄 DATASET")
print("=" * 100)
print(DATASET_PATH)

print("\n📊 DATASET INFORMATION")
print("=" * 100)
print(f"Rows    : {len(df)}")
print(f"Columns : {len(df.columns)}")

checks = {}

# ============================================================
# 3. COLUMNS
# ============================================================

columns_ok = (
    len(df.columns) == 13
    and df.columns.tolist() == EXPECTED_COLUMNS
)

checks["13 columns"] = columns_ok

print("\n🧪 COLUMN VALIDATION")
print("=" * 100)

print(
    "✅ All 13 required columns exist."
    if columns_ok
    else "❌ Column validation failed."
)

if columns_ok:
    print("✅ Column order is correct.")

# ============================================================
# 4. ROW COUNT
# ============================================================

row_count_ok = len(df) == EXPECTED_ROWS

checks["448 questions"] = row_count_ok

print("\n📦 ROW COUNT")
print("=" * 100)
print(f"Total questions: {len(df)}")

print(
    "✅ Exactly 448 questions."
    if row_count_ok
    else "❌ Expected 448 questions."
)

# ============================================================
# 5. REQUIRED FIELDS
# ============================================================

required_ok = True

print("\n🧪 REQUIRED FIELD VALIDATION")
print("=" * 100)

for col in EXPECTED_COLUMNS:

    missing = int(df[col].isna().sum())

    empty = int(
        df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    ok = (
        missing == 0
        and empty == 0
    )

    print(
        f"{'✅' if ok else '❌'} "
        f"{col:<15}"
        f"Missing: {missing:<3}"
        f"Empty: {empty:<3}"
    )

    if not ok:
        required_ok = False

checks["Required fields"] = required_ok

# ============================================================
# 6. FAQ IDs
# ============================================================

print("\n🆔 FAQ ID VALIDATION")
print("=" * 100)

valid_id_format = (
    df["faq_id"]
    .astype(str)
    .str.fullmatch(r"FAQ\d{4}")
    .all()
)

duplicate_ids = df[
    df["faq_id"].duplicated(keep=False)
]

numeric_ids = pd.to_numeric(
    df["faq_id"]
    .astype(str)
    .str.extract(r"FAQ(\d{4})")[0],
    errors="coerce"
)

expected_numeric_ids = np.arange(
    1,
    EXPECTED_ROWS + 1
)

continuous_ids = (
    numeric_ids.notna().all()
    and numeric_ids.astype(int).tolist()
    == expected_numeric_ids.tolist()
)

checks["Valid FAQ IDs"] = bool(valid_id_format)
checks["No duplicate IDs"] = duplicate_ids.empty
checks["Continuous IDs"] = bool(continuous_ids)

print(
    "✅ All FAQ IDs have valid format."
    if valid_id_format
    else "❌ Invalid FAQ ID format."
)

print(
    "✅ No duplicate FAQ IDs."
    if duplicate_ids.empty
    else f"❌ Duplicate IDs found: {len(duplicate_ids)}"
)

print(
    "✅ IDs continuous: FAQ0001 → FAQ0448"
    if continuous_ids
    else "❌ IDs are not continuous."
)

# ============================================================
# 7. NORMALIZATION
# ============================================================

def normalize_text(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


df["_normalized_question"] = (
    df["question"]
    .map(normalize_text)
)

# ============================================================
# 8. EXACT DUPLICATES
# ============================================================

exact_duplicates = df[
    df["question"].duplicated(keep=False)
]

checks["No exact duplicates"] = (
    exact_duplicates.empty
)

print("\n📝 EXACT DUPLICATE CHECK")
print("=" * 100)

if exact_duplicates.empty:
    print("✅ No exact duplicate questions.")
else:
    print(
        exact_duplicates[
            ["faq_id", "intent_id", "language", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 9. NORMALIZED DUPLICATES
# ============================================================

normalized_duplicates = df[
    df["_normalized_question"].duplicated(
        keep=False
    )
]

checks["No normalized duplicates"] = (
    normalized_duplicates.empty
)

print("\n🔎 NORMALIZED DUPLICATE CHECK")
print("=" * 100)

if normalized_duplicates.empty:
    print("✅ No normalized duplicate questions.")
else:
    print(
        normalized_duplicates[
            ["faq_id", "intent_id", "language", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 10. SAME-INTENT DUPLICATES
# ============================================================

same_intent_duplicates = df[
    df.duplicated(
        subset=[
            "intent_id",
            "_normalized_question"
        ],
        keep=False
    )
]

checks["No same-intent duplicates"] = (
    same_intent_duplicates.empty
)

print("\n🔍 DUPLICATES INSIDE SAME INTENT")
print("=" * 100)

if same_intent_duplicates.empty:
    print("✅ No normalized duplicates inside the same intent.")
else:
    print(
        same_intent_duplicates[
            ["faq_id", "intent_id", "language", "question"]
        ].to_string(index=False)
    )

# ============================================================
# 11. INTENT MAPPING
# ============================================================

intent_mapping_ok = True

print("\n🔗 INTENT ID ↔ INTENT NAME")
print("=" * 100)

for intent_id, intent_name in EXPECTED_INTENTS.items():

    subset = df[
        df["intent_id"] == intent_id
    ]

    actual_names = set(
        subset["intent_name"].astype(str)
    )

    ok = (
        len(subset) == 56
        and actual_names == {intent_name}
    )

    if not ok:
        intent_mapping_ok = False

    print(
        f"{'✅' if ok else '❌'} "
        f"{intent_id} → {intent_name} "
        f"| rows={len(subset)}"
    )

checks["Intent mappings"] = intent_mapping_ok

# ============================================================
# 12. LANGUAGE DISTRIBUTION
# ============================================================

language_counts = (
    df["language"]
    .value_counts()
)

language_balance_ok = (
    set(language_counts.index)
    == EXPECTED_LANGUAGES
    and all(
        language_counts.get(
            language,
            0
        ) == 112
        for language in EXPECTED_LANGUAGES
    )
)

checks["Language balance"] = language_balance_ok

print("\n🌍 LANGUAGE DISTRIBUTION")
print("=" * 100)
print(language_counts.to_string())

print(
    "✅ 112 questions × 4 languages."
    if language_balance_ok
    else "❌ Language balance failed."
)

# ============================================================
# 13. INTENT DISTRIBUTION
# ============================================================

intent_counts = (
    df["intent_id"]
    .value_counts()
)

intent_balance_ok = all(
    int(
        intent_counts.get(
            intent_id,
            0
        )
    ) == 56
    for intent_id in EXPECTED_INTENTS
)

checks["Intent balance"] = intent_balance_ok

print("\n🎯 INTENT DISTRIBUTION")
print("=" * 100)

print(
    df.groupby(
        ["intent_id", "intent_name"]
    )
    .size()
    .rename("questions")
    .reset_index()
    .to_string(index=False)
)

print(
    "✅ 56 questions × 8 intents."
    if intent_balance_ok
    else "❌ Intent balance failed."
)

# ============================================================
# 14. LANGUAGE × INTENT
# ============================================================

language_intent = pd.crosstab(
    df["intent_id"],
    df["language"]
)

language_intent_ok = True

for intent_id in EXPECTED_INTENTS:

    for language in EXPECTED_LANGUAGES:

        count = int(
            language_intent.loc[
                intent_id,
                language
            ]
        )

        if count != 14:
            language_intent_ok = False

checks["Language × Intent"] = language_intent_ok

print("\n📊 LANGUAGE × INTENT MATRIX")
print("=" * 100)
print(language_intent.to_string())

print(
    "✅ Every intent has 14 questions per language."
    if language_intent_ok
    else "❌ Language × Intent balance failed."
)

# ============================================================
# 15. QUESTION TYPES
# ============================================================

question_type_counts = (
    df["question_type"]
    .value_counts()
)

unexpected_types = (
    set(df["question_type"])
    - VALID_TYPES
)

question_types_ok = not unexpected_types

checks["Question types"] = question_types_ok

print("\n📝 QUESTION TYPE DISTRIBUTION")
print("=" * 100)
print(question_type_counts.to_string())

print(
    "✅ All question types are valid."
    if question_types_ok
    else f"❌ Invalid question types: {unexpected_types}"
)

# ============================================================
# 16. STATUS
# ============================================================

status_counts = (
    df["status"]
    .value_counts()
)

unexpected_status = (
    set(df["status"])
    - VALID_STATUS
)

status_ok = not unexpected_status

checks["Status"] = status_ok

print("\n📌 STATUS DISTRIBUTION")
print("=" * 100)
print(status_counts.to_string())

print(
    "✅ Status values are valid."
    if status_ok
    else f"❌ Invalid statuses: {unexpected_status}"
)

# ============================================================
# 17. DIFFICULTY
# ============================================================

difficulty_counts = (
    df["difficulty"]
    .value_counts()
)

unexpected_difficulty = (
    set(df["difficulty"])
    - VALID_DIFFICULTY
)

difficulty_ok = not unexpected_difficulty

checks["Difficulty"] = difficulty_ok

print("\n📌 DIFFICULTY DISTRIBUTION")
print("=" * 100)
print(difficulty_counts.to_string())

print(
    "✅ Difficulty values are valid."
    if difficulty_ok
    else f"❌ Invalid difficulty: {unexpected_difficulty}"
)

# ============================================================
# 18. GROUND TRUTH / ANSWERS
# ============================================================

ground_truth_ok = (
    df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

answers_ok = (
    df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

checks["Ground truth"] = bool(ground_truth_ok)
checks["Answers"] = bool(answers_ok)

print("\n🎯 GROUND TRUTH / ANSWER")
print("=" * 100)

print(
    "✅ Ground truth complete."
    if ground_truth_ok
    else "❌ Ground truth incomplete."
)

print(
    "✅ Answers complete."
    if answers_ok
    else "❌ Answers incomplete."
)

# ============================================================
# 19. GROUND TRUTH / ANSWER WORDING
# ============================================================

gt_norm = df["ground_truth"].map(normalize_text)
answer_norm = df["answer"].map(normalize_text)

wording_difference = (
    gt_norm != answer_norm
)

print("\n🔗 GROUND TRUTH ↔ ANSWER")
print("=" * 100)

print(
    "Rows with different wording:",
    int(wording_difference.sum())
)

print(
    "ℹ️ Different wording alone is not treated as a failure."
)

# ============================================================
# 20. ATTEMPTS
# ============================================================

attempts_numeric = pd.to_numeric(
    df["attempts"],
    errors="coerce"
)

attempts_ok = (
    attempts_numeric.notna().all()
    and (attempts_numeric >= 0).all()
)

checks["Attempts"] = attempts_ok

print("\n🔢 ATTEMPTS VALIDATION")
print("=" * 100)

print(
    "✅ Attempts are numeric and non-negative."
    if attempts_ok
    else "❌ Invalid attempts values."
)

print(f"Min    : {attempts_numeric.min()}")
print(f"Max    : {attempts_numeric.max()}")
print(f"Average: {attempts_numeric.mean():.2f}")

# ============================================================
# 21. QUESTION LENGTH
# ============================================================

question_lengths = (
    df["question"]
    .astype(str)
    .str.len()
)

very_short = df[
    question_lengths < 10
]

question_length_ok = very_short.empty

checks["Question length"] = question_length_ok

print("\n📏 QUESTION LENGTH")
print("=" * 100)

print(f"Shortest : {question_lengths.min()}")
print(f"Longest  : {question_lengths.max()}")
print(f"Average  : {question_lengths.mean():.2f}")

print(
    "✅ No suspiciously short questions."
    if question_length_ok
    else "⚠️ Suspiciously short questions found."
)

# ============================================================
# 22. FULL PAIRWISE SIMILARITY
# ============================================================

print("\n🔍 FULL QUESTION SIMILARITY")
print("=" * 100)

questions = (
    df["question"]
    .astype(str)
    .tolist()
)

total_pairs = (
    EXPECTED_ROWS
    * (EXPECTED_ROWS - 1)
    // 2
)

print(
    f"Comparing {total_pairs:,} question pairs..."
)

similarity_pairs = []

for i, j in combinations(
    range(EXPECTED_ROWS),
    2
):

    score = SequenceMatcher(
        None,
        normalize_text(questions[i]),
        normalize_text(questions[j])
    ).ratio()

    if score >= 0.80:

        similarity_pairs.append({
            "similarity": round(score, 3),
            "faq_id_1": df.iloc[i]["faq_id"],
            "intent_1": df.iloc[i]["intent_id"],
            "language_1": df.iloc[i]["language"],
            "question_1": questions[i],
            "faq_id_2": df.iloc[j]["faq_id"],
            "intent_2": df.iloc[j]["intent_id"],
            "language_2": df.iloc[j]["language"],
            "question_2": questions[j],
        })

similarity_df = pd.DataFrame(
    similarity_pairs
)

# ============================================================
# 23. THRESHOLD SUMMARY
# ============================================================

print("\n📊 SIMILARITY THRESHOLD SUMMARY")
print("=" * 100)

for threshold in [
    0.80,
    0.85,
    0.90,
    0.95
]:

    if similarity_df.empty:
        count = 0
    else:
        count = int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )

    print(
        f"{threshold:.2f}+ : {count} pairs"
    )

# ============================================================
# 24. SAME / CROSS INTENT
# ============================================================

if similarity_df.empty:

    same_intent = pd.DataFrame()
    cross_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["intent_1"]
        ==
        similarity_df["intent_2"]
    ]

    cross_intent = similarity_df[
        similarity_df["intent_1"]
        !=
        similarity_df["intent_2"]
    ]

print("\n🎯 SAME-INTENT SIMILARITY")
print("=" * 100)
print(
    f"Pairs >= 0.80: {len(same_intent)}"
)

print("\n⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)
print(
    f"Pairs >= 0.80: {len(cross_intent)}"
)

# ============================================================
# 25. TOP SIMILAR
# ============================================================

print("\n📋 TOP 30 MOST SIMILAR QUESTIONS")
print("=" * 100)

if similarity_df.empty:

    print("None.")

else:

    print(
        similarity_df
        .sort_values(
            "similarity",
            ascending=False
        )
        .head(30)
        .to_string(index=False)
    )

# ============================================================
# 26. HIGH SIMILARITY
# ============================================================

high_similarity = (
    similarity_df[
        similarity_df["similarity"]
        >= SIMILARITY_THRESHOLD
    ]
    if not similarity_df.empty
    else pd.DataFrame()
)

checks["Similarity < 0.85"] = (
    high_similarity.empty
)

print("\n🚨 HIGH-SIMILARITY REVIEW")
print("=" * 100)

if high_similarity.empty:

    print("✅ No similarity >= 0.85.")

else:

    print(
        f"❌ {len(high_similarity)} pairs >= 0.85"
    )

    print(
        high_similarity
        .sort_values(
            "similarity",
            ascending=False
        )
        .to_string(index=False)
    )

# ============================================================
# 27. FINAL SCORE
# ============================================================

print("\n" + "=" * 100)
print("🏆 FINAL VALIDATION SCORE")
print("=" * 100)

for name, result in checks.items():

    print(
        f"{'✅' if result else '❌'} {name}"
    )

passed = sum(checks.values())
total = len(checks)

score = (
    passed / total * 100
)

print("-" * 100)

print(
    f"Passed checks      : {passed}/{total}"
)

print(
    f"Validation Score   : {score:.1f}%"
)

if score == 100:

    print(
        "🟢 DATASET IS FULLY VALIDATED"
    )

else:

    print(
        "🟡 DATASET NEEDS REVIEW"
    )

# ============================================================
# 28. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("📊 FINAL SUMMARY")
print("=" * 100)

print(f"Questions : {len(df)}")

print("\nLanguages:")
print(
    language_counts.to_string()
)

print("\nIntents:")
print(
    intent_counts.to_string()
)

print("\nLanguage × Intent:")
print(
    language_intent.to_string()
)

print("\nQuestion Types:")
print(
    question_type_counts.to_string()
)

print("\nStatus:")
print(
    status_counts.to_string()
)

print("\nDifficulty:")
print(
    difficulty_counts.to_string()
)

print(
    f"\nSimilarity >= 0.85: "
    f"{len(high_similarity)}"
)

# ============================================================
# 29. READ-ONLY
# ============================================================

print("\n" + "=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print("✅ hotel_faq_final.csv was NOT modified.")
print("✅ No rows added.")
print("✅ No rows deleted.")
print("✅ No values changed.")

# ============================================================
# 30. CLEAN MEMORY ONLY
# ============================================================

df.drop(
    columns=["_normalized_question"],
    inplace=True,
    errors="ignore"
)

print("\n" + "=" * 100)
print("🎉 COMPLETE 448-QUESTION AUDIT FINISHED")
print("=" * 100)
````

````python
# ============================================================
# 🏨 HOTEL AI - ULTRA COMPLETE 448 DATASET AUDIT
# FULL STRUCTURAL + DATA QUALITY + DUPLICATE + SIMILARITY
# READ-ONLY - NO DATA MODIFICATION
# ============================================================

from pathlib import Path
from itertools import combinations
from difflib import SequenceMatcher

import pandas as pd
import numpy as np
import re
import unicodedata


# ============================================================
# 1. CONFIGURATION
# ============================================================

PROJECT_ROOT = Path(
    "/Users/sondosahmed/Desktop/HotelAI-Intelligent-Agent"
)

DATASET_PATH = (
    PROJECT_ROOT
    / "03_Structured_Data"
    / "CSV"
    / "final"
    / "hotel_faq_final.csv"
)

EXPECTED_ROWS = 448

EXPECTED_COLUMNS = [
    "faq_id",
    "intent_id",
    "language",
    "question_type",
    "question",
    "status",
    "intent_name",
    "ground_truth",
    "category",
    "source",
    "attempts",
    "difficulty",
    "answer",
]

EXPECTED_INTENTS = {
    "INT001": "hotel_location",
    "INT002": "hotel_category",
    "INT003": "reception_availability",
    "INT004": "check_in_time",
    "INT005": "check_out_time",
    "INT006": "early_check_in",
    "INT007": "late_check_out",
    "INT008": "identification_required",
}

EXPECTED_LANGUAGES = {
    "Arabic",
    "English",
    "French",
    "Russian",
}

VALID_QUESTION_TYPES = {
    "Direct",
    "Paraphrased",
    "Scenario",
}

VALID_STATUS = {
    "valid",
    "generated",
}

VALID_DIFFICULTY = {
    "Easy",
}

# Similarity review thresholds
SIMILARITY_REVIEW = 0.80
SIMILARITY_HIGH = 0.85
SIMILARITY_VERY_HIGH = 0.90
SIMILARITY_EXTREME = 0.95


# ============================================================
# 2. HEADER
# ============================================================

print("=" * 100)
print("🏨 HOTEL AI - ULTRA COMPLETE 448 DATASET AUDIT")
print("=" * 100)


# ============================================================
# 3. LOAD DATASET
# ============================================================

assert PROJECT_ROOT.exists(), (
    f"❌ Project root not found:\n{PROJECT_ROOT}"
)

assert DATASET_PATH.exists(), (
    f"❌ Dataset not found:\n{DATASET_PATH}"
)

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print("\n" + "=" * 100)
print("📂 PROJECT / DATASET")
print("=" * 100)

print(f"Project : {PROJECT_ROOT}")
print(f"Dataset : {DATASET_PATH}")
print(f"Rows    : {len(df)}")
print(f"Columns : {len(df.columns)}")


# ============================================================
# 4. CHECK REGISTRY
# ============================================================

checks = {}


# ============================================================
# 5. COLUMN CHECK
# ============================================================

columns_ok = (
    len(df.columns) == 13
    and df.columns.tolist() == EXPECTED_COLUMNS
)

checks["13 columns"] = columns_ok

print("\n" + "=" * 100)
print("🧪 1. COLUMN VALIDATION")
print("=" * 100)

if columns_ok:
    print("✅ All 13 required columns exist.")
    print("✅ Column order is correct.")
else:
    print("❌ Column validation failed.")


# ============================================================
# 6. ROW COUNT
# ============================================================

rows_ok = len(df) == EXPECTED_ROWS

checks["448 questions"] = rows_ok

print("\n" + "=" * 100)
print("📦 2. ROW COUNT")
print("=" * 100)

print(f"Total questions: {len(df)}")

if rows_ok:
    print("✅ Exactly 448 questions.")
else:
    print("❌ Dataset does not contain exactly 448 questions.")


# ============================================================
# 7. REQUIRED FIELD VALIDATION
# ============================================================

required_fields_ok = True

print("\n" + "=" * 100)
print("🧪 3. REQUIRED FIELD VALIDATION")
print("=" * 100)

for col in EXPECTED_COLUMNS:

    missing = int(
        df[col].isna().sum()
    )

    empty = int(
        df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    ok = (
        missing == 0
        and empty == 0
    )

    print(
        f"{'✅' if ok else '❌'} "
        f"{col:<15}"
        f"Missing: {missing:<4}"
        f"Empty: {empty:<4}"
    )

    if not ok:
        required_fields_ok = False

checks["Required fields complete"] = required_fields_ok


# ============================================================
# 8. NORMALIZATION FUNCTION
# ============================================================

def normalize_text(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


# ============================================================
# 9. FAQ ID FORMAT
# ============================================================

print("\n" + "=" * 100)
print("🆔 4. FAQ ID VALIDATION")
print("=" * 100)

valid_id_format = (
    df["faq_id"]
    .astype(str)
    .str.fullmatch(r"FAQ\d{4}")
    .all()
)

duplicate_id_rows = df[
    df["faq_id"].duplicated(
        keep=False
    )
]

numeric_ids = pd.to_numeric(
    df["faq_id"]
    .astype(str)
    .str.extract(r"FAQ(\d{4})")[0],
    errors="coerce"
)

expected_ids = np.arange(
    1,
    EXPECTED_ROWS + 1
)

continuous_ids = (
    numeric_ids.notna().all()
    and
    numeric_ids.astype(int).tolist()
    == expected_ids.tolist()
)

checks["Valid FAQ ID format"] = bool(
    valid_id_format
)

checks["No duplicate FAQ IDs"] = (
    duplicate_id_rows.empty
)

checks["Continuous FAQ IDs"] = bool(
    continuous_ids
)

print(
    "✅ All FAQ IDs have valid format."
    if valid_id_format
    else "❌ Invalid FAQ IDs found."
)

print(
    "✅ No duplicate FAQ IDs."
    if duplicate_id_rows.empty
    else "❌ Duplicate FAQ IDs found."
)

print(
    "✅ IDs continuous: FAQ0001 → FAQ0448"
    if continuous_ids
    else "❌ IDs are not continuous."
)


# ============================================================
# 10. PREPARE NORMALIZED QUESTIONS
# ============================================================

df["_normalized_question"] = (
    df["question"]
    .map(normalize_text)
)


# ============================================================
# 11. EXACT DUPLICATES
# ============================================================

print("\n" + "=" * 100)
print("📝 5. EXACT QUESTION DUPLICATE CHECK")
print("=" * 100)

exact_duplicates = df[
    df["question"].duplicated(
        keep=False
    )
]

checks["No exact question duplicates"] = (
    exact_duplicates.empty
)

if exact_duplicates.empty:

    print("✅ No exact duplicate questions.")

else:

    print(
        exact_duplicates[
            [
                "faq_id",
                "intent_id",
                "language",
                "question",
            ]
        ].to_string(index=False)
    )


# ============================================================
# 12. NORMALIZED DUPLICATES
# ============================================================

print("\n" + "=" * 100)
print("🔎 6. NORMALIZED QUESTION DUPLICATE CHECK")
print("=" * 100)

normalized_duplicates = df[
    df["_normalized_question"].duplicated(
        keep=False
    )
]

checks["No normalized duplicates"] = (
    normalized_duplicates.empty
)

if normalized_duplicates.empty:

    print("✅ No normalized duplicate questions.")

else:

    print(
        normalized_duplicates[
            [
                "faq_id",
                "intent_id",
                "language",
                "question",
            ]
        ].to_string(index=False)
    )


# ============================================================
# 13. SAME-INTENT NORMALIZED DUPLICATES
# ============================================================

print("\n" + "=" * 100)
print("🔍 7. DUPLICATES INSIDE SAME INTENT")
print("=" * 100)

same_intent_duplicates = df[
    df.duplicated(
        subset=[
            "intent_id",
            "_normalized_question",
        ],
        keep=False
    )
]

checks["No same-intent normalized duplicates"] = (
    same_intent_duplicates.empty
)

if same_intent_duplicates.empty:

    print(
        "✅ No normalized duplicates inside the same intent."
    )

else:

    print(
        same_intent_duplicates[
            [
                "faq_id",
                "intent_id",
                "language",
                "question",
            ]
        ].to_string(index=False)
    )


# ============================================================
# 14. INTENT ID ↔ NAME
# ============================================================

print("\n" + "=" * 100)
print("🔗 8. INTENT ID ↔ INTENT NAME")
print("=" * 100)

intent_mapping_ok = True

for intent_id, expected_name in EXPECTED_INTENTS.items():

    subset = df[
        df["intent_id"] == intent_id
    ]

    actual_names = set(
        subset["intent_name"]
        .astype(str)
    )

    ok = (
        len(subset) == 56
        and actual_names == {
            expected_name
        }
    )

    if not ok:
        intent_mapping_ok = False

    print(
        f"{'✅' if ok else '❌'} "
        f"{intent_id} → {expected_name}"
        f" | rows={len(subset)}"
    )

checks["Intent ID ↔ Name mappings"] = (
    intent_mapping_ok
)


# ============================================================
# 15. LANGUAGE DISTRIBUTION
# ============================================================

print("\n" + "=" * 100)
print("🌍 9. LANGUAGE DISTRIBUTION")
print("=" * 100)

language_counts = (
    df["language"]
    .value_counts()
)

language_balance_ok = (
    set(language_counts.index)
    == EXPECTED_LANGUAGES
    and
    all(
        int(
            language_counts.get(
                language,
                0
            )
        ) == 112
        for language in EXPECTED_LANGUAGES
    )
)

checks["Language balance"] = (
    language_balance_ok
)

print(
    language_counts.to_string()
)

print(
    "✅ 112 × 4 languages."
    if language_balance_ok
    else "❌ Language balance failed."
)


# ============================================================
# 16. INTENT DISTRIBUTION
# ============================================================

print("\n" + "=" * 100)
print("🎯 10. INTENT DISTRIBUTION")
print("=" * 100)

intent_counts = (
    df["intent_id"]
    .value_counts()
)

intent_balance_ok = all(
    int(
        intent_counts.get(
            intent_id,
            0
        )
    ) == 56
    for intent_id in EXPECTED_INTENTS
)

checks["Intent balance"] = (
    intent_balance_ok
)

print(
    df.groupby(
        [
            "intent_id",
            "intent_name",
        ]
    )
    .size()
    .rename("questions")
    .reset_index()
    .to_string(index=False)
)

print(
    "✅ 56 × 8 intents."
    if intent_balance_ok
    else "❌ Intent balance failed."
)


# ============================================================
# 17. LANGUAGE × INTENT
# ============================================================

print("\n" + "=" * 100)
print("📊 11. LANGUAGE × INTENT MATRIX")
print("=" * 100)

language_intent = pd.crosstab(
    df["intent_id"],
    df["language"]
)

language_intent_ok = True

for intent_id in EXPECTED_INTENTS:

    for language in EXPECTED_LANGUAGES:

        value = int(
            language_intent.loc[
                intent_id,
                language
            ]
        )

        if value != 14:
            language_intent_ok = False

print(
    language_intent.to_string()
)

print(
    "✅ Every intent has 14 questions per language."
    if language_intent_ok
    else "❌ Language × Intent is not balanced."
)

checks["Language × Intent balance"] = (
    language_intent_ok
)


# ============================================================
# 18. QUESTION TYPES
# ============================================================

print("\n" + "=" * 100)
print("📝 12. QUESTION TYPE VALIDATION")
print("=" * 100)

question_type_counts = (
    df["question_type"]
    .value_counts()
)

invalid_question_types = (
    set(df["question_type"])
    - VALID_QUESTION_TYPES
)

question_types_ok = (
    not invalid_question_types
)

checks["Question types valid"] = (
    question_types_ok
)

print(
    question_type_counts.to_string()
)

print(
    "✅ All question types are valid."
    if question_types_ok
    else
    f"❌ Invalid question types: {invalid_question_types}"
)


# ============================================================
# 19. STATUS
# ============================================================

print("\n" + "=" * 100)
print("📌 13. STATUS VALIDATION")
print("=" * 100)

status_counts = (
    df["status"]
    .value_counts()
)

invalid_statuses = (
    set(df["status"])
    - VALID_STATUS
)

status_ok = (
    not invalid_statuses
)

checks["Status valid"] = (
    status_ok
)

print(
    status_counts.to_string()
)

print(
    "✅ Status values are valid."
    if status_ok
    else
    f"❌ Invalid status values: {invalid_statuses}"
)


# ============================================================
# 20. DIFFICULTY
# ============================================================

print("\n" + "=" * 100)
print("📌 14. DIFFICULTY VALIDATION")
print("=" * 100)

difficulty_counts = (
    df["difficulty"]
    .value_counts()
)

invalid_difficulties = (
    set(df["difficulty"])
    - VALID_DIFFICULTY
)

difficulty_ok = (
    not invalid_difficulties
)

checks["Difficulty valid"] = (
    difficulty_ok
)

print(
    difficulty_counts.to_string()
)

print(
    "✅ Difficulty values are valid."
    if difficulty_ok
    else
    f"❌ Invalid difficulty values: {invalid_difficulties}"
)


# ============================================================
# 21. GROUND TRUTH
# ============================================================

print("\n" + "=" * 100)
print("🎯 15. GROUND TRUTH / ANSWER COMPLETENESS")
print("=" * 100)

ground_truth_ok = (
    df["ground_truth"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

answer_ok = (
    df["answer"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

checks["Ground truth complete"] = (
    bool(ground_truth_ok)
)

checks["Answers complete"] = (
    bool(answer_ok)
)

print(
    "✅ Ground truth complete."
    if ground_truth_ok
    else "❌ Ground truth incomplete."
)

print(
    "✅ Answers complete."
    if answer_ok
    else "❌ Answers incomplete."
)


# ============================================================
# 22. GROUND TRUTH / ANSWER CONSISTENCY
# ============================================================

print("\n" + "=" * 100)
print("🔗 16. GROUND TRUTH ↔ ANSWER")
print("=" * 100)

gt_normalized = (
    df["ground_truth"]
    .map(normalize_text)
)

answer_normalized = (
    df["answer"]
    .map(normalize_text)
)

different_wording = (
    gt_normalized
    !=
    answer_normalized
)

same_exact_content = (
    gt_normalized
    ==
    answer_normalized
)

print(
    f"Rows with different normalized wording: "
    f"{int(different_wording.sum())}"
)

print(
    f"Rows with matching normalized wording : "
    f"{int(same_exact_content.sum())}"
)

print(
    "ℹ️ Different wording is informational only."
)


# ============================================================
# 23. ATTEMPTS
# ============================================================

print("\n" + "=" * 100)
print("🔢 17. ATTEMPTS VALIDATION")
print("=" * 100)

attempts_numeric = pd.to_numeric(
    df["attempts"],
    errors="coerce"
)

attempts_ok = (
    attempts_numeric.notna().all()
    and
    (attempts_numeric >= 0).all()
)

checks["Attempts valid"] = (
    bool(attempts_ok)
)

print(
    "✅ Attempts are numeric and non-negative."
    if attempts_ok
    else "❌ Invalid Attempts values."
)

print(
    f"Min    : {attempts_numeric.min()}"
)

print(
    f"Max    : {attempts_numeric.max()}"
)

print(
    f"Average: {attempts_numeric.mean():.2f}"
)


# ============================================================
# 24. QUESTION LENGTH
# ============================================================

print("\n" + "=" * 100)
print("📏 18. QUESTION LENGTH VALIDATION")
print("=" * 100)

question_lengths = (
    df["question"]
    .astype(str)
    .str.len()
)

very_short_questions = df[
    question_lengths < 10
]

question_length_ok = (
    very_short_questions.empty
)

checks["Question length valid"] = (
    question_length_ok
)

print(
    f"Shortest : {question_lengths.min()}"
)

print(
    f"Longest  : {question_lengths.max()}"
)

print(
    f"Average  : {question_lengths.mean():.2f}"
)

print(
    "✅ No suspiciously short questions."
    if question_length_ok
    else "⚠️ Suspiciously short questions found."
)


# ============================================================
# 25. WHITESPACE CHECK
# ============================================================

print("\n" + "=" * 100)
print("🧹 19. WHITESPACE / TEXT CLEANLINESS")
print("=" * 100)

leading_trailing = (
    df["question"]
    .astype(str)
    .map(
        lambda x: x != x.strip()
    )
)

double_spaces = (
    df["question"]
    .astype(str)
    .str.contains(
        r"\s{2,}",
        regex=True,
        na=False
    )
)

whitespace_ok = not (
    leading_trailing.any()
    or
    double_spaces.any()
)

checks["Whitespace cleanliness"] = (
    whitespace_ok
)

print(
    "✅ No obvious leading/trailing whitespace."
    if not leading_trailing.any()
    else
    f"⚠️ Leading/trailing whitespace rows: "
    f"{int(leading_trailing.sum())}"
)

print(
    "✅ No obvious repeated spaces."
    if not double_spaces.any()
    else
    f"⚠️ Repeated-space rows: "
    f"{int(double_spaces.sum())}"
)


# ============================================================
# 26. FULL PAIRWISE SIMILARITY
# ============================================================

print("\n" + "=" * 100)
print("🔍 20. FULL QUESTION SIMILARITY")
print("=" * 100)

questions = (
    df["question"]
    .astype(str)
    .tolist()
)

total_pairs = (
    EXPECTED_ROWS
    * (EXPECTED_ROWS - 1)
    // 2
)

print(
    f"Comparing {total_pairs:,} question pairs..."
)

similarity_pairs = []

for i, j in combinations(
    range(EXPECTED_ROWS),
    2
):

    score = SequenceMatcher(
        None,
        normalize_text(questions[i]),
        normalize_text(questions[j])
    ).ratio()

    if score >= SIMILARITY_REVIEW:

        similarity_pairs.append({
            "similarity": round(score, 3),

            "faq_id_1": df.iloc[i]["faq_id"],
            "intent_1": df.iloc[i]["intent_id"],
            "language_1": df.iloc[i]["language"],
            "question_1": questions[i],

            "faq_id_2": df.iloc[j]["faq_id"],
            "intent_2": df.iloc[j]["intent_id"],
            "language_2": df.iloc[j]["language"],
            "question_2": questions[j],
        })

similarity_df = pd.DataFrame(
    similarity_pairs
)


# ============================================================
# 27. THRESHOLDS
# ============================================================

print("\n" + "=" * 100)
print("📊 21. SIMILARITY THRESHOLD SUMMARY")
print("=" * 100)

for threshold in [
    SIMILARITY_REVIEW,
    SIMILARITY_HIGH,
    SIMILARITY_VERY_HIGH,
    SIMILARITY_EXTREME,
]:

    if similarity_df.empty:

        count = 0

    else:

        count = int(
            (
                similarity_df["similarity"]
                >= threshold
            ).sum()
        )

    print(
        f"{threshold:.2f}+ : {count} pairs"
    )


# ============================================================
# 28. SAME INTENT / CROSS INTENT
# ============================================================

if similarity_df.empty:

    same_intent = pd.DataFrame()
    cross_intent = pd.DataFrame()

else:

    same_intent = similarity_df[
        similarity_df["intent_1"]
        ==
        similarity_df["intent_2"]
    ]

    cross_intent = similarity_df[
        similarity_df["intent_1"]
        !=
        similarity_df["intent_2"]
    ]

print("\n" + "=" * 100)
print("🎯 SAME-INTENT SIMILARITY")
print("=" * 100)

print(
    f"Pairs >= 0.80: {len(same_intent)}"
)

print("\n" + "=" * 100)
print("⚠️ CROSS-INTENT SIMILARITY")
print("=" * 100)

print(
    f"Pairs >= 0.80: {len(cross_intent)}"
)


# ============================================================
# 29. TOP SIMILAR QUESTIONS
# ============================================================

print("\n" + "=" * 100)
print("📋 TOP 30 MOST SIMILAR QUESTIONS")
print("=" * 100)

if similarity_df.empty:

    print("None.")

else:

    print(
        similarity_df
        .sort_values(
            "similarity",
            ascending=False
        )
        .head(30)
        .to_string(index=False)
    )


# ============================================================
# 30. HIGH SIMILARITY CHECK
# ============================================================

high_similarity = (
    similarity_df[
        similarity_df["similarity"]
        >= SIMILARITY_HIGH
    ]
    if not similarity_df.empty
    else pd.DataFrame()
)

checks["Similarity < 0.85"] = (
    high_similarity.empty
)

print("\n" + "=" * 100)
print("🚨 HIGH-SIMILARITY REVIEW")
print("=" * 100)

if high_similarity.empty:

    print(
        "✅ No similarity >= 0.85."
    )

else:

    print(
        f"❌ {len(high_similarity)} pairs "
        f"have similarity >= 0.85."
    )

    print(
        high_similarity
        .sort_values(
            "similarity",
            ascending=False
        )
        .to_string(index=False)
    )


# ============================================================
# 31. CROSS-INTENT HIGH SIMILARITY
# ============================================================

cross_intent_high = (
    cross_intent[
        cross_intent["similarity"]
        >= SIMILARITY_REVIEW
    ]
    if not cross_intent.empty
    else pd.DataFrame()
)

cross_intent_ok = (
    cross_intent_high.empty
)

checks["No cross-intent >= 0.80"] = (
    cross_intent_ok
)

print("\n" + "=" * 100)
print("⚠️ CROSS-INTENT REVIEW")
print("=" * 100)

if cross_intent_high.empty:

    print(
        "✅ No cross-intent similarity >= 0.80."
    )

else:

    print(
        f"❌ {len(cross_intent_high)} "
        f"cross-intent pairs >= 0.80."
    )

    print(
        cross_intent_high
        .sort_values(
            "similarity",
            ascending=False
        )
        .to_string(index=False)
    )


# ============================================================
# 32. FINAL SCORE
# ============================================================

print("\n" + "=" * 100)
print("🏆 FINAL VALIDATION SCORE")
print("=" * 100)

for name, result in checks.items():

    print(
        f"{'✅' if result else '❌'} {name}"
    )

passed_checks = sum(
    bool(value)
    for value in checks.values()
)

total_checks = len(checks)

validation_score = (
    passed_checks
    / total_checks
    * 100
)

print("-" * 100)

print(
    f"Passed checks      : "
    f"{passed_checks}/{total_checks}"
)

print(
    f"Validation Score   : "
    f"{validation_score:.1f}%"
)

if validation_score == 100:

    print(
        "🟢 DATASET IS FULLY VALIDATED"
    )

else:

    print(
        "🟡 DATASET NEEDS REVIEW"
    )


# ============================================================
# 33. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("📊 FINAL DATASET SUMMARY")
print("=" * 100)

print(
    f"Questions : {len(df)}"
)

print("\nLanguages:")
print(
    language_counts.to_string()
)

print("\nIntents:")
print(
    intent_counts.to_string()
)

print("\nLanguage × Intent:")
print(
    language_intent.to_string()
)

print("\nQuestion Types:")
print(
    question_type_counts.to_string()
)

print("\nStatus:")
print(
    status_counts.to_string()
)

print("\nDifficulty:")
print(
    difficulty_counts.to_string()
)

print(
    f"\nSimilarity >= 0.85 : "
    f"{len(high_similarity)}"
)

print(
    f"Cross-intent >= 0.80 : "
    f"{len(cross_intent_high)}"
)


# ============================================================
# 34. READ-ONLY CONFIRMATION
# ============================================================

print("\n" + "=" * 100)
print("🔒 READ-ONLY CONFIRMATION")
print("=" * 100)

print(
    "✅ hotel_faq_final.csv was NOT modified."
)

print(
    "✅ No rows were added."
)

print(
    "✅ No rows were deleted."
)

print(
    "✅ No values were changed."
)


# ============================================================
# 35. CLEAN TEMP COLUMN
# ============================================================

df.drop(
    columns=[
        "_normalized_question"
    ],
    inplace=True,
    errors="ignore"
)


print("\n" + "=" * 100)
print("🎉 COMPLETE 448-QUESTION ULTRA AUDIT FINISHED")
print("=" * 100)
````
